# Question 4: Can you explain your MindSense AI project?

## Answer:

“MindSense AI is a machine learning-based mental health risk prediction web application.

The main purpose of the application is to predict the risk level based on the user's assessment data.

I worked on the complete machine learning workflow. First, the data is preprocessed and the required features are prepared. Then the machine learning model is trained and evaluated. I used Logistic Regression for the risk classification.

I also integrated SHAP, which is used for Explainable AI. It helps us understand which features contributed to the model's prediction.

Apart from the machine learning part, the application also provides risk assessment, analytics, recommendations, assessment history and PDF reports.

The application was developed using Python and Flask, with SQLite for storing assessment history.

Overall, MindSense AI helped me understand how to connect a machine learning model with a web application and build an end-to-end AI application.

## 🧠 Memory Trick:

Remember the flow:

**Problem → Data → Model → Explain → Application → Store → Report**

- **Problem:** Mental health risk prediction
- **Data:** Preprocessing + feature engineering
- **Model:** Logistic Regression
- **Explain:** SHAP
- **Application:** Flask + risk assessment + analytics + recommendations
- **Store:** SQLite + assessment history
- **Report:** PDF generation

### Short version to remember:

**“MindSense AI = Risk Prediction + Logistic Regression + SHAP + Flask + SQLite + Reports.”**

# Question 5: Why did you choose Logistic Regression for MindSense AI?

## Answer:

“I chose Logistic Regression because my project is a classification problem. The model needs to classify the user's risk into categories such as Low Risk or High Risk.

Logistic Regression is a simple and effective model for binary classification. It is also easy to understand and interpret compared with many more complex models.

Another reason was that I wanted to use Explainable AI with SHAP. Logistic Regression works well for this because we can understand how different features contribute to the prediction.

After training the model, I evaluated its performance using classification metrics and selected it for the application based on its performance and suitability for the problem.”

## 🧠 Memory Trick:

Remember:

**Classification → Simple → Interpretable → SHAP → Performance**

- **Classification:** Low Risk / High Risk
- **Simple:** Easy to train and understand
- **Interpretable:** Easier to explain predictions
- **SHAP:** Helps explain feature contributions
- **Performance:** Evaluated using classification metrics

### One-line memory:

**“I chose Logistic Regression because it was suitable for my binary classification problem, simple to understand, and worked well with Explainable AI.”**
