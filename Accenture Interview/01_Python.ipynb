Q2. What is the difference between a compiled language and an interpreted language?

A compiled language generally translates source code into a lower-level or machine-executable form before execution, while an interpreted approach executes code through an interpreter at runtime. Python is commonly described as an interpreted language, while Java source code is compiled into bytecode and executed by the JVM.

Q3. What is the difference between a list, tuple, set, and dictionary in Python?

List, tuple, set and dictionary are built-in collection data types in Python. A list is ordered and mutable, while a tuple is ordered and immutable. A set is mutable and stores unique elements without relying on positional indexing. A dictionary is mutable and stores data as key-value pairs.

Q4. Why would you use a tuple instead of a list in Python?

I would use a tuple instead of a list when I want the data to remain unchanged after creation. A tuple is immutable, whereas a list is mutable. If I want to store only unique values, I would use a set instead.

Q6. What is the difference between == and is in Python?

== checks whether two objects have equal values, whereas is checks whether two variables refer to the same object in memory.
