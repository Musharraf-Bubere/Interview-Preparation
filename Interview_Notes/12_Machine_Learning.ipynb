# 🟢 PART 1 — MACHINE LEARNING FUNDAMENTALS

# Q1. What is Machine Learning?

## Answer

Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every individual rule.

Traditional programming:

    Rules + Data
        ↓
      Program
        ↓
      Output

Machine Learning:

    Data + Expected Outcomes
        ↓
       Model
        ↓
    Learned Patterns
        ↓
      Prediction

### Example

Suppose we want to predict house prices.

Input features:

    Area
    Number of bedrooms
    Location
    Age of house

Target:

    House Price

The ML model learns the relationship between these features and prices.

### Interview Answer

Machine Learning is a field of AI where algorithms learn patterns from data and use those learned patterns to make predictions or decisions on unseen data.

# Q9. Why do we split data into train and test sets?

## Answer

The primary reason is to estimate how well the model generalizes to unseen data.

If we evaluate the model only on training data, the model may perform extremely well simply because it has learned the training examples.

### Problem

    Training Data
         ↓
       Model
         ↓
    Training Accuracy = 99%

This does not necessarily mean:

    Real-world performance = 99%

Therefore:

    Training Data
         ↓
       Model
         ↓
    Unseen Test Data
         ↓
    Generalization Performance

### Interview Answer

We split data so that we can evaluate the model on unseen examples and estimate its ability to generalize beyond the training data.

# Q10. What is overfitting?

## Answer

Overfitting occurs when a model learns the training data too closely, including noise and accidental patterns, and therefore performs poorly on unseen data.

Typical pattern:

    Training Performance → Very High
    Validation/Test Performance → Much Lower

Example:

    Training Accuracy = 99%
    Test Accuracy = 72%

This is a strong sign of overfitting.

### Causes

- Model too complex
- Too many features
- Small dataset
- Noise
- Excessive training
- Weak regularization

### Solutions

- More training data
- Regularization
- Feature selection
- Simpler model
- Cross-validation
- Early stopping
- Data augmentation where appropriate
- Pruning for tree-based models

### Interview Answer

Overfitting occurs when a model learns training-specific patterns or noise instead of generalizable patterns, resulting in strong training performance but poor performance on unseen data.


---

# Q11. What is underfitting?

## Answer

Underfitting occurs when the model is too simple to capture the underlying relationship in the data.

Typical pattern:

    Training Performance → Poor
    Test Performance → Poor

### Causes

- Model too simple
- Insufficient features
- Excessive regularization
- Poor feature engineering
- Insufficient training

### Solutions

- Use a more expressive model
- Add useful features
- Reduce excessive regularization
- Train longer where appropriate
- Improve feature engineering

### Interview Answer

Underfitting occurs when a model is unable to learn the important patterns in the data, resulting in poor performance on both training and unseen data.

# Q12. Explain the bias-variance tradeoff.

## Answer

Bias and variance represent two major sources of model error.

### Bias

Bias is error caused by overly simplistic assumptions.

High bias often leads to:

    Underfitting

### Variance

Variance refers to how sensitive the model is to changes in the training data.

High variance often leads to:

    Overfitting

### Relationship

    High Bias
        ↓
    Underfitting

    High Variance
        ↓
    Overfitting

We want a balance.

    Model Complexity
          ↑
    Bias generally ↓
    Variance generally ↑

### Goal

Find a model with sufficiently low bias and variance that generalizes well.

### Interview Answer

The bias-variance tradeoff is the balance between underfitting caused by high bias and overfitting caused by high variance. Good models aim to minimize total generalization error rather than minimizing either component independently.

### Interview Answer

# Q14. What is data leakage?

## Answer

Data leakage occurs when information that would not legitimately be available at prediction time enters the training process.

This can make validation or test performance look artificially good.

### Example

Suppose we predict whether a customer will default.

Feature:

    "Loan recovery status"

If this information becomes available only after default, using it during training creates leakage.

The model is effectively receiving future information.

### Another Example

Performing preprocessing on the entire dataset before splitting can sometimes leak information from validation/test data into training.

### Prevention

- Split data appropriately
- Fit preprocessing only on training data
- Use Pipelines
- Avoid future information
- Carefully define feature availability time

### Interview Answer

Data leakage occurs when information unavailable at real prediction time influences model training or evaluation, causing overly optimistic performance estimates.

A baseline is a simple reference solution used to determine whether a more sophisticated model provides meaningful improvement.

# Q18. What are MCAR, MAR, and MNAR?

## Answer

These describe mechanisms behind missing data.

### MCAR

    Missing Completely At Random

Missingness is unrelated to observed or unobserved data.

Example:

    Random technical failure causes some values to disappear.

---

### MAR

    Missing At Random

Missingness depends on observed variables.

Example:

    Income is missing more frequently for younger customers,
    and age is available.

---

### MNAR

    Missing Not At Random

Missingness depends on the missing value itself or unobserved information.

Example:

People with very high income may be less willing to report their income.

### Interview Answer

MCAR means missingness is unrelated to the data, MAR means missingness depends on observed variables, and MNAR means missingness depends on the missing value itself or unobserved factors.

# Q19. What is feature scaling?

## Answer

Feature scaling transforms numerical features to comparable scales.

Example:

    Age        → 20–60
    Salary     → 20,000–2,00,000
    Experience → 0–20

Salary has much larger numerical magnitude.

Some algorithms can be affected by these scale differences.

### Important for

- KNN
- K-Means
- SVM
- Logistic Regression
- Neural Networks
- PCA
- Gradient-based algorithms

Tree-based algorithms generally do not require feature scaling.

### Interview Answer

Feature scaling transforms numerical variables to comparable ranges so that algorithms sensitive to feature magnitude are not disproportionately influenced by larger-scale features.

# Q20. What is normalization?

## Answer

Normalization commonly refers to scaling values to a bounded range, often:

    [0, 1]

A common min-max formula is:

    x' = (x - min(x)) / (max(x) - min(x))

Example:

    Original:
    10, 20, 30

    Normalized:
    0, 0.5, 1

### Useful for

- Neural networks
- Distance-based algorithms
- Some optimization problems

### Important

Normalization is different from standardization.

### Interview Answer

Normalization commonly rescales values to a fixed range such as [0,1], often using min-max scaling.

# Q21. What is standardization?

## Answer

Standardization transforms a feature so that it has approximately:

    Mean = 0
    Standard Deviation = 1

Formula:

    z = (x - μ) / σ

where:

    μ = mean
    σ = standard deviation

### Example

If:

    x = 70
    mean = 50
    std = 10

then:

    z = (70 - 50) / 10
      = 2

### Important

Standardization does not necessarily constrain values between 0 and 1.

### Interview Answer

Standardization transforms values using their mean and standard deviation so the resulting feature has approximately zero mean and unit variance.


---

# Q22. Normalization vs standardization?

## Answer

| Normalization | Standardization |
|---|---|
| Often maps to fixed range | Centers around zero |
| Commonly [0,1] | Mean ≈ 0, std ≈ 1 |
| Min-max formula | Z-score formula |
| Sensitive to min/max | Can still be influenced by outliers |
| Useful in many distance/NN settings | Common for linear models, SVM, PCA, etc. |

### Important

There is no universal rule saying one is always better.

The appropriate choice depends on:

- Data distribution
- Algorithm
- Outliers
- Optimization behavior

### Interview Answer

Normalization typically maps values to a fixed range, while standardization centers and scales features based on mean and standard deviation.

# Q26. What is Linear Regression?

## Answer

Linear Regression models the relationship between input variables and a continuous target using a linear function.

For one feature:

    y = β₀ + β₁x

For multiple features:

    y = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ

where:

    y  = predicted value
    β₀ = intercept
    βᵢ = coefficients
    xᵢ = features

### Example

Predict salary:

    Salary =
    β₀
    + β₁(Experience)
    + β₂(Education)
    + β₃(Location)

### Interview Answer

Linear Regression predicts a continuous target by modeling it as a linear combination of input features.


---

# Q27. What are the assumptions of Linear Regression?

## Answer

Important classical assumptions include:

### 1. Linearity

The expected target should have an appropriate linear relationship with predictors.

### 2. Independence

Observations/errors should be appropriately independent for the intended statistical inference.

### 3. Homoscedasticity

Residual variance should be approximately constant across fitted values.

### 4. Normality of Residuals

For classical statistical inference, residual normality can be relevant, especially for confidence intervals and hypothesis tests.

It is not required for obtaining predictions in the same strict sense.

### 5. Low/No Perfect Multicollinearity

Predictors should not contain exact linear dependencies.

### Interview Answer

Classical Linear Regression assumptions include linearity, independence, homoscedasticity, and absence of perfect multicollinearity; residual normality is mainly important for classical statistical inference rather than basic prediction.

# Q28. What is Logistic Regression?

## Answer

Logistic Regression is a supervised learning algorithm commonly used for classification.

Despite its name, it is primarily a classification algorithm.

It models the probability of a class using the sigmoid function.

### Sigmoid

    σ(z) = 1 / (1 + e⁻ᶻ)

Output:

    0 to 1

Example:

    Probability of fraud = 0.87

A threshold can then be used to produce a class prediction.

### Typical binary classification

    P(y=1 | X)

### Interview Answer

Logistic Regression is a classification algorithm that models class probabilities using a logistic/sigmoid function and converts those probabilities into class predictions.

# Q31. What is the difference between loss function and cost function?

## Answer

Terminology varies across textbooks and libraries.

A common distinction is:

### Loss

Error for an individual training example.

    L(y, ŷ)

### Cost

Average or aggregate loss across the training dataset.

    J(θ) = 1/m Σ Lᵢ

Therefore:

    Individual Error → Loss
    Dataset-Level Aggregate → Cost

However, many sources use the terms interchangeably.

### Interview Answer

Typically, loss refers to error for an individual example, while cost refers to an aggregate such as the average loss across the dataset, although terminology varies.

# Q32. What is Mean Squared Error (MSE)?

## Answer

MSE measures the average squared difference between actual and predicted values.

Formula:

    MSE = (1/n) Σ(yᵢ - ŷᵢ)²

### Example

Actual:

    [10, 20]

Predicted:

    [12, 18]

Errors:

    [-2, 2]

Squared errors:

    [4, 4]

MSE:

    (4 + 4) / 2
    = 4

### Important Property

Because errors are squared, large errors receive disproportionately high penalty.

### Interview Answer

MSE is the average squared prediction error and strongly penalizes large errors.


---

# Q33. What is MAE?

## Answer

MAE stands for:

    Mean Absolute Error

Formula:

    MAE = (1/n) Σ|yᵢ - ŷᵢ|

It calculates the average absolute difference between actual and predicted values.

### Example

Actual:

    [10, 20]

Predicted:

    [12, 18]

Absolute errors:

    [2, 2]

MAE:

    2

### Advantage

MAE is less sensitive to extreme errors than MSE because it does not square the error.

### Interview Answer

MAE is the average absolute difference between actual and predicted values and is generally more robust to large errors than MSE.


---

# Q34. What is RMSE?

## Answer

RMSE stands for:

    Root Mean Squared Error

Formula:

    RMSE = √MSE

Because it takes the square root of MSE, RMSE has the same units as the target variable.

### Example

If:

    MSE = 25

then:

    RMSE = 5

### Interpretation

If predicting house prices in lakhs:

    RMSE = 5 lakh

roughly indicates the model's typical error scale, though RMSE is not simply the average absolute error.

### Interview Answer

RMSE is the square root of MSE. It penalizes large errors more strongly than MAE and is expressed in the same units as the target.


---

# Q35. What is R² score?

## Answer

R², or coefficient of determination, measures how much of the variation in the target is explained by the model relative to a baseline that predicts the mean.

Formula:

    R² =
    1 - SS_res / SS_tot

where:

    SS_res = residual sum of squares
    SS_tot = total sum of squares

### Interpretation

If:

    R² = 0.80

the model explains 80% of the variance relative to the mean-baseline under the standard formulation.

### Important

R² can be negative on unseen data if the model performs worse than the mean baseline.

### Interview Answer

R² measures how much target variance is explained by the model relative to a mean-prediction baseline. Higher values generally indicate better fit, but it should not be used alone.


# 🟠 PART 4 — CLASSIFICATION


# Q36. What is a confusion matrix?

## Answer

A confusion matrix summarizes classification predictions.

For binary classification:

                    Actual
                 Positive Negative

Predicted Positive    TP       FP

Predicted Negative    FN       TN

### TP — True Positive

Predicted positive and actually positive.

### TN — True Negative

Predicted negative and actually negative.

### FP — False Positive

Predicted positive but actually negative.

### FN — False Negative

Predicted negative but actually positive.

### Example

Fraud detection:

    TP → Fraud correctly detected
    TN → Genuine transaction correctly identified
    FP → Genuine transaction flagged as fraud
    FN → Fraud missed

### Interview Answer

A confusion matrix summarizes classification results using true positives, true negatives, false positives, and false negatives.


---

# Q37. What is accuracy?

## Answer

Accuracy is the proportion of all predictions that are correct.

Formula:

    Accuracy =
    (TP + TN) / (TP + TN + FP + FN)

### Example

Suppose:

    TP = 80
    TN = 90
    FP = 10
    FN = 20

Accuracy:

    (80 + 90) / 200
    = 85%

### Problem

Accuracy can be misleading with imbalanced datasets.

Example:

    99% non-fraud
    1% fraud

A model predicting "non-fraud" for every transaction gets:

    99% accuracy

but detects:

    0% fraud

### Interview Answer

Accuracy is the proportion of correct predictions, but it can be misleading when classes are highly imbalanced.


---

# Q38. What is precision?

## Answer

Precision measures how many predicted positives are actually positive.

Formula:

    Precision =
    TP / (TP + FP)

### Example

Suppose:

    TP = 90
    FP = 10

Precision:

    90 / 100
    = 90%

### Meaning

Of all cases predicted as positive:

    90% were actually positive.

### Important Use Case

Precision is important when false positives are costly.

Example:

    Spam detection

If legitimate emails are incorrectly classified as spam, users may miss important messages.

### Interview Answer

Precision measures the correctness of positive predictions and is especially important when false positives are costly.


---

# Q39. What is recall?

## Answer

Recall measures how many actual positive cases the model successfully identifies.

Formula:

    Recall =
    TP / (TP + FN)

### Example

Suppose:

    TP = 90
    FN = 10

Recall:

    90 / 100
    = 90%

### Important Use Case

Recall is important when false negatives are costly.

Example:

    Disease detection

Missing a genuinely sick patient can be more costly than sending a healthy patient for further testing.

### Interview Answer

Recall measures the proportion of actual positive cases correctly identified and is especially important when false negatives are costly.

    +

    1.0

and recall:

    0.0

F1 becomes:

    0

### Interview Answer

F1-score combines precision and recall using their harmonic mean and is useful when both false positives and false negatives matter.

ROC-AUC measures a classifier's ability to discriminate between classes across thresholds using the ROC curve, though PR-AUC is often more informative for severe class imbalance.

Regularization adds a penalty for model complexity to the training objective, helping reduce overfitting and improve generalization.

# Q47. What is L1 regularization?

## Answer

L1 regularization adds the absolute values of model coefficients to the objective.

Conceptually:

    Loss + λ Σ|wᵢ|

where:

    λ = regularization strength

A key property is that L1 can drive some coefficients exactly to zero.

Therefore, it can perform a form of:

    Feature Selection

### Example

Before:

    Feature A → 2.5
    Feature B → 1.2
    Feature C → 0.8

After strong L1:

    Feature A → 2.1
    Feature B → 0
    Feature C → 0.3

Feature B may effectively be removed.

### Interview Answer

L1 regularization penalizes the absolute magnitude of coefficients and can produce sparse models by driving some coefficients to zero.


---

# Q48. What is L2 regularization?

## Answer

L2 regularization adds squared coefficient values to the objective.

Conceptually:

    Loss + λ Σwᵢ²

It discourages very large weights but generally does not force them exactly to zero.

### Effect

    Large coefficients
          ↓
    Penalized
          ↓
    Smaller coefficients

### Common Example

Ridge Regression uses L2 regularization.

### Interview Answer

L2 regularization penalizes squared coefficient magnitudes, shrinking weights toward zero and helping reduce model complexity without typically making them exactly zero.

# Q49. L1 vs L2 regularization?

|---|---|
| Absolute values | Squared values |
| Can produce exact zero weights | Usually shrinks weights without exact zeros |
| Encourages sparsity | Encourages distributed small weights |
| Useful for feature selection | Useful for controlling coefficient magnitude |
| Lasso | Ridge |

### Interview Answer

L1 encourages sparse coefficients and can perform feature selection, whereas L2 shrinks coefficients toward zero and is commonly used to control model complexity.

# Q50. What is gradient descent?

## Answer

Gradient Descent is an optimization algorithm used to minimize a differentiable objective function.

The model calculates the gradient of the loss with respect to parameters and updates them in the opposite direction.

        ↓

    Calculate Gradient
        ↓

where:

    η = learning rate
    ∇J = gradient

### Goal

Move toward a parameter configuration that minimizes the objective.

### Interview Answer

Gradient Descent iteratively updates model parameters in the direction that reduces the objective function.

# Q56. What is Gini impurity?

## Answer

Gini impurity measures how mixed the classes are in a node.

Formula:

    Gini = 1 - Σ pᵢ²

where:

    pᵢ = proportion of class i

### Pure Node

Suppose all examples belong to one class:

    p = 1

Then:

    Gini = 1 - 1²
         = 0

So:

    Gini = 0
    ↓
    Pure node

### Mixed Node

If classes are evenly distributed in binary classification:

    p₁ = 0.5
    p₂ = 0.5

Then:

    Gini = 1 - 0.25 - 0.25
         = 0.5

### Interview Answer

Gini impurity measures class impurity in a decision-tree node. Lower values indicate purer nodes, with zero representing a completely pure node.

### Information Gain

Information Gain measures how much uncertainty is reduced after splitting.

Conceptually:

    Information Gain
    =

A good split generally produces a large reduction in impurity.

### Interview Answer

Entropy measures uncertainty in a node, while information gain measures how much that uncertainty decreases after a proposed split.

# Q73. How does PCA work?

### Step 7

    High-Dimensional Data
          ↓

    Find Principal Directions
          ↓
    Rank by Explained Variance

SHAP provides feature-attribution values for individual predictions and can also be aggregated to understand global importance.

### Important

Feature importance does not automatically mean causal importance.

### Interview Answer

Feature importance measures how much a model relies on features for prediction, using methods such as tree-based importance, permutation importance, or SHAP.

# Q84. What is SHAP?

## Answer

SHAP stands for:

    SHapley Additive exPlanations

It is an explainability framework based on Shapley values from cooperative game theory.

SHAP attempts to attribute a prediction among the input features.

Conceptually:

    Base Prediction
         +
    Feature Contributions
         ↓
    Final Prediction

### Example

Suppose:

    Base Risk = 40%

Features contribute:

    High debt      → +20%
    Good income    → -5%
    Late payments  → +15%

Final prediction:

    70%

### Benefits

- Local explanations
- Global analysis
- Feature contribution analysis
- Model debugging

### Interview Answer

SHAP explains model predictions by assigning contribution values to features based on Shapley-value principles, helping understand why individual predictions were made.

Parameters are learned from training data, whereas hyperparameters are configuration choices set by the practitioner or optimization procedure.

# Q86. What is early stopping?

## Answer

Early stopping is a regularization technique that stops training when validation performance stops improving.

Typical pattern:

    Training Loss ↓ continuously

but:

    Validation Loss ↓
          ↓
       minimum
          ↓
       starts ↑

At this point, continued training may cause overfitting.

### Process

    Train
      ↓
    Monitor validation metric
      ↓
    Improvement?
      ↓
    Yes → Continue
    No  → Stop

### Common Use

- Neural Networks
- Gradient Boosting

### Interview Answer

Early stopping terminates training when validation performance stops improving, helping prevent overfitting and unnecessary computation.

# Q89. What is out-of-bag (OOB) evaluation?

## Answer

In bagging methods such as Random Forest, each tree is trained on a bootstrap sample.

Because bootstrap sampling does not include every training example in every sample, some observations are left out for a given tree.

These are called:

    Out-of-Bag samples

They can be used to estimate model performance without a separate validation set in certain settings.

### Interview Answer

Out-of-bag evaluation uses observations excluded from a tree's bootstrap sample as an internal validation mechanism for bagging-based models such as Random Forest.

# Q90. What is calibration in classification?

## Answer

Calibration measures whether predicted probabilities correspond to observed frequencies.

Suppose a model predicts:

    Probability = 0.8

For many similar predictions, approximately:

    80%

should actually be positive for a well-calibrated model.

### Discrimination vs Calibration

Discrimination asks:

    Can the model rank positives above negatives?

Calibration asks:

    Do predicted probabilities correspond to actual frequencies?

A model can have strong discrimination but poor calibration.

### Interview Answer

Calibration evaluates whether predicted probabilities accurately reflect observed event frequencies, which is important when probabilities are used for decision-making.

# Q91. What is imbalanced learning and how can you handle it properly?

## Answer

Imbalanced learning occurs when one class is significantly underrepresented.

A proper approach includes:

### Step 1

Use suitable metrics:

    Precision
    Recall
    F1
    PR-AUC
    ROC-AUC
    Cost-based metrics

### Step 2

Use class weighting where appropriate.

### Step 3

Consider resampling.

    Oversampling
    Undersampling
    SMOTE

### Step 4

Tune the classification threshold.

### Step 5

Evaluate using stratified cross-validation.

### Important

Do not apply oversampling to the entire dataset before cross-validation because synthetic or duplicated observations can leak into validation folds.

### Interview Answer

Imbalanced learning requires appropriate metrics, class weighting or carefully performed resampling, threshold tuning, and leakage-safe cross-validation.

# Q92. What is SMOTE?

## Answer

SMOTE stands for:

    Synthetic Minority Over-sampling Technique

It creates synthetic minority-class examples rather than simply duplicating existing examples.

Conceptually:

    Minority Point A
          +
    Minority Point B
          ↓
    Synthetic Point

### Why?

To increase minority-class representation.

### Benefits

- Can reduce severe class imbalance
- Avoids simple duplication

### Risks

- Can create unrealistic synthetic observations
- May amplify noise
- Must be applied only to training data/folds

### Interview Answer

SMOTE generates synthetic minority-class samples using neighboring minority examples and is used to address class imbalance, but it must be applied carefully within training data to avoid leakage.

- Sparse data
- Difficult visualization

### Solutions

- PCA
- Regularization
- Domain-based feature engineering
- More training data

### Interview Answer

The curse of dimensionality describes how high-dimensional feature spaces become increasingly sparse and difficult to model effectively, often requiring more data and careful dimensionality reduction.

# Q94. What is concept drift?

## Answer

Concept drift occurs when the relationship between inputs and the target changes over time.

Example:

A fraud detection model learns:

    Old Fraud Patterns

Attackers change their behavior.

Now:

    New Fraud Patterns

The old relationship no longer works as well.

### Types

- Sudden drift
- Gradual drift
- Incremental drift
- Recurring drift

### Detection

Monitor:

- Model performance
- Feature distributions
- Prediction distributions
- Business outcomes

### Solutions

- Retraining
- Online learning
- Recent-data weighting
- Drift monitoring

### Interview Answer

Concept drift occurs when the underlying relationship between features and the target changes over time, causing previously learned patterns to become less effective.

# Q95. What is data drift?

## Answer

Data drift occurs when the distribution of input features changes between training and production.

Example:

Training:

    Average customer age = 35

Production:

    Average customer age = 52

The input distribution has changed.

### Difference

    Data Drift
       ↓
    P(X) changes

    Concept Drift
       ↓
    P(y|X) changes

### Interview Answer

Data drift occurs when the distribution of input features changes between training and deployment, while concept drift refers to changes in the relationship between inputs and targets.

Model drift refers to degradation in model performance over time due to changes in:

- Relationships
- Environment
- Business processes

It is often a practical production concern rather than one single mathematical definition.

A recommendation model performs:

    90% business KPI target

Initially.

After six months:

    72%

Possible causes:

    Data Drift
    Concept Drift
    Product Changes
    User Behavior Changes

### Monitoring

Track:

- Prediction quality
- Input distributions
- Output distributions
- Business KPIs

### Interview Answer

Model drift is the degradation of model effectiveness over time, often caused by changing data distributions, relationships, user behavior, or the production environment.
