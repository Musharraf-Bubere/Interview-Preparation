# 🤖 Machine Learning — Interview Questions

---

## Q1. What is Machine Learning?

### Interview Answer

Machine Learning is a branch of Artificial Intelligence that allows computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every possible case.

For example, a machine learning model can learn from historical customer data and predict whether a new customer is likely to leave a service.

### Memory Trick

**ML = Learn patterns from data → Make predictions**

### One-Line Answer

Machine Learning enables computers to learn patterns from data and use them to make predictions or decisions.

### Interview Answer

The three main types are:

1. **Supervised Learning**
2. **Unsupervised Learning**
3. **Reinforcement Learning**

Supervised learning uses labeled data.

Unsupervised learning works with unlabeled data.

Reinforcement learning learns through interactions with an environment using rewards and penalties.

### Examples

Supervised:
- Classification
- Regression

Unsupervised:
- Clustering
- Dimensionality reduction

Reinforcement:
- Game playing
- Robotics

### Memory Trick

**Supervised → Labels**

**Unsupervised → No labels**

**Reinforcement → Rewards**

### One-Line Answer

The main types of machine learning are supervised, unsupervised, and reinforcement learning.

## Q3. What is Supervised Learning?

### Interview Answer

Supervised learning is a machine learning approach where the model learns from **labeled data**.

The training data contains both input features and the expected output or target.

The model learns the relationship between them and uses it to predict outputs for new data.

### Example

Predicting whether an email is spam:

    Input → Email features
    Output → Spam / Not Spam

### Common Algorithms

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine
- K-Nearest Neighbors

### Memory Trick

**Supervised = Teacher provides answers**

### One-Line Answer

Supervised learning trains a model using labeled input-output examples.

---

## Q4. What is Unsupervised Learning?

### Interview Answer

Unsupervised learning works with **unlabeled data**.

The model tries to discover hidden patterns, structures, or groups in the data without having a predefined target variable.

### Example

Customer segmentation:

    Customer Data
         ↓
    Clustering
         ↓
    Customer Groups

### Common Algorithms

- K-Means
- Hierarchical Clustering
- DBSCAN
- PCA

### Memory Trick

**Unsupervised = Discover patterns without labels**

### One-Line Answer

Unsupervised learning finds patterns or structures in data without using labeled target values.

## Q5. What is the difference between Classification and Regression?

### Interview Answer

Both are supervised learning problems.

**Classification** predicts a categorical output.

Examples:

    Spam / Not Spam
    Fraud / Not Fraud
    Disease / No Disease

**Regression** predicts a continuous numerical value.

Examples:

    House price
    Temperature
    Salary

### Memory Trick

**Classification → Category**

**Regression → Number**

### One-Line Answer

Classification predicts categories, while regression predicts continuous numerical values.

The target variable is the output that the model is trying to predict.

### Example

Suppose we want to predict house prices.

    Features:
    - Area
    - Number of bedrooms
    - Location
    - Age

    Target:
    - House price

### Memory Trick

**Features → Input**

**Target → Output**

### One-Line Answer

Features are input variables, while the target is the output variable the model learns to predict.

## Q7. Why do we split data into training and testing sets?

### Interview Answer

We split data so that we can evaluate how well the trained model performs on **unseen data**.

The training set is used to learn model parameters.

The test set is kept separate and used for final evaluation.

This helps us understand whether the model generalizes beyond the training data.

### Example

    Dataset
       ↓
    Train → Model learns
       ↓
    Test → Final evaluation

### Memory Trick

**Train = Learn**

**Test = Evaluate**

### One-Line Answer

We split data so the model can be trained on one portion and evaluated on unseen data.

## Q8. What is validation data?

### Interview Answer

Validation data is used during model development to compare configurations, tune hyperparameters, and make decisions about the model.

A common workflow is:

    Training Data → Learn parameters
    Validation Data → Tune/select model
    Test Data → Final evaluation

For smaller datasets, techniques such as cross-validation can reduce the need for a separate fixed validation set.

### Memory Trick

**Train → Learn**

**Validation → Tune**

**Test → Final check**

### One-Line Answer

Validation data is used during model development for tuning and model selection before final testing.

## Q9. What is overfitting?

### Interview Answer

Overfitting occurs when a model learns the training data too closely, including noise or irrelevant patterns, and therefore performs poorly on unseen data.

Typically:

    Training performance → Very high
    Test performance → Much lower

### Causes

- Model too complex
- Too little training data
- Too many irrelevant features
- Excessive training

### Solutions

- Cross-validation
- Regularization
- Feature selection
- More training data
- Simpler model
- Early stopping for suitable models

### Memory Trick

**Overfitting = Memorizing instead of generalizing**

### One-Line Answer

Overfitting occurs when a model performs very well on training data but poorly on unseen data because it has learned noise or overly specific patterns.

---

## Q10. What is underfitting?

### Interview Answer

Underfitting occurs when a model is too simple to capture the important patterns in the data.

As a result, it performs poorly on both training and unseen data.

### Typical Pattern

    Training performance → Poor
    Test performance → Poor

### Causes

- Model too simple
- Insufficient features
- Excessive regularization
- Insufficient training

### Solutions

- Use a more suitable model
- Add useful features
- Reduce excessive regularization
- Train appropriately

### Memory Trick

**Underfitting = Model hasn't learned enough**

### One-Line Answer

Underfitting occurs when a model is too simple to capture the important patterns in the data.

---

## Q11. What is the bias-variance tradeoff?

### Interview Answer

Bias represents error caused by overly simplistic assumptions.

Variance represents how sensitive a model is to changes in the training data.

A high-bias model may underfit.

A high-variance model may overfit.

The goal is to find a suitable balance that gives good generalization.

### Memory Trick

**High Bias → Underfitting**

**High Variance → Overfitting**

### One-Line Answer

The bias-variance tradeoff is the balance between a model being too simple and too sensitive to training data.

It may include:

- Handling missing values
- Removing duplicates
- Encoding categorical variables
- Scaling numerical features
- Handling outliers
- Correcting data types
- Cleaning inconsistent data

### Memory Trick

**Raw Data → Clean Data → Model-ready Data**

### One-Line Answer

Data preprocessing transforms raw data into a clean and suitable format for machine learning.

## Q13. How do you handle missing values?

### Interview Answer

The approach depends on the reason and amount of missing data.

Possible approaches include:

- Removing rows or columns when appropriate
- Mean or median imputation for numerical data
- Mode or suitable-category imputation for categorical data
- Using a model-based imputation method
- Treating missingness as a meaningful category in some cases

I would first understand why the values are missing before choosing a method.

### Memory Trick

**Understand → Choose → Impute/Remove → Validate**

### One-Line Answer

Missing values can be removed or imputed using an appropriate strategy based on the data and the reason for missingness.

## Q15. What is feature scaling and why is it needed?

### Interview Answer

Feature scaling transforms numerical features to comparable scales.

It is particularly important for algorithms that are sensitive to feature magnitude or distance.

Common methods include:

- Standardization
- Min-Max scaling

For example, if one feature ranges from 0–1 and another ranges from 0–1,000,000, scaling can prevent the larger-scale feature from dominating certain algorithms.

### Memory Trick

**Scaling = Comparable ranges**

### One-Line Answer

Feature scaling puts numerical features on comparable scales and is important for many distance- or optimization-based algorithms.

---

## Q16. What is the difference between normalization and standardization?

### Interview Answer

Normalization often refers to scaling values to a fixed range, commonly 0 to 1.

Standardization transforms data so that it has approximately:

    Mean = 0
    Standard deviation = 1

### Formula

Min-Max normalization:

    x_scaled = (x - min) / (max - min)

Standardization:

    z = (x - mean) / standard_deviation

The appropriate technique depends on the algorithm and data.

### Memory Trick

**Normalization → Range**

**Standardization → Mean 0, Standard deviation 1**

### One-Line Answer

Normalization commonly scales values to a fixed range, while standardization centers data around zero with unit variance.

## Q17. What is data leakage?

### Interview Answer

Data leakage occurs when information that should not be available during training is unintentionally used by the model.

This can produce unrealistically high evaluation performance.

Common causes include:

- Using test data during training
- Performing preprocessing before splitting data
- Including future information in features
- Target leakage

### Example

If we calculate an imputation value using the entire dataset before splitting into train and test, information from the test set can influence the training process.

### Memory Trick

**Leakage = Future/unavailable information enters training**

### One-Line Answer

Data leakage occurs when information that should be unavailable during training influences the model.

## Q18. What is cross-validation?

### Interview Answer

Cross-validation is a technique used to evaluate a model by dividing the training data into multiple folds.

For example, in 5-fold cross-validation:

1. Divide data into five parts.
2. Train on four parts.
3. Validate on the remaining part.
4. Repeat until each part has been used for validation.
5. Calculate the average performance.

### Memory Trick

**Cross-validation = Multiple train/validation splits**

### One-Line Answer

Cross-validation evaluates a model across multiple train-validation splits to obtain a more reliable estimate of performance.

## Q19. What is precision, recall, and F1-score?

### Interview Answer

These are commonly used classification metrics.

### Precision

Out of the samples predicted as positive, how many were actually positive?

    Precision = TP / (TP + FP)

### Recall

Out of the actual positive samples, how many did the model correctly identify?

    Recall = TP / (TP + FN)

### F1-score

F1-score is the harmonic mean of precision and recall.

    F1 = 2 × Precision × Recall / (Precision + Recall)

### Memory Trick

**Precision → Predicted positive: How many correct?**

**Recall → Actual positive: How many found?**

**F1 → Balance of both**

### One-Line Answer

Precision measures the correctness of positive predictions, recall measures how many actual positives were found, and F1 balances both.

## Q20. What is a confusion matrix?

### Interview Answer

A confusion matrix is a table used to evaluate classification predictions.

It contains four main values:

- True Positive
- True Negative
- False Positive
- False Negative

### Structure

                    Predicted
                 Positive   Negative

    Actual Pos    TP         FN
    Actual Neg    FP         TN

These values can be used to calculate metrics such as accuracy, precision, recall, and F1-score.

### Memory Trick

**TP → Correct Positive**

**TN → Correct Negative**

**FP → Wrong Positive**

**FN → Missed Positive**

### One-Line Answer

A confusion matrix summarizes classification predictions using true positives, true negatives, false positives, and false negatives.

## Q21. What is the difference between parameters and hyperparameters?

### Interview Answer

Parameters are values learned by the model from the training data.

For example, the weights of a linear regression model are parameters.

Hyperparameters are configuration values chosen before or during model training and are not directly learned from the training data in the same way.

Examples include:

- Learning rate
- Number of trees
- Maximum tree depth
- Regularization strength
- Number of neighbors

### Memory Trick

**Parameters → Learned**

**Hyperparameters → Set/Tuned**

### One-Line Answer

Parameters are learned from data, while hyperparameters are configuration values chosen to control the learning process.

## Q22. What is regularization?

### Interview Answer

Regularization is a technique used to reduce overfitting by adding a penalty for model complexity.

Two common forms are:

- L1 regularization
- L2 regularization

L1 can encourage some coefficients to become exactly zero, which can provide a form of feature selection.

L2 generally shrinks coefficients toward zero.

### Memory Trick

**Regularization = Control model complexity**

### One-Line Answer

Regularization reduces overfitting by penalizing model complexity.

### Memory Trick

**Ensemble = Many models → Stronger prediction**

### One-Line Answer

Ensemble learning combines multiple models or learners to improve predictive performance and robustness.

## Q24. How do you choose a Machine Learning algorithm?

### Interview Answer

I would not choose an algorithm only because it is popular.

I would consider:

1. Type of problem
2. Size of dataset
3. Feature types
4. Relationship between features and target
5. Interpretability requirements
6. Training and inference constraints
7. Performance requirements
8. Risk of overfitting

I would establish a suitable baseline, evaluate candidate models using appropriate validation, and select the model based on both performance and practical requirements.

### Memory Trick

**Problem → Data → Model → Evaluate → Select**

### One-Line Answer

I choose a model based on the problem, data characteristics, requirements, and validated performance rather than popularity alone.

## Q25. How would you troubleshoot a Machine Learning model that is not performing well?

### Interview Answer

I would follow a systematic approach:

1. Verify the problem definition and target variable.
2. Check data quality.
3. Check missing values and duplicates.
4. Check feature distributions and outliers.
5. Verify preprocessing.
6. Check for data leakage.
7. Check train-test splitting.
8. Establish or compare with a baseline.
9. Analyze evaluation metrics and errors.
10. Perform feature engineering or selection.
11. Tune hyperparameters where appropriate.
12. Try alternative models.
13. Use cross-validation.
14. Re-evaluate on unseen data.

I would focus on identifying the root cause rather than simply changing the algorithm.

### Memory Trick

**Problem → Data → Features → Leakage → Model → Metrics → Tune → Test**

### One-Line Answer

I would troubleshoot ML performance by checking the problem, data, preprocessing, leakage, features, model, metrics, and validation before making changes.
