# 🟢 BASIC LEVEL

## Q1. What is Machine Learning?

### Interview Answer

Machine Learning is a branch of Artificial Intelligence in which algorithms learn patterns from data and use those patterns to make predictions or decisions on unseen data.

Traditional programming generally follows:

```text
Rules + Data → Output
```

Machine Learning follows:

```text
Data + Expected Output
        ↓
      Model
        ↓
     Learned Rules
        ↓
   Prediction on New Data
```

### Example

Suppose we want to predict house prices.

Features:

```text
Area
Bedrooms
Location
Age
```

Target:

```text
House Price
```

The model learns the relationship between these features and price from historical examples.

### Memory Trick

**ML = Learn patterns from data → predict unseen data.**

### One-Line Answer

> Machine Learning enables systems to learn patterns from historical data and use those learned patterns to make predictions or decisions on new data.

---

## Q2. What are the main types of Machine Learning?

### Interview Answer

The three major categories are:

```text
Machine Learning
│
├── Supervised Learning
├── Unsupervised Learning
└── Reinforcement Learning
```

### 1. Supervised Learning

The dataset contains input features and known target labels.

Examples:

- Classification
- Regression

```text
X + y → Model → Prediction
```

### 2. Unsupervised Learning

The data does not contain target labels.

Examples:

- Clustering
- Dimensionality reduction
- Anomaly detection

```text
X → Model → Patterns / Groups
```

### 3. Reinforcement Learning

An agent learns through interaction with an environment using rewards and penalties.

```text
Agent → Action → Environment
  ↑                 ↓
  └──── Reward ─────┘
```

### One-Line Answer

> The major learning paradigms are supervised, unsupervised, and reinforcement learning.

# 🟡 INTERMEDIATE LEVEL

## Q11. Why do we split data into train, validation, and test sets?

### Interview Answer

Data splitting helps estimate how well a model will perform on unseen data and prevents us from tuning directly against the final evaluation set.

Typical structure:

```text
Dataset
   ↓
Train ─────→ Learn Parameters
   ↓
Validation → Tune Hyperparameters
   ↓
Test ──────→ Final Evaluation
```

### Example

```text
70% Train
15% Validation
15% Test
```

The exact split depends on the dataset and problem.

### Important Point

For small datasets, cross-validation can be used instead of maintaining a large separate validation set.

### One-Line Answer

> Training data fits the model, validation data supports model selection and tuning, and the test set provides an unbiased final evaluation when used only at the end.

## Q12. What is overfitting?

### Interview Answer

Overfitting occurs when a model learns the training data too closely, including noise and random patterns, and therefore performs poorly on unseen data.

```text
Training Performance → Very High
Validation Performance → Poor
```

### Example

A very deep decision tree can memorize training examples.

### Symptoms

```text
Train error ↓↓↓
Validation error ↑
```

### Solutions

- Regularization
- Simpler model
- More training data
- Cross-validation
- Feature selection
- Early stopping
- Pruning
- Data augmentation where appropriate

### One-Line Answer

> Overfitting occurs when a model captures training-specific noise and fails to generalize to unseen data.

---

## Q13. What is underfitting?

### Interview Answer

Underfitting occurs when a model is too simple to capture important patterns in the data.

```text
Training Performance → Poor
Validation Performance → Poor
```

### Causes

- Model too simple
- Insufficient features
- Excessive regularization
- Poor feature representation

### Solutions

- Increase model complexity
- Add useful features
- Reduce excessive regularization
- Train appropriately

### One-Line Answer

> Underfitting occurs when the model is too simple to capture the underlying patterns in the data.

## Q14. What is the bias-variance tradeoff?

### Interview Answer

Bias represents error caused by overly strong simplifying assumptions, while variance represents sensitivity to the training data.

```text
High Bias
→ Underfitting

High Variance
→ Overfitting
```

The goal is to find a suitable balance that generalizes well.

### Conceptual Relationship

```text
Model Complexity
      →
Bias decreases
Variance increases
```

This is a conceptual relationship, not a universal rule for every model.

### One-Line Answer

> The bias-variance tradeoff is the balance between underfitting from overly simple assumptions and overfitting from excessive sensitivity to training data.

## Q15. What is cross-validation?

### Interview Answer

Cross-validation evaluates a model across multiple train-validation splits instead of relying on one split.

### K-Fold Cross-Validation

```text
Dataset

Fold 1 → Validation
Folds 2-5 → Training

Fold 2 → Validation
Folds 1,3-5 → Training

...

Average Scores
```

### Benefits

- More reliable performance estimate
- Better use of limited data
- Useful for model and hyperparameter comparison

### Important Point

Cross-validation should be performed without leaking information from validation folds into training transformations.

### One-Line Answer

> Cross-validation repeatedly trains and validates a model on different partitions to obtain a more robust estimate of generalization performance.

## Q16. What is data leakage?

### Interview Answer

Data leakage occurs when information that would not be legitimately available at prediction time influences model training.

This can produce unrealistically strong evaluation results.

### Example

Suppose we predict whether a customer will default.

If we include:

```text
"Recovery Status After Default"
```

as a feature, it may contain future information unavailable when making the original prediction.

That is target leakage.

### Another Example

Scaling the complete dataset before train-test splitting can allow information from the test set to influence preprocessing.

### Prevention

```text
Split first
 ↓
Fit transformations only on training data
 ↓
Apply to validation/test
```

### One-Line Answer

> Data leakage occurs when information from outside the legitimate training information enters model training or evaluation.

## Q18. What is feature scaling?

### Interview Answer

Feature scaling transforms numerical variables to comparable ranges.

Two common methods are:

### Standardization

```text
z = (x - mean) / standard deviation
```

Result generally has:

```text
Mean ≈ 0
Std ≈ 1
```

### Min-Max Scaling

```text
x_scaled = (x - min) / (max - min)
```

Typically maps values into:

```text
0 to 1
```

### Important

Scaling is particularly important for distance- or gradient-based algorithms.

Examples:

```text
KNN
K-Means
SVM
Neural Networks
```

Tree-based models generally do not require feature scaling.

### One-Line Answer

> Feature scaling puts numerical features on comparable scales and is especially important for algorithms sensitive to feature magnitude.

# 🔴 ADVANCED LEVEL

## Q21. What is regularization?

### Interview Answer

Regularization adds a penalty to the objective function to discourage overly complex models and improve generalization.

### L1 Regularization

```text
Loss + λ Σ|w|
```

L1 can encourage some coefficients to become exactly zero.

### L2 Regularization

```text
Loss + λ Σw²
```

L2 discourages large coefficients.

### Concept

```text
Without Regularization
→ Model may become too complex

With Regularization
→ Complexity is controlled
```

### One-Line Answer

> Regularization controls model complexity by penalizing model parameters and can reduce overfitting.

---

## Q22. What is the difference between L1 and L2 regularization?

### Interview Answer

| L1 | L2 |
|---|---|
| Uses absolute weights | Uses squared weights |
| Can produce exact zeros | Usually shrinks weights toward zero |
| Useful for sparsity | Useful for smooth shrinkage |

### Example

If we have 100 features and many are irrelevant, L1 may drive some coefficients to zero.

### Memory Trick

**L1 = feature selection tendency.  
L2 = weight shrinkage.**

### One-Line Answer

> L1 encourages sparse solutions, while L2 generally shrinks parameters without forcing as many exact zeros.

## Q23. What is gradient descent?

### Interview Answer

Gradient descent is an optimization algorithm used to minimize a loss function.

Conceptually:

```text
Initialize Parameters
        ↓
Make Prediction
        ↓
Calculate Loss
        ↓
Calculate Gradient
        ↓
Update Parameters
        ↓
Repeat
```

Update:

```text
parameter = parameter - learning_rate × gradient
```

### Example

If the gradient points toward increasing loss, the algorithm moves parameters in the opposite direction.

### One-Line Answer

> Gradient descent iteratively updates model parameters in the direction that reduces the loss function.

### One-Line Answer

## Q33. Explain the confusion matrix.

### Interview Answer

For binary classification, a confusion matrix contains:

```text
                    Predicted
                  Positive Negative

Actual Positive     TP       FN

Actual Negative     FP       TN
```

Where:

```text
TP = True Positive
TN = True Negative
FP = False Positive
FN = False Negative
```

### Metrics

```text
Precision = TP / (TP + FP)

Recall = TP / (TP + FN)

F1 = 2 × Precision × Recall / (Precision + Recall)
```

### Example

In disease screening:

```text
FN = Sick patient predicted healthy
```

This may be much more serious than a false positive.

### One-Line Answer

> A confusion matrix summarizes correct and incorrect classification decisions and provides the basis for precision, recall, F1, and related metrics.

## Q34. What is ROC-AUC?

### Interview Answer

ROC-AUC summarizes how well a classifier ranks positive examples above negative examples across classification thresholds.

The ROC curve plots:

```text
True Positive Rate
vs
False Positive Rate
```

AUC means:

```text
Area Under the Curve
```

### Interpretation

An AUC around 0.5 indicates performance close to random ranking, while higher values indicate better discrimination.

### Important Point

ROC-AUC can be less informative for highly imbalanced datasets, where PR-AUC may better reflect positive-class performance.

### One-Line Answer

> ROC-AUC measures a classifier's ability to discriminate between classes across thresholds.

## Q35. What is precision-recall tradeoff?

### Interview Answer

Precision measures how many predicted positives are actually positive.

```text
Precision = TP / (TP + FP)
```

Recall measures how many actual positives are detected.

```text
Recall = TP / (TP + FN)
```

Changing the classification threshold can increase one while decreasing the other.

### Example

Spam filtering:

```text
Higher recall
→ Catch more spam
→ Possibly flag more legitimate emails

Higher precision
→ Fewer legitimate emails incorrectly flagged
→ Potentially miss more spam
```

### One-Line Answer

> Precision-recall tradeoff reflects the balance between avoiding false positives and avoiding false negatives as the classification threshold changes.

Resampling should be performed within the training process/folds to avoid validation leakage.

### One-Line Answer

> Class imbalance requires appropriate evaluation metrics and, when necessary, techniques such as class weighting, resampling, and threshold optimization.

## Q39. What is PCA?

### Interview Answer

PCA, or Principal Component Analysis, is a linear dimensionality-reduction technique that transforms correlated variables into a set of orthogonal principal components ordered by explained variance.

### Concept

```text
Original Features
X1 X2 X3 X4 X5
        ↓
       PCA
        ↓
PC1 PC2 PC3
```

### Important Steps

```text
Scale where appropriate
 ↓
Compute covariance structure / SVD
 ↓
Find principal directions
 ↓
Project data
```

### Important Point

PCA is sensitive to feature scale, so standardization is commonly appropriate when features have different units.

### One-Line Answer

> PCA projects data onto orthogonal directions that capture decreasing amounts of variance, allowing dimensionality reduction.

## Q42. How do you prevent data leakage in an ML pipeline?

### Interview Answer

I make sure every transformation that learns parameters is fitted only on training data.

Incorrect:

```text
Complete Dataset
 ↓
Scale
 ↓
Split
```

Correct:

```text
Complete Dataset
 ↓
Split
 ↓
Train → Fit Scaler
Validation/Test → Transform
```

For cross-validation, preprocessing should be included inside the CV pipeline so each fold learns transformations only from its training portion.

### Example

Use a pipeline conceptually:

```text
Scaler
 ↓
Feature Transformer
 ↓
Model
```

and evaluate the whole pipeline through cross-validation.

### One-Line Answer

> I prevent leakage by splitting before fitting learned transformations and by placing preprocessing inside the cross-validation/training pipeline.

## Q45. What is SHAP?

### Interview Answer

SHAP stands for **SHapley Additive exPlanations**.

It uses concepts from cooperative game theory to attribute a model prediction among input features.

Conceptually:

```text
Baseline Prediction
       +
Feature Contributions
       ↓
Final Prediction
```

### Example

```text
Prediction = High Risk

Feature Contributions:
Debt Ratio      → increases risk
Low Income      → increases risk
Good Credit     → decreases risk
```

### Uses

- Local explanations
- Global feature importance
- Model debugging
- Bias investigation

### One-Line Answer

> SHAP assigns feature contributions to predictions using Shapley-value-based explanations.

## Q46. What is concept drift and data drift?

### Interview Answer

**Data drift** means the distribution of input data changes over time.

Example:

```text
Training customers
→ Mostly age 20–40

Production
→ Mostly age 40–60
```

**Concept drift** means the relationship between inputs and target changes.

Example:

```text
Customer behavior changes
→ Old relationship between features and churn is no longer valid
```

### Production Flow

```text
Production Data
 ↓
Monitor Distributions
 ↓
Detect Drift
 ↓
Evaluate Model
 ↓
Retrain / Investigate
```

### One-Line Answer

> Data drift changes the input distribution, while concept drift changes the relationship between inputs and the target.

# ⭐ IMPORTANT MACHINE LEARNING COMPARISONS

## Supervised vs Unsupervised Learning

```text
Supervised
→ Has target
→ Prediction

Unsupervised
→ No target
→ Pattern discovery
```

---

## Classification vs Regression

```text
Classification
→ Discrete classes

Regression
→ Continuous values
```

---

## Overfitting vs Underfitting

```text
Overfitting
→ Train performance high
→ Generalization poor

Underfitting
→ Train performance poor
→ Generalization poor
```

---

## Bagging vs Boosting

```text
Bagging
→ Parallel / more independent learners
→ Reduces variance

Boosting
→ Sequential learners
→ Corrects previous errors
```

---

## Random Forest vs Gradient Boosting

```text
Random Forest
→ Many randomized trees
→ Strong baseline
→ Often robust

Gradient Boosting
→ Sequential trees
→ Often highly predictive
→ Requires careful tuning
```

---

## L1 vs L2

```text
L1
→ |weights|
→ Sparse solutions

L2
→ weights²
→ Shrinks weights
```

---

## Parameter vs Hyperparameter

```text
Parameter
→ Learned during training

Hyperparameter
→ Configured/tuned outside parameter learning
```

---

## Data Drift vs Concept Drift

```text
Data Drift
→ P(X) changes

Concept Drift
→ P(Y|X) changes
```
