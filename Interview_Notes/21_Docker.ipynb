# 🐳 DOCKER INTERVIEW PREPARATION
# 25 Questions | Generic Technical Interview Preparation

---

## Q1. What is Docker?

### Interview Answer

Docker is a platform used to package an application along with its dependencies into a container.

A container provides a consistent environment in which the application can run.

For example, instead of installing Python, libraries, and other dependencies manually on every machine, we can define them in a Docker image and run the application inside a container.

### Example

    Application
        +
    Dependencies
        +
    Configuration
        ↓
      Docker Image
        ↓
      Container

### Memory Trick

**Docker = Application + Dependencies + Container**

### One-Line Answer

Docker packages an application and its dependencies into a container so it can run consistently across different environments.

---

## Q2. What is a container?

### Interview Answer

A container is a lightweight, isolated environment in which an application runs along with its required dependencies.

Containers share the host operating system's kernel, which generally makes them lighter and faster to start than virtual machines.

### Example

    Host Machine
        ↓
    Docker Engine
        ↓
    ┌──────────────┐
    │ Container    │
    │ Python App   │
    │ Dependencies │
    └──────────────┘

### Memory Trick

**Container = Isolated Application Environment**

### One-Line Answer

A container is a lightweight isolated environment used to run an application and its dependencies.

---

## Q3. What is the difference between Docker and a Virtual Machine?

### Interview Answer

The main difference is how they use the operating system.

A virtual machine includes a complete guest operating system along with the application.

A Docker container shares the host operating system's kernel and contains the application and its dependencies.

Therefore, containers are generally lighter and faster to start than virtual machines.

### Example

    Virtual Machine:

    Hardware
       ↓
    Host OS
       ↓
    Hypervisor
       ↓
    Guest OS
       ↓
    Application


    Docker:

    Hardware
       ↓
    Host OS
       ↓
    Docker Engine
       ↓
    Container
       ↓
    Application

### Memory Trick

**VM → Full OS**

**Container → Shared Kernel**

### One-Line Answer

A VM includes a complete guest OS, while a Docker container shares the host kernel and is generally more lightweight.

---

## Q4. What is a Docker image?

### Interview Answer

A Docker image is a read-only template used to create containers.

It contains the application code, dependencies, libraries, configuration, and instructions required to run the application.

A container is created from an image.

### Example

    Dockerfile
        ↓
    Docker Image
        ↓
    Container

### Memory Trick

**Image = Template**

### One-Line Answer

A Docker image is a template containing everything required to create and run a container.

---

## Q5. What is a Dockerfile?

### Interview Answer

A Dockerfile is a text file containing instructions for building a Docker image.

It defines things such as:

- Base image
- Working directory
- Dependencies
- Application files
- Environment configuration
- Command used to start the application

### Example

    FROM python:3.14

    WORKDIR /app

    COPY requirements.txt .

    RUN pip install -r requirements.txt

    COPY . .

    CMD ["python", "app.py"]

### Memory Trick

**Dockerfile = Recipe for Image**

### One-Line Answer

A Dockerfile contains instructions that Docker uses to build an image.

---

## Q6. What is the difference between a Docker image and a container?

### Interview Answer

A Docker image is the template, while a container is the running instance created from that image.

One image can be used to create multiple containers.

### Example

    Docker Image
       ↓
    ┌───────────┐
    ↓           ↓
    Container  Container

### Memory Trick

**Image = Blueprint**

**Container = Running Instance**

### One-Line Answer

An image is a template, while a container is a running instance created from that image.

## Q7. What is Docker Hub?

### Interview Answer

Docker Hub is a registry where Docker images can be stored and shared.

Developers can pull existing public images from Docker Hub or push their own images to a repository.

For example, a Python base image can be pulled from Docker Hub and used as the starting point for an application image.

### Memory Trick

**Docker Hub = Image Repository**

### One-Line Answer

Docker Hub is a registry used to store, share, and retrieve Docker images.

## Q8. What is a Docker registry?

### Interview Answer

A Docker registry is a system used to store and distribute Docker images.

Docker Hub is one example of a registry.

Organizations can also use private registries to store their internal application images.

### Memory Trick

**Registry = Store + Distribute Images**

### One-Line Answer

A Docker registry stores and distributes Docker images.

Docker Engine is the core runtime that builds and manages Docker images and containers.

## Q11. What is the difference between docker run and docker start?

### Interview Answer

`docker run` creates a new container from an image and starts it.

`docker start` starts an existing stopped container.

### Example

    docker run image_name

    → Creates + Starts a new container


    docker start container_name

    → Starts an existing container

### Memory Trick

**run = Create + Start**

**start = Start Existing**

### One-Line Answer

`docker run` creates and starts a new container, while `docker start` starts an existing stopped container.

---

## Q12. What is the difference between CMD and ENTRYPOINT?

### Interview Answer

Both are used to define what should run when a container starts.

`CMD` provides a default command or default arguments and can generally be overridden when running the container.

`ENTRYPOINT` is used when we want the container to behave like a specific executable, making the main command less likely to be replaced accidentally.

### Example

    CMD ["python", "app.py"]

    ENTRYPOINT ["python"]

### Memory Trick

**CMD = Default**

**ENTRYPOINT = Main Executable**

### One-Line Answer

CMD defines a default command, while ENTRYPOINT defines the main executable behavior of a container.

## Q13. What is a Docker volume?

### Interview Answer

A Docker volume is a mechanism used to persist data outside the container's writable layer.

Containers are generally treated as temporary, so data stored only inside a container can be lost when the container is removed.

Volumes allow important data to survive container recreation.

### Example

    Container
       ↓
    Application Data
       ↓
    Docker Volume
       ↓
    Persistent Storage

### Memory Trick

**Volume = Persistent Data**

### One-Line Answer

A Docker volume provides persistent storage so important data can survive container recreation.

---

## Q14. What is Docker networking?

### Interview Answer

Docker networking allows containers and other systems to communicate with each other.

For example, in a backend application we might have:

    FastAPI Container
          ↓
    PostgreSQL Container

Both containers can communicate through a Docker network.

Docker provides different network drivers, with bridge networking being commonly used for containers on the same Docker host.

### Memory Trick

**Docker Network = Container Communication**

### One-Line Answer

Docker networking allows containers and external systems to communicate with each other.

---

## Q15. What is Docker Compose?

### Interview Answer

Docker Compose is a tool used to define and manage multi-container applications.

Instead of manually running each container, we can define services in a Compose configuration file.

For example, an application may have:

    FastAPI
    PostgreSQL
    Redis

Docker Compose can define how these services should run and communicate.

### Memory Trick

**Compose = Multiple Containers Together**

### One-Line Answer

Docker Compose is used to define and run applications consisting of multiple containers.

---

## Q16. Why would you use Docker Compose for a FastAPI application?

### Interview Answer

A FastAPI application may depend on other services such as PostgreSQL.

Instead of installing and configuring every service manually, Docker Compose can define them as separate services.

For example:

    FastAPI
       ↓
    PostgreSQL

Both can be started together using the Compose configuration.

This makes local development and environment setup more consistent.

### Memory Trick

**FastAPI + DB = Compose**

### One-Line Answer

Docker Compose can simplify running a FastAPI application together with dependencies such as PostgreSQL.

## Q17. What is a multi-stage Docker build?

### Interview Answer

A multi-stage Docker build uses multiple stages in a Dockerfile.

One stage can be used to build or prepare the application, while another stage contains only what is required to run it.

This can reduce the final image size and avoid including unnecessary build dependencies.

### Example

    Stage 1
    Build / Install
         ↓
    Stage 2
    Runtime Image
         ↓
    Smaller Final Image

### Memory Trick

**Build Stage → Runtime Stage**

### One-Line Answer

Multi-stage builds separate the build environment from the runtime environment to produce cleaner and often smaller images.

---

## Q18. What is Docker image layering?

### Interview Answer

Docker images are built using layers.

Each instruction in a Dockerfile can create a layer or contribute to the image's filesystem layers.

Docker can reuse unchanged layers when rebuilding an image, which can make builds more efficient.

### Example

    Base Python Image
          ↓
    Dependencies Layer
          ↓
    Application Layer
          ↓
    Final Image

### Memory Trick

**Image = Layers**

### One-Line Answer

Docker images are built from layers, allowing Docker to reuse unchanged parts during image builds.

## Q19. How would you reduce Docker image size?

### Interview Answer

I would use several approaches.

First, I would choose an appropriate lightweight base image where practical.

Then I would avoid unnecessary packages and files.

I could use a `.dockerignore` file to prevent unnecessary files from being copied into the image.

I could also use multi-stage builds when appropriate.

### Memory Trick

**Small Base + Clean Files + .dockerignore + Multi-stage**

### One-Line Answer

I would reduce image size by using an appropriate base image, removing unnecessary dependencies, using `.dockerignore`, and applying multi-stage builds where useful.

---

## Q20. What is .dockerignore?

### Interview Answer

`.dockerignore` is a file used to specify files and directories that should not be sent to the Docker build context.

For example, we generally do not want to copy:

    .git
    __pycache__
    .pytest_cache
    .venv
    local environment files
    temporary files

This can reduce the build context and prevent unnecessary files from entering the image.

### Memory Trick

**.dockerignore = Don't Copy**

### One-Line Answer

`.dockerignore` prevents unnecessary files from being included in the Docker build context.

---

## Q21. How should secrets such as API keys be handled in Docker?

### Interview Answer

I should not hard-code API keys or passwords inside the Dockerfile or source code.

Instead, sensitive configuration should be provided through environment variables or an appropriate secret-management mechanism.

For production systems, I would prefer a proper cloud or platform secret-management solution.

### Example

    API_KEY
    DATABASE_PASSWORD
    SECRET_KEY

    ↓

    Environment / Secret Manager

### Memory Trick

**Secrets ≠ Source Code**

### One-Line Answer

Secrets should be injected securely through environment variables or secret-management systems rather than hard-coded into images or source code.

## Q22. A Docker container starts and immediately stops. How would you troubleshoot it?

### Interview Answer

I would first check the container status and logs.

For example:

    docker ps -a

    docker logs <container>

Then I would check:

- Startup command
- Dockerfile
- Application errors
- Environment variables
- Missing dependencies
- Port configuration
- File paths

A container usually stops when its main process exits, so I would focus on why the main process terminated.

### Memory Trick

**Status → Logs → Command → Config → Dependencies**

### One-Line Answer

I would inspect the container status and logs first, then verify the startup command, configuration, dependencies, and application errors.

---

## Q23. Your application works locally but fails inside Docker. What would you check?

### Interview Answer

I would compare the local environment with the container environment.

I would check:

- Python version
- Installed dependencies
- Environment variables
- File paths
- Working directory
- Network configuration
- Database connection
- Port configuration
- Operating-system differences

I would then reproduce the issue inside the container and inspect the logs.

### Memory Trick

**Local vs Container → Compare Everything**

### One-Line Answer

I would compare dependencies, configuration, paths, networking, database connectivity, and runtime versions between local and Docker environments.

---

## Q24. How would you troubleshoot a Dockerized API that cannot connect to PostgreSQL?

### Interview Answer

I would troubleshoot the connection step by step.

First, I would check whether both containers are running.

Then I would verify that both services are on the same Docker network.

I would check the database hostname, port, username, password, and database name.

An important point is that from one Docker container, the database host is usually the database service name rather than `localhost`.

Then I would inspect application and PostgreSQL logs.

### Memory Trick

**Containers → Network → Host → Port → Credentials → Logs**

### One-Line Answer

I would verify container status, Docker networking, database service name, port, credentials, and logs.

---

## Q25. Explain how you would Dockerize a Python/FastAPI application.

### Interview Answer

I would start by creating a Dockerfile.

The basic process would be:

    1. Choose a Python base image.
    2. Set the working directory.
    3. Copy the dependency file.
    4. Install dependencies.
    5. Copy the application code.
    6. Expose the application port if appropriate.
    7. Define the startup command.
    8. Build the Docker image.
    9. Run the container.
    10. Test the API.

For example:

    Dockerfile
        ↓
    Python Base Image
        ↓
    Install Dependencies
        ↓
    Copy Application
        ↓
    Start FastAPI
        ↓
    Docker Container

For a real application with PostgreSQL or other services, I could use Docker Compose to manage the complete local environment.

### Memory Trick

**Base → Dependencies → Code → Build → Run → Test**

### One-Line Answer

To Dockerize FastAPI, I would create a Dockerfile, install dependencies, copy the application, define the startup command, build the image, run the container, and test the API.

# ⭐ INTERVIEW GOLDEN POINT
