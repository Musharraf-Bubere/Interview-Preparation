## Q1. What is LangChain?

### Interview Answer

LangChain is a framework for building applications powered by Large Language Models. It provides abstractions and integrations for connecting LLMs with prompts, tools, retrievers, structured outputs, document loaders, vector stores, agents, and other application components.

Instead of writing every integration from scratch, LangChain provides reusable building blocks.

### Basic Flow

```text
User
 ↓
Prompt
 ↓
LLM
 ↓
Response
```

A more useful application can become:

```text
User
 ↓
Prompt
 ↓
LLM
 ↓
Tool / Retriever
 ↓
Result
 ↓
LLM
 ↓
Final Answer
```

### Example

A chatbot can use LangChain to:

- Create prompts
- Call an LLM
- Retrieve documents
- Call tools
- Produce structured output

### Memory Trick

**LangChain = Building blocks for LLM applications.**

### One-Line Answer

> LangChain is a framework and ecosystem for building LLM applications by connecting models with prompts, tools, retrieval, structured outputs, and workflows.

## Q2. Why do we need LangChain?

### Interview Answer

Calling an LLM directly is enough for simple applications, but production applications often need more than one model call.

They may need:

```text
Prompt Management
+
Model Integration
+
Tools
+
Retrieval
+
Structured Output
+
Memory
+
Tracing
```

LangChain provides reusable abstractions for these components.

### Example

Instead of manually implementing:

```text
Prompt → OpenAI API → Parse JSON → Tool → Another LLM Call
```

LangChain provides components that can be composed into a workflow.

### Important Interview Point

I would not use LangChain automatically for every project. If a simple direct model API call solves the problem, additional framework abstraction may not be necessary.

### Memory Trick

**Use abstractions when application complexity increases.**

### One-Line Answer

> LangChain becomes valuable when an LLM application needs reusable integrations, tools, retrieval, structured outputs, or composable workflows.

## Q4. What is LCEL?

### Interview Answer

LCEL stands for **LangChain Expression Language**.

It allows LangChain components to be composed using a declarative pipeline style.

For example:

```text
Prompt
   ↓
Model
   ↓
Parser
```

Conceptually:

```python
chain = prompt | model | parser
```

The `|` operator represents composition between runnable components.

### Why Useful?

LCEL makes pipelines:

- Composable
- Reusable
- Easier to read
- Easier to stream
- Easier to execute asynchronously
- Easier to trace

### Example

```text
Input
 ↓
Prompt Template
 ↓
Chat Model
 ↓
String Parser
 ↓
Output
```

### Memory Trick

**LCEL = Connect components into runnable pipelines.**

### One-Line Answer

> LCEL is LangChain's expression syntax for composing runnable components into executable pipelines.

## Q6. What is a prompt template?

### Interview Answer

A prompt template is a reusable prompt definition containing variables that are filled dynamically at runtime.

Example:

```text
You are a helpful assistant.

Answer the question:
{question}
```

At runtime:

```text
question =
"What is RAG?"
```

becomes:

```text
You are a helpful assistant.

Answer the question:
What is RAG?
```

### Benefit

It separates prompt structure from runtime data.

### Memory Trick

**Template = Prompt + Variables.**

### One-Line Answer

> A prompt template is a reusable prompt with dynamic variables that are populated at runtime.

- Database operations
- Classification
- Routing
- Form extraction
- Tool arguments

### Memory Trick

**Structured output = LLM response that software can reliably consume.**

### One-Line Answer

> Structured output constrains an LLM response to a predefined schema so downstream application code can process it reliably.

# 🟡 INTERMEDIATE LEVEL

## Q11. What is LangGraph?

### Interview Answer

LangGraph is a framework for building stateful, controllable agent and workflow applications using graph-based execution.

Instead of representing an application only as a linear chain, LangGraph represents it as:

```text
Nodes + Edges + State
```

### Example

```text
START
  ↓
Planner
  ↓
Research
  ↓
Evaluate
  ├──→ More Research
  └──→ Generate
          ↓
         END
```

The graph can contain branches, loops, persistence, and human intervention.

### Memory Trick

**LangGraph = State + Nodes + Edges + Control.**

### One-Line Answer

> LangGraph is a framework for building stateful, graph-based LLM workflows and agents with explicit control over execution.

---

## Q12. Why use LangGraph instead of a simple chain?

### Interview Answer

A simple chain is appropriate when execution is mostly linear.

For example:

```text
Input → Prompt → LLM → Output
```

But real agents often need:

```text
Loops
Branches
State
Retries
Human approval
Tool calls
Persistence
Conditional routing
```

LangGraph is designed for these workflows.

### Example

```text
Question
 ↓
Research
 ↓
Quality Check
 ├── Poor → Research Again
 └── Good → Answer
```

A graph represents this naturally.

### Memory Trick

**Chain = Straight road. Graph = Road network.**

### One-Line Answer

> LangGraph is preferable when an LLM workflow needs explicit state, branching, loops, persistence, or human-in-the-loop control.

## Q20. What is tool calling?

### Interview Answer

Tool calling is a mechanism where the model produces a structured request for an external tool instead of directly generating the final answer.

### Flow

```text
User
 ↓
LLM
 ↓
Tool Call
 ↓
Application Executes Tool
 ↓
Tool Result
 ↓
LLM
 ↓
Final Answer
```

### Example

```text
User:
"What's the weather in Mumbai?"

LLM:
call weather_tool(city="Mumbai")

Tool:
32°C, clear

LLM:
"It's currently 32°C and clear."
```

### Important Security Point

The application, not the model, should ultimately control whether and how a tool is executed.

### Memory Trick

**LLM requests; application executes.**

### One-Line Answer

> Tool calling allows an LLM to request structured execution of external functions while the application remains responsible for execution and authorization.

## Q25. What is an agent loop?

### Interview Answer

An agent loop is the repeated cycle through which an agent decides what action to take, executes it, observes the result, and decides whether more work is needed.

```text
        ┌───────────────┐
        ↓               │
     Decide             │
        ↓               │
      Action            │
        ↓               │
    Observation         │
        ↓               │
     Evaluate ──────────┘
        ↓
      Finish
```

### Why Important?

Without proper termination conditions, an agent could continue unnecessarily.

### Controls

- Maximum iterations
- Timeout
- Budget limits
- Tool limits
- Error handling
- Stop conditions

### Memory Trick

**Agent loop needs a stop button.**

### One-Line Answer

> An agent loop repeatedly decides, acts, observes, and evaluates until a defined completion condition is reached.
