# 🔗 LangChain, LangGraph & Agentic AI — 25 Interview Questions

---

## Q1. What is LangChain?

### Interview Answer

LangChain is a framework used to build applications powered by Large Language Models.

It provides components for working with:

- LLMs
- Prompts
- Structured outputs
- Tools
- Retrievers
- RAG
- Agents
- Memory and application workflows

Instead of writing every integration from scratch, LangChain provides reusable building blocks.

### Memory Trick

**LangChain = Framework for building LLM applications**

### One-Line Answer

LangChain is a framework that provides reusable components for building applications with LLMs.

---

## Q2. Why would you use LangChain instead of directly calling an LLM API?

### Interview Answer

We can directly call an LLM API for simple applications.

LangChain becomes useful when the application contains multiple components such as:

- Prompt templates
- Structured outputs
- Tools
- Retrieval
- Multiple model calls
- Complex workflows
- Agents

It helps organize these components into a reusable application pipeline.

### Memory Trick

**Simple app → Direct API**

**Complex LLM app → Framework can help**

### One-Line Answer

LangChain is useful when an LLM application requires multiple reusable components such as prompts, tools, retrieval, and structured workflows.

## Q3. What is a prompt template in LangChain?

### Interview Answer

A prompt template is a reusable template for creating prompts dynamically.

Instead of hardcoding a complete prompt every time, we define variables that can be filled with actual values.

### Example

    Template:
    "Summarize the following text:
    {text}"

    Input:
    text = "Artificial Intelligence is..."

The application fills the variable and sends the resulting prompt to the model.

### Memory Trick

**Prompt Template = Reusable Prompt + Variables**

### One-Line Answer

A prompt template is a reusable prompt structure containing variables that can be populated dynamically.

---

## Q4. What is a chain in LangChain?

### Interview Answer

A chain represents a sequence of operations where the output of one component can be passed to another component.

For example:

    Input
      ↓
    Prompt
      ↓
    LLM
      ↓
    Parser
      ↓
    Output

Chains help organize multiple processing steps into a pipeline.

### Memory Trick

**Chain = Step A → Step B → Step C**

### One-Line Answer

A chain is a sequence of connected operations used to process data through an LLM application.

---

## Q5. What is LCEL?

### Interview Answer

LCEL stands for **LangChain Expression Language**.

It provides a way to compose LangChain components into runnable pipelines.

For example, conceptually:

    Prompt → LLM → Parser

The components can be connected so the output of one becomes the input of the next.

### Memory Trick

**LCEL = Connect LangChain components**

### One-Line Answer

LCEL is a syntax and framework within LangChain for composing runnable components into pipelines.

## Q6. What is LangGraph?

### Interview Answer

LangGraph is a framework for building stateful, multi-step, and agentic workflows.

It represents an application as a graph containing:

- Nodes
- Edges
- State

Nodes perform tasks, while edges determine how execution moves between nodes.

It is especially useful when a workflow requires:

- Conditional execution
- Loops
- Multiple agents
- State management
- Human intervention
- Iterative processing

### Memory Trick

**LangGraph = Nodes + Edges + State**

### One-Line Answer

LangGraph is a framework for building stateful and controllable multi-step or agentic workflows using graphs.

---

## Q7. What is the difference between LangChain and LangGraph?

### Interview Answer

LangChain provides components and abstractions for building LLM applications.

LangGraph focuses on orchestrating more complex workflows using graphs and state.

### Simple Comparison

    LangChain
    → Components and LLM application building blocks

    LangGraph
    → Workflow orchestration using nodes, edges, and state

LangChain components can also be used inside LangGraph workflows.

### Memory Trick

**LangChain = Building Blocks**

**LangGraph = Workflow**

### One-Line Answer

LangChain provides LLM application components, while LangGraph focuses on orchestrating stateful and complex workflows.

## Q8. What is Agentic AI?

### Interview Answer

Agentic AI refers to AI systems that can reason about a task, decide what actions are needed, use tools, and continue working toward a goal.

Instead of only generating a response, an agent can interact with external systems.

### Example

A travel agentic system may:

    User Request
        ↓
    Decide what information is needed
        ↓
    Search destinations
        ↓
    Check hotels
        ↓
    Check weather
        ↓
    Find activities
        ↓
    Generate itinerary

### Memory Trick

**Agent = Reason + Decide + Act**

### One-Line Answer

Agentic AI refers to AI systems that can reason about tasks, make decisions, use tools, and perform multiple actions toward a goal.

## Q9. What is an AI agent?

### Interview Answer

An AI agent is a system that uses an AI model to decide what actions should be taken to accomplish a goal.

An agent can typically:

- Understand the task
- Decide what information it needs
- Select tools
- Execute actions
- Observe results
- Continue or change its approach
- Produce a final result

### Memory Trick

**Agent = Goal + Reasoning + Tools + Actions**

### One-Line Answer

An AI agent is a system that uses an AI model to decide and execute actions to accomplish a goal.

### Interview Answer

A simple LLM application may follow a fixed flow:

    Input → Prompt → LLM → Output

An agentic application can dynamically decide what action or tool to use.

    Input
      ↓
    Agent
      ↓
    Decide Action
      ↓
    Tool
      ↓
    Observe Result
      ↓
    Decide Next Action
      ↓
    Final Answer

The key difference is the level of decision-making and dynamic action.

### Memory Trick

**LLM App → Fixed flow**

**Agent → Can decide actions**

### One-Line Answer

A simple LLM application follows a predefined process, while an agent can dynamically decide actions and use tools to achieve a goal.

## Q11. What is state in LangGraph?

### Interview Answer

State is the shared information maintained during workflow execution.

It can contain information such as:

- User input
- Intermediate results
- Tool results
- Agent outputs
- Validation results
- Attempt counts
- Final response

Each node can read relevant state and update it.

### Example

    State:
    {
        user_query
        retrieved_data
        tool_results
        validation
        final_response
    }

### Memory Trick

**State = Workflow's shared memory**

### One-Line Answer

State stores the information that nodes need to read and update during a LangGraph workflow.

## Q12. What are nodes and edges in LangGraph?

### Interview Answer

A **node** represents an operation or function performed by the workflow.

An **edge** determines where execution goes next.

### Example

    Node A
      ↓
    Node B
      ↓
    Node C

Here:

- A, B, C are nodes.
- The arrows are edges.

Conditional edges can choose different paths based on the current state.

### Memory Trick

**Node = Work**

**Edge = Direction**

### One-Line Answer

Nodes perform tasks, while edges control the flow of execution between those tasks.

---

## Q13. What is a conditional edge?

### Interview Answer

A conditional edge allows a workflow to choose the next node based on a condition or the current state.

### Example

    Validator
        ↓
    Is response valid?
       / \
     Yes  No
      ↓    ↓
    Final Refine

If the response is valid, the workflow moves to the final response.

If it is invalid, it goes to a refinement step.

### Memory Trick

**Conditional Edge = Decision in workflow**

### One-Line Answer

A conditional edge selects the next workflow step based on a condition or state.

### Interview Answer

A sequential workflow executes tasks one after another in a predefined order.

### Example

    Research
       ↓
    Generate
       ↓
    Validate
       ↓
    Final Response

Each step depends on the previous step.

### Memory Trick

**Sequential = One after another**

### One-Line Answer

A sequential workflow executes tasks in a fixed order, one step after another.

## Q15. What is a parallel workflow?

### Interview Answer

A parallel workflow executes independent tasks at the same time or concurrently when the framework and runtime support it.

For example, after destination research, we may independently collect:

- Hotel information
- Weather information
- Activity information

These tasks do not necessarily need to wait for each other.

### Example

    Destination Research
          ↓
    ┌─────┼─────┐
    ↓     ↓     ↓
    Hotel Weather Activities
    └─────┼─────┘
          ↓
      Itinerary

### Memory Trick

**Parallel = Independent tasks together**

### One-Line Answer

A parallel workflow executes independent tasks concurrently to reduce unnecessary waiting.

---

## Q16. What is an iterative workflow?

### Interview Answer

An iterative workflow repeats a process until a condition is satisfied or a maximum number of attempts is reached.

### Example

    Generate
       ↓
    Validate
       ↓
    Valid?
     /   \
   Yes    No
    ↓      ↓
   End   Refine
            ↓
         Validate

This is useful when the first generated result may need improvement.

### Memory Trick

**Iterative = Generate → Check → Improve → Check**

### One-Line Answer

An iterative workflow repeatedly executes steps until the result meets a condition or a maximum attempt limit is reached.

## Q17. What is multi-agent AI?

### Interview Answer

Multi-agent AI uses multiple specialized agents where each agent is responsible for a particular task.

Instead of asking one agent to perform everything, responsibilities are divided.

### Example

    User
      ↓
    Orchestrator
      ↓
    ┌──────┬──────┬──────┐
    ↓      ↓      ↓
    Research Hotel Weather
    Agent   Agent   Agent
    └──────┬──────┬──────┘
           ↓
       Final Agent

This can make complex systems easier to organize and maintain.

### Memory Trick

**Multi-Agent = Divide responsibilities**

### One-Line Answer

Multi-agent AI uses multiple specialized agents that collaborate to solve a complex task.

## Q18. Why use multiple specialized agents instead of one agent?

### Interview Answer

Multiple specialized agents can improve separation of responsibilities.

For example:

- Research Agent → Research
- Weather Agent → Weather
- Hotel Agent → Hotels
- Validation Agent → Validation

Benefits include:

- Clear responsibilities
- Easier debugging
- Easier testing
- Better modularity
- Ability to use different tools for different tasks

However, multi-agent systems also introduce additional complexity, latency, and coordination overhead, so they should be used only when they provide a real benefit.

### Memory Trick

**Specialization → Modularity → Easier maintenance**

### One-Line Answer

Multiple agents are useful when a problem can be divided into clearly defined specialized responsibilities.

## Q20. How do you prevent an agent from getting stuck in an infinite loop?

### Interview Answer

I would put explicit control mechanisms around the workflow.

For example:

- Maximum iteration count
- Maximum tool calls
- Timeout limits
- State-based stopping conditions
- Validation conditions
- Error handling

### Example

    Attempt 1 → Validate → Fail
    Attempt 2 → Validate → Fail
    Attempt 3 → Validate → Stop

Even if validation continues to fail, the workflow should stop after the configured limit.

### Memory Trick

**Loop = Condition + Limit**

### One-Line Answer

I would control agent loops using stopping conditions, maximum iterations, tool-call limits, and timeouts.

## Q21. How do you handle failures in an agentic workflow?

### Interview Answer

I would identify which component failed first.

I would check:

1. Input
2. Agent decision
3. Tool call
4. External API
5. Tool response
6. State update
7. Next workflow transition
8. Final output

Depending on the failure, I might use:

- Exception handling
- Retries
- Timeouts
- Fallback logic
- Validation
- Logging
- Maximum retry limits

### Memory Trick

**Find failed node → Find root cause → Recover safely**

### One-Line Answer

I would trace the workflow node by node, identify the failed component, handle the error appropriately, and prevent uncontrolled retries.

### Example

          ↓
    Human Approval
       /      \
     Yes       No
      ↓         ↓

**AI decides → Human verifies → Action**

### One-Line Answer

Human-in-the-loop means requiring human review or approval at selected points in an AI workflow.

## Q24. How would you debug an Agentic AI or LangGraph application?

### Interview Answer

I would debug it step by step instead of looking only at the final answer.

I would inspect:

1. User input
2. Initial state
3. Node execution
4. LLM input
5. LLM output
6. Tool calls
7. Tool results
8. State updates
9. Conditional decisions
10. Final response

I would use logs and traces to understand where the workflow behavior differs from the expected behavior.

If one node produces an incorrect result, I would isolate and test that node independently.

### Memory Trick

**State → Node → LLM → Tool → Decision → Output**

### One-Line Answer

I would trace state changes, node execution, model calls, tool calls, and conditional transitions to identify the failing component.

## Q25. What is the difference between deterministic workflows and agentic workflows?

### Interview Answer

A deterministic workflow follows predefined rules and paths.

For example:

    Step A → Step B → Step C

An agentic workflow can make decisions dynamically based on the current state, available tools, and task requirements.

For example:

    Agent
      ↓
    Decide
      ↓
    Tool A or Tool B
      ↓
    Observe Result
      ↓
    Decide Next Step

Deterministic workflows provide more predictability, while agentic workflows provide more flexibility.

In real applications, a hybrid approach is often useful.

### Memory Trick

**Deterministic = Fixed**

**Agentic = Dynamic**

### One-Line Answer

Deterministic workflows follow predefined paths, while agentic workflows can dynamically decide actions based on the task and current state.

# 🔥 FOUR IMPORTANT WORKFLOW PATTERNS

## 1. Sequential

**A → B → C**

One after another.

## 2. Parallel

**A → B + C + D**

Independent tasks can execute concurrently.

## 3. Conditional

**A → Decision → B OR C**

Choose a path based on a condition.

## 4. Iterative

**A → Check → Improve → Check**

Repeat until successful or until a limit is reached.
