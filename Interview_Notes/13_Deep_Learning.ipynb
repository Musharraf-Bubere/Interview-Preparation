# 🟢 BASIC LEVEL

## Q1. What is Deep Learning?

### Interview Answer

Deep Learning is a subfield of Machine Learning that uses neural networks with multiple layers to automatically learn hierarchical representations from data.

Traditional Machine Learning often relies heavily on manually designed features:

```text
Raw Data
   ↓
Feature Engineering
   ↓
ML Model
   ↓
Prediction
```

Deep Learning can learn representations directly from raw or relatively lightly processed data:

```text
Raw Data
   ↓
Neural Network
   ↓
Learned Representations
   ↓
Prediction
```

### Example

For image classification:

```text
Image
 ↓
Edges
 ↓
Shapes
 ↓
Object Parts
 ↓
Object
 ↓
Class
```

### One-Line Answer

> Deep Learning uses multi-layer neural networks to automatically learn increasingly complex representations from data.

---

## Q2. What is an Artificial Neural Network?

### Interview Answer

An Artificial Neural Network (ANN) is a computational model inspired loosely by biological neural networks.

A basic neural network contains:

```text
Input Layer
     ↓
Hidden Layer(s)
     ↓
Output Layer
```

Each neuron typically performs:

```text
z = w₁x₁ + w₂x₂ + ... + b

a = activation(z)
```

### Example

```text
Age ─────┐
Income ──┼──→ Hidden Layers ──→ Risk
Credit ──┘
```

### One-Line Answer

> An ANN is a network of interconnected computational units that learn parameters to map inputs to outputs.

---

## Q3. What are neurons, weights, and biases?

### Interview Answer

A neuron receives inputs, multiplies them by weights, adds a bias, and applies an activation function.

```text
Inputs
x1 x2 x3
 ↓  ↓  ↓
Weights
w1 w2 w3
 ↓
Weighted Sum + Bias
 ↓
Activation
 ↓
Output
```

Mathematically:

```text
z = Σ(wᵢxᵢ) + b
a = f(z)
```

### Weight

Controls the influence of an input.

### Bias

Allows the neuron to shift its activation.

### One-Line Answer

> Weights determine the influence of inputs, while the bias shifts the neuron's activation before the activation function is applied.

---

## Q4. What are activation functions?

### Interview Answer

Activation functions introduce nonlinearity into neural networks.

Without nonlinear activations, stacking linear layers would still produce an overall linear transformation.

Common activation functions:

```text
Sigmoid
Tanh
ReLU
Leaky ReLU
GELU
Softmax
```

### Example

ReLU:

```text
ReLU(x) = max(0, x)
```

### Why Needed?

```text
Linear Layers
     ↓
No meaningful nonlinear representation

Nonlinear Activations
     ↓
Complex functions can be learned
```

### One-Line Answer

> Activation functions introduce nonlinearity so neural networks can learn complex relationships.

---

## Q5. Explain ReLU.

### Interview Answer

ReLU stands for Rectified Linear Unit.

```text
ReLU(x) = max(0, x)
```

Therefore:

```text
x < 0 → 0
x > 0 → x
```

### Advantages

- Simple
- Computationally efficient
- Usually helps optimization compared with sigmoid/tanh in deep networks

### Problem

A neuron can become permanently inactive if it consistently receives negative inputs and gradients become zero.

This is known as the:

```text
Dying ReLU problem
```

### Alternative

```text
Leaky ReLU
```

### One-Line Answer

> ReLU outputs zero for negative inputs and the input itself for positive inputs, providing a simple nonlinear activation widely used in hidden layers.

---

## Q6. What is the difference between Sigmoid, Tanh, and ReLU?

### Interview Answer

| Function | Range | Common Use |
|---|---|---|
| Sigmoid | 0 to 1 | Binary probability output |
| Tanh | -1 to 1 | Some recurrent architectures |
| ReLU | 0 to ∞ | Hidden layers |

### Sigmoid

```text
σ(x) = 1 / (1 + e^-x)
```

### Tanh

```text
tanh(x)
```

### ReLU

```text
max(0, x)
```

Sigmoid and tanh can suffer from saturation and vanishing gradients, especially in deep networks.

### One-Line Answer

> ReLU is commonly used in hidden layers, while sigmoid is useful for binary probability outputs and tanh appears in some recurrent architectures.

## Q7. What is forward propagation?

### Interview Answer

Forward propagation is the process of passing input data through the network to calculate predictions.

```text
Input
 ↓
Layer 1
 ↓
Activation
 ↓
Layer 2
 ↓
Activation
 ↓
Output
```

For each layer:

```text
z = Wx + b
a = f(z)
```

The final activation produces the model prediction.

### One-Line Answer

> Forward propagation passes inputs through all network layers to compute the prediction.

---

## Q8. What is a loss function?

### Interview Answer

A loss function measures how different the model's predictions are from the target values.

```text
Prediction
    +
Actual Target
    ↓
Loss Function
    ↓
Loss Value
```

Examples:

### Regression

```text
MSE
MAE
Huber Loss
```

### Classification

```text
Binary Cross-Entropy
Categorical Cross-Entropy
```

### One-Line Answer

> A loss function quantifies prediction error and provides the optimization objective used to train the neural network.

---

## Q9. What is an epoch, batch, and iteration?

### Interview Answer

### Epoch

One complete pass through the training dataset.

### Batch

A subset of training examples processed together.

### Iteration

One parameter update, usually corresponding to processing one batch.

Example:

```text
10,000 training samples
Batch size = 100

Iterations per epoch = 10,000 / 100 = 100
```

If training for:

```text
20 epochs
```

then approximately:

```text
2,000 parameter-update iterations
```

### One-Line Answer

> An epoch is one complete pass through the dataset, a batch is a subset processed together, and an iteration is typically one parameter update.

---

## Q10. What is backpropagation?

### Interview Answer

Backpropagation computes gradients of the loss with respect to model parameters by applying the chain rule from the output layer backward through the network.

```text
Forward Pass
     ↓
Prediction
     ↓
Loss
     ↓
Backward Pass
     ↓
Gradients
     ↓
Parameter Update
```

The optimizer then uses these gradients to update weights.

### Important Interview Point

Backpropagation computes gradients; the optimizer performs the parameter update.

### One-Line Answer

> Backpropagation efficiently computes gradients of the loss with respect to network parameters so an optimizer can update them.

# 🟡 INTERMEDIATE LEVEL

## Q11. What is gradient descent in Deep Learning?

### Interview Answer

Gradient descent minimizes the loss by updating parameters in the direction opposite to the gradient.

```text
θ_new = θ_old - η ∇L(θ)
```

Where:

```text
θ = Parameters
η = Learning Rate
∇L = Gradient of Loss
```

### Training Loop

```text
Input
 ↓
Forward Pass
 ↓
Loss
 ↓
Backpropagation
 ↓
Gradient
 ↓
Optimizer Update
 ↓
Repeat
```

### One-Line Answer

> Gradient descent updates neural-network parameters in the direction that reduces the training loss.

---

## Q12. What is the difference between Batch, Stochastic, and Mini-Batch Gradient Descent?

### Interview Answer

### Batch Gradient Descent

Uses the entire training dataset for each update.

```text
Entire Dataset
 ↓
One Update
```

### Stochastic Gradient Descent

Uses one training example per update.

```text
One Example
 ↓
One Update
```

### Mini-Batch Gradient Descent

Uses a small batch of examples.

```text
Mini-Batch
 ↓
One Update
```

Modern deep-learning training commonly uses mini-batches.

### One-Line Answer

> Batch GD uses all samples per update, SGD uses one sample, and mini-batch GD uses a small subset per update.

### Interview Answer

An optimizer uses gradients to update neural-network parameters.

Common optimizers include:

```text
SGD
Momentum
RMSprop
Adam
AdamW
```

### Adam

Adam combines ideas related to momentum and adaptive learning rates.

### AdamW

AdamW decouples weight decay from the gradient-based update, which can improve regularization behavior in many settings.

### One-Line Answer

> An optimizer determines how model parameters are updated using gradients during training.

### Interview Answer

Adam is popular because it adapts the effective step size for parameters and incorporates momentum-like behavior.

It generally requires less manual tuning than plain SGD for many problems.

### Advantages

- Fast practical convergence
- Adaptive updates
- Works well across many architectures
- Useful default for many deep-learning tasks

### Important Interview Point

Adam is not universally best. SGD with momentum can sometimes provide better generalization, especially in certain computer-vision settings.

### One-Line Answer

## Q15. What is the vanishing gradient problem?

### Interview Answer

The vanishing gradient problem occurs when gradients become extremely small as they propagate backward through many layers or time steps.

```text
Output
 ↓
Small Gradient
 ↓
Smaller Gradient
 ↓
Very Small Gradient
 ↓
Early Layers Learn Slowly
```

This was especially problematic with saturating activations such as sigmoid and tanh in deep/recurrent networks.

### Solutions

- ReLU-family activations
- Proper initialization
- Residual connections
- Normalization
- LSTM/GRU for long sequences
- Transformer architectures in sequence tasks

### One-Line Answer

> Vanishing gradients cause early layers to receive extremely small updates, making learning slow or ineffective.

---

## Q16. What is the exploding gradient problem?

### Interview Answer

Exploding gradients occur when gradients become extremely large during backpropagation.

```text
Gradient
 ↓
Large
 ↓
Very Large
 ↓
Unstable Updates
```

### Symptoms

- Loss becomes NaN
- Training becomes unstable
- Parameters become extremely large

### Solutions

- Gradient clipping
- Better initialization
- Appropriate learning rate
- Normalization
- Stable architectures

### One-Line Answer

> Exploding gradients occur when gradients grow excessively large, causing unstable parameter updates.

---

## Q17. What is weight initialization?

### Interview Answer

Weight initialization determines the starting values of neural-network parameters.

Poor initialization can cause:

```text
Vanishing gradients
Exploding activations
Slow convergence
```

Common strategies:

```text
Xavier / Glorot Initialization
He Initialization
```

### Typical Use

```text
Xavier → often associated with tanh/sigmoid
He → commonly used with ReLU-family activations
```

### One-Line Answer

> Weight initialization chooses suitable starting parameters so activations and gradients remain in useful ranges during training.

---

## Q18. What is Batch Normalization?

### Interview Answer

Batch Normalization normalizes activations using statistics computed from a mini-batch during training, followed by learnable scale and shift parameters.

Conceptually:

```text
Layer Output
 ↓
Normalize
 ↓
Scale + Shift
 ↓
Next Layer
```

### Benefits

- Can stabilize training
- Often allows larger learning rates
- Can provide some regularization effect

### Important Point

Behavior differs between training and inference because inference uses stored running statistics.

### One-Line Answer

> Batch Normalization normalizes intermediate activations using batch statistics during training and learned parameters to transform them.

---

## Q19. What is Dropout?

### Interview Answer

Dropout is a regularization technique that randomly disables a fraction of activations during training.

```text
Before Dropout
● ● ● ● ● ●

After Dropout
● ○ ● ○ ● ●
```

This reduces reliance on specific neurons and can improve generalization.

During inference, dropout is disabled and activations are appropriately scaled according to the implementation.

### One-Line Answer

> Dropout randomly removes activations during training to reduce over-reliance on particular neurons and help prevent overfitting.

---

## Q20. What is Early Stopping?

### Interview Answer

Early stopping stops training when validation performance stops improving.

```text
Epoch
 ↓
Training Loss ↓
Validation Loss ↓
          ↓
Validation stops improving
          ↓
Stop Training
```

Often a patience value is used:

```text
No improvement for N epochs
→ Stop
```

### Benefits

- Reduces overfitting
- Saves training time
- Helps select a useful checkpoint

### One-Line Answer

> Early stopping terminates training when validation performance no longer improves according to a chosen criterion.

# 🔴 ADVANCED LEVEL

## Q21. What is a CNN?

### Interview Answer

A Convolutional Neural Network (CNN) is a neural architecture especially effective for grid-like data such as images.

Typical structure:

```text
Image
 ↓
Convolution
 ↓
Activation
 ↓
Pooling / Downsampling
 ↓
Convolution
 ↓
Activation
 ↓
Global Pooling / Flatten
 ↓
Dense Layer
 ↓
Output
```

CNNs exploit local connectivity and shared weights.

### One-Line Answer

> CNNs use learned convolutional filters to extract hierarchical local patterns and are widely used for image and spatial data.

---

## Q22. What is convolution in CNNs?

### Interview Answer

A convolution applies a learnable kernel over local regions of an input to produce feature maps.

Conceptually:

```text
Input Image
[ pixels ]
    ↓
Kernel
    ↓
Sliding Operation
    ↓
Feature Map
```

Early filters may learn patterns such as:

```text
Edges
Textures
Corners
```

Deeper layers can learn more complex structures.

### One-Line Answer

> Convolution applies learnable local filters across an input to detect spatial patterns and generate feature maps.

---

## Q23. What are stride and padding?

### Interview Answer

### Stride

Stride controls how far the convolution kernel moves between positions.

```text
Stride = 1
→ Moves one position

Stride = 2
→ Moves two positions
```

Larger stride usually reduces spatial dimensions.

### Padding

Padding adds values around the input, often zeros, to control output dimensions.

Common modes:

```text
Valid
Same
```

### One-Line Answer

> Stride controls the movement of the kernel, while padding controls how the input boundaries are handled and influences output dimensions.

---

## Q24. What is pooling?

### Interview Answer

Pooling reduces spatial dimensions of feature maps.

### Max Pooling

Takes the maximum value in a local region.

```text
[1 5]
[2 3]

Max = 5
```

### Average Pooling

Calculates the average.

### Benefits

- Reduces computation
- Reduces spatial resolution
- Can provide some translation robustness

Modern architectures may use strided convolutions or global average pooling instead of traditional pooling in some stages.

### One-Line Answer

> Pooling downsamples feature maps to reduce spatial dimensions and computation while retaining useful information.

---

## Q25. What is transfer learning?

### Interview Answer

Transfer learning uses knowledge learned from one task or dataset as a starting point for another related task.

```text
Large Dataset
 ↓
Pretrained Model
 ↓
Fine-Tune
 ↓
Target Dataset
 ↓
Target Task
```

### Example

A CNN pretrained on a large image dataset can be adapted for:

```text
Medical Image Classification
```

### Benefits

- Less training data required
- Faster convergence
- Often better performance

### One-Line Answer

> Transfer learning reuses representations learned from a pretrained model for a related downstream task.

---

## Q26. What is fine-tuning?

### Interview Answer

Fine-tuning means continuing training of a pretrained model on a target dataset.

Possible strategies:

```text
Freeze most layers
Train new head
        ↓
Gradually unfreeze layers
        ↓
Fine-tune with small learning rate
```

### Why Small Learning Rate?

A large learning rate can destroy useful pretrained representations through excessive updates.

### One-Line Answer

> Fine-tuning adapts a pretrained model to a new task by updating some or all of its parameters on target data.

---

## Q27. What is an RNN?

### Interview Answer

A Recurrent Neural Network processes sequential data while maintaining a hidden state that carries information from previous time steps.

```text
x₁ → RNN → h₁
x₂ → RNN → h₂
x₃ → RNN → h₃
```

Conceptually:

```text
hₜ = f(xₜ, hₜ₋₁)
```

### Applications

- Time series
- Sequence modeling
- Earlier NLP systems
- Speech-related tasks

### Limitation

Vanishing and exploding gradients can make learning long-term dependencies difficult.

### One-Line Answer

> RNNs process sequences recurrently, using a hidden state to carry information from earlier time steps.

---

## Q28. What is LSTM?

### Interview Answer

LSTM stands for Long Short-Term Memory.

It is a recurrent architecture designed to better preserve information across long sequences.

It uses gates:

```text
Forget Gate
Input Gate
Output Gate
```

Conceptually:

```text
Previous State
      ↓
Forget / Input / Output Gates
      ↓
Cell State
      ↓
Hidden State
```

The cell state provides a pathway for information to persist over time.

### One-Line Answer

> LSTM is an RNN architecture that uses gated cell-state updates to better model long-term dependencies.

---

## Q29. What is GRU?

### Interview Answer

GRU stands for Gated Recurrent Unit.

It is a gated recurrent architecture with fewer gates and parameters than a standard LSTM.

Main concepts include:

```text
Update Gate
Reset Gate
```

### LSTM vs GRU

```text
LSTM
→ More complex
→ Separate cell and hidden states

GRU
→ Simpler
→ Fewer parameters
```

GRUs can be faster in some settings, but the best choice depends on the task and data.

### One-Line Answer

> GRU is a simpler gated recurrent architecture that uses update and reset mechanisms to control information flow.

---

## Q30. Why were Transformers introduced and what problem do they solve?

### Interview Answer

Transformers were introduced to model sequences using attention rather than relying primarily on recurrence.

RNNs process sequences sequentially:

```text
x1 → x2 → x3 → x4
```

This can make long-range dependencies and parallel training difficult.

Transformers use attention:

```text
x1 ─┐
x2 ─┼→ Attention → Contextual Representations
x3 ─┤
x4 ─┘
```

This allows tokens to directly interact through attention and enables substantial parallelization during training.

### One-Line Answer

> Transformers use attention to model relationships between sequence elements directly, reducing the sequential limitations of recurrent architectures.

# 🟣 EXPERT LEVEL

## Q31. What is attention?

### Interview Answer

Attention allows a model to dynamically assign different importance to different elements when computing a representation.

The standard scaled dot-product attention is:

```text
Attention(Q,K,V)
=
softmax(QKᵀ / √dₖ)V
```

Where:

```text
Q = Queries
K = Keys
V = Values
```

### Intuition

For a token:

```text
Query
 ↓
Compare with Keys
 ↓
Attention Scores
 ↓
Weighted Values
 ↓
Contextual Representation
```

### Example

In a sentence, a token can attend strongly to another token that helps determine its meaning.

### One-Line Answer

> Attention computes context-aware representations by weighting information from other elements according to their relevance.

---

## Q32. What is self-attention?

### Interview Answer

Self-attention is attention where queries, keys, and values are derived from the same sequence.

```text
Input Sequence
      ↓
Q, K, V from same sequence
      ↓
Self-Attention
      ↓
Contextualized Sequence
```

This allows each token to consider other tokens in the same sequence.

### Example

In:

```text
"The animal didn't cross the road because it was tired."
```

Attention can help the model relate "it" to relevant context.

### One-Line Answer

> Self-attention allows elements within the same sequence to attend to one another to build contextual representations.

---

## Q33. What is multi-head attention?

### Interview Answer

Multi-head attention performs multiple attention operations in parallel using different learned projections.

```text
Input
 ↓
Head 1
Head 2
Head 3
...
Head N
 ↓
Concatenate
 ↓
Linear Projection
```

Different heads can learn different relationships or representation subspaces.

### One-Line Answer

> Multi-head attention runs several attention mechanisms in parallel so the model can capture different relationships and representation patterns.

---

## Q34. Why do Transformers need positional encoding?

### Interview Answer

Self-attention by itself does not inherently encode token order.

For example:

```text
A B C
```

and:

```text
C B A
```

contain the same elements but different order.

Positional information tells the model where elements occur.

Approaches include:

```text
Sinusoidal Positional Encoding
Learned Position Embeddings
Rotary Position Embeddings (RoPE)
Relative Position Methods
```

### One-Line Answer

> Positional information gives Transformers information about sequence order, which pure self-attention does not inherently provide.

---

## Q35. What is the difference between encoder-only, decoder-only, and encoder-decoder Transformers?

### Interview Answer

### Encoder-Only

Designed primarily for understanding representations.

Examples of tasks:

```text
Classification
Embedding
Token Classification
```

Architecture:

```text
Input → Encoder → Representation
```

### Decoder-Only

Designed primarily for autoregressive generation.

```text
Previous Tokens
      ↓
Decoder
      ↓
Next Token
```

Used by many modern LLMs.

### Encoder-Decoder

Uses an encoder to represent input and decoder to generate output.

```text
Input
 ↓
Encoder
 ↓
Context
 ↓
Decoder
 ↓
Output
```

Useful for sequence-to-sequence tasks such as translation and summarization.

### One-Line Answer

> Encoder-only models focus on understanding, decoder-only models on autoregressive generation, and encoder-decoder models map one sequence to another.

## Q36. What is a learning-rate scheduler?

### Interview Answer

A learning-rate scheduler changes the learning rate during training.

Examples:

```text
Step Decay
Cosine Decay
Exponential Decay
Warmup
Reduce on Plateau
```

### Why?

A relatively larger learning rate can help early optimization, while smaller rates later can help fine-tune parameters.

Transformer training often uses warmup followed by a decay schedule.

### One-Line Answer

> A learning-rate scheduler changes the learning rate during training to improve optimization and convergence.

## Q37. What is gradient clipping?

### Interview Answer

Gradient clipping limits the magnitude of gradients before the optimizer applies updates.

For example, if the gradient norm becomes too large:

```text
Large Gradient
      ↓
Clip
      ↓
Controlled Gradient
      ↓
Optimizer
```

### Why?

Primarily to prevent unstable updates caused by exploding gradients.

### One-Line Answer

> Gradient clipping limits excessively large gradients to stabilize training.

---

## Q38. What is residual learning and why are skip connections useful?

### Interview Answer

Residual learning allows a layer or block to learn a residual function while providing a shortcut path for the input.

Instead of:

```text
x → F(x)
```

a residual block computes:

```text
x → F(x)
 \     ↓
  └→ Add → Output
```

Conceptually:

```text
Output = F(x) + x
```

### Benefits

- Easier optimization of deep networks
- Better gradient flow
- Enables very deep architectures

Residual connections are fundamental to architectures such as ResNet and also appear widely in Transformers.

### One-Line Answer

> Skip connections provide shortcut paths that improve gradient flow and make deep networks easier to optimize.

---

## Q39. What is data augmentation?

### Interview Answer

Data augmentation creates varied training examples from existing data while preserving the underlying label.

For images:

```text
Original Image
 ↓
Flip
Rotate
Crop
Resize
Color Variation
 ↓
Augmented Samples
```

For text, augmentation must preserve meaning and should be applied carefully.

### Benefits

- Reduces overfitting
- Improves robustness
- Effectively increases training diversity

### Important Point

Augmentation must be appropriate to the task. An augmentation that changes the label is harmful.

### One-Line Answer

> Data augmentation increases training diversity through label-preserving transformations to improve generalization.

---

## Q40. What is class imbalance in Deep Learning?

### Interview Answer

Class imbalance occurs when some classes have substantially fewer examples than others.

Example:

```text
Normal = 95%
Fraud = 5%
```

A deep classifier may become biased toward the majority class.

### Solutions

```text
Class-weighted loss
Oversampling
Targeted augmentation
Focal Loss
Threshold optimization
Appropriate evaluation metrics
```

### One-Line Answer

> Class imbalance can bias a deep-learning model toward majority classes and can be addressed through loss weighting, sampling, augmentation, and threshold strategies.

## Q41. How do you prevent overfitting in Deep Learning?

### Interview Answer

Common techniques include:

```text
More / better data
↓
Data Augmentation
↓
Dropout
↓
Weight Decay
↓
Early Stopping
↓
Smaller Model
↓
Transfer Learning
↓
Cross-Validation where appropriate
```

### Example

If training accuracy reaches:

```text
99%
```

but validation accuracy remains:

```text
82%
```

the model may be overfitting.

I would investigate data quality and leakage first, then consider regularization, augmentation, model capacity, and training strategy.

### One-Line Answer

> I control overfitting through better data, augmentation, regularization, appropriate model capacity, early stopping, and reliable validation.

---

## Q42. What is weight decay?

### Interview Answer

Weight decay is a regularization technique that discourages excessively large model weights.

Conceptually:

```text
Objective
=
Task Loss
+
Regularization
```

In modern optimizers such as AdamW, weight decay is decoupled from the gradient-based adaptive update.

### Benefit

It can improve generalization by discouraging unnecessarily large parameters.

### One-Line Answer

> Weight decay regularizes a model by discouraging large parameter values and can improve generalization.

---

## Q43. What is mixed-precision training?

### Interview Answer

Mixed-precision training uses different numerical precisions, commonly lower precision for many computations and higher precision where needed for stability.

Typical formats include:

```text
FP32
FP16
BF16
```

### Benefits

- Lower memory usage
- Faster computation on supported hardware
- Larger effective batch sizes in some cases

### Important Point

Numerical stability techniques such as loss scaling may be required for some FP16 workflows.

### One-Line Answer

> Mixed-precision training combines numerical precisions to reduce memory usage and accelerate training while maintaining sufficient numerical stability.

---

## Q44. What is transfer learning vs training from scratch?

### Interview Answer

### Transfer Learning

```text
Pretrained Model
 ↓
Fine-Tune
 ↓
Target Task
```

Advantages:

- Faster
- Less data required
- Often stronger starting point

### From Scratch

```text
Random Initialization
 ↓
Train Entire Model
```

Useful when:

- Domain differs substantially
- Suitable pretrained model is unavailable

### One-Line Answer

> Transfer learning starts from learned representations, while training from scratch learns the model entirely from randomly initialized parameters.

## Q46. How do you debug a Deep Learning model that is not learning?

### Interview Answer

I would use a systematic debugging process.

```text
Check Data
 ↓
Check Labels
 ↓
Check Shapes
 ↓
Check Preprocessing
 ↓
Check Loss Function
 ↓
Check Output Activation
 ↓
Check Learning Rate
 ↓
Check Gradients
 ↓
Try Small Dataset
 ↓
Overfit Tiny Batch
 ↓
Check Initialization
 ↓
Check Architecture
```

### Powerful Debugging Test

Try to overfit a very small dataset.

If the model cannot achieve near-perfect training performance on a tiny batch, there may be a problem with:

```text
Data
Labels
Loss
Model
Optimization
Implementation
```

### One-Line Answer

> I first verify data, labels, tensor shapes, loss, outputs, gradients, and optimization, then use the tiny-dataset overfitting test to isolate implementation problems.

## Q48. What is model quantization?

### Interview Answer

Quantization represents model parameters or computations using lower numerical precision.

For example:

```text
FP32 → INT8
```

### Benefits

- Smaller model size
- Lower memory usage
- Potentially faster inference
- Lower deployment cost

### Tradeoff

Aggressive quantization can reduce model accuracy.

### One-Line Answer

> Quantization reduces numerical precision to make models smaller and potentially faster and cheaper to serve, with some possible accuracy tradeoff.

## Q49. What is knowledge distillation?

### Interview Answer

Knowledge distillation trains a smaller student model using information from a larger teacher model.

```text
Large Teacher
     ↓
Soft Predictions / Knowledge
     ↓
Small Student
     ↓
Efficient Model
```

The student learns from the teacher's outputs in addition to, or sometimes alongside, the original labels.

### Benefits

- Smaller model
- Faster inference
- Lower memory requirements
- Better deployment efficiency

### One-Line Answer

> Knowledge distillation transfers useful behavior from a large teacher model to a smaller student model.

# ⭐ IMPORTANT DEEP LEARNING COMPARISONS

## Machine Learning vs Deep Learning

```text
Machine Learning
→ Often requires more manual feature engineering
→ Strong on structured/tabular data
→ Often lower compute requirement

Deep Learning
→ Learns hierarchical representations
→ Excellent for images, audio, text and complex unstructured data
→ Often requires more data and compute
```

---

## CNN vs RNN vs Transformer

```text
CNN
→ Spatial patterns
→ Images / spatial data

RNN
→ Sequential processing
→ Sequence data
→ Long dependencies can be difficult

Transformer
→ Attention-based
→ Strong sequence modeling
→ Highly parallelizable during training
```

---

## LSTM vs GRU

```text
LSTM
→ More gates
→ Separate cell state
→ More parameters

GRU
→ Fewer gates
→ Simpler
→ Fewer parameters
```

---

## Sigmoid vs Softmax

```text
Sigmoid
→ Independent probability for each output
→ Common for binary or multilabel classification

Softmax
→ Produces a probability distribution across classes
→ Common for mutually exclusive multiclass classification
```

---

## BatchNorm vs Dropout

```text
BatchNorm
→ Normalizes activations
→ Can stabilize optimization
→ Training/inference behavior differs

Dropout
→ Randomly removes activations during training
→ Regularization
```

---

## Epoch vs Batch vs Iteration

```text
Epoch
→ One pass through dataset

Batch
→ Subset of data

Iteration
→ Usually one batch update
```
