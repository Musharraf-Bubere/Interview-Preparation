# Deep Learning Interview Questions & Answers — 100 Questions

## Basic Level

### Q1. What is Deep Learning?

**Answer:**

Deep Learning is a subset of Machine Learning that uses artificial neural networks with multiple layers to automatically learn complex patterns from data.

Traditional machine learning often requires manually designed features, whereas deep learning can learn useful representations directly from raw or minimally processed data.

**Example:**

For image classification:

```text
Image
  ↓
Convolutional layers
  ↓
Feature extraction
  ↓
Fully connected layers
  ↓
Prediction
```

For example, a neural network can learn edges in early layers, shapes in middle layers, and complete objects in deeper layers.

**Interview Answer:**

> Deep Learning is a branch of machine learning that uses multi-layer neural networks to learn hierarchical representations from data. It is especially effective for unstructured data such as images, text, audio, and video.

---

### Q2. What is an Artificial Neural Network (ANN)?

**Answer:**

An Artificial Neural Network is a computational model inspired by biological neural networks. It consists of interconnected neurons organized into layers.

The main components are:

- Input layer
- Hidden layers
- Output layer
- Weights
- Biases
- Activation functions

A neuron generally computes:

```text
z = w1x1 + w2x2 + ... + wnxn + b
output = activation(z)
```

**Example:**

A neural network predicting whether a customer will churn could take:

```text
Age
Monthly charges
Contract duration
Support calls
      ↓
Neural Network
      ↓
Churn / No Churn
```

---

### Q3. What is a neuron in a neural network?

**Answer:**

A neuron is the basic computational unit of a neural network.

It receives input values, multiplies them by weights, adds a bias, and passes the result through an activation function.

```text
Inputs → Weighted Sum → Activation Function → Output
```

Mathematically:

```text
z = Σ(wixi) + b
a = f(z)
```

Where:

- `xi` = input
- `wi` = weight
- `b` = bias
- `f` = activation function
- `a` = neuron output

---

### Q4. What are weights in a neural network?

**Answer:**

Weights determine how important each input is to a neuron.

During training, the model adjusts these weights to reduce prediction error.

**Example:**

Suppose a house-price model receives:

```text
Area = 2000 sq ft
Bedrooms = 3
Location score = 8
```

The network learns weights representing how strongly each feature contributes to the prediction.

---

### Q5. What is bias in a neural network?

**Answer:**

Bias is an additional learnable parameter that allows a neuron to shift its activation function.

Without bias:

```text
z = wx
```

With bias:

```text
z = wx + b
```

Bias increases the flexibility of the model.

---

### Q6. What is an activation function?

**Answer:**

An activation function determines whether and how strongly a neuron should activate.

It introduces non-linearity into the network.

Common activation functions include:

- ReLU
- Sigmoid
- Tanh
- Softmax
- GELU

Without activation functions, multiple neural-network layers would effectively behave like a single linear transformation.

**Interview Answer:**

> Activation functions introduce non-linearity, allowing neural networks to learn complex relationships rather than only linear relationships.

---

### Q7. What is ReLU?

**Answer:**

ReLU stands for Rectified Linear Unit.

```text
ReLU(x) = max(0, x)
```

Therefore:

```text
x < 0 → 0
x > 0 → x
```

**Example:**

```text
ReLU(-5) = 0
ReLU(3) = 3
```

ReLU is widely used in hidden layers because it is computationally simple and generally helps optimization compared with saturating activations.

---

### Q8. What is the sigmoid activation function?

**Answer:**

Sigmoid maps a value to the range `(0, 1)`.

```text
σ(x) = 1 / (1 + e^-x)
```

It is commonly used for binary classification output.

**Example:**

If the model outputs:

```text
0.92
```

we can interpret this as a high estimated probability for the positive class, assuming the model is configured accordingly.

---

### Q9. What is the tanh activation function?

**Answer:**

Tanh maps values approximately to:

```text
(-1, 1)
```

Formula:

```text
tanh(x) = (e^x - e^-x) / (e^x + e^-x)
```

Unlike sigmoid, tanh is zero-centered.

However, it can still suffer from vanishing gradients for large positive or negative inputs.

---

### Q10. What is Softmax?

**Answer:**

Softmax converts a vector of logits into a probability distribution.

For class `i`:

```text
softmax(zi) = e^zi / Σe^zj
```

The probabilities sum to 1.

**Example:**

```text
Cat     = 0.70
Dog     = 0.20
Horse   = 0.10
```

Softmax is commonly used for multi-class classification.

---

### Q11. What is forward propagation?

**Answer:**

Forward propagation is the process of passing input data through the neural network to produce a prediction.

```text
Input
  ↓
Layer 1
  ↓
Activation
  ↓
Layer 2
  ↓
Activation
  ↓
Output
```

No parameter updates occur during forward propagation itself.

---

### Q12. What is backward propagation?

**Answer:**

Backpropagation is the algorithm used to calculate gradients of the loss with respect to model parameters.

It applies the chain rule of calculus to propagate error information backward through the network.

```text
Prediction
   ↓
Loss
   ↓
Gradients
   ↓
Weights updated
```

Backpropagation calculates gradients; an optimizer such as SGD or Adam uses those gradients to update parameters.

---

### Q13. What is a loss function?

**Answer:**

A loss function measures how different the model's prediction is from the actual target.

Examples:

- Mean Squared Error (MSE)
- Mean Absolute Error (MAE)
- Binary Cross-Entropy
- Categorical Cross-Entropy

**Example:**

For regression:

```text
Actual = 100
Predicted = 90
```

MSE contributes:

```text
(100 - 90)^2 = 100
```

The optimizer attempts to minimize the loss over training.

---

### Q14. What is an optimizer?

**Answer:**

An optimizer updates model parameters using gradients to minimize the loss.

Common optimizers include:

- SGD
- Momentum
- RMSprop
- Adam
- AdamW

A simplified gradient-descent update is:

```text
w_new = w_old - learning_rate × gradient
```

---

### Q15. What is Gradient Descent?

**Answer:**

Gradient Descent is an optimization algorithm that updates model parameters in the direction that reduces the loss.

```text
New parameter = Old parameter - Learning Rate × Gradient
```

The gradient indicates the direction of increasing loss, so moving in the opposite direction reduces it.

---

### Q16. What is learning rate?

**Answer:**

Learning rate controls the size of parameter updates.

If it is too large:

```text
Loss may oscillate or diverge
```

If it is too small:

```text
Training can become extremely slow
```

A suitable learning rate is important for stable and efficient training.

---

### Q17. What is an epoch?

**Answer:**

One epoch means the model has processed the entire training dataset once.

**Example:**

If there are 10,000 training examples and:

```text
epochs = 20
```

the model processes the training dataset 20 times, subject to the batching setup.

---

### Q18. What is a batch?

**Answer:**

A batch is a subset of the training data processed together before an optimizer update.

For example:

```text
Dataset = 10,000 samples
Batch size = 100
```

Approximately 100 batches are processed per epoch.

---

### Q19. What is batch size?

**Answer:**

Batch size specifies the number of samples processed in one training step.

Common values include:

```text
16
32
64
128
```

Larger batches can improve hardware utilization but require more memory.

---

### Q20. What is an iteration or training step?

**Answer:**

One training step usually consists of:

```text
Take one batch
    ↓
Forward pass
    ↓
Calculate loss
    ↓
Backpropagation
    ↓
Optimizer update
```

The number of steps per epoch is approximately:

```text
ceil(number_of_training_samples / batch_size)
```

---

## Intermediate Level

### Q21. What is overfitting in deep learning?

**Answer:**

Overfitting occurs when a model learns the training data too closely, including noise or accidental patterns, and performs poorly on unseen data.

Typical pattern:

```text
Training loss ↓
Validation loss ↓ initially
Validation loss ↑ later
```

Solutions include:

- More data
- Data augmentation
- Dropout
- Weight decay
- Early stopping
- Smaller model
- Cross-validation where appropriate

---

### Q22. What is underfitting?

**Answer:**

Underfitting occurs when a model is too simple or insufficiently trained to capture the underlying patterns.

Typical symptoms:

```text
Training performance = poor
Validation performance = poor
```

Possible solutions:

- Increase model capacity
- Train longer
- Improve features or representations
- Reduce excessive regularization
- Improve data quality

---

### Q23. What is the bias-variance tradeoff?

**Answer:**

Bias represents error caused by overly simplistic assumptions.

Variance represents sensitivity to the particular training data.

```text
High Bias → Underfitting
High Variance → Overfitting
```

A good model balances both.

---

### Q24. What is dropout?

**Answer:**

Dropout is a regularization technique in which randomly selected activations are dropped during training.

Example:

```text
Before:
[1.2, 0.8, 2.1, 1.4]

After dropout:
[1.2, 0, 2.1, 0]
```

This reduces reliance on particular neurons and can improve generalization.

Dropout is normally disabled during inference.

---

### Q25. What is batch normalization?

**Answer:**

Batch Normalization normalizes activations using statistics computed from mini-batches during training, followed by learnable scale and shift parameters.

A simplified form is:

```text
x_normalized = (x - mean) / sqrt(variance + epsilon)
output = gamma * x_normalized + beta
```

It can improve optimization and make training more stable.

---

### Q26. What is normalization in deep learning?

**Answer:**

Normalization generally refers to transforming values to a controlled numerical range or distribution.

For input features, common approaches include:

```text
Standardization:
x' = (x - mean) / standard_deviation
```

or min-max scaling:

```text
x' = (x - min) / (max - min)
```

Proper input scaling can make optimization easier.

---

### Q27. What is vanishing gradient?

**Answer:**

Vanishing gradients occur when gradients become extremely small as they are propagated backward through many layers.

As a result, early layers may learn very slowly.

It is associated particularly with saturating activations such as sigmoid and tanh in deep networks.

Possible solutions include:

- ReLU-family activations
- Proper initialization
- Residual connections
- Normalization techniques
- Suitable architectures

---

### Q28. What is exploding gradient?

**Answer:**

Exploding gradients occur when gradients become extremely large during training.

This can cause:

- Unstable training
- Extremely large parameter updates
- NaN losses

A common solution is gradient clipping.

---

### Q29. What is gradient clipping?

**Answer:**

Gradient clipping limits the magnitude of gradients before the optimizer updates parameters.

Example:

```text
If gradient norm > threshold
→ scale gradient down
```

It is particularly useful for stabilizing recurrent neural networks and other models where exploding gradients can occur.

---

### Q30. What is weight initialization?

**Answer:**

Weight initialization determines the initial values of trainable weights.

Poor initialization can cause:

- Vanishing gradients
- Exploding activations
- Slow convergence

Common methods include:

- Xavier/Glorot initialization
- He initialization

He initialization is commonly used with ReLU-family activations.

---

### Q31. What is Xavier initialization?

**Answer:**

Xavier/Glorot initialization is designed to keep the variance of activations and gradients reasonably controlled across layers.

It is commonly associated with sigmoid, tanh, and related activation settings.

---

### Q32. What is He initialization?

**Answer:**

He initialization is designed particularly for networks using ReLU-like activations.

It uses a variance related to the number of input units, helping maintain activation scale through layers.

---

### Q33. What is regularization?

**Answer:**

Regularization reduces overfitting by discouraging overly complex solutions.

Common approaches:

- L1 regularization
- L2 regularization / weight decay
- Dropout
- Data augmentation
- Early stopping

---

### Q34. What is L1 regularization?

**Answer:**

L1 adds the absolute values of weights to the objective:

```text
Loss_total = Loss + λΣ|w|
```

It can encourage some weights to become exactly zero, producing sparse solutions.

---

### Q35. What is L2 regularization?

**Answer:**

L2 regularization penalizes large weights:

```text
Loss_total = Loss + λΣw²
```

It generally encourages smaller parameter values and can improve generalization.

---

### Q36. What is early stopping?

**Answer:**

Early stopping stops training when validation performance stops improving.

Example:

```text
Epoch 10 → validation loss = 0.35
Epoch 20 → validation loss = 0.29
Epoch 30 → validation loss = 0.30
Epoch 40 → validation loss = 0.34
```

The best checkpoint may be around epoch 20.

---

### Q37. What is a CNN?

**Answer:**

CNN stands for Convolutional Neural Network.

CNNs are especially effective for spatial data such as images.

Typical architecture:

```text
Image
 ↓
Convolution
 ↓
Activation
 ↓
Pooling
 ↓
Convolution
 ↓
Pooling
 ↓
Flatten / Global Pooling
 ↓
Dense
 ↓
Output
```

---

### Q38. What is convolution?

**Answer:**

Convolution applies a learnable filter/kernel across an input to detect local patterns.

For images, early filters may learn patterns resembling:

```text
Edges
Corners
Textures
```

Deeper layers can combine these into higher-level structures.

---

### Q39. What is a kernel/filter in CNN?

**Answer:**

A kernel is a small matrix of learnable weights that slides across the input.

Example:

```text
3 × 3 kernel
```

The kernel interacts with local pixels to produce a feature map.

During training, the network learns useful filters automatically.

---

### Q40. What is a feature map?

**Answer:**

A feature map is the output produced by applying a convolutional filter to an input.

Different filters can learn different patterns.

For example:

```text
Filter 1 → vertical edges
Filter 2 → horizontal edges
Filter 3 → textures
```

---

### Q41. What is padding in CNN?

**Answer:**

Padding adds extra values, often zeros, around the input before convolution.

Common types:

```text
Valid → no padding
Same → padding chosen to preserve spatial dimensions under the usual stride-1 setup
```

Padding helps control output dimensions and allows border information to participate more fully.

---

### Q42. What is stride?

**Answer:**

Stride determines how far the convolution kernel moves at each step.

For example:

```text
Stride = 1 → move one pixel
Stride = 2 → move two pixels
```

Larger stride usually reduces spatial dimensions.

---

### Q43. What is pooling?

**Answer:**

Pooling reduces spatial dimensions while retaining important information.

Common types:

- Max pooling
- Average pooling

Example:

```text
2 × 2 Max Pooling

[1  5]
[3  2]

Output = 5
```

---

### Q44. What is transfer learning?

**Answer:**

Transfer learning uses knowledge learned from one task or dataset as a starting point for another related task.

Example:

```text
ImageNet-pretrained CNN
        ↓
Fine-tune on
medical image classification
```

It is especially useful when the target dataset is relatively small.

---

### Q45. What is fine-tuning?

**Answer:**

Fine-tuning means adapting a pretrained model to a new task by training some or all of its parameters on the target dataset.

A common strategy is:

```text
Freeze most layers
↓
Train new task-specific head
↓
Unfreeze selected deeper layers
↓
Train with a small learning rate
```

---

### Q46. What is feature extraction in transfer learning?

**Answer:**

Feature extraction uses a pretrained network as a fixed representation generator.

The pretrained layers are frozen and only a new task-specific head is trained.

```text
Pretrained model → Features → New classifier
```

---

### Q47. What is data augmentation?

**Answer:**

Data augmentation creates varied training examples by applying transformations that preserve the target semantics.

For images:

- Rotation
- Cropping
- Flipping
- Translation
- Scaling
- Color adjustments

It can improve generalization and reduce overfitting.

---

### Q48. What is a validation set?

**Answer:**

A validation set is used during model development to:

- Tune hyperparameters
- Compare architectures
- Perform early stopping
- Select checkpoints

It should not be treated as the final unbiased evaluation set.

---

### Q49. What is a test set?

**Answer:**

The test set is reserved for final evaluation after model development decisions have been made.

It estimates how well the final selected model generalizes to unseen data.

---

### Q50. Why should we not repeatedly tune a model on the test set?

**Answer:**

Repeatedly using the test set for decisions causes information leakage.

The test set gradually becomes part of the development process, making reported performance overly optimistic.

A cleaner workflow is:

```text
Train → Validation → Final Test
```

---

## Advanced Level

### Q51. What is an RNN?

**Answer:**

RNN stands for Recurrent Neural Network.

RNNs process sequential data while maintaining a hidden state that carries information from previous time steps.

```text
x1 → RNN → h1
          ↓
x2 → RNN → h2
          ↓
x3 → RNN → h3
```

They have historically been used for:

- Time series
- Language
- Speech
- Sequential signals

---

### Q52. What is the problem with vanilla RNNs?

**Answer:**

Vanilla RNNs can struggle with long-term dependencies because repeated multiplication through time can cause gradients to vanish or explode.

As a result, they may have difficulty remembering information from far earlier in a sequence.

---

### Q53. What is LSTM?

**Answer:**

LSTM stands for Long Short-Term Memory.

It is a type of recurrent neural network designed to better preserve useful information over long sequences.

It uses gates such as:

- Forget gate
- Input gate
- Output gate

These gates regulate information flow through the cell.

---

### Q54. What is GRU?

**Answer:**

GRU stands for Gated Recurrent Unit.

It is another gated recurrent architecture.

Compared with LSTM, GRU has a simpler structure and fewer gates/parameters while still helping address long-term dependency problems.

---

### Q55. LSTM vs GRU?

**Answer:**

| LSTM | GRU |
|---|---|
| More complex | Simpler |
| More parameters | Fewer parameters |
| Uses cell state and hidden state | Uses a combined state |
| Can be useful for complex sequence dependencies | Often faster and efficient |

The best choice depends on the dataset, task, and computational constraints.

---

### Q56. What is attention?

**Answer:**

Attention allows a model to assign different importance to different parts of an input when producing an output.

Instead of treating every token equally, the model learns which tokens are most relevant.

Example:

```text
"The animal didn't cross the road because it was tired."
```

Attention helps the model determine which earlier words are relevant to interpreting "it".

---

### Q57. What is self-attention?

**Answer:**

Self-attention calculates relationships between tokens within the same sequence.

For each token, the model creates:

```text
Query (Q)
Key (K)
Value (V)
```

A simplified attention equation is:

```text
Attention(Q,K,V)
= softmax(QKᵀ / √dk)V
```

---

### Q58. What are Query, Key, and Value?

**Answer:**

In self-attention:

- Query represents what the current token is looking for.
- Key represents what each token offers for matching.
- Value contains the information that will be aggregated.

The similarity between Query and Key determines how much attention each Value receives.

---

### Q59. What is a Transformer?

**Answer:**

A Transformer is a neural-network architecture based primarily on attention mechanisms rather than recurrence.

A typical Transformer block contains:

```text
Multi-Head Self-Attention
        ↓
Add & Norm
        ↓
Feed-Forward Network
        ↓
Add & Norm
```

Transformers are widely used in:

- NLP
- Computer vision
- Speech
- Multimodal AI
- Large language models

---

### Q60. Why are Transformers better suited to parallel training than RNNs?

**Answer:**

RNNs process tokens sequentially because the hidden state at time `t` depends on time `t-1`.

Transformers can process the tokens of a sequence largely in parallel during training because self-attention operates over the sequence collectively.

This makes Transformers much more scalable on modern accelerators.

---

### Q61. What is multi-head attention?

**Answer:**

Multi-head attention runs several attention operations in parallel.

Different heads can learn different relationships.

Conceptually:

```text
Input
 ↓
Head 1 → relationship type A
Head 2 → relationship type B
Head 3 → relationship type C
 ↓
Concatenate
 ↓
Linear projection
```

---

### Q62. What is positional encoding?

**Answer:**

Self-attention itself does not inherently know token order.

Positional information is therefore added to token representations.

Traditional Transformers used sinusoidal positional encodings, while many modern architectures use learned or rotary positional representations.

The purpose is to provide information about token position.

---

### Q63. What is an embedding?

**Answer:**

An embedding is a dense numerical representation of an object such as a word, token, sentence, image, or user.

Example:

```text
"king" → [0.21, -0.45, 0.83, ...]
```

Similar objects often have representations that are closer in embedding space.

Embeddings are widely used in:

- NLP
- Search
- Recommendation systems
- RAG
- Classification

---

### Q64. What is an autoencoder?

**Answer:**

An autoencoder is a neural network that learns to reconstruct its input.

Architecture:

```text
Input
  ↓
Encoder
  ↓
Latent representation
  ↓
Decoder
  ↓
Reconstructed input
```

It can be used for:

- Dimensionality reduction
- Representation learning
- Denoising
- Anomaly detection

---

### Q65. What is a variational autoencoder (VAE)?

**Answer:**

A VAE is a generative model that learns a probabilistic latent representation.

Instead of mapping an input to a single deterministic latent vector, the encoder learns parameters of a distribution, typically a mean and variance.

The objective combines:

```text
Reconstruction loss
+
KL divergence
```

The model can then sample latent vectors and generate new data.

---

### Q66. What is a GAN?

**Answer:**

GAN stands for Generative Adversarial Network.

It contains two neural networks:

```text
Generator → creates synthetic data
Discriminator → distinguishes real from generated data
```

They are trained in an adversarial process.

Example:

```text
Random noise
    ↓
Generator
    ↓
Fake image
    ↓
Discriminator
    ↓
Real / Fake
```

---

### Q67. What is the generator in GAN?

**Answer:**

The generator learns to create synthetic samples that resemble real training data.

It receives a latent noise vector and maps it into the data space.

```text
Noise z → Generator → Synthetic sample
```

---

### Q68. What is the discriminator in GAN?

**Answer:**

The discriminator attempts to distinguish real samples from samples generated by the generator.

During adversarial training, the generator tries to produce increasingly realistic samples.

---

### Q69. What is mode collapse in GANs?

**Answer:**

Mode collapse occurs when the generator produces limited varieties of samples instead of covering the diversity of the real data distribution.

For example, a face generator might repeatedly generate faces with very similar characteristics.

It is a major GAN training challenge.

---

### Q70. What is a learning-rate scheduler?

**Answer:**

A learning-rate scheduler changes the learning rate during training.

Examples include:

- Step decay
- Exponential decay
- Cosine annealing
- Reduce-on-plateau
- Warmup followed by decay

A common strategy is to start with a suitable learning rate and gradually reduce it as training progresses.

---

### Q71. What is Adam optimizer?

**Answer:**

Adam combines ideas related to momentum and adaptive learning rates.

It maintains moving estimates of:

```text
First moment → gradient mean
Second moment → squared-gradient mean
```

This allows different parameters to receive adaptive updates.

Adam is widely used because it often provides fast and stable optimization with relatively little tuning.

---

### Q72. What is AdamW?

**Answer:**

AdamW is a variant of Adam that decouples weight decay from the adaptive gradient update.

This gives more direct control over weight decay and is commonly used in modern deep-learning training, especially for Transformer-based models.

---

### Q73. What is cross-entropy loss?

**Answer:**

Cross-entropy measures the difference between the target distribution and predicted probability distribution.

For a single multi-class example:

```text
Loss = -Σ yi log(pi)
```

where:

- `yi` = target indicator
- `pi` = predicted probability

For a one-hot target, the loss mainly depends on the probability assigned to the correct class.

---

### Q74. Binary cross-entropy vs categorical cross-entropy?

**Answer:**

**Binary Cross-Entropy:**

Used for binary classification or independent binary labels.

```text
Example:
Spam / Not Spam
```

**Categorical Cross-Entropy:**

Commonly used for single-label multi-class classification.

```text
Cat / Dog / Horse
```

For multiple independent labels, binary cross-entropy is often appropriate even when there are more than two labels.

---

### Q75. What is class imbalance in deep learning?

**Answer:**

Class imbalance occurs when some classes have substantially more examples than others.

Example:

```text
Normal transactions = 99%
Fraud transactions = 1%
```

Accuracy alone can be misleading.

Possible solutions:

- Class-weighted loss
- Oversampling
- Undersampling
- Data augmentation
- Focal loss
- Threshold tuning
- Precision/Recall-based evaluation

---

### Q76. What is focal loss?

**Answer:**

Focal loss modifies cross-entropy to reduce the contribution of easy examples and focus training more on difficult examples.

It is particularly useful in highly imbalanced classification tasks, such as object detection.

---

### Q77. What is precision in classification?

**Answer:**

Precision answers:

> Of all samples predicted as positive, how many were actually positive?

```text
Precision = TP / (TP + FP)
```

High precision means relatively few false positives.

---

### Q78. What is recall?

**Answer:**

Recall answers:

> Of all actual positive samples, how many did the model identify?

```text
Recall = TP / (TP + FN)
```

High recall means relatively few false negatives.

---

### Q79. What is F1-score?

**Answer:**

F1-score is the harmonic mean of precision and recall.

```text
F1 = 2 × Precision × Recall
     -------------------------
       Precision + Recall
```

It is useful when both false positives and false negatives matter.

---

### Q80. What is data leakage in deep learning?

**Answer:**

Data leakage occurs when information unavailable at prediction time influences training or model selection.

Examples:

- Using future information
- Normalizing before splitting data
- Duplicates appearing across train and validation sets
- Using test labels during tuning

Leakage can produce unrealistically high evaluation performance.

---

### Q81. What is distributed training?

**Answer:**

Distributed training uses multiple GPUs or machines to train a model.

Common approaches include:

- Data parallelism
- Model parallelism
- Pipeline parallelism

The goal is to train large models or datasets more efficiently.

---

### Q82. What is data parallelism?

**Answer:**

In data parallelism, each device has a copy of the model but processes different batches of data.

Conceptually:

```text
              Model
                ↓
      ┌─────────┼─────────┐
     GPU1      GPU2      GPU3
   Batch A    Batch B    Batch C
      └─────────┼─────────┘
          Gradient sync
                ↓
         Model update
```

---

### Q83. What is model parallelism?

**Answer:**

Model parallelism divides the model itself across devices.

For example:

```text
GPU 1 → Layers 1–10
GPU 2 → Layers 11–20
GPU 3 → Layers 21–30
```

It is useful when the model is too large to fit into one device's memory.

---

### Q84. What is mixed precision training?

**Answer:**

Mixed precision training uses different numerical precisions, such as FP16/BF16 and FP32, during training.

Benefits can include:

- Lower memory usage
- Faster computation
- Better accelerator utilization

Careful scaling or modern framework support is used to maintain numerical stability.

---

### Q85. What is quantization?

**Answer:**

Quantization reduces the numerical precision used to represent model parameters and/or activations.

For example:

```text
FP32 → INT8
```

Benefits:

- Smaller model
- Lower memory usage
- Faster inference on supported hardware

There can be a tradeoff with accuracy.

---

### Q86. What is pruning?

**Answer:**

Pruning removes or reduces the importance of selected model parameters or structures.

Example:

```text
Dense model
   ↓
Remove low-importance weights
   ↓
Sparse/smaller model
```

It can reduce model size and potentially inference cost, depending on hardware and implementation.

---

### Q87. What is knowledge distillation?

**Answer:**

Knowledge distillation trains a smaller **student** model using information from a larger **teacher** model.

```text
Large Teacher
      ↓
Soft predictions
      ↓
Small Student
```

The student attempts to retain much of the teacher's useful behavior while being cheaper to run.

---

### Q88. What are logits?

**Answer:**

Logits are raw model outputs before applying an output transformation such as softmax.

Example:

```text
Logits:
[2.5, 1.0, -0.5]

Softmax:
[0.79, 0.18, 0.03]  approximately
```

Cross-entropy implementations often accept logits directly for numerical stability.

---

### Q89. What is a confusion matrix?

**Answer:**

A confusion matrix summarizes classification predictions.

For binary classification:

```text
                 Predicted
              Positive Negative

Actual Pos       TP       FN
Actual Neg       FP       TN
```

It is used to derive:

- Accuracy
- Precision
- Recall
- F1-score
- Specificity

---

### Q90. What is ROC-AUC?

**Answer:**

ROC-AUC measures the ability of a classifier to rank positive examples above negative examples across classification thresholds.

The ROC curve plots:

```text
True Positive Rate
vs
False Positive Rate
```

AUC summarizes the area under this curve.

## Expert Level

### Q91. What is the difference between generative and discriminative models?

**Answer:**

Discriminative models focus on predicting labels or decision boundaries.

```text
P(y | x)
```

Generative models attempt to model how data can be generated, often involving the data distribution itself.

Examples:

```text
Discriminative:
Logistic Regression
CNN classifier

Generative:
VAE
GAN
Diffusion model
Language model
```

---

### Q92. What is a diffusion model?

**Answer:**

A diffusion model is a generative model trained around a process that gradually adds noise to data and learns to reverse that process.

Conceptually:

```text
Real image
   ↓
Add noise repeatedly
   ↓
Nearly random noise

Then learn:

Noise
  ↓
Denoising steps
  ↓
Generated image
```

Diffusion models are widely used for image, audio, video, and other generative tasks.

---

### Q93. What is self-supervised learning?

**Answer:**

Self-supervised learning creates training signals from the data itself rather than requiring manually labeled examples for every training instance.

Examples:

- Predicting masked tokens
- Predicting missing portions of data
- Contrastive learning

It is important for learning representations from large unlabeled datasets.

---

### Q94. What is contrastive learning?

**Answer:**

Contrastive learning learns representations by bringing related examples closer together in embedding space and pushing unrelated examples farther apart.

Conceptually:

```text
Positive pair   → closer
Negative pair   → farther
```

It is widely used in modern representation learning.

---

### Q95. What is a foundation model?

**Answer:**

A foundation model is a large pretrained model trained on broad data that can be adapted to many downstream tasks.

Examples include large language models and multimodal models.

The general workflow is:

```text
Large-scale pretraining
        ↓
General representation
        ↓
Fine-tuning / prompting / adaptation
        ↓
Specific applications
```

---

### Q96. What is fine-tuning vs feature extraction?

**Answer:**

**Feature extraction:**

```text
Pretrained layers → Frozen
New head → Train
```

**Fine-tuning:**

```text
Pretrained layers → Some/all trainable
New head → Train
```

Feature extraction is often safer with small datasets, while fine-tuning can achieve better task-specific performance when enough suitable data is available.

### Q97. How would you debug a deep-learning model that has poor performance?

**Answer:**

I would follow a systematic process:

```text
1. Verify data quality
2. Check labels
3. Check train/validation/test split
4. Check preprocessing
5. Check for leakage
6. Build a simple baseline
7. Verify the loss function
8. Check output activation
9. Check learning rate
10. Monitor training/validation curves
11. Check class imbalance
12. Inspect predictions
13. Tune architecture
14. Tune regularization
15. Evaluate on an untouched test set
```

I would avoid changing many things simultaneously because that makes debugging difficult.

---

### Q98. How would you handle a model that has high training accuracy but low validation accuracy?

**Answer:**

This is a classic sign of overfitting.

I would investigate:

```text
Training accuracy ↑
Validation accuracy ↓
```

Potential actions:

- Add or improve data augmentation
- Increase regularization
- Use dropout where appropriate
- Apply weight decay
- Use early stopping
- Reduce model complexity
- Collect more representative data
- Check for train/validation distribution differences
- Check for duplicate leakage

I would compare validation curves rather than relying on one metric from one epoch.

### Q99. How would you deploy a deep-learning model in production?

**Answer:**

A typical workflow is:

```text
Train model
   ↓
Evaluate model
   ↓
Save/version model
   ↓
Create inference pipeline
   ↓
Expose API
   ↓
Containerize
   ↓
Deploy
   ↓
Monitor
```

For example:

```text
Client
  ↓
FastAPI
  ↓
Preprocessing
  ↓
Deep Learning Model
  ↓
Postprocessing
  ↓
Prediction
```

Production monitoring should include:

- Latency
- Throughput
- Error rates
- Resource usage
- Data drift
- Prediction quality when labels become available
- Model/version tracking

### Q100. Explain a complete deep-learning project end-to-end as you would in an interview.

**Answer:**

A strong answer should explain the project as a complete ML lifecycle rather than only describing the neural network.

Example:

> I first understood the business problem and defined the target and evaluation metric. Then I collected and inspected the data, handled missing values and duplicates, and performed exploratory analysis. I split the data appropriately into training, validation, and test sets while avoiding leakage.
>
> Next, I prepared the inputs using the preprocessing required by the model. I selected a suitable architecture based on the data type. During training, I used an appropriate loss function and optimizer and monitored both training and validation performance.
>
> I handled overfitting using techniques such as regularization, data augmentation, early stopping, or transfer learning where appropriate. I evaluated the final model using task-specific metrics rather than relying only on accuracy.
>
> After selecting the best model using validation performance, I evaluated it once on the untouched test set. I then saved and versioned the model, built the inference pipeline, exposed it through an API, containerized the application, and deployed it. Finally, I monitored inference latency, errors, data drift, and model performance after deployment.

**Example project architecture:**

```text
Raw Data
   ↓
Data Validation
   ↓
Preprocessing
   ↓
Train / Validation / Test Split
   ↓
Model Architecture
   ↓
Training
   ↓
Validation
   ↓
Hyperparameter Tuning
   ↓
Final Evaluation
   ↓
Model Versioning
   ↓
Inference API
   ↓
Docker
   ↓
Deployment
   ↓
Monitoring
```
