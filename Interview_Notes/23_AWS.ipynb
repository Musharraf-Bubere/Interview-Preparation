## Q1. What is AWS?

### Interview Answer

AWS stands for Amazon Web Services.

It is a cloud computing platform provided by Amazon that offers services for computing, storage, databases, networking, security, monitoring, AI, and application deployment.

Instead of managing physical servers ourselves, we can use AWS services based on the application's requirements.

### Example

    Application
        ↓
       AWS
        ↓
    Compute + Storage + Database + Networking

### Memory Trick

**AWS = Cloud Services**

### One-Line Answer

AWS is Amazon's cloud platform that provides services for building, deploying, and managing applications.

---

## Q2. What is cloud computing?

### Interview Answer

Cloud computing means using computing resources such as servers, storage, databases, networking, and other services over the internet.

Instead of purchasing and maintaining physical infrastructure, organizations can use resources provided by cloud providers such as AWS.

### Example

    Traditional:
    Company → Physical Servers → Maintenance

    Cloud:
    Company → AWS → Required Resources

### Memory Trick

**Cloud = Resources over Internet**

### One-Line Answer

Cloud computing provides computing resources and services over the internet without requiring organizations to manage all physical infrastructure themselves.

---

## Q3. What are the main benefits of AWS?

### Interview Answer

Some important benefits of AWS are:

- Scalability
- Flexibility
- High availability
- Large number of managed services
- Global infrastructure
- Security and access-control features
- Monitoring and logging
- Pay-as-you-go pricing for many services

The exact benefits depend on which AWS services are used.

### Memory Trick

**Scale + Flexibility + Availability + Managed Services**

### One-Line Answer

AWS provides scalable, flexible, and managed cloud services for building and operating applications.

---

## Q4. What is an AWS Region?

### Interview Answer

An AWS Region is a geographical area that contains multiple AWS infrastructure locations.

When deploying an application, we can select a suitable region based on factors such as latency, availability, compliance, and data-location requirements.

### Example

    AWS
      ↓
    Region
      ↓
    Multiple Availability Zones

### Memory Trick

**Region = Geographic Area**

### One-Line Answer

An AWS Region is a geographical area containing AWS infrastructure where cloud resources can be deployed.

---

## Q5. What is an Availability Zone?

### Interview Answer

An Availability Zone, or AZ, is an isolated location within an AWS Region.

A region normally contains multiple Availability Zones.

Using multiple Availability Zones can improve application availability because the application does not have to depend on a single infrastructure location.

### Example

    Region
    ├── AZ-1
    ├── AZ-2
    └── AZ-3

### Memory Trick

**Region → Multiple AZs**

### One-Line Answer

An Availability Zone is an isolated infrastructure location within an AWS Region.

---

## Q6. What is EC2?

### Interview Answer

Amazon EC2 stands for Elastic Compute Cloud.

It provides virtual servers in the AWS cloud.

We can choose the operating system, instance type, storage, networking configuration, and other settings according to application requirements.

### Example

    AWS
      ↓
    EC2
      ↓
    Virtual Machine
      ↓
    Application

### Memory Trick

**EC2 = Cloud Virtual Machine**

### One-Line Answer

EC2 is AWS's service for running configurable virtual servers in the cloud.

---

## Q7. What is the difference between EC2 and a Docker container?

### Interview Answer

EC2 provides a virtual machine.

Docker provides containerization, where an application and its dependencies are packaged into a container.

We can actually use Docker containers inside an EC2 instance.

So they are not necessarily alternatives.

### Example

    AWS EC2
       ↓
    Virtual Machine
       ↓
    Docker
       ↓
    Container
       ↓
    Application

### Memory Trick

**EC2 = VM**

**Docker = Container**

### One-Line Answer

EC2 provides a virtual machine, while Docker packages applications into containers, and Docker can run inside EC2.

## Q8. What is S3?

### Interview Answer

Amazon S3 stands for Simple Storage Service.

It is an object storage service used to store files and other unstructured data.

It can be used for documents, images, videos, backups, logs, datasets, and generated files.

### Example

    Application
        ↓
       S3
        ↓
      Bucket
        ↓
    Files / Objects

### Memory Trick

**S3 = Object Storage**

### One-Line Answer

S3 is AWS's object storage service for storing files and other unstructured data.

---

## Q9. What is an S3 bucket?

### Interview Answer

An S3 bucket is a container used to store objects in Amazon S3.

For example, an application could have a bucket for uploaded documents and another for generated reports.

Objects inside a bucket can be organized using keys that represent their paths or names.

### Example

    Bucket
    ├── resumes/
    │   ├── resume1.pdf
    │   └── resume2.pdf
    │
    └── reports/
        └── report.pdf

### Memory Trick

**Bucket = Container for S3 Objects**

### One-Line Answer

An S3 bucket is a logical container used to store objects such as files in Amazon S3.

---

## Q10. What is RDS?

### Interview Answer

Amazon RDS stands for Relational Database Service.

It is a managed service for running relational databases.

It supports database engines such as PostgreSQL, MySQL, MariaDB, Oracle, and SQL Server.

AWS manages many infrastructure and operational tasks while the application continues to use the database normally.

### Example

    FastAPI
       ↓
      RDS
       ↓
    PostgreSQL

### Memory Trick

**RDS = Managed Relational Database**

### One-Line Answer

RDS is AWS's managed relational database service.

---

## Q11. What is the difference between RDS and S3?

### Interview Answer

RDS is designed for relational structured data and supports SQL databases.

S3 is object storage designed for files and other unstructured data.

For example:

    PostgreSQL Data
        → RDS

    PDF / Image / CSV File
        → S3

### Memory Trick

**RDS = Database**

**S3 = Files**

### One-Line Answer

RDS stores structured relational data, while S3 stores objects such as files and documents.

---

## Q12. What is AWS Lambda?

### Interview Answer

AWS Lambda is a serverless compute service.

It allows us to run code without managing servers directly.

A Lambda function runs in response to events or requests and AWS manages the underlying infrastructure.

It is useful for short-running event-driven workloads.

### Example

    Event
      ↓
    Lambda
      ↓
    Python Function
      ↓
    Result

### Memory Trick

**Lambda = Run Code Without Managing Servers**

### One-Line Answer

AWS Lambda is a serverless service that runs code in response to events or requests without requiring direct server management.

---

## Q13. What is serverless computing?

### Interview Answer

Serverless computing means using a cloud service where the provider manages the underlying server infrastructure.

It does not mean that servers do not exist.

The developer focuses mainly on application code and configuration while the cloud provider handles much of the infrastructure management.

AWS Lambda is an example of a serverless service.

### Memory Trick

**Serverless ≠ No Servers**

**Serverless = No Server Management**

### One-Line Answer

Serverless means the cloud provider manages the underlying infrastructure while developers focus mainly on application code.

---

## Q14. What is IAM in AWS?

### Interview Answer

IAM stands for Identity and Access Management.

It controls who can access AWS resources and what actions they are allowed to perform.

IAM uses concepts such as users, roles, policies, and permissions.

### Example

    Identity
       ↓
     Policy
       ↓
    Permissions
       ↓
    AWS Resource

### Memory Trick

**IAM = Who Can Do What**

### One-Line Answer

AWS IAM manages identities and controls access to AWS resources through permissions and policies.

---

## Q15. What is an IAM role?

### Interview Answer

An IAM role is an identity with a set of permissions that can be assumed by trusted entities such as AWS services, applications, or users.

Roles are commonly used when an application needs to access another AWS service.

For example, an EC2 instance or Lambda function can use an IAM role instead of storing long-term access keys in application code.

### Memory Trick

**Role = Temporary / Assumed Permissions**

### One-Line Answer

An IAM role provides permissions that can be assumed by an AWS service, application, or authorized identity.

---

## Q16. What is the principle of least privilege in AWS?

### Interview Answer

Least privilege means granting only the permissions that are required to perform a task.

For example, if an application only needs to read files from a specific S3 bucket, it should not receive full S3 administrative access.

This reduces the impact of accidental or unauthorized actions.

### Memory Trick

**Minimum Required Access**

### One-Line Answer

Least privilege means giving an identity only the permissions it actually needs.

---

## Q17. What is a VPC?

### Interview Answer

VPC stands for Virtual Private Cloud.

It is a logically isolated network environment in AWS where we can deploy and control cloud resources.

A VPC allows us to configure networking components such as subnets, route tables, security groups, and network access rules.

### Example

    AWS
      ↓
     VPC
      ↓
    ┌───────────┬───────────┐
    ↓           ↓
    Subnet     Subnet
    ↓           ↓
    EC2        Database

### Memory Trick

**VPC = Private Cloud Network**

### One-Line Answer

A VPC is a logically isolated network environment used to control networking for AWS resources.

---

## Q18. What is a security group?

### Interview Answer

A security group is a virtual firewall associated with AWS resources such as EC2 instances.

It controls inbound and outbound network traffic based on rules.

For example, we could allow HTTP or HTTPS traffic to a web server while restricting other unnecessary access.

### Memory Trick

**Security Group = Network Firewall**

### One-Line Answer

A security group acts as a virtual firewall that controls network traffic to and from supported AWS resources.

---

## Q19. What is CloudWatch?

### Interview Answer

Amazon CloudWatch is a monitoring and observability service.

It can collect and provide access to metrics, logs, and other operational information.

It can be used to monitor things such as:

- CPU usage
- Request counts
- Latency
- Errors
- Application logs
- Resource behavior

It can also be used to create alarms based on defined conditions.

### Memory Trick

**CloudWatch = Metrics + Logs + Alarms**

### One-Line Answer

CloudWatch is AWS's monitoring and observability service for metrics, logs, and alarms.

---

## Q20. What is Auto Scaling?

### Interview Answer

AWS Auto Scaling allows resources to automatically adjust based on workload and configured policies.

For example, if application traffic increases, additional EC2 instances can be launched.

When demand decreases, instances can be reduced.

This helps applications handle changing workloads efficiently.

### Example

    Low Traffic
        ↓
    2 Instances

    High Traffic
        ↓
    5 Instances

### Memory Trick

**Demand ↑ → Resources ↑**

**Demand ↓ → Resources ↓**

### One-Line Answer

Auto Scaling automatically adjusts application resources according to workload and configured scaling policies.

## Q21. How would you deploy a FastAPI application on AWS?

### Interview Answer

There are multiple ways to deploy a FastAPI application on AWS.

One possible approach is:

    FastAPI
       ↓
    Dockerfile
       ↓
    Docker Image
       ↓
    Container Registry
       ↓
    AWS Container Service
       ↓
    Running Application

Another approach is to run the application directly on an EC2 instance.

The choice depends on the application's requirements, scale, infrastructure control, and operational needs.

### Memory Trick

**FastAPI → Container → AWS → Application**

### One-Line Answer

I can deploy FastAPI using containers on AWS or run it on EC2, depending on the application's requirements.

## Q22. Your AWS application is returning a 500 error. How would you troubleshoot it?

### Interview Answer

I would first reproduce the issue and inspect the application logs.

Then I would check:

- Request data
- Application traceback
- Environment variables
- Database connectivity
- External API calls
- IAM permissions
- Network configuration
- Recent deployment changes

I would use CloudWatch logs and metrics where applicable.

Then I would identify the root cause, apply the fix, and retest.

### Memory Trick

**Reproduce → Logs → Dependencies → IAM/Network → Fix → Test**

### One-Line Answer

I would reproduce the error, inspect logs and metrics, trace the request through its dependencies, identify the root cause, and retest.

---

## Q23. Your EC2 instance is running but the application is not accessible. What would you check?

### Interview Answer

I would troubleshoot from infrastructure to application.

First, I would check whether the EC2 instance is actually healthy.

Then I would check:

- Application process
- Application port
- Security group rules
- Network configuration
- Route configuration
- Public or private connectivity
- Operating system firewall
- Application logs

For example, if the FastAPI application listens on port 8000 but the required network rule does not allow traffic to that port, external access may fail.

### Memory Trick

**Instance → Process → Port → Security Group → Network → Logs**

### One-Line Answer

I would check the instance, application process, port, security group, network configuration, firewall, and logs.

---

## Q24. How would you secure an application deployed on AWS?

### Interview Answer

I would use multiple security layers.

First, I would apply least privilege through IAM.

I would avoid hard-coding credentials in source code or Docker images.

I would use appropriate secret-management mechanisms.

I would restrict network access using VPC configuration and security groups.

I would secure application APIs using authentication and authorization.

I would also keep dependencies updated and use monitoring and logging to identify unusual behavior.

### Memory Trick

**IAM + Secrets + Network + Auth + Monitoring**

### One-Line Answer

I would secure the application using least-privilege IAM, secret management, network controls, API authentication, updated dependencies, and monitoring.

### Interview Answer

A typical architecture could look like:

    Developer
        ↓
       Git
        ↓
      Docker
        ↓
    Docker Image
        ↓
    Container Registry
        ↓
    AWS Container Service
        ↓
      FastAPI
       /   \
      /     \
    RDS      S3
    DB      Files
      \
       \
     CloudWatch
     Monitoring
     Logs

The exact AWS services depend on the application requirements.

For example, a FastAPI application could be containerized using Docker, stored in a container registry, and deployed using an AWS container service.

A relational database could be hosted using RDS, files could be stored in S3, and CloudWatch could be used for monitoring and logs.

### Memory Trick

**Code → Docker → Registry → Compute → RDS/S3 → CloudWatch**

### One-Line Answer

A typical architecture can use Docker for packaging, an AWS compute or container service for FastAPI, RDS for relational data, S3 for files, and CloudWatch for monitoring.
