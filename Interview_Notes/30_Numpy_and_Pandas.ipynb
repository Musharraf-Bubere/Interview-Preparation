# 🎯 Interview Answer Framework

For a normal question:

```text
1. Definition
       ↓
2. How it works
       ↓
3. Example
       ↓
4. Real-world use
       ↓
5. Trade-off / limitation
```

For an advanced question:

```text
Problem
  ↓
Approach
  ↓
Technical reasoning
  ↓
Example
  ↓
Performance / Memory
  ↓
Trade-off
  ↓
Production consideration
```

# 🔥 Final Interview Rule

Do not answer only with a definition.

A strong interview answer should naturally move through:

```text
WHAT is it?
      ↓
WHY is it used?
      ↓
HOW does it work?
      ↓
EXAMPLE
      ↓
WHERE would I use it?
      ↓
WHAT are the limitations/trade-offs?
```

For senior-level questions, also be ready for:

```text
Why did you choose this approach?
Why not another approach?
How does it affect memory?
How does it affect performance?
How would you test it?
How would you handle large data?
How would you prevent data-quality issues?
```
