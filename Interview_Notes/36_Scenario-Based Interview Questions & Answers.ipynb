# 🟢 BASIC SCENARIO-BASED QUESTIONS

## Q1. Sales suddenly dropped by 20% this month. How would you analyze the problem?

### 🎤 Interview Answer

I would not immediately conclude that customers are buying less. I would first validate the data and then break the problem down by time, geography, product, customer segment, channel, and other relevant dimensions.

My approach would be:

1. Confirm that the 20% decline is real.
2. Compare it with the previous period and the same period last year.
3. Check whether the decline is caused by fewer orders, lower average order value, or both.
4. Segment the decline by region, product, channel, and customer type.
5. Investigate operational or business events that could explain the change.
6. Quantify the major drivers and communicate the findings.

### 🔍 Analysis

A useful decomposition is:

```text
Revenue
  ↓
Orders × Average Order Value
  ↓
Customers × Orders per Customer × Average Order Value
```

### 💻 Example

```text
Previous Month Revenue = ₹10 Crore
Current Month Revenue  = ₹8 Crore
Decline                = 20%
```

If orders fell from 1,00,000 to 80,000 while AOV remained ₹1,000, order volume is the primary driver.

### 🧠 Memory Trick

**Validate → Decompose → Segment → Investigate → Quantify**

### ⚡ One-Line Interview Answer

> I would validate the decline, decompose revenue into its drivers, segment the change, identify the root cause, and quantify its business impact.

## Q2. A stakeholder says revenue increased, but profit decreased. How would you investigate?

### 🎤 Interview Answer

I would compare revenue and cost components rather than assuming higher revenue means higher profitability.

I would analyze:

- Revenue
- Cost of goods sold
- Discounts
- Shipping costs
- Marketing costs
- Returns
- Operational expenses
- Product-level margins

### Diagram

```text
Revenue ↑
   ↓
Gross Profit
   ↓
COGS ↑?
Discounts ↑?
Shipping ↑?
Returns ↑?
Marketing Cost ↑?
   ↓
Net Profit ↓
```

### Example

```text
Revenue:       ₹10M → ₹12M
COGS:           ₹6M → ₹8M
Discounts:      ₹0.5M → ₹1M
Shipping:       ₹0.5M → ₹0.8M
```

Revenue increased, but costs increased faster.

### ⚡ One-Line Interview Answer

> I would decompose the profit equation and identify which cost or margin component increased faster than revenue.

## Q3. A dataset contains missing values. What would you do?

### 🎤 Interview Answer

I would not automatically replace every missing value with the mean.

First, I would understand:

- Why the value is missing.
- How much data is missing.
- Whether missingness is random.
- Whether missingness itself carries business meaning.
- Whether the column is numeric or categorical.

Then I would choose an appropriate treatment.

### Diagram

```text
Missing Values
      ↓
Understand Cause
      ↓
Measure Percentage
      ↓
Check Pattern
      ↓
Choose Treatment
```

Possible approaches include:

```text
Remove rows
Fill with median
Fill with mean
Use mode
Forward/backward fill
Model-based imputation
Create "Unknown" category
Keep missing as meaningful
```

### ⚡ One-Line Interview Answer

> I first investigate why values are missing and then choose an imputation or exclusion strategy based on the business and statistical context.

## Q4. You find duplicate records in a sales dataset. How would you handle them?

### 🎤 Interview Answer

I would first determine whether the records are true duplicates or legitimate repeated transactions.

I would identify a business key such as Order ID, Order Item ID, or Customer ID plus Transaction Date. Then I would compare duplicate records and understand why they occurred before removing anything.

### Diagram

```text
Duplicate Rows
      ↓
Identify Business Key
      ↓
Compare Records
      ↓
True Duplicate?
   ↙          ↘
 Yes           No
 ↓              ↓
Remove       Investigate
```

### ⚡ One-Line Interview Answer

> I verify whether duplicates are genuine duplicates using business keys before removing them.

## Q5. A stakeholder asks, "What are our best-selling products?" How would you answer?

### 🎤 Interview Answer

I would first clarify what "best-selling" means. It could mean highest revenue, highest units sold, highest profit, highest number of orders, or highest growth.

After defining the metric, I would aggregate sales by product and rank the products.

### Example

```text
Product     Units    Revenue    Profit
A           10,000   ₹20M       ₹2M
B            8,000   ₹25M       ₹8M
C           15,000   ₹18M       ₹6M
```

If "best-selling" means revenue, Product B ranks first.

### ⚡ One-Line Interview Answer

> I would clarify the business definition of "best-selling" before selecting the metric and ranking products.

## Q6. How would you analyze customer churn?

### 🎤 Interview Answer

I would first define churn precisely, such as customers who have not purchased within a specified period. Then I would calculate the churn rate and segment churned customers by meaningful dimensions.

### Formula

```text
Churn Rate =
Churned Customers / Customers at Start of Period × 100
```

### Analysis

```text
Churn
 ↓
Region
Product
Customer Segment
Acquisition Channel
Tenure
Usage
Support Issues
```

### ⚡ One-Line Interview Answer

> I would define churn clearly, calculate the churn rate, segment churned customers, and identify factors associated with churn.

## Q13. How would you determine whether a marketing campaign was successful?

### 🎤 Interview Answer

I would not evaluate a campaign only by revenue. I would compare campaign performance against a suitable baseline or control group and analyze conversion rate, incremental revenue, customer acquisition, cost per acquisition, return on marketing spend, and profit impact.

### Formula

```text
ROAS = Revenue Attributed to Campaign / Campaign Cost
```

For stronger causal inference, I would prefer an experiment or control group where available.

### ⚡ One-Line Interview Answer

> I would measure incremental business impact rather than relying only on campaign-attributed revenue.

## Q16. How would you calculate customer lifetime value?

### 🎤 Interview Answer

The exact calculation depends on the business model.

A simple approximation is:

```text
CLV ≈
Average Purchase Value
× Purchase Frequency
× Customer Lifetime
× Margin
```

I would clearly document the assumptions because CLV can be modeled in several ways.

### ⚡ One-Line Interview Answer

> CLV estimates the economic value a customer generates over the relationship, and the calculation should match the business model and available data.

### ⚡ One-Line Interview Answer

# 🟠 ADVANCED SCENARIO-BASED QUESTIONS

## Q26. A product's sales increased after a discount campaign. How do you determine whether the campaign actually caused the increase?

### 🎤 Interview Answer

I would distinguish correlation from causation.

If possible, I would use a randomized control group or A/B test. If an experiment is unavailable, I would compare against a carefully selected baseline and control for major confounding factors.

### Diagram

```text
Campaign
   ↓
Observed Sales Increase
   ↓
Was Campaign the Cause?
   ↓
Experiment / Control / Quasi-experiment
```

### ⚡ One-Line Interview Answer

> I would use an experiment or appropriate control strategy to estimate incremental impact rather than assuming the observed increase was caused by the campaign.

---

## Q27. How would you design an A/B test?

### 🎤 Interview Answer

I would define the hypothesis, primary metric, control and treatment groups, experiment population, duration, and statistical decision criteria before looking at results.

### Example

```text
H0: New Checkout = Old Checkout
H1: New Checkout ≠ Old Checkout
```

I would also define guardrail metrics such as cancellation rate or payment failure rate.

### ⚡ One-Line Interview Answer

> I would define the hypothesis, randomize users, choose a primary metric and guardrails, run the test appropriately, and analyze statistical and business significance.

---

## Q28. A/B test conversion improved from 10% to 10.5%. Is the experiment successful?

### 🎤 Interview Answer

Not necessarily.

The relative improvement is:

```text
(10.5 - 10) / 10 × 100 = 5%
```

But I would still check statistical significance, confidence intervals, sample size, experiment validity, and business impact.

### ⚡ One-Line Interview Answer

> A numerical improvement is not enough; I would check statistical significance, uncertainty, experiment quality, and practical business impact.

---

## Q29. How would you identify the root cause of a business KPI decline?

### 🎤 Interview Answer

I would use a structured root-cause analysis.

```text
KPI Decline
    ↓
Validate
    ↓
Decompose
    ↓
Segment
    ↓
Find Largest Driver
    ↓
Investigate Driver
    ↓
Business Event
    ↓
Root Cause Hypothesis
    ↓
Validate Hypothesis
```

I would avoid stopping at the first correlation I find.

### ⚡ One-Line Interview Answer

> I decompose the KPI, identify the largest driver, form a hypothesis, and validate it against data and business events.

---

## Q30. How would you detect whether a dataset has data leakage?

### 🎤 Interview Answer

I would check whether information used as a feature would actually have been available at the time the prediction or business decision was made.

For example, using a customer's cancellation status to predict whether the customer will cancel would be leakage if cancellation happens after the prediction point.

### ⚡ One-Line Interview Answer

> I look for features that contain future information unavailable at prediction time because they can make analytical or predictive results misleading.

---

### 🎤 Interview Answer

I would first understand the variable types.

For numerical variables, I could inspect scatter plots and correlation. For categorical variables, I could compare rates or use appropriate statistical tests.

I would remember that correlation does not establish causation.

### ⚡ One-Line Interview Answer

> I choose the analytical method based on variable types and use both statistical evidence and business context to interpret relationships.

### 🎤 Interview Answer

I would investigate whether the correlation is caused by confounding variables, time trends, data leakage, aggregation effects, selection bias, or coincidence.

### ⚡ One-Line Interview Answer

> I would investigate confounding factors and data-generation processes before interpreting a strong correlation as meaningful.

## Q36. Why would you use median instead of mean?

### 🎤 Interview Answer

I would use the median when the distribution is skewed or contains significant outliers.

For example:

```text
Delivery:
2, 3, 3, 4, 5, 100
```

The mean is heavily influenced by 100, while the median better represents the typical observation.

### ⚡ One-Line Interview Answer

> Median is useful for skewed data because it is less sensitive to extreme values than the mean.

---

## Q37. A dataset contains 90% records from one category and 10% from another. How would this affect analysis?

### 🎤 Interview Answer

I would recognize this as an imbalanced distribution. I would avoid relying only on overall averages because the majority category can dominate the result.

I would analyze each category separately and consider weighted or stratified approaches where appropriate.

### ⚡ One-Line Interview Answer

> I would account for imbalance because overall metrics can hide important behavior in smaller groups.

I would start with the business conclusion rather than the technical process.

### ⚡ One-Line Interview Answer

## Q48. How would you handle Simpson's paradox in an analysis?

### 🎤 Interview Answer

Simpson's paradox occurs when an overall relationship differs from the relationship observed within important subgroups.

I would therefore inspect aggregated results alongside relevant segments.

### Example

```text
Overall:
Treatment A looks better

Within Group 1:
Treatment B better

Within Group 2:
Treatment B better
```

### ⚡ One-Line Interview Answer

> I would check aggregated results against meaningful subgroups because aggregation can reverse or hide relationships.

---

## Q49. A stakeholder asks you to prove that one factor caused another. What would you say?

### 🎤 Interview Answer

I would explain that observational data may show association but does not automatically establish causality.

I would ask whether we have randomized experiments, control groups, temporal evidence, appropriate causal design, or confounder controls.

I would clearly separate what the data demonstrates from what it suggests.

### ⚡ One-Line Interview Answer

> I would distinguish association from causation and explain what evidence is required before making a causal claim.

---

## Q50. You discover that the analysis contradicts what management expected. What do you do?

### 🎤 Interview Answer

I would not change the analysis to match expectations.

I would first validate the data, definitions, transformations, and methodology.

If the result remains valid, I would present the finding respectfully, explain the evidence, identify limitations, and discuss what additional analysis could test alternative explanations.

### ⚡ One-Line Interview Answer

> I would validate the analysis objectively and communicate evidence-based findings even when they differ from expectations.

# 🏆 DATA ANALYSIS CASE STUDY QUESTIONS

## Q51. An e-commerce company's revenue fell 15%. Walk me through your complete analysis.

### 🎤 Interview Answer

I would start by validating the revenue metric and comparing it with the previous period and the same period last year.

Then I would decompose:

```text
Revenue
 ↓
Orders × AOV
```

Next I would analyze:

```text
Orders
 ↓
Customers × Purchase Frequency

AOV
 ↓
Price × Product Mix × Discount
```

Then I would segment the decline by product, category, region, channel, customer segment, device, and payment method.

Finally, I would connect the largest data-driven driver to business events such as pricing changes, campaigns, inventory issues, delivery problems, or website changes.

### Example Conclusion

```text
Revenue ↓ 15%

Main Driver:
Orders ↓ 12%

Largest Segment:
Mobile Customers

Likely Operational Issue:
Checkout Conversion ↓ 20%

Recommendation:
Investigate mobile checkout and payment failures.
```

### ⚡ One-Line Interview Answer

> I would decompose revenue into volume and value, segment the decline, identify the largest contributor, validate the root cause, and recommend an action.

---

## Q52. A subscription company's churn increased from 5% to 8%. How would you analyze it?

### 🎤 Interview Answer

I would first validate the churn definition and measurement window.

Then I would compare 5% to 8% and segment churn by customer tenure, plan, acquisition channel, region, usage, payment failures, support interactions, and product engagement.

I would perform cohort analysis to determine whether newer or older customers are experiencing the increase.

### ⚡ One-Line Interview Answer

# 🧩 SQL / PYTHON / EXCEL / POWER BI SCENARIO QUESTIONS

## Q56. When would you use SQL instead of Python for analysis?

### 🎤 Interview Answer

I would generally use SQL when the data is already stored in a relational database and the task involves filtering, joining, aggregating, or preparing data close to the source.

I would use Python when I need more advanced statistical analysis, machine learning, custom processing, or visualization workflows.

### Decision

```text
Database Filtering / Joining / Aggregation
              ↓
             SQL

Advanced Analysis / Statistics / ML
              ↓
            Python
```

### ⚡ One-Line Interview Answer

> I use SQL for efficient database-side data manipulation and Python for deeper analytical, statistical, and modeling workflows.

---

## Q57. When would you use Excel instead of Python?

### 🎤 Interview Answer

Excel is useful for quick exploratory analysis, business users, smaller datasets, ad hoc calculations, and lightweight reporting.

Python becomes more appropriate when the analysis needs automation, scalability, reproducibility, advanced statistics, or machine learning.

### ⚡ One-Line Interview Answer

> Excel is excellent for quick business analysis and collaboration, while Python is better for scalable, automated, and advanced analytical workflows.

---

## Q58. When would you use Power BI instead of a Python visualization?

### 🎤 Interview Answer

I would use Power BI when the goal is interactive business reporting, self-service analytics, dashboarding, scheduled refresh, security, and distribution to business users.

Python is more suitable for custom analytical workflows and specialized visualizations.

### ⚡ One-Line Interview Answer

> I use Power BI for interactive business reporting and distribution, while Python is useful for custom analytical and statistical workflows.

---

## Q59. How would you automate a recurring data analysis?

### 🎤 Interview Answer

I would first identify the repeatable steps and separate configuration from logic.

A possible architecture is:

```text
Data Source
    ↓
SQL / Python
    ↓
Validation
    ↓
Transformation
    ↓
Analysis
    ↓
Report / Dashboard
    ↓
Scheduled Execution
```

I would also add logging and data-quality checks.

### ⚡ One-Line Interview Answer

> I would convert repeatable analysis into a reproducible pipeline with validation, logging, and scheduled execution.

## Q60. A SQL query returns an unexpectedly high number of rows after a JOIN. What would you check?

### 🎤 Interview Answer

I would check the relationship cardinality.

The most common cause is an unintended one-to-many or many-to-many join.

I would inspect:

```text
Join Keys
Duplicate Keys
Null Keys
Relationship Cardinality
Join Type
```

### Example

```text
Customer
CustomerID = 101

Orders
CustomerID = 101 → 5 rows
```

Joining the customer to orders produces 5 rows for that customer, which may be correct.

### ⚡ One-Line Interview Answer

> I would inspect join keys and relationship cardinality because duplicate keys can multiply rows during joins.

# 💬 STAKEHOLDER & BUSINESS SCENARIOS

## Q61. A stakeholder gives you a vague request: "Analyze sales." What do you do?

### 🎤 Interview Answer

I would clarify the business question before starting.

I would ask:

```text
What decision are we trying to make?
What time period?
Which sales metric?
Which business segments?
What is the expected output?
Who will use it?
```

### ⚡ One-Line Interview Answer

> I convert vague requests into specific analytical questions, metrics, dimensions, time periods, and decisions.

---

## Q62. A stakeholder asks for 20 charts. Would you create all of them?

### 🎤 Interview Answer

Not automatically.

I would understand what decisions the charts are intended to support and remove redundant visuals.

The goal is not to maximize the number of charts but to maximize useful information.

### ⚡ One-Line Interview Answer

> I focus on decision usefulness rather than the number of charts.

---

## Q63. A stakeholder disagrees with your analysis. How would you handle it?

### 🎤 Interview Answer

I would first understand their concern and ask whether we have different metric definitions, assumptions, or business context.

Then I would review the analysis and evidence collaboratively.

If the analysis remains valid, I would explain the reasoning transparently.

### ⚡ One-Line Interview Answer

> I treat disagreement as an opportunity to validate definitions and assumptions rather than becoming defensive about the analysis.

---

## Q64. How do you decide whether an insight is actionable?

### 🎤 Interview Answer

An insight is actionable when it identifies something the business can influence and provides enough evidence to support a decision.

### Framework

```text
Observation
   ↓
Explanation
   ↓
Business Impact
   ↓
Action
   ↓
Expected Outcome
```

### Example

Weak:

> Sales are lower in Region A.

Actionable:

> Region A sales declined mainly because repeat purchases fell among customers acquired through Channel X; increasing retention activity for this segment should be investigated.

### ⚡ One-Line Interview Answer

> An actionable insight connects a finding to a controllable business action and expected impact.

---

## Q65. What is your overall approach to solving a data analysis problem in an interview?

### 🎤 Interview Answer

My approach is:

```text
1. Understand the business problem
2. Define the metric
3. Understand the data
4. Validate data quality
5. Explore the data
6. Segment the problem
7. Form hypotheses
8. Test hypotheses
9. Identify drivers
10. Quantify impact
11. Recommend actions
12. Communicate assumptions and limitations
```

I avoid jumping directly into coding because the most important part of analysis is understanding what decision the analysis needs to support.

### 🧠 Memory Trick

**Understand → Define → Validate → Explore → Hypothesize → Test → Quantify → Recommend → Communicate**

### ⚡ One-Line Interview Answer

> My approach is business-first: understand the decision, validate the data, investigate systematically, quantify the drivers, and turn evidence into actionable recommendations.
