## Q1. What is NLP?

### Interview Answer

NLP (Natural Language Processing) is a branch of AI that enables computers to understand, analyze, process, and generate human language.

### Detailed Explanation

### Example

```text
Suppose a customer writes, `I am very happy with this product.` An NLP sentiment-analysis system can process the sentence and predict **Positive** sentiment. Chatbots, translation, search, NER, summarization, and LLMs are other NLP applications.
```

### Real-World Use Cases

- Chatbots
- Machine translation
- Search engines
- Sentiment analysis
- Document summarization

### Memory Trick

**NLP = Understand + Analyze + Generate Language**

### One-Line Answer

> NLP (Natural Language Processing) is a branch of AI that enables computers to understand, analyze, process, and generate human language.

## Q3. What is tokenization?

### Interview Answer

Tokenization breaks text into smaller units such as words, subwords, characters, or sentences.

### Detailed Explanation

Models need these units before language can be converted into numerical representations.

### Example

```text
The sentence `I love Python` can be word-tokenized as `['I', 'love', 'Python']`. A subword tokenizer may split an uncommon word into smaller pieces rather than requiring the complete word to exist in its vocabulary.
```

### Real-World Use Cases

- LLM preprocessing
- Search
- Text classification
- Document processing
- Information extraction

### Memory Trick

**Text → Tokens**

### One-Line Answer

> Tokenization breaks text into smaller units such as words, subwords, characters, or sentences.

## Q10. What is Bag of Words?

### Interview Answer

Bag of Words (BoW) represents text using vocabulary-word occurrence or frequency.

### Detailed Explanation

It is simple and interpretable but ignores word order and deeper semantics.

### Example

```text
For documents `I love Python` and `I love NLP`, vocabulary could be `[I, love, Python, NLP]`. The vectors become `[1,1,1,0]` and `[1,1,0,1]`. `Python is great` and `great Python is` receive the same BoW representation.
```

### Real-World Use Cases

- Spam classification
- Sentiment analysis
- Topic classification
- Baseline NLP models

### Memory Trick

**Words + Counts, no context**

### One-Line Answer

> Bag of Words (BoW) represents text using vocabulary-word occurrence or frequency.

## Q11. What is TF-IDF?

### Interview Answer

TF-IDF stands for Term Frequency-Inverse Document Frequency.

### Detailed Explanation

It gives higher weight to terms that are important in a document but relatively rare across the collection. A common formulation is `TF-IDF = TF × IDF`.

### Example

```text
If a collection contains 1,000 documents and `the` appears in 990, its IDF is low. If `transformer` appears in only 20 documents, its IDF is much higher, so it can receive more weight in relevant documents.
```

### Real-World Use Cases

- Search engines
- Information retrieval
- Keyword extraction
- Text classification

### Memory Trick

**Important here, rare globally**

### One-Line Answer

> TF-IDF stands for Term Frequency-Inverse Document Frequency.

## Q17. What is Word2Vec?

### Interview Answer

Word2Vec learns word embeddings from word-context relationships.

### Detailed Explanation

Its two main architectures are CBOW, which predicts a target word from surrounding context, and Skip-gram, which predicts context words from a target word.

### Example

```text
For `The cat sits on the mat`, CBOW may use surrounding words to predict `sits`. Skip-gram can take `cat` as the target and learn to predict nearby words such as `the` and `sits`.
```

### Real-World Use Cases

- Semantic similarity
- Recommendation systems
- Text clustering
- Classical NLP representation

### Memory Trick

**CBOW: context → word; Skip-gram: word → context**

### One-Line Answer

> Word2Vec learns word embeddings from word-context relationships.

## Q21. What is cosine similarity?

### Interview Answer

### Detailed Explanation

### Example

```text
If an embedding for `car` is compared with `automobile`, cosine similarity may be high. Comparing `car` with an unrelated concept such as `banana` would generally produce a lower semantic similarity.
```

### Real-World Use Cases

- RAG retrieval
- Duplicate detection
- Recommendation systems

### Memory Trick

**Compare vector directions**

### One-Line Answer

> Cosine similarity compares vectors using the angle between them: `cos(A,B) = (A·B)/(||A|| ||B||)`.

## Q31. What is self-attention?

### Interview Answer

Self-attention allows every token to consider other tokens in the same sequence.

### Detailed Explanation

### Example

```text
In `The dog chased the ball because it was fast`, the representation of `it` can attend to other tokens to determine which context is relevant. The attention weights determine how strongly information from each token contributes.
```

### Real-World Use Cases

- Transformer models
- LLMs
- Question answering
- Contextual language understanding

### Memory Trick

**Q asks, K matches, V provides**

### One-Line Answer

> Self-attention allows every token to consider other tokens in the same sequence.

## Q32. Why is the scaling factor √dₖ used in attention?

### Interview Answer

As the key/query dimensionality increases, dot products can become large.

### Detailed Explanation

Large logits can make softmax excessively peaked and gradients less stable. Dividing by `√dₖ` keeps the values at a more suitable scale.

### Example

```text
If query/key vectors have many dimensions, their dot product can grow in magnitude. Scaling before softmax prevents the attention distribution from becoming unnecessarily extreme.
```

### Real-World Use Cases

- Transformer training
- Attention implementation
- Deep learning optimization

### Memory Trick

**Scaling stabilizes softmax**

### One-Line Answer

> As the key/query dimensionality increases, dot products can become large.

## Q41. What is perplexity?

### Interview Answer

Perplexity measures how well a language model predicts a sequence.

### Detailed Explanation

A common formulation is `PPL = exp(-1/N × Σ log P(token_i))`. Lower perplexity generally indicates better predictive likelihood on the evaluated data, but it is not a complete quality metric.

### Example

```text
If Model A assigns higher probability to the actual next tokens than Model B on the same test corpus, Model A can have lower perplexity. A lower perplexity does not automatically mean fewer hallucinations or better user experience.
```

### Real-World Use Cases

- Language-model evaluation
- Model comparison
- Research benchmarking

### Memory Trick

**Perplexity = prediction uncertainty**

### One-Line Answer

> Perplexity measures how well a language model predicts a sequence.

## Q43. What is the difference between temperature, top-k, and top-p sampling?

### Interview Answer

Temperature changes the sharpness of the probability distribution.

### Detailed Explanation

Top-k restricts sampling to the k highest-probability tokens. Top-p (nucleus sampling) dynamically chooses the smallest set of tokens whose cumulative probability reaches p.

### Example

```text
Suppose the next-token probabilities are `cat=.45, dog=.30, car=.15, tree=.05, book=.05`. With top-k=2, only `cat` and `dog` are considered. With top-p=.75, the candidate set expands until cumulative probability reaches about 0.75.
```

### Real-World Use Cases

- LLM generation
- Creative writing
- Chat applications
- Response-control tuning

### Memory Trick

**Temperature = sharpness; top-k = count; top-p = probability mass**

### One-Line Answer

> Temperature changes the sharpness of the probability distribution.

## Q44. What is the attention complexity problem for long documents?

### Interview Answer

Standard self-attention compares every token with every other token, creating an attention matrix with roughly n² entries for sequence length n.

### Detailed Explanation

This quadratic scaling makes long contexts expensive.

### Example

```text
With 1,000 tokens there are about 1,000,000 token-pair positions; with 10,000 tokens there are about 100,000,000. Long-document systems may therefore use chunking, retrieval, sliding windows, sparse attention, or efficient attention methods.
```

### Real-World Use Cases

- Long-document NLP
- Long-context LLMs
- RAG systems
- Efficient attention

### Memory Trick

**Longer context → quadratic attention cost**

### One-Line Answer

# 🚀 BEST WAY TO ANSWER AN NLP INTERVIEW QUESTION

Use this structure when speaking:

```text
1. Define the concept
        ↓
2. Explain how it works
        ↓
3. Give a simple example
        ↓
4. Explain why it is used
        ↓
5. Mention limitations / trade-offs
        ↓
6. Connect it to a real-world application
```

For advanced and expert questions, add:

```text
Concept
   ↓
How it works
   ↓
Mathematical / technical intuition
   ↓
Trade-offs
   ↓
Evaluation
   ↓
Production considerations
```
