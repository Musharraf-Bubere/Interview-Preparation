# 🟢 TEXT PREPROCESSING


## Q6. What is tokenization? What are its different types?

### Answer

Tokenization is the process of breaking text into smaller units called **tokens**.

### Types

#### 1. Sentence Tokenization

    "I love Python. NLP is interesting."

    ↓

    [
      "I love Python.",
      "NLP is interesting."
    ]

#### 2. Word Tokenization

    "I love Python"

    ↓

    ["I", "love", "Python"]

#### 3. Subword Tokenization

A word can be divided into smaller meaningful or reusable pieces.

    "playing"

    ↓

    Possible subwords:
    "play" + "ing"

The exact result depends on the tokenizer.

#### 4. Character Tokenization

Text is divided into individual characters.

    "cat"

    ↓

    ["c", "a", "t"]

### One-Line Interview Answer

Tokenization divides text into smaller units such as sentences, words, subwords, or characters.


---

## Q7. What is the difference between word tokenization and subword tokenization?

### Answer

### Word Tokenization

Treats complete words as tokens.

    "I love Python"

    ↓

    ["I", "love", "Python"]

### Subword Tokenization

Can divide words into smaller pieces.

    "unhappiness"

    ↓

    Possible:
    ["un", "happi", "ness"]

The exact segmentation depends on the tokenizer.

### Why Subword Tokenization?

It helps handle:

- Rare words
- New words
- Misspellings
- Morphological variations
- Out-of-vocabulary problems

### Example

Suppose the vocabulary does not contain:

    "unhappiness"

A subword tokenizer may still represent it using known pieces.

### One-Line Interview Answer

Word tokenization uses complete words, while subword tokenization represents words using smaller reusable units and handles rare or unseen words better.

## Q8. What are stopwords? Should we always remove stopwords?

### Answer

Stopwords are common words that may provide relatively little information for certain NLP tasks.

Examples:

    the
    is
    a
    an
    and
    of

However, **we should not always remove them**.

### Example

    "This movie is not good."

If we remove:

    "not"

we get:

    "This movie is good."

The meaning changes completely.

### Important Point

Stopword removal depends on the task.

For some traditional ML pipelines, it can reduce noise and dimensionality.

For sentiment analysis, question answering, and modern Transformer models, blindly removing stopwords can be harmful.

### One-Line Interview Answer

Stopwords are common words that may be removed in some NLP tasks, but they should not always be removed because they can carry important contextual meaning.

## Q9. What is stemming? What are its common algorithms?

### Answer

Stemming is the process of reducing words to a root-like form using heuristic rules, usually by removing prefixes or suffixes.

### Example

    playing
    played
    plays

may be reduced to:

    play

However, stemming can sometimes produce a form that is not a valid dictionary word.

### Common Stemming Algorithms

- Porter Stemmer
- Snowball Stemmer
- Lancaster Stemmer
- RegexpStemmer

### Advantages

- Simple
- Fast
- Computationally inexpensive

### Disadvantage

The resulting stem may not be a valid word.

### One-Line Interview Answer

Stemming reduces words to root-like forms using heuristic rules, with common algorithms such as Porter, Snowball, Lancaster, and RegexpStemmer.


---

## Q10. What is lemmatization, and how is it different from stemming?

### Answer

Lemmatization reduces a word to its linguistically appropriate dictionary base form, called a **lemma**.

It can use:

- Vocabulary
- Morphology
- Part of Speech
- Linguistic rules

### Example

    better
       ↓
    good

A lemmatizer can understand that "better" has "good" as its lemma in the appropriate context.

### Stemming vs Lemmatization

| Stemming | Lemmatization |
|---|---|
| Heuristic | Linguistically informed |
| Usually faster | Usually more computationally expensive |
| Root-like output | Dictionary base form |
| May produce non-words | Aims for valid linguistic forms |

### Memory Trick

    Stemming
    → Shortcut

    Lemmatization
    → Linguistic

### One-Line Interview Answer

Stemming uses heuristic rules to produce root-like forms, while lemmatization uses linguistic information to produce appropriate dictionary base forms.

POS tagging assigns grammatical categories to tokens based on context and supports tasks such as lemmatization, parsing, and information extraction.

# 🟡 TEXT REPRESENTATION


## Q13. What is One-Hot Encoding in NLP? What are its limitations?

### Answer

One-Hot Encoding represents each vocabulary token as a binary vector.

Suppose vocabulary is:

    ["cat", "dog", "fish"]

Then:

    cat
    → [1, 0, 0]

    dog
    → [0, 1, 0]

    fish
    → [0, 0, 1]

### Limitations

1. High dimensionality
2. Sparse vectors
3. High memory usage for large vocabularies
4. No semantic relationship
5. All words are equally distant in the basic representation

For example:

    cat → [1,0,0]
    dog → [0,1,0]

and:

    cat → [1,0,0]
    car → [0,0,1]

The representation itself does not know that "cat" and "dog" may be semantically related.

### One-Line Interview Answer

One-hot encoding represents tokens using sparse binary vectors but does not capture semantic relationships and becomes inefficient for large vocabularies.

## Q14. What is Bag of Words (BoW)? How does it work?

### Answer

Bag of Words represents text using the words present in a vocabulary while ignoring word order.

### Example

Documents:

    D1 = "I love Python"

    D2 = "I love NLP"

Vocabulary:

    [I, love, Python, NLP]

Representations:

    D1 → [1, 1, 1, 0]

    D2 → [1, 1, 0, 1]

The values can represent word counts or binary presence depending on the implementation.

### Why "Bag"?

Because word order is ignored.

### Example

    "Dog bites man"

and:

    "Man bites dog"

can have the same basic Bag-of-Words representation.

### One-Line Interview Answer

Bag of Words represents a document using vocabulary word occurrence or frequency while ignoring word order.

## Q15. What are the limitations of Bag of Words?

### Answer

### 1. Ignores Word Order

    "Dog bites man"

and:

    "Man bites dog"

can have the same representation.

### 2. Sparse Representation

Most vocabulary positions are zero.

### 3. High Dimensionality

Large vocabulary → large vectors.

### 4. No Deep Semantic Understanding

BoW does not inherently understand relationships such as:

    king ↔ queen

### 5. Vocabulary Problem

Words not represented in the vocabulary require an appropriate handling strategy.

### One-Line Interview Answer

BoW ignores word order, produces sparse high-dimensional vectors, and does not capture deep semantic relationships.

An N-Gram is a contiguous sequence of `N` tokens.

### Unigram

One token.

    "I love NLP"

    ↓

    I
    love
    NLP

### Bigram

Two consecutive tokens.

    I love
    love NLP

### Trigram

Three consecutive tokens.

### Why N-Grams?

They capture local word-order information.

For example:

    "machine learning"

as a bigram preserves the relationship between:

    machine + learning

### Limitation

- Vocabulary grows
- Sparsity increases
- Memory requirements increase
- Data requirements increase

### One-Line Interview Answer

N-Grams are contiguous sequences of N tokens, such as unigrams, bigrams, and trigrams, used to capture local word-order information.

## Q17. What is TF-IDF? Explain TF and IDF.

### Answer

TF-IDF stands for:

    Term Frequency – Inverse Document Frequency

It measures how important a term is to a document relative to a collection of documents.

### TF — Term Frequency

Measures how frequently a term appears in a document.

A simple formulation is:

    TF(t,d) =
    Number of occurrences of t in d
    ---------------------------------
    Total number of terms in d

### IDF — Inverse Document Frequency

Measures how rare a term is across the corpus.

A common conceptual formulation is:

    IDF(t) =
    log(
        Total Number of Documents
        --------------------------------
        Number of Documents Containing t
    )

Implementations may use smoothing or other variations.

### TF-IDF

Conceptually:

    TF-IDF = TF × IDF

### Example

If:

    "Python"

appears many times in one document but rarely across the entire corpus,

its TF-IDF score can be high.

If:

    "the"

appears in almost every document,

its IDF is low.

### Memory Trick

    TF
    → Important in this document

    IDF
    → Rare across documents

### One-Line Interview Answer

TF-IDF measures term importance by combining how frequently a word occurs in a document with how rare it is across the corpus.


---

## Q18. What are the limitations of TF-IDF?

### Answer

TF-IDF is useful but has several limitations.

### Limitations

1. Does not understand deep semantics
2. Does not inherently capture word order
3. Produces sparse vectors
4. Can create high-dimensional representations
5. Treats terms largely as independent features
6. Does not naturally understand contextual meaning

### Example

    "car"

and:

    "automobile"

may be semantically similar, but basic TF-IDF treats them as different features.

### One-Line Interview Answer

TF-IDF captures statistical term importance but does not provide deep semantic or contextual understanding and typically produces sparse high-dimensional vectors.


# 🟡 WORD EMBEDDINGS


## Q19. What are word embeddings, and why are they better than One-Hot Encoding?

### Answer

Word embeddings represent words as **dense numerical vectors** in a continuous vector space.

The vectors can capture useful semantic and syntactic relationships.

### One-Hot

    cat → [1, 0, 0, 0, 0]

### Embedding

    cat → [0.21, -0.43, 0.67, 0.18, ...]

The exact values are learned from data.

### Advantages of Embeddings

- Dense representation
- Lower dimensionality
- Can capture semantic relationships
- Can represent similarity between words
- More useful for neural NLP models

### Example

Semantically related words may have nearby representations:

    king
    queen
    prince
    princess

### One-Line Interview Answer

Word embeddings represent words as dense vectors that can capture semantic relationships, unlike sparse one-hot vectors.

## Q20. What is Word2Vec? Explain CBOW and Skip-Gram.

### Answer

Word2Vec is a family of neural methods for learning word embeddings from contextual relationships between words.

It has two classic architectures:

    Word2Vec
       ├── CBOW
       └── Skip-Gram

### CBOW

CBOW stands for **Continuous Bag of Words**.

It predicts a target word from surrounding context words.

    Context Words
         ↓
       CBOW
         ↓
    Target Word

Example:

    "The cat ___ on the mat."

Context can be used to predict:

    "sat"

### Skip-Gram

Skip-Gram does the opposite.

It uses a target word to predict surrounding context words.

    Target Word
         ↓
     Skip-Gram
         ↓
    Context Words

### Memory Trick

    CBOW
    → Context → Word

    Skip-Gram
    → Word → Context

### One-Line Interview Answer

Word2Vec learns word embeddings using contextual relationships, with CBOW predicting a target from context and Skip-Gram predicting context from a target word.


---

## Q21. What is the difference between CBOW and Skip-Gram?

### Answer

### CBOW

    Context
       ↓
    Target Word

CBOW predicts the target word using surrounding context.

### Skip-Gram

    Target Word
       ↓
    Context

Skip-Gram predicts surrounding context words from the target word.

### Comparison

| CBOW | Skip-Gram |
|---|---|
| Context → Target | Target → Context |
| Often computationally efficient | Can be useful for rare words |
| Predicts target from surrounding words | Predicts surrounding words from target |
| Usually efficient with frequent words | Often better suited to learning representations for less frequent words |

### Example

Sentence:

    "I love machine learning"

For target:

    "machine"

CBOW:

    "I", "love", "learning" → machine

Conceptually, Skip-Gram:

    machine → "love"
    machine → "learning"

depending on the context window.

### One-Line Interview Answer

CBOW predicts a target word from surrounding context, while Skip-Gram predicts surrounding context words from a target word.

## Q22. What is the difference between Word2Vec, GloVe, and FastText?

### Answer

All three are methods for learning word representations, but they use different approaches.

### Word2Vec

Learns embeddings using local contextual prediction.

Main approaches:

    CBOW
    Skip-Gram

### GloVe

GloVe stands for:

    Global Vectors for Word Representation

It learns embeddings using global word co-occurrence statistics.

### FastText

FastText represents words using subword character information in addition to word-level information.

This helps with:

- Rare words
- Morphologically rich languages
- Unseen word forms

### Comparison

| Model | Main Idea |
|---|---|
| Word2Vec | Context-based prediction |
| GloVe | Global co-occurrence statistics |
| FastText | Word + subword information |

### One-Line Interview Answer

Word2Vec learns embeddings through contextual prediction, GloVe uses global co-occurrence statistics, and FastText incorporates subword information.


# 🔴 DEEP LEARNING & MODERN NLP


## Q23. What is an RNN, and what problems does it have in NLP?

### Answer

RNN stands for **Recurrent Neural Network**.

It is designed for sequential data and maintains a hidden state that carries information from previous time steps.

### Basic Flow

    Token 1
       ↓
      RNN
       ↓
    Hidden State
       ↓
    Token 2
       ↓
      RNN
       ↓
    Hidden State
       ↓
      ...

The hidden state carries information from earlier tokens.

### Why RNNs are useful for NLP?

Language is sequential.

For example:

    "I went to the bank to deposit money."

Previous context helps understand the meaning of "bank."

### Major Problems

#### 1. Vanishing Gradient

Gradients can become extremely small during backpropagation through long sequences.

This makes learning long-range dependencies difficult.

#### 2. Exploding Gradient

Gradients can become extremely large, causing unstable training.

#### 3. Sequential Processing

Traditional RNNs process tokens sequentially, which limits parallelization.

#### 4. Long-Term Dependencies

Standard RNNs can struggle to retain information over long sequences.

### One-Line Interview Answer

An RNN processes sequential data using a hidden state, but it can suffer from vanishing/exploding gradients, difficulty with long-term dependencies, and limited parallelization.


---

## Q24. What are LSTM and GRU? Why were they introduced?

### Answer

LSTM and GRU are gated recurrent architectures designed to improve upon standard RNNs, especially for learning longer-term dependencies.

---

### LSTM

LSTM stands for:

    Long Short-Term Memory

It uses a memory cell and gates.

Main gates:

    Forget Gate
    Input Gate
    Output Gate

### LSTM Concept

    Previous Information
           ↓
      Forget Gate
           ↓
    Keep / Discard

    New Information
           ↓
       Input Gate
           ↓
      Update Memory

    Memory
       ↓
    Output Gate
       ↓
      Output

---

### GRU

GRU stands for:

    Gated Recurrent Unit

It uses a simpler gating mechanism.

Common gates:

    Update Gate
    Reset Gate

### LSTM vs GRU

| LSTM | GRU |
|---|---|
| More complex | Simpler |
| Uses cell state | No separate cell state in standard formulation |
| More parameters | Fewer parameters |
| More gates | Fewer gates |
| Can be computationally heavier | Often faster |

### Why introduced?

They were introduced to improve the ability of recurrent models to learn dependencies over longer sequences and reduce the limitations of standard RNNs.

### One-Line Interview Answer

LSTM and GRU are gated RNN architectures introduced to better handle long-term dependencies and training problems found in standard RNNs.

## Q25. What is a Transformer? Explain Self-Attention, BERT, GPT, and why Transformers became important.

### Answer

A Transformer is a neural network architecture based primarily on **attention mechanisms** rather than recurrence.

It was introduced in the paper:

    "Attention Is All You Need"

Transformers became extremely important because they can efficiently model relationships between tokens and allow highly parallelizable training.

---

# 1. What is Self-Attention?

Self-attention allows each token to interact with other tokens in the same sequence.

### Example

    "The animal did not cross the road because it was tired."

To understand:

    "it"

the model can use information from other tokens in the sentence.

### Query, Key, Value

Self-attention uses:

    Query
    Key
    Value

Conceptually:

    Query
    → What information am I looking for?

    Key
    → What information does this token represent?

    Value
    → What information should be passed forward?

A common formulation is:

    Attention(Q,K,V)
    =
    softmax(QKᵀ / √dₖ)V

---

# 2. What is Multi-Head Attention?

Instead of performing one attention operation, Transformer models can use multiple attention heads.

    Head 1 → Relationship A
    Head 2 → Relationship B
    Head 3 → Relationship C
       ...
    Head N → Relationship N

The outputs are combined.

### One-Line Concept

Multi-head attention allows the model to learn different relationships between tokens simultaneously.


---

# 3. Why does a Transformer need positional information?

Self-attention does not inherently encode the order of tokens.

Therefore, Transformers need positional information.

Possible approaches include:

- Positional Encoding
- Learned Positional Embeddings
- Rotary Positional Embeddings and other positional mechanisms

### Example

These sentences have different meanings:

    "Dog bites man."

    "Man bites dog."

The model needs positional information to distinguish the order.


---

# 4. What is BERT?

BERT stands for:

    Bidirectional Encoder Representations from Transformers

BERT is primarily an **encoder-based Transformer model**.

It uses contextual information from both directions when creating token representations.

### Common Uses

- Text Classification
- Named Entity Recognition
- Question Answering
- Sentence Representation

### BERT

    Input
      ↓
    Transformer Encoder
      ↓
    Contextual Representations
      ↓
    NLP Task


---

# 5. What is GPT?

GPT stands for:

    Generative Pre-trained Transformer

GPT-style models are generally **decoder-only Transformer language models** trained primarily using autoregressive next-token prediction.

### Example

    "Machine learning is"

The model predicts a likely next token based on previous context.

### GPT

    Previous Tokens
         ↓
    Transformer Decoder
         ↓
    Next Token
         ↓
    Next Token
         ↓
       ...


---

# 6. BERT vs GPT

| BERT | GPT |
|---|---|
| Encoder-based | Decoder-only |
| Bidirectional contextual representations | Causal/autoregressive context |
| Strong for representation tasks | Strong for generation |
| Originally pretrained with masked language modeling | Autoregressive next-token prediction |

### Memory Trick

    BERT
    → Understand / Represent

    GPT
    → Generate


---

# 7. Why did Transformers become important?

### 1. Self-Attention

They can directly model relationships between tokens.

### 2. Long-Range Dependencies

Attention allows information to flow between distant tokens.

### 3. Parallel Training

Unlike traditional RNNs, Transformer training can process sequence positions much more efficiently in parallel.

### 4. Scalability

Transformer architectures scale effectively with large datasets and computational resources.

### 5. Foundation for Modern AI

Transformers power many modern systems including:

- BERT-style models
- GPT-style models
- Machine Translation Systems
- Large Language Models
- RAG Systems
- Conversational AI
- Multimodal Models

### One-Line Interview Answer

A Transformer is an attention-based architecture that models relationships between tokens efficiently; BERT uses an encoder for bidirectional contextual representations, while GPT uses a decoder-only architecture for autoregressive generation, making Transformers a foundation of modern NLP and LLM systems.
