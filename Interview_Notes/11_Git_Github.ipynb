# 🔀 Git & GitHub — Interview Questions

---

## Q1. What is Git?

### Interview Answer

Git is a **distributed version control system** used to track changes in source code and other files.

It helps developers:

- Track code changes
- Create branches
- Collaborate with other developers
- Revert changes
- Maintain different versions of a project

### Memory Trick

**Git = Track + Manage + Collaborate**

### One-Line Answer

Git is a distributed version control system used to track and manage changes in a project.

---

## Q2. What is GitHub?

### Interview Answer

GitHub is a cloud-based platform that provides hosting and collaboration features for Git repositories.

It allows developers to:

- Store repositories remotely
- Collaborate with teams
- Review code
- Create pull requests
- Track issues
- Manage project workflows

### Memory Trick

**Git = Version control**

**GitHub = Platform for collaboration**

### One-Line Answer

GitHub is a platform for hosting Git repositories and collaborating on software projects.

---

## Q3. What is the difference between Git and GitHub?

### Interview Answer

Git is the **version control system** that runs on our computer.

GitHub is a **platform** that can host Git repositories remotely and provide collaboration features.

### Example

I can use Git without GitHub:

    git init

And I can push the repository to GitHub:

    git push origin main

### Memory Trick

**Git → Tool**

**GitHub → Platform**

### One-Line Answer

Git is a version control tool, while GitHub is a platform for hosting Git repositories and collaboration.

## Q4. What is a Git repository?

### Interview Answer

A Git repository is a project directory that Git uses to track files and their history.

It contains Git's internal metadata, typically inside the `.git` directory.

### Example

        .git/

### Memory Trick

**Repository = Project + Git history**

### One-Line Answer

A Git repository is a project managed by Git that contains files and their version history.

## Q7. What is the staging area in Git?

### Interview Answer

The staging area is an intermediate area where we select changes that we want to include in the next commit.

We add changes to the staging area using `git add`.

### Example

    git add app.py

Now the changes to `app.py` are staged for the next commit.

### Memory Trick

**Working → Staging → Commit**

### One-Line Answer

The staging area contains selected changes that are prepared for the next commit.

---

## Q8. What is a Git commit?

### Interview Answer

A commit is a recorded snapshot of staged changes in the Git repository.

A commit usually contains:

- Changes
- Author information
- Timestamp
- Commit message
- Reference to previous history

### Example

    git add app.py
    git commit -m "Add API endpoint"

### Memory Trick

**Commit = Save a snapshot**

### One-Line Answer

A commit records a snapshot of staged changes in Git history.

## Q9. What is the difference between `git add` and `git commit`?

### Interview Answer

`git add` moves selected changes from the working directory to the staging area.

`git commit` records the staged changes in the repository history.

### Flow

    Working Directory
          ↓
       git add
          ↓
    Staging Area
          ↓
      git commit
          ↓
    Repository History

### Memory Trick

**add = Prepare**

**commit = Record**

### One-Line Answer

`git add` stages changes, while `git commit` records those staged changes in Git history.

## Q13. What is the difference between `git merge` and `git rebase`?

### Interview Answer

Both are used to integrate changes from one branch into another, but they work differently.

`git merge` combines histories and usually creates a merge commit when needed.

`git rebase` moves or replays commits onto another base commit, creating a more linear history.

### Example

Merge:

    git switch main
    git merge feature

Rebase:

    git switch feature
    git rebase main

### Important Point

Rebase rewrites commit history, so it should be used carefully, especially with commits that have already been shared with others.

### Memory Trick

**Merge = Combine histories**

**Rebase = Replay on new base**

### One-Line Answer

Merge combines branch histories, while rebase rewrites the branch history by replaying commits onto a new base.

---

## Q14. What is a merge conflict?

### Interview Answer

A merge conflict occurs when Git cannot automatically determine which changes should be kept.

This commonly happens when different branches modify the same part of a file.

### Typical Process

1. Run merge or rebase.
2. Git reports a conflict.
3. Open the conflicting file.
4. Decide which changes to keep.
5. Remove conflict markers.
6. Stage the resolved file.
7. Complete the merge or rebase.

### Memory Trick

**Conflict = Git needs human decision**

### One-Line Answer

A merge conflict occurs when Git cannot automatically reconcile conflicting changes from different branches.

## Q16. What is `git pull`?

### Interview Answer

`git pull` is used to retrieve changes from a remote repository and integrate them into the current local branch.

Conceptually, it commonly performs:

    git fetch
    +
    git merge

depending on configuration.

### Example

    git pull origin main

### Memory Trick

**pull = Get + Integrate**

### One-Line Answer

`git pull` fetches remote changes and integrates them into the current local branch.

---

## Q17. What is `git fetch`?

### Interview Answer

`git fetch` downloads new commits and other relevant references from a remote repository without automatically integrating those changes into the current branch.

It is useful when we want to inspect remote changes before merging or rebasing them.

### Example

    git fetch origin

### Memory Trick

**fetch = Download, don't integrate**

### One-Line Answer

`git fetch` downloads remote changes without automatically applying them to the current working branch.

## Q20. What is `.gitignore`?

### Interview Answer

`.gitignore` is a file that specifies files and directories that Git should normally ignore when detecting untracked files.

It is commonly used for:

- Virtual environments
- Environment variables
- Cache files
- Build files
- IDE files
- Secrets

### Example

    .env
    __pycache__/
    .venv/
    *.pyc

### Important Point

`.gitignore` does not automatically remove files that are already tracked by Git.

### Memory Trick

**gitignore = Don't track these files**

### One-Line Answer

`.gitignore` specifies files and directories that Git should normally exclude from tracking as untracked files.

## Q24. How would you undo changes in Git?

### Interview Answer

It depends on which stage the changes are in.

If changes are only in the working directory, we can use commands such as:

    git restore <file>

If changes are staged and we want to unstage them:

    git restore --staged <file>

If a commit has already been made and we want to create a new commit that reverses it:

    git revert <commit>

The correct command depends on whether the changes are unstaged, staged, or committed.

### Memory Trick

**Working → restore**

**Staged → restore --staged**

**Committed → revert**

### One-Line Answer

The appropriate Git command depends on whether the change is unstaged, staged, or already committed.

## Q25. What is your typical Git workflow when working on a project?

### Interview Answer

My typical workflow is:

1. Pull the latest changes.
2. Create or switch to a feature branch.
3. Make the required changes.
4. Test the changes.
5. Check the differences.
6. Stage the required files.
7. Commit with a meaningful message.
8. Push the branch to the remote repository.
9. Create a pull request if the team workflow requires it.
10. Address review feedback and merge the changes.

### Example

    git pull origin main

    git switch -c feature-name

    # Make changes

    git status
    git diff

    git add .
    git commit -m "Add feature"

    git push -u origin feature-name

### Memory Trick

**Pull → Branch → Code → Test → Add → Commit → Push → PR → Merge**

### One-Line Answer

My typical Git workflow is pull, create a branch, make and test changes, review, commit, push, and create a pull request when required.
