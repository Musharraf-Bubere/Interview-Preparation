## Q1. Can you explain your AI Travel Planner project?

### Answer:

Sure.

AI Travel Planner is a multi-agent travel planning system that helps users create personalized travel itineraries.

### Problem Statement:

Planning a trip manually can require collecting information about destinations, hotels, activities, restaurants and weather from different sources.

The goal of this project was to build an AI system that can understand the user's travel requirements and generate a personalized and practical travel plan.

### Solution:

I built the project using a multi-agent architecture.

The user provides requirements such as the destination, travel dates, duration, number of travelers, budget and preferences.

The request is processed through a LangGraph-based workflow, where different specialized agents handle different parts of the travel-planning process.

For example, the Destination Agent researches the destination, the Stay Agent finds accommodation options, the Activity Agent researches activities, the Weather Agent checks weather information, and the Food Agent researches restaurants.

The Itinerary Agent then combines this information and generates the complete itinerary.

After that, an Itinerary Validator checks the generated plan for issues such as incorrect duration, duplicate places, geographical conflicts, budget problems, weather considerations and preference matching.

If problems are found, the Refine Itinerary Agent improves the itinerary and sends it back for validation.

Once the itinerary is valid, the Final Response Agent prepares the final travel plan.

I used Gemini as the LLM and integrated external tools and APIs such as Tavily, SerpApi and WeatherAPI to collect travel information.

LangGraph is used to orchestrate the workflow and handle sequential, parallel, conditional and iterative execution.

### Architecture:

```User

↓

FastAPI Backend

↓

LangGraph Workflow

↓

Destination Agent

↓

┌────────────┬────────────┬────────────┐
│            │            │
Stay       Activity     Weather
Agent        Agent        Agent
│            │            │
└────────────┴────────────┴────────────┘

↓

Food Agent

↓

Itinerary Agent

↓

Itinerary Validator

↓

Valid → Final Response

Invalid → Refine Itinerary → Validator

↓

Personalized Travel Itinerary
```

### Result:


### Why LangGraph?

I used LangGraph because it allows me to represent the workflow as nodes and edges, maintain shared state between agents, and implement sequential, parallel, conditional, and iterative workflows.

This gives me better control over the execution flow compared with simply sending everything to one LLM.

### Memory Trick:

User → Destination → Parallel Research → Food → Itinerary → Validate → Refine/Final

### Short Answer:

"LangGraph controls the complete workflow. It manages the agents, shared state, parallel research, validation, and the refinement loop. If the itinerary is valid, it goes to the final response; otherwise, it is refined and validated again."

## Q9. Why did you use LangGraph instead of using a single LLM?

### Answer:

I used LangGraph because the travel-planning problem has multiple different tasks, and each task has a different responsibility.

If I used a single LLM, it would have to research the destination, find hotels, find activities, check weather, find restaurants, generate the itinerary, and validate everything.

That would make the system harder to debug, test, and control.

With LangGraph, I can divide the problem into specialized agents. For example, the Stay Agent handles accommodation research, the Weather Agent handles weather information, and the Activity Agent handles activities.

I can also run independent tasks in parallel, use conditional decisions after validation, and create an iterative refinement loop when the itinerary has problems.

Another important advantage is that the agents share a common state, so information collected by one part of the workflow can be used by another part.

So, I chose LangGraph because it gives me better workflow control, modularity, debugging, testing, and scalability for a multi-step Agentic AI application.

### Example:

Instead of:

User → One LLM → Final Itinerary

I use:

User → LangGraph → Specialized Agents → Itinerary → Validation → Refinement → Final Response

### Memory Trick:

Single LLM = Everything Together  
LangGraph = Specialized + Controlled + Connected

### Short Answer:

"I used LangGraph because my application has multiple specialized tasks. It allows me to separate responsibilities, run independent tasks in parallel, maintain shared state, apply conditional logic, and create a validation-refinement loop. This makes the system easier to control, test, and debug."

## Q10. What is the role of LangGraph in your AI Travel Planner?

### Answer:

LangGraph is mainly used to control and orchestrate the complete workflow of my multi-agent system.

In my project, each agent is represented as a part of the workflow, and LangGraph controls how information moves from one component to another.

It also maintains a shared state called TravelState. This state contains the user's travel requirements and the information collected by different agents, such as accommodation options, activities, weather, restaurants, itinerary and validation results.

LangGraph also helps me implement different workflow patterns.

For example:

- Sequential workflow for tasks that need to happen one after another.
- Parallel workflow for independent tasks such as accommodation, activities and weather research.
- Conditional workflow to decide whether the itinerary should go to the final response or refinement.
- Iterative workflow to repeatedly refine and validate the itinerary.

So, LangGraph acts as the workflow engine that coordinates the different agents and controls the execution of the travel-planning process.

### Example:

User Request
↓
LangGraph
↓
Agents
↓
Shared TravelState
↓
Itinerary
↓
Validation
↓
Refinement or Final Response

### Memory Trick:

LangGraph = **Orchestrate + State + Control + Decisions**

### One-Line Answer:

"LangGraph is the orchestration layer of my project. It manages the agents, shared state, workflow execution, conditional decisions, parallel processing and the validation-refinement loop."

## Q11. What is TravelState in your AI Travel Planner, and why did you use it?

### Answer:

TravelState is the shared state used by my LangGraph workflow.

It stores the information required by different agents during the travel-planning process.

For example, it contains the user's inputs such as:

- Destination
- Travel dates
- Duration
- Number of travelers
- Budget
- Preferences

It also stores the information generated or collected during the workflow, such as:

- Destination information
- Accommodation options
- Activities
- Weather information
- Restaurants
- Generated itinerary
- Validation results
- Validation attempts
- Final response

Each agent can read the information it needs from the shared state and update the state with its results.

I used shared state because the agents need to exchange information with each other. For example, the Itinerary Agent needs the accommodation, activity, weather and restaurant information collected by previous agents.

So, TravelState acts as a common information layer that connects the different parts of my LangGraph workflow.

### Simple Example:

User Input
↓
TravelState
↓
Destination Agent → updates destination data
↓
Stay / Activity / Weather Agents → update their results
↓
Food Agent
↓
Itinerary Agent → reads all collected information
↓
Validator → updates validation result
↓
Refinement / Final Response

### Memory Trick:

**TravelState = Shared Memory of the Workflow**

### One-Line Answer:

"TravelState is the shared state in my LangGraph application. It stores user requirements and the results produced by different agents, allowing the agents to communicate and use information collected earlier in the workflow."

## Q12. What are the different workflow patterns you used in LangGraph?

### Answer:

I used four main workflow patterns in my AI Travel Planner: sequential, parallel, conditional, and iterative workflows.

### 1. Sequential Workflow:

In a sequential workflow, one task is completed before the next task starts.

For example:

Destination → Food → Itinerary → Validation → Final Response

This is useful when the next step depends on the output of the previous step.

### 2. Parallel Workflow:

In a parallel workflow, independent tasks can run at the same time.

For example, after destination research:

Stay Agent
Activity Agent
Weather Agent

can work independently.

This can reduce unnecessary waiting because these agents do not need each other's results at that stage.

### 3. Conditional Workflow:

In a conditional workflow, the next step depends on a condition.

In my project, after the itinerary is validated:

If valid → Final Response

If invalid → Refine Itinerary

So the validator controls which path the workflow follows.

### 4. Iterative Workflow:

In an iterative workflow, a process can repeat until a condition is satisfied or a maximum number of attempts is reached.

In my project, if the itinerary is invalid, the Refine Itinerary Agent fixes the problems and sends it back to the Validator.

This loop continues until the itinerary becomes valid or the maximum number of validation attempts is reached.

### Simple Example:

Sequential:
Destination → Food → Itinerary

Parallel:
Stay + Activity + Weather

Conditional:
Valid → Final
Invalid → Refine

Iterative:
Refine → Validate → Refine → Validate

### Memory Trick:

**S → P → C → I**

- S = Sequential
- P = Parallel
- C = Conditional
- I = Iterative

### One-Line Answer:

"My project uses sequential workflows for dependent tasks, parallel workflows for independent research, conditional workflows for decision-making, and iterative workflows for the validation and refinement loop."

## Q13. How does the validation and refinement process work in your AI Travel Planner?

### Answer:

After the Itinerary Agent generates the travel plan, I don't directly send it to the user.

First, the itinerary goes to the Itinerary Validator.

The validator checks different aspects of the itinerary, such as:

- Correct trip duration
- Required fields
- Duplicate places
- Geographic consistency
- Whether the accommodation and activities are based on researched information
- User preferences
- Weather considerations
- Budget

I use a hybrid validation approach.

Some checks are performed programmatically using deterministic rules, such as checking the number of days and detecting duplicates.

Other checks require semantic understanding, such as whether the itinerary properly matches the user's preferences and whether the plan is logically suitable. These checks are handled using the LLM.

If the itinerary passes validation, it moves to the Final Response Agent.

If validation fails, the Refine Itinerary Agent receives the validation issues and modifies the itinerary to fix them.

The refined itinerary is then sent back to the Validator.

This process continues until the itinerary becomes valid or the maximum number of refinement attempts is reached.

### Simple Flow:

```
Itinerary Generated
        ↓
    Validator
        ↓
   ┌────┴────┐
   ↓         ↓
 Valid     Invalid
   ↓         ↓
 Final     Refine
 Response    ↓
           Validator
```

### Why did I use this approach?

Because an LLM-generated itinerary may contain problems such as duplicate locations, incorrect duration, poor geographic planning, or mismatch with user preferences.

The validation and refinement loop provides an additional layer of control instead of blindly trusting the first LLM-generated response.

### Memory Trick:

**Generate → Validate → Valid = Final / Invalid = Refine → Validate Again**

### One-Line Answer:

"After generating the itinerary, I validate it using both deterministic checks and LLM-based semantic checks. If it is invalid, the refinement agent fixes the issues and the itinerary is validated again until it passes or reaches the maximum number of attempts."

## Q14. How do the agents use external tools and APIs in your AI Travel Planner?

### Answer:

In my project, the agents use external tools and APIs when they need real-world travel information.

The basic flow is:

User Request
↓
Agent
↓
Tool Call
↓
External API
↓
Tool Result
↓
Agent
↓
Structured Output

For example, the Destination Agent uses Tavily to research destination information.

The Stay Agent uses SerpApi Google Hotels to find accommodation options such as hotels, prices, ratings and locations.

The Activity Agent and Food Agent use SerpApi Google Maps to research activities and restaurants.

The Weather Agent uses WeatherAPI to get weather information for the travel period.

The agents then process these results and provide structured information to the next part of the LangGraph workflow.

I used external tools because the LLM itself may not have the latest real-world information about hotels, restaurants, weather or activities.

### Examples:

- Destination Agent → Tavily
- Stay Agent → SerpApi Google Hotels
- Activity Agent → SerpApi Google Maps
- Food Agent → SerpApi Google Maps
- Weather Agent → WeatherAPI

### Why are tools important?

Tools allow the agents to access external information instead of relying only on the knowledge of the LLM.

This makes the travel plan more grounded in researched information.

### Memory Trick:

**Agent → Tool → API → Result → Agent**

### One-Line Answer:

"I use external tools and APIs to provide real-world travel information to the agents. For example, Tavily is used for destination research, SerpApi for hotels, activities and restaurants, and WeatherAPI for weather information."

## Q15. How does the Itinerary Agent generate the travel itinerary?

### Answer:

The Itinerary Agent is responsible for generating the complete travel itinerary using the information collected by the other agents.

Before generating the itinerary, it receives the user's travel requirements such as:

- Destination
- Travel dates
- Duration
- Number of travelers
- Budget
- Preferences

It also uses the research collected by the other agents, such as:

- Destination information
- Accommodation options
- Activities
- Weather information
- Restaurants

The agent then combines this information and creates a day-by-day itinerary.

While generating the itinerary, it considers factors such as the user's preferences, budget, weather, and geographical consistency.

An important part of the design is that the itinerary should use researched locations and restaurants instead of inventing unsupported places or information.

After the itinerary is generated, it is not immediately returned to the user. It is passed to the Itinerary Validator for further checking.

### Simple Flow:

User Requirements
        ↓
Research from Agents
        ↓
Itinerary Agent
        ↓
Day-by-Day Itinerary
        ↓
Itinerary Validator
        ↓
Valid / Refine

### Example:

If the user requests:

- Goa
- 3 days
- Budget of ₹30,000
- Beaches, food and relaxation

The Itinerary Agent uses the researched hotels, activities, restaurants and weather information to create a three-day plan while considering those preferences and the available budget.

### Memory Trick:

**User Requirements + Agent Research → Itinerary Agent → Day-by-Day Plan → Validation**

### One-Line Answer:

"The Itinerary Agent combines the user's travel requirements with the information collected by the research agents and generates a personalized day-by-day itinerary, which is then passed to the validator for checking."

## Q16. Why did you use Pydantic in your AI Travel Planner?

### Answer:

I used Pydantic for structured data validation and structured outputs in my project.

Since different agents produce information that is passed to other parts of the workflow, I wanted the data to follow a defined structure.

For example, I used Pydantic models for things such as:

- Destination analysis
- Accommodation information
- Activity information
- Restaurant information
- Itinerary validation

This helps make sure that the data returned by the agents follows the expected format.

Pydantic is also useful when working with FastAPI because it can validate API request and response data.

In my project, structured outputs make the communication between the agents and the rest of the application more predictable and easier to validate.

### Example:

Instead of allowing an agent to return completely unstructured text:

"Here are some hotels in Goa..."

I can define an expected structure containing fields such as:

- Hotel name
- Location
- Price
- Rating

The application can then validate that structure before using the information.

### Why is this important in an Agentic AI system?

LLMs can generate flexible responses, so relying only on free-form text can make downstream processing difficult.

Using structured output with Pydantic gives the application a predictable format and helps catch invalid data early.

### Memory Trick:

**Pydantic = Structure + Validation + Reliability**

### One-Line Answer:

"I used Pydantic to define and validate structured data exchanged between the agents and the backend. It makes LLM outputs and API data more predictable, reliable, and easier to process."

## Q17. What happens if one of the external APIs or tools fails in your AI Travel Planner?

### Answer:

If an external API or tool fails, I would first identify which part of the workflow is affected instead of assuming that the entire application has failed.

I would follow a troubleshooting approach:

1. Reproduce the issue and identify the failing agent.
2. Check the API request, parameters and input data.
3. Check the application logs and error response.
4. Verify whether the external API is available and whether the API configuration is correct.
5. Check whether the tool returned valid or incomplete data.
6. Identify the root cause.
7. Fix the issue and test the affected workflow again.

For example, if the Weather Agent is not receiving weather data, I would check the WeatherAPI request, destination, travel dates, API configuration and returned response.

I would also make sure that invalid or missing tool results are not blindly passed to the next agent.

### Simple Troubleshooting Flow:

API Failure
↓
Identify Agent
↓
Check Request
↓
Check Logs / Error
↓
Check API Response
↓
Find Root Cause
↓
Fix
↓
Test Workflow

### Example:

If the Stay Agent fails:

Stay Agent
↓
SerpApi Google Hotels
↓
Error
↓
Check request + API configuration + response
↓
Fix
↓
Run Stay Agent again
↓
Continue workflow

### Memory Trick:

**Reproduce → Check → Root Cause → Fix → Test**

### One-Line Answer:

"If an external API fails, I would identify the affected agent, reproduce the issue, check the request, logs, API response and configuration, find the root cause, fix it, and test the workflow again."

## Q18. How did you handle hallucinations in your AI Travel Planner?

### Answer:

Since the project uses LLMs to generate travel information and itineraries, there is a possibility that the model could generate unsupported or incorrect information.

To reduce this problem, I designed the system so that the itinerary is based on information collected from external tools and APIs.

The Itinerary Agent receives researched information from the other agents and is expected to use those researched locations, restaurants and accommodation options instead of inventing unsupported places.

I also added an Itinerary Validator after itinerary generation.

The validator checks whether the generated itinerary is grounded in the researched information and whether there are other issues such as duplicate places, incorrect duration, geographical conflicts, budget problems and preference mismatches.

If the validator finds problems, the Refine Itinerary Agent fixes them and the itinerary goes through validation again.

So, instead of completely trusting the LLM's first response, I use external research, structured outputs, deterministic checks, LLM-based validation and iterative refinement to improve reliability.

### Simple Flow:

External Research
↓
LLM Generates Itinerary
↓
Validation
↓
Valid → Final Response
↓
Invalid → Refine
↓
Validate Again

### Memory Trick:

**Ground → Generate → Validate → Refine**

### One-Line Answer:

"I reduce hallucinations by grounding the itinerary in external research, using structured outputs, validating the generated plan, and refining it when unsupported or inconsistent information is detected."

## Q19. What was the biggest challenge you faced while building the AI Travel Planner?

### Answer:

The biggest challenge was making the AI-generated itinerary reliable and logically consistent.

An LLM can generate a good itinerary, but the first response may contain problems such as duplicate places, incorrect trip duration, geographical conflicts, budget issues, or information that was not present in the research data.

To handle this, I designed a validation and refinement process.

First, the Itinerary Agent generates the itinerary using the information collected by the research agents.

Then, the Itinerary Validator checks the itinerary using both deterministic programmatic checks and LLM-based semantic validation.

If problems are found, the Refine Itinerary Agent fixes those issues and sends the itinerary back to the validator.

This creates a controlled loop where the itinerary is improved before the final response is given to the user.

Another challenge was coordinating multiple agents and passing information between them. I handled this using LangGraph and a shared TravelState.

### My Approach:

Problem
↓
Understand the issue
↓
Add validation
↓
Identify the validation failure
↓
Refine the itinerary
↓
Validate again
↓
Final Response

### Memory Trick:

**Challenge → Validation → Refinement → Reliable Result**

### One-Line Answer:

"The biggest challenge was making the LLM-generated itinerary reliable. I addressed this using external research, shared state, structured outputs, hybrid validation, and an iterative refinement loop."

## Q19. What was the biggest challenge you faced while building the AI Travel Planner?

### Answer:

The biggest challenge was making the AI-generated itinerary reliable and logically consistent.

An LLM can generate a good itinerary, but the first response may contain problems such as duplicate places, incorrect trip duration, geographical conflicts, budget issues, or information that was not present in the research data.

To handle this, I designed a validation and refinement process.

First, the Itinerary Agent generates the itinerary using the information collected by the research agents.

Then, the Itinerary Validator checks the itinerary using both deterministic programmatic checks and LLM-based semantic validation.

If problems are found, the Refine Itinerary Agent fixes those issues and sends the itinerary back to the validator.

This creates a controlled loop where the itinerary is improved before the final response is given to the user.

Another challenge was coordinating multiple agents and passing information between them. I handled this using LangGraph and a shared TravelState.

### My Approach:

```

```

### Memory Trick:

**Reproduce → Node → State → Logs → Root Cause → Fix → Test**

### One-Line Answer:

"If the LangGraph workflow fails, I first reproduce the issue and identify the failed node. Then I check its inputs, shared state, logs, API or LLM response, and workflow routing to find the root cause. After fixing it, I test the affected node and the complete workflow again."

## Q23. What would you do if an agent gives incorrect or incomplete output?

### Answer:

If an agent gives incorrect or incomplete output, I would first identify which agent produced the problem and reproduce the issue.

Then I would check the input given to the agent and the information available in the shared TravelState.

After that, I would check the external tool or API response, because incomplete research data could affect the agent's output.

I would also check the prompt, LLM response, and Pydantic structured output to see whether the response follows the expected format.

For the itinerary specifically, the Itinerary Validator provides another layer of checking. It can identify problems such as missing information, duplicate places, incorrect duration, geographical conflicts, budget issues, or preference mismatches.

If the output is invalid, the Refine Itinerary Agent can use the validation issues to improve the itinerary.

After making the correction, I would test the agent again and then test the complete workflow to make sure the issue is resolved.

### Debugging Flow:

Incorrect Output
↓
Identify Agent
↓
Check Input / TravelState
↓
Check Tool / API Result
↓
Check Prompt / LLM Response
↓
Check Pydantic Output
↓
Validate
↓
Refine / Fix
↓
Retest

### Example:

If the Activity Agent returns incomplete activity information, I would check:

1. Was the destination passed correctly?
2. Did the external API return enough information?
3. Did the agent process the API response correctly?
4. Did the LLM follow the expected structure?
5. Did Pydantic validation detect missing fields?
6. Can the issue be corrected and tested again?

### Memory Trick:

**Agent → Input → Tool → LLM → Structure → Validate → Fix → Test**

### One-Line Answer:

"If an agent gives incorrect or incomplete output, I would trace the problem from the input and shared state through the tool response, LLM output and structured validation, identify the root cause, fix it, and test the workflow again."

## Q24. How would you improve your AI Travel Planner in the future?

### Answer:

There are several areas I would improve in the future.

First, I would complete the planned MCP integration so that tools can be connected through a more standardized tool interface.

Second, I would add persistence so that users can save and retrieve their previous travel plans.

I would also add Docker containerization, CI/CD, and deployment so the application can be packaged and deployed more easily.

Another improvement would be observability, so that I can monitor agent execution, API failures, latency, validation attempts, and other workflow information.

I would also improve the agents and validation system further by adding better evaluation and testing for itinerary quality.

For example, I could evaluate whether generated itineraries consistently satisfy user preferences, budget, geography, weather and duration requirements.

### Future Improvements:

- MCP integration
- Persistence
- Docker
- CI/CD
- Deployment
- Observability
- Better agent evaluation
- More advanced travel-planning capabilities

### Memory Trick:

**MCP → Persistence → Docker → CI/CD → Deployment → Observability → Evaluation**

### One-Line Answer:

"In the future, I would focus on completing MCP integration, adding persistence, Docker, CI/CD and deployment, and improving observability and evaluation so the system becomes more production-ready."

## Q25. Can you explain your AI Travel Planner project in 1–2 minutes?

### Answer:

Sure.

AI Travel Planner is a multi-agent travel planning system that generates personalized travel itineraries based on the user's destination, dates, duration, budget, number of travelers and preferences.

The main idea was to divide the travel-planning process into specialized agents instead of depending on a single LLM.

I used LangGraph to orchestrate the workflow. The system has specialized components for destination research, accommodation, activities, weather and restaurants. These agents use external tools and APIs such as Tavily, SerpApi and WeatherAPI to collect relevant information.

The Itinerary Agent then combines this information with the user's requirements and generates a day-by-day itinerary.

After that, the Itinerary Validator checks the itinerary for issues such as incorrect duration, duplicate places, geographical conflicts, budget problems and preference mismatches.

If the itinerary is invalid, the Refine Itinerary Agent fixes the identified issues and sends it back for validation. Once it passes validation, the Final Response Agent prepares the final travel plan.

I used Gemini for LLM capabilities, Pydantic for structured outputs, and FastAPI for the backend API.

My main contribution was on the backend and AI side, especially the LangGraph workflow, agents, LLM integration, external tool/API integration, itinerary generation, validation and refinement.

Overall, this project gave me practical experience in building a controlled multi-agent AI application and troubleshooting a multi-step workflow.

### Memory Trick:

**Problem → Multi-Agent → LangGraph → Tools → Itinerary → Validate → Refine → Final → My Contribution**

### One-Line Closing:

"Overall, the project combines Agentic AI, workflow orchestration, external tools, structured outputs and validation to generate more reliable personalized travel plans."
