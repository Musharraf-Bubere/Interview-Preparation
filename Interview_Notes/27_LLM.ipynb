# 🤖 LLM INTERVIEW PREPARATION
# 25 Questions | Generic Technical Interview Preparation

---

## Q1. What is an LLM?

### Interview Answer

LLM stands for Large Language Model.

It is a type of neural network model trained on a very large amount of text data to understand and generate human-like language.

Modern LLMs are generally based on the Transformer architecture and learn statistical patterns in language.

Examples of LLM capabilities include:

- Text generation
- Question answering
- Summarization
- Translation
- Classification
- Information extraction
- Code generation
- Reasoning and tool use

### Memory Trick

**LLM = Large-Scale Language Understanding + Generation**

### One-Line Answer

An LLM is a large neural network trained on massive amounts of text to process and generate human-like language.

These concepts are related but represent different levels.

    Artificial Intelligence
          ↓
    Machine Learning
          ↓
    Deep Learning
          ↓
    Transformer-based Models
          ↓
          LLM

AI is the broader field of building systems that perform tasks requiring intelligence.

Machine Learning allows systems to learn patterns from data.

Deep Learning uses neural networks with multiple layers.

LLMs are large deep learning models specialized in processing and generating language.

### Memory Trick

**AI → ML → DL → LLM**

### One-Line Answer

AI is the broad field, ML is learning from data, deep learning uses neural networks, and LLMs are large deep learning models designed primarily for language tasks.

## Q3. What is a Transformer?

### Interview Answer

A Transformer is a neural network architecture introduced for sequence processing and now widely used in language models.

Its key idea is the self-attention mechanism, which allows the model to consider relationships between tokens in a sequence.

Transformers can process sequence information efficiently and support highly parallel training.

### Basic Flow

    Input Text
       ↓
    Tokenization
       ↓
    Embeddings
       ↓
    Transformer Layers
       ↓
    Output

### Memory Trick

**Transformer = Attention + Parallel Processing**

### One-Line Answer

A Transformer is a neural network architecture based heavily on attention mechanisms and is the foundation of most modern LLMs.

## Q4. What is a token?

### Interview Answer

A token is a unit of text processed by an LLM.

Depending on the tokenizer, a token can represent a complete word, part of a word, punctuation, or another text unit.

For example, a word may be represented as one token or several subword tokens.

### Flow

    Text
      ↓
    Tokenizer
      ↓
    Tokens
      ↓
    Token IDs
      ↓
    Model

### Memory Trick

**Text → Tokens → IDs**

### One-Line Answer

A token is a unit of text produced by a tokenizer that the LLM processes.

---

## Q5. What is tokenization?

### Interview Answer

Tokenization is the process of converting text into smaller units called tokens.

The tokenizer then maps these tokens to numerical token IDs that can be processed by the model.

### Example

    "I love AI"

         ↓

    ["I", "love", "AI"]

         ↓

    [token_id_1, token_id_2, token_id_3]

The exact tokens depend on the tokenizer.

### Memory Trick

**Tokenization = Text → Tokens**

### One-Line Answer

Tokenization converts text into tokens and then maps those tokens to numerical IDs that the model can process.

---

## Q6. What are embeddings?

### Interview Answer

Embeddings are numerical vector representations of data such as words, tokens, sentences, documents, or other objects.

They represent semantic or learned relationships in a continuous vector space.

Similar meanings can often have vectors that are closer together according to an appropriate similarity measure.

### Example

    "King"
       ↓
    [0.21, -0.14, 0.73, ...]

    "Queen"
       ↓
    [0.19, -0.12, 0.70, ...]

### Memory Trick

**Embedding = Meaning as Numbers**

### One-Line Answer

An embedding represents text or another object as a numerical vector that captures learned semantic information.

## Q7. What is positional encoding?

### Interview Answer

Transformers process tokens in parallel, so they need information about token order.

Positional information allows the model to distinguish between different positions in a sequence.

Depending on the Transformer architecture, this information can be provided using techniques such as positional encodings or learned/rotary positional representations.

### Example

    "Dog bites man"

    is different from

    "Man bites dog"

The order matters.

### Memory Trick

**Attention sees relationships**

**Position provides order**

### One-Line Answer

Positional information gives Transformer models information about where tokens occur in a sequence.

---

## Q8. What is self-attention?

### Interview Answer

Self-attention allows each token to determine which other tokens in the same sequence are relevant to it.

It helps the model capture relationships between words or tokens, even when they are far apart in the sequence.

### Conceptual Flow

    Tokens
      ↓
    Query
    Key
    Value
      ↓
    Attention Scores
      ↓
    Weighted Information
      ↓
    Context-Aware Representation

### Example

In:

    "The animal didn't cross the road because it was tired."

The model needs to understand what "it" refers to from the surrounding context.

### Memory Trick

**Self-Attention = Which Tokens Matter to Me?**

### One-Line Answer

Self-attention allows each token to weigh the importance of other tokens in the same sequence when creating its representation.

---

## Q9. What are Query, Key, and Value in attention?

### Interview Answer

In the attention mechanism, each input representation is transformed into three vectors:

- Query
- Key
- Value

The Query represents what a token is looking for.

The Key represents what information each token offers for matching.

The Value contains the information that is aggregated based on the attention scores.

Conceptually:

    Query × Keys
         ↓
    Attention Scores
         ↓
    Weighted Values
         ↓
    Attention Output

### Memory Trick

**Query = What am I looking for?**

**Key = What do I represent?**

**Value = What information do I provide?**

### One-Line Answer

Queries and keys determine attention relevance, while values provide the information that is combined according to those attention weights.

---

## Q10. What is multi-head attention?

### Interview Answer

Multi-head attention runs multiple attention mechanisms, called heads, in parallel.

Different heads can learn different relationships or patterns in the input.

Their outputs are then combined.

### Flow

    Input
      ↓
    ┌───────────────┐
    │ Attention Head 1 │
    │ Attention Head 2 │
    │ Attention Head 3 │
    │      ...         │
    └───────────────┘
      ↓
    Concatenate
      ↓
    Linear Projection
      ↓
    Output

### Memory Trick

**Multiple Heads = Multiple Relationship Views**

### One-Line Answer

Multi-head attention uses multiple attention heads in parallel so the model can learn different relationships within the sequence.

## Q11. What is the difference between encoder and decoder architectures?

### Interview Answer

A Transformer encoder is designed to create contextual representations of an input sequence.

A Transformer decoder generates output tokens and, in encoder-decoder architectures, can attend to encoder outputs.

Common architectural patterns include:

**Encoder-only**
→ BERT-style models

**Decoder-only**
→ GPT-style language models

**Encoder-decoder**
→ T5-style models

### Memory Trick

**Encoder = Understand**

**Decoder = Generate**

### One-Line Answer

Encoder models primarily build representations of input text, decoder models generate text autoregressively, and encoder-decoder models combine both roles.

---

## Q12. What is autoregressive language modeling?

### Interview Answer

Autoregressive language modeling generates tokens sequentially.

The model predicts the next token based on the tokens that have already been provided.

### Example

    "The cat is"

        ↓

    Predict next token

        ↓

    "sleeping"

        ↓

    Predict next token

        ↓

    Continue...

Mathematically, the model learns to estimate the probability of the next token given the previous tokens.

### Memory Trick

**Previous Tokens → Next Token**

### One-Line Answer

Autoregressive language modeling predicts the next token based on the preceding context.

## Q13. What is pretraining?

### Interview Answer

Pretraining is the initial large-scale training phase where an LLM learns general language patterns from a massive dataset.

Depending on the model architecture, the objective can involve predicting missing tokens, next tokens, or another self-supervised objective.

The resulting model learns general representations and language capabilities.

### Flow

    Large Dataset
         ↓
    Tokenization
         ↓
    Pretraining
         ↓
    Base Model
         ↓
    Fine-Tuning / Alignment
         ↓
    Application

### Memory Trick

**Pretraining = Learn General Patterns**

### One-Line Answer

Pretraining teaches an LLM general language patterns using large-scale datasets and self-supervised learning objectives.

## Q14. What is fine-tuning?

### Interview Answer

Fine-tuning is additional training performed on a pretrained model using a more specific dataset or objective.

It can adapt the model to a particular task, domain, behavior, or style.

Examples include adapting a model for:

- Domain-specific language
- Classification
- Instruction following
- Specialized generation

### Memory Trick

**Pretraining = General**

**Fine-Tuning = Specific**

### One-Line Answer

Fine-tuning adapts a pretrained model to a specific task, domain, or behavior using additional training data.

---

## Q15. What is instruction tuning?

### Interview Answer

Instruction tuning is a form of fine-tuning where the model is trained on examples that pair instructions with desired responses.

The goal is to improve the model's ability to follow natural-language instructions.

### Example

    Instruction:
    "Summarize this paragraph."

    Expected Response:
    "The paragraph explains..."

The model learns the relationship between instructions and appropriate outputs.

### Memory Trick

**Instruction Tuning = Learn to Follow Instructions**

### One-Line Answer

Instruction tuning fine-tunes a pretrained model on instruction-response examples to improve instruction following.

RLHF stands for Reinforcement Learning from Human Feedback.

It is a method used to align model behavior with human preferences.

A simplified workflow is:

    Pretrained Model
          ↓
    Instruction Tuning
          ↓
    Human Preference Data
          ↓
    Reward Modeling
          ↓
    Reinforcement Learning
          ↓
    Aligned Model

The exact alignment pipeline can vary between models.

### Memory Trick

**RLHF = Human Preferences → Better-Aligned Behavior**

### One-Line Answer

RLHF is an alignment approach that uses human preference feedback to optimize model behavior.

## Q17. What is inference in an LLM?

### Interview Answer

Inference is the process of using a trained model to generate predictions or outputs for new input.

For a generative LLM, inference generally involves:

    Prompt
      ↓
    Tokenization
      ↓
    Model
      ↓
    Next-Token Probabilities
      ↓
    Token Selection
      ↓
    Generated Text

Inference does not update the model's parameters.

### Memory Trick

**Training = Learn**

**Inference = Use**

### One-Line Answer

Inference is the process of using a trained LLM to generate an output for new input without updating its parameters.

## Q18. What are temperature and top-p?

### Interview Answer

Temperature and top-p are decoding parameters that influence how the model selects tokens during generation.

**Temperature**

Controls how sharply or broadly the model's probability distribution is used during sampling.

Lower temperature generally makes output more predictable, while higher temperature can increase randomness.

**Top-p**

Nucleus sampling limits candidate tokens to the smallest set whose cumulative probability reaches a specified threshold.

### Example

    Low Temperature
        ↓
    More Predictable

    Higher Temperature
        ↓
    More Variable

### Memory Trick

**Temperature = Randomness Level**

**Top-p = Probability Nucleus**

### One-Line Answer

Temperature controls sampling sharpness, while top-p limits sampling to a probability mass containing the most likely candidate tokens.

## Q19. What is hallucination in LLMs?

### Interview Answer

Hallucination occurs when an LLM generates information that is unsupported, incorrect, or fabricated while presenting it as if it were reliable.

This can happen because language models generate outputs based on learned patterns and probabilities rather than inherently verifying every factual statement.

### Example

    User:
    "Give me the source of this claim."

    Model:
    Produces a citation that does not actually exist.

This is a hallucination.

### Ways to Reduce Hallucinations

- Retrieval-Augmented Generation
- Grounding responses in trusted data
- Tool use
- Better prompting
- Structured outputs
- Verification
- Human review for high-risk applications

### Memory Trick

**Hallucination = Confident but Unsupported Output**

### One-Line Answer

An LLM hallucination is an unsupported or incorrect generated statement presented as though it were factual.

## Q20. What is context window?

### Interview Answer

The context window is the amount of tokenized information a model can consider within a particular request or conversation context.

It can include:

- User input
- System instructions
- Conversation history
- Retrieved documents
- Tool results
- Other supplied context

If the supplied context exceeds the model's supported limit, information may need to be truncated, summarized, or otherwise managed.

### Memory Trick

**Context Window = Model's Available Context**

### One-Line Answer

A context window is the maximum amount of tokenized context that a model can process for a given interaction.

## Q22. What is RAG and why is it used with LLMs?

### Interview Answer

RAG stands for Retrieval-Augmented Generation.

It combines information retrieval with LLM generation.

Instead of relying only on the model's internal learned knowledge, relevant external documents are retrieved and provided as context to the model.

### Flow

    User Query
        ↓
    Embedding / Retrieval
        ↓
    Relevant Documents
        ↓
    Context
        ↓
    LLM
        ↓
    Grounded Answer

### Why Use RAG?

RAG can help with:

- Private company documents
- Frequently changing information
- Domain-specific knowledge
- Source-grounded responses
- Reducing unsupported answers

### Memory Trick

**RAG = Retrieve → Augment → Generate**

### One-Line Answer

RAG retrieves relevant external information and supplies it to an LLM as context so the model can generate a more grounded response.

## Q23. What is the difference between RAG and fine-tuning?

### Interview Answer

RAG and fine-tuning solve different problems.

**RAG**

Adds external information at inference time.

Useful when knowledge changes frequently or comes from private documents.

**Fine-Tuning**

Changes the model through additional training.

Useful for adapting behavior, style, task performance, or domain-specific patterns.

### Comparison

    RAG
    → Change Knowledge Context

    Fine-Tuning
    → Change Model Behavior / Learned Parameters

### Example

If a company wants an assistant to answer questions from constantly changing internal policies:

    RAG
    → Usually a strong approach

If a company wants the model to consistently follow a specialized output style:

    Fine-Tuning
    → May be appropriate

### Memory Trick

**RAG = Give Information**

**Fine-Tuning = Train Behavior**

### One-Line Answer

RAG supplies external knowledge during inference, while fine-tuning updates model parameters using additional training data.

# ⭐ INTERVIEW GOLDEN QUESTION

## "How does an LLM generate text?"

### Strong Answer

An LLM first converts the input text into tokens and token IDs.

These tokens are transformed into embeddings and processed through Transformer layers containing attention and other neural-network components.

The model produces a probability distribution over possible next tokens.

A decoding strategy then selects the next token.

That token is added to the context, and the process repeats until the model reaches a stopping condition.

### Short Version

**Text → Tokens → Embeddings → Transformer → Probabilities → Next Token → Repeat**


# ⭐ INTERVIEW GOLDEN QUESTION

## "Why can an LLM hallucinate?"

### Strong Answer

An LLM is fundamentally trained to model patterns in data and generate likely sequences of tokens. It does not automatically verify every generated statement against a trusted source.

Therefore, it can produce fluent but unsupported information.

For applications requiring factual grounding, techniques such as RAG, tool use, source verification, structured outputs, and evaluation can help reduce this risk.
