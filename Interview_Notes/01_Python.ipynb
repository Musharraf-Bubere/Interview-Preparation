## Python — Q3. What is the difference between mutable and immutable objects in Python?

### Answer:

The main difference is whether the value of an object can be changed after the object is created.

### Mutable Objects:

A mutable object can be changed after it is created.

For example, a list is mutable.

Example:

numbers = [1, 2, 3]
numbers[0] = 10

Now the list becomes:

[10, 2, 3]

The same list object was modified.

Common mutable types include:

- `list`
- `set`
- `dict`

### Immutable Objects:

An immutable object cannot be changed after it is created.

For example, strings are immutable.

Example:

name = "Python"

We cannot directly change one character of the existing string.

If we do:

name = "Java"

Python creates a new string object and the variable `name` refers to the new object.

Common immutable types include:

- `int`
- `float`
- `bool`
- `str`
- `tuple`
- `frozenset`

### Simple Difference:

Mutable → Can be modified after creation.

Immutable → Cannot be modified after creation.

### Example:

Mutable:

numbers = [1, 2, 3]
numbers.append(4)

The list is changed.

Immutable:

name = "Python"
name = name + " Programming"

A new string is created rather than modifying the original string.

### Why is this important?

Understanding mutability is important because it affects how objects behave when they are modified or passed between different parts of a program.

For example, if multiple variables refer to the same mutable list, changing the list through one variable can affect what is seen through another variable.

### Quick Table:

| Type | Mutable / Immutable |
|---|---|
| `list` | Mutable |
| `set` | Mutable |
| `dict` | Mutable |
| `int` | Immutable |
| `float` | Immutable |
| `bool` | Immutable |
| `str` | Immutable |
| `tuple` | Immutable |
| `frozenset` | Immutable |

### Memory Trick:

Mutable = Modify

Immutable = Cannot Modify

### One-Line Answer:

"Mutable objects can be changed after creation, such as lists, sets, and dictionaries, whereas immutable objects cannot be changed after creation, such as integers, strings, booleans, and tuples."

## Python — Q4. What is the difference between a list, tuple, set, and dictionary in Python?

### Answer:

All four are built-in collection data types in Python, but they store and organize data differently.

### 1. List

A list is an ordered and mutable collection.

It allows duplicate values and supports indexing.

Example:

numbers = [10, 20, 30, 20]

Here, `20` can appear more than once.

### 2. Tuple

A tuple is an ordered and immutable collection.

It allows duplicate values and supports indexing.

Example:

coordinates = (10, 20, 10)

Once created, the tuple cannot be modified.

### 3. Set

A set is an unordered collection of unique elements.

It does not allow duplicate values.

Example:

numbers = {10, 20, 30, 20}

The duplicate `20` is removed.

Sets are useful when we mainly need unique values.

### 4. Dictionary

A dictionary stores data as key-value pairs.

Each key is used to access its corresponding value.

Example:

student = {
    "name": "Musharraf",
    "age": 25
}

Here, `"name"` and `"age"` are keys, while `"Musharraf"` and `25` are their values.

### Quick Comparison:

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Ordered | Yes | Yes | No* | Yes** |
| Mutable | Yes | No | Yes | Yes |
| Duplicates | Allowed | Allowed | Not allowed | Keys must be unique |
| Indexing | Yes | Yes | No | Access using keys |
| Main Use | General collection | Fixed collection | Unique values | Key-value data |

* Sets are conceptually unordered; do not rely on their iteration order.

** Dictionaries preserve insertion order in modern Python.

### Example:

List:

skills = ["Python", "SQL", "Python"]

Tuple:

skills = ("Python", "SQL", "Python")

Set:

skills = {"Python", "SQL", "Python"}

Dictionary:

skills = {
    "primary": "Python",
    "database": "SQL"
}

### When would you use them?

- Use a **list** when you need an ordered, changeable collection.
- Use a **tuple** when you need an ordered collection that should not be changed.
- Use a **set** when you need unique elements.
- Use a **dictionary** when you need to store and retrieve data using key-value pairs.

### Memory Trick:

List → Ordered + Mutable

Tuple → Ordered + Immutable

Set → Unique Values

Dictionary → Key + Value

### One-Line Answer:

"A list is an ordered and mutable collection, a tuple is ordered and immutable, a set stores unique elements, and a dictionary stores data as key-value pairs."

## Python — Q5. What is type casting in Python?

### Answer:

Type casting means converting a value from one data type to another data type.

Python provides built-in functions such as `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, and `set()` for type conversion.

### Example 1: String to Integer

age = "25"
age = int(age)

Now `age` is an integer.

### Example 2: Integer to String

age = 25
age = str(age)

Now `age` is a string.

### Example 3: Integer to Float

number = 10
number = float(number)

Now `number` becomes `10.0`.

### Example 4: List to Set

numbers = [1, 2, 2, 3]
unique_numbers = set(numbers)

The result contains only unique values.

### Two Types of Type Casting:

1. Implicit Type Casting

Python automatically converts one compatible data type into another when required.

Example:

a = 10
b = 2.5

result = a + b

Here, Python converts the integer `10` to a floating-point value during the operation, and the result is a float.

2. Explicit Type Casting

The programmer manually converts the data type using functions such as `int()`, `float()`, or `str()`.

Example:

age = "25"
age = int(age)

### Important Point:

Not every conversion is possible.

For example:

value = "hello"
number = int(value)

This produces a `ValueError` because `"hello"` cannot be converted into an integer.

### Memory Trick:

Type Casting = **Convert Data Type**

Implicit → Python converts automatically

Explicit → Programmer converts manually

### One-Line Answer:

"Type casting is the process of converting a value from one data type to another. Python supports both implicit conversion, which happens automatically, and explicit conversion using functions such as `int()`, `float()`, and `str()`."

## Python — Q6. What is the difference between `==` and `is` in Python?

### Answer:

The main difference is:

- `==` checks whether two objects have the same value.
- `is` checks whether two variables refer to the same object in memory.

### Example:

a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
Output: True

Here, `a` and `b` contain the same values.

But:

print(a is b)
Output: False

Here, `a` and `b` are two different list objects, even though their contents are the same.

### Another Example:

a = [1, 2, 3]
b = a

print(a == b)
Output: True

print(a is b)
Output: True

In this case, `b` refers to the same object as `a`.

### Common Use of `is`:

`is` is commonly used when checking for `None`.

Example:

result = None

if result is None:
    print("No result")

This is preferred over using:

if result == None:

### Quick Comparison:

| Operator | Checks | Example |
|---|---|---|
| `==` | Value equality | `a == b` |
| `is` | Object identity | `a is b` |

### Memory Trick:

`==` → Same Value?

`is` → Same Object?

### One-Line Answer:

"`==` is used to compare the values of two objects, while `is` is used to check whether two variables refer to the same object in memory."

## Python — Q7. What are conditional statements in Python?

### Answer:

Conditional statements are used to make decisions in a Python program based on whether a condition is true or false.

Python mainly provides:

- `if`
- `elif`
- `else`

### 1. `if`

The `if` statement executes a block of code when a condition is `True`.

Example:

age = 20

if age >= 18:
    print("Adult")

### 2. `if-else`

The `else` block executes when the `if` condition is `False`.

Example:

age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")

### 3. `if-elif-else`

`elif` is used when we need to check multiple conditions.

Example:

marks = 75

if marks >= 90:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")

Python checks the conditions from top to bottom and executes the first condition that is true.

### Nested Conditions:

We can also place one conditional statement inside another.

Example:

age = 25
has_id = True

if age >= 18:
    if has_id:
        print("Allowed")

### Why are conditional statements important?

They allow a program to make decisions and execute different code depending on the situation.

For example, a program might check:

- Whether a user is authenticated.
- Whether a value is valid.
- Whether a file exists.
- Whether an API response was successful.
- Whether a particular condition has been satisfied.

### Memory Trick:

`if` → Check

`elif` → Check Another

`else` → Otherwise

### One-Line Answer:

"Conditional statements are used to make decisions in Python. The main conditional statements are `if`, `elif`, and `else`, which allow different blocks of code to execute depending on whether conditions are true or false."

## Python — Q8. What are loops in Python?

### Answer:

Loops are used to execute a block of code repeatedly.

Python mainly provides two types of loops:

1. `for` loop
2. `while` loop

### 1. `for` Loop:

A `for` loop is generally used when we want to iterate over a sequence or collection, such as a list, tuple, string, dictionary, or range.

Example:

numbers = [1, 2, 3, 4]

for number in numbers:
    print(number)

The loop processes each element of the list one by one.

### 2. `while` Loop:

A `while` loop repeatedly executes a block of code as long as a condition remains `True`.

Example:

count = 1

while count <= 3:
    print(count)
    count += 1

The loop stops when the condition becomes `False`.

### Difference:

| `for` Loop | `while` Loop |
|---|---|
| Used to iterate over a sequence or iterable | Used while a condition is true |
| Often used when iterating through known data | Useful when the number of iterations depends on a condition |
| Example: list, tuple, range | Example: condition-based repetition |

### Loop Control Statements:

Python also provides:

- `break` → stops the loop completely.
- `continue` → skips the current iteration and moves to the next iteration.
- `pass` → does nothing; it can be used as a placeholder.

Example:

for number in range(5):
    if number == 3:
        break
    print(number)

Here, the loop stops when `number` becomes `3`.

### Why are loops important?

Loops help avoid writing the same code repeatedly.

For example, instead of writing separate statements to process every record in a dataset, we can use a loop to process the records one by one.

### Memory Trick:

`for` → Iterate

`while` → Condition

`break` → Stop

`continue` → Skip

`pass` → Do Nothing

### One-Line Answer:

"Loops are used to execute a block of code repeatedly. Python mainly provides `for` loops for iterating over iterables and `while` loops for repeating code while a condition is true."

## Python — Q9. What is the difference between `break`, `continue`, and `pass` in Python?

### Answer:

`break`, `continue`, and `pass` are control statements used to control the execution of a Python program, especially inside loops.

### 1. `break`

`break` immediately terminates the loop.

Example:

for number in range(5):
    if number == 3:
        break
    print(number)

Output:

0
1
2

When `number` becomes `3`, the loop stops completely.

### 2. `continue`

`continue` skips the current iteration and moves to the next iteration of the loop.

Example:

for number in range(5):
    if number == 3:
        continue
    print(number)

Output:

0
1
2
4

When `number` is `3`, that iteration is skipped, but the loop continues.

### 3. `pass`

`pass` does nothing.

It is mainly used as a placeholder when Python requires a statement but we don't want to execute any code yet.

Example:

def my_function():
    pass

The function does nothing for now, but the code is syntactically valid.

### Quick Comparison:

| Statement | Purpose | Effect |
|---|---|---|
| `break` | Stop the loop | Terminates the loop |
| `continue` | Skip current iteration | Moves to next iteration |
| `pass` | Do nothing | Acts as a placeholder |

### Simple Example:

Imagine processing 10 records:

- `break` → Stop processing all remaining records.
- `continue` → Skip the current record and process the next one.
- `pass` → Do nothing for the current code block.

### Memory Trick:

`break` → **STOP**

`continue` → **SKIP**

`pass` → **NOTHING**

### One-Line Answer:

"`break` terminates the loop, `continue` skips the current iteration and moves to the next one, while `pass` does nothing and is mainly used as a placeholder."

def greet(name):

   Examples: `print()`, `len()`, `type()`, `sum()`

   Small anonymous functions created using the `lambda` keyword.

### Why are functions important?

Functions provide:

- Easier debugging
- Reduced code duplication

### Memory Trick:

## Python — Q11. What are `*args` and `**kwargs` in Python?

### Answer:

`*args` and `**kwargs` are used when we want a function to accept a variable number of arguments.

### 1. `*args`

`*args` allows a function to accept any number of positional arguments.

Inside the function, `args` is stored as a tuple.

Example:

def add_numbers(*args):
    return sum(args)

result = add_numbers(10, 20, 30, 40)

print(result)

Output:

100

Here, we can pass any number of positional arguments.

### 2. `**kwargs`

`**kwargs` allows a function to accept any number of keyword arguments.

Inside the function, `kwargs` is stored as a dictionary.

Example:

def display_info(**kwargs):
    print(kwargs)

display_info(name="Musharraf", age=25)

Output:

{"name": "Musharraf", "age": 25}

Here, `name` and `age` are keyword arguments.

### Main Difference:

| `*args` | `**kwargs` |
|---|---|
| Accepts variable positional arguments | Accepts variable keyword arguments |
| Stored as a tuple | Stored as a dictionary |
| Uses positional values | Uses key-value pairs |

### Example Together:

def student_info(*args, **kwargs):
    print("Subjects:", args)
    print("Details:", kwargs)

student_info("Python", "SQL", name="Musharraf", age=25)

Here:

`args` contains:

("Python", "SQL")

`kwargs` contains:

{"name": "Musharraf", "age": 25}

### Why are they useful?

They make functions more flexible when we don't know in advance how many arguments the function will receive.

They are commonly useful when designing reusable functions and wrappers.

### Memory Trick:

`*args` → Many **positional** arguments → Tuple

`**kwargs` → Many **keyword** arguments → Dictionary

### One-Line Answer:

"`*args` allows a function to accept a variable number of positional arguments and stores them as a tuple, while `**kwargs` accepts a variable number of keyword arguments and stores them as a dictionary."

## Python — Q13. What is inheritance in Python?

### Answer:

Inheritance is an OOP concept where one class can acquire the properties and methods of another class.

The class that provides the properties and methods is called the parent class or base class.

The class that inherits them is called the child class or derived class.

### Example:

class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

dog = Dog()

dog.speak()
dog.bark()

Output:

Animal makes a sound
Dog barks

Here:

- `Animal` is the parent class.
- `Dog` is the child class.
- `Dog` inherits the `speak()` method from `Animal`.
- `Dog` can also have its own method, such as `bark()`.

### Why is inheritance useful?

Inheritance helps with:

- Code reusability
- Reducing code duplication
- Extending existing functionality
- Creating relationships between classes
- Making code easier to maintain

### Types of Inheritance in Python:

Python supports different forms of inheritance:

1. Single Inheritance
   One child class inherits from one parent class.

2. Multiple Inheritance
   One child class inherits from multiple parent classes.

3. Multilevel Inheritance
   A class inherits from a class that itself inherits from another class.

4. Hierarchical Inheritance
   Multiple child classes inherit from the same parent class.

5. Hybrid Inheritance
   A combination of multiple inheritance patterns.

### Important Interview Point:

A child class can also override a method inherited from the parent class. This is called method overriding and is commonly associated with polymorphism.

### Memory Trick:

**Parent → Child → Reuse + Extend**

### One-Line Answer:

"Inheritance is an OOP concept where a child class inherits properties and methods from a parent class. It promotes code reusability and allows existing functionality to be extended."

## Python — Q14. What is polymorphism in Python?

### Answer:

Polymorphism means "many forms."

In OOP, polymorphism allows the same method, function, or interface to behave differently depending on the object or data being used.

### Example: Method Overriding

class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()

Output:

Dog barks
Cat meows

Here, the same `speak()` method is called for both objects, but it produces different behavior.

This is an example of polymorphism through method overriding.

### Another Example: Built-in Functions

The `len()` function can work with different types of objects.

Example:

len("Python")
Output: 6

len([10, 20, 30])
Output: 3

The same `len()` function behaves appropriately for different types of objects.

### Common Forms of Polymorphism in Python:

1. Method Overriding
   A child class provides its own implementation of a method inherited from the parent class.

2. Duck Typing
   Python focuses on whether an object supports the required behavior rather than its exact type.

3. Operator Overloading
   Operators such as `+` can behave differently depending on the objects involved.

Example:

5 + 3
Output: 8

"Hello " + "Python"
Output: "Hello Python"

### Why is polymorphism useful?

Polymorphism helps us write flexible and reusable code.

We can work with different objects through a common interface without writing completely separate code for each object.

### Memory Trick:

**Polymorphism = One Interface → Many Behaviors**

### One-Line Answer:

"Polymorphism allows the same method, function, or interface to behave differently for different objects or data types, making code more flexible and reusable."

## Python — Q15. What is encapsulation in Python?

### Answer:

Encapsulation is an OOP concept where data and the methods that operate on that data are bundled together inside a class.

It also helps control how the internal data of an object is accessed or modified.

For example, instead of allowing code outside a class to directly modify an important value, we can provide methods that control how that value is changed.

### Example:

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)

account.deposit(500)

print(account.get_balance())

Output:

1500

Here:

- `__balance` is treated as a private attribute using Python's name-mangling mechanism.
- The balance is accessed through methods such as `deposit()` and `get_balance()`.
- The `deposit()` method can validate the amount before modifying the balance.

### Access Conventions in Python:

Python does not have traditional private access modifiers in exactly the same way as some languages such as Java.

Instead, Python commonly uses naming conventions:

- `name` → public
- `_name` → protected-style convention; indicates internal use
- `__name` → triggers name mangling and is used to make accidental direct access harder

### Why is encapsulation useful?

Encapsulation helps us:

- Protect internal data from unintended modification
- Control how data is accessed or changed
- Keep related data and behavior together
- Make code easier to maintain
- Improve the organization of classes

### Real-World Example:

A bank account should not allow any part of the program to directly set the balance to an invalid value.

Instead:

User → `deposit()` → Validation → Balance Update

This gives the class control over how its internal data changes.

### Memory Trick:

**Encapsulation = Bundle + Control Access**

### One-Line Answer:

"Encapsulation is the OOP concept of bundling data and related methods inside a class while controlling how the internal data is accessed or modified."

## Python — Q16. What is abstraction in Python?

### Answer:

Abstraction is an OOP concept where we hide unnecessary implementation details and expose only the essential functionality to the user.

The main idea is to focus on **what an object does** rather than **how it does it internally**.

### Example:

Imagine we have a payment system.

The user only needs to call:

`make_payment()`

They do not need to know the internal steps such as:

- Connecting to the payment service
- Validating the transaction
- Processing the payment
- Updating the transaction status

The internal implementation can be hidden behind the method.

### Abstraction Using Abstract Classes:

Python provides the `abc` module for creating abstract base classes.

Example:

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):

    def speak(self):
        print("Dog barks")

dog = Dog()
dog.speak()

Output:

Dog barks

Here:

- `Animal` is an abstract base class.
- `speak()` is an abstract method.
- The abstract method defines what the child class should provide.
- `Dog` provides the actual implementation.

### Why is abstraction useful?

Abstraction helps us:

- Hide unnecessary implementation details
- Expose only required functionality
- Reduce complexity
- Create clear interfaces
- Make large applications easier to maintain

### Encapsulation vs Abstraction:

Encapsulation:
Focuses on **bundling data and methods together and controlling access to data**.

Abstraction:
Focuses on **hiding implementation details and exposing essential functionality**.

### Simple Example:

Encapsulation:

"How do I control access to this data?"

Abstraction:

"How do I hide the internal implementation and show only what the user needs?"

### Memory Trick:

**Abstraction = Hide Complexity → Show Essential**

### One-Line Answer:

"Abstraction is the process of hiding unnecessary implementation details and exposing only the essential functionality, helping reduce complexity and provide a clear interface."

## Python — Q21. What is list comprehension in Python?

### Answer:

List comprehension is a concise way to create a new list from an existing iterable.

It allows us to write a loop and, optionally, a condition in a single expression.

### Normal Approach:

numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)

Output:

[1, 4, 9, 16, 25]

### Using List Comprehension:

numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)

Output:

[1, 4, 9, 16, 25]

Both approaches produce the same result, but list comprehension is more concise.

### List Comprehension with a Condition:

We can also include an `if` condition.

Example:

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)

Output:

[2, 4, 6]

Here:

- `number` → value added to the new list
- `for number in numbers` → iterates through the existing list
- `if number % 2 == 0` → keeps only even numbers

### General Syntax:

[expression for item in iterable if condition]

The `if` condition is optional.

### Why is list comprehension useful?

It can:

- Make simple transformations more concise
- Filter data easily
- Improve readability when the logic is simple
- Reduce the amount of code required for basic list operations

### Important Interview Point:

List comprehension should not be used just to make code shorter.

If the logic becomes complicated, a normal `for` loop is often easier to read and maintain.

### Memory Trick:

**Expression → For → If**

### One-Line Answer:

"List comprehension is a concise way to create a new list by applying an expression to items from an iterable, optionally using a condition to filter the items."

## Python — Q23. What are iterators and generators in Python?

### Answer:

An iterator is an object that allows us to access elements one at a time, rather than getting all elements at once.

An object can be used as an iterator when it provides the `__iter__()` and `__next__()` methods.

### Example of an Iterator:

numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))

Output:

10
20
30

Here:

- `iter()` creates an iterator from the list.
- `next()` retrieves the next value from the iterator.
- When there are no more values, `next()` raises `StopIteration`.

### What is a Generator?

A generator is a special type of iterator that produces values one at a time using the `yield` keyword.

Example:

def count_numbers():
    yield 1
    yield 2
    yield 3

numbers = count_numbers()

print(next(numbers))
print(next(numbers))
print(next(numbers))

Output:

1
2
3

Unlike a normal function that uses `return`, a generator using `yield` pauses its execution and remembers its state until the next value is requested.

### Why are Generators Useful?

Generators are useful when working with large amounts of data because they produce values one at a time instead of storing the entire sequence in memory.

Example:

def generate_numbers(n):
    for i in range(n):
        yield i

for number in generate_numbers(5):
    print(number)

The values are generated as they are needed.

### Iterator vs Generator:

| Iterator | Generator |
|---|---|
| Object used to iterate through values | Special type of iterator |
| Can be created using `iter()` | Usually created using a function with `yield` |
| Uses `__iter__()` and `__next__()` | Python automatically provides iterator behavior |
| Can require more implementation | Simpler to create |
| Useful for sequential access | Especially useful for lazy and memory-efficient processing |

### What is Lazy Evaluation?

Lazy evaluation means values are generated only when they are needed rather than generating everything in advance.

This can save memory when processing large datasets or streams of data.

### Memory Trick:

**Iterator → One Value at a Time**

**Generator → Iterator + `yield` + Lazy Evaluation**

### One-Line Answer:

"An iterator allows us to access elements one at a time using `next()`, while a generator is a convenient way to create an iterator using `yield`, producing values lazily and helping with memory-efficient processing."

## Python — Q24. What is a decorator in Python?

### Answer:

A decorator is a function that allows us to add or modify the behavior of another function without changing the original function's code.

In simple words, a decorator takes a function, adds some extra functionality, and returns a new function.

Decorators are commonly used for things such as logging, authentication, authorization, timing, caching, and validation.

### Basic Example:

def my_decorator(func):

    def wrapper():
        print("Before function")
        func()
        print("After function")

    return wrapper


@my_decorator
def greet():
    print("Hello")

greet()

Output:

Before function
Hello
After function

Here:

- `my_decorator` is the decorator function.
- `greet` is the original function.
- `wrapper` adds extra behavior around `greet`.
- `@my_decorator` applies the decorator to `greet`.

### What happens internally?

When we write:

@my_decorator
def greet():
    print("Hello")

Python essentially treats it like:

def greet():
    print("Hello")

greet = my_decorator(greet)

So, the original function is passed to the decorator.

### Decorator with Arguments:

If the original function accepts arguments, the wrapper can use `*args` and `**kwargs`.

Example:

def log_function(func):

    def wrapper(*args, **kwargs):
        print("Function is being called")
        return func(*args, **kwargs)

    return wrapper


@log_function
def add(a, b):
    return a + b

print(add(10, 20))

Output:

Function is being called
30

### Real-World Example:

Suppose we have an API endpoint and want to check whether a user is authenticated before executing the endpoint.

Instead of writing the same authentication logic inside every function, we can create an authentication decorator and apply it to the required functions.

### Why are decorators useful?

Decorators provide:

- Code reusability
- Separation of concerns
- Cleaner code
- Ability to add functionality without modifying the original function

### Important Interview Point:

When creating decorators, `functools.wraps` is commonly used so that the wrapper preserves metadata such as the original function's name and documentation.

### Memory Trick:

**Decorator = Function + Extra Behavior**

## Python — Q25. What is the difference between shallow copy and deep copy in Python?

### Answer:

The difference is related to how Python copies nested objects.

### Shallow Copy:

A shallow copy creates a new outer object, but nested objects inside it are still referenced from the original object.

So, changes to a nested mutable object can affect both the original and the shallow copy.

Example:

import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)

shallow[0].append(5)

print(original)
print(shallow)

Output:

original → [[1, 2, 5], [3, 4]]
shallow  → [[1, 2, 5], [3, 4]]

The outer lists are different objects, but the inner lists are shared.

### Deep Copy:

A deep copy creates a new outer object and recursively creates copies of the nested objects as well.

Therefore, changes to nested objects in the deep copy do not affect the original object.

Example:

import copy

original = [[1, 2], [3, 4]]

deep = copy.deepcopy(original)

deep[0].append(5)

print(original)
print(deep)

Output:

original → [[1, 2], [3, 4]]
deep     → [[1, 2, 5], [3, 4]]

Here, the nested lists are also copied.

### Quick Comparison:

| Shallow Copy | Deep Copy |
|---|---|
| Creates a new outer object | Creates a new outer object |
| Nested objects are shared | Nested objects are copied recursively |
| Uses `copy.copy()` | Uses `copy.deepcopy()` |
| Changes to nested mutable objects can affect the original | Changes to nested mutable objects do not affect the original |

### Simple Way to Remember:

Suppose we have:

original = [[1, 2], [3, 4]]

Shallow copy:

New outer box → Same inner boxes

Deep copy:

New outer box → New inner boxes

### Important Interview Point:

For simple objects without nested mutable objects, the difference may not be noticeable.

The difference becomes important when the object contains nested mutable objects such as lists or dictionaries.

### Memory Trick:

Shallow → **New Outer, Shared Inner**

Deep → **New Outer, New Inner**

### One-Line Answer:

"A shallow copy creates a new outer object but shares nested objects with the original, while a deep copy recursively creates copies of nested objects, so changes to the copied object's nested data do not affect the original."
