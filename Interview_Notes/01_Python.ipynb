## Q18. Explain variable scope and LEGB.

### Answer:

Python resolves names using the LEGB rule:

```text
L → Local
E → Enclosing
G → Global
B → Built-in
```

Example:

```python
x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print(x)

    inner()

outer()
```

Python first checks the local scope, then enclosing scopes, then global scope, then built-ins.

### Interview Answer:

> "Python uses the LEGB rule for name resolution: Local, Enclosing, Global and Built-in. The interpreter searches these scopes in that order."

### 🧠 Memory Trick:

**LEGB = Local → Enclosing → Global → Built-in**

## Q31. Explain polymorphism.

```python
class Dog:

    def speak(self):
        return "Bark"


class Cat:

    def speak(self):
        return "Meow"


animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
```

## Q33. Explain abstraction.

### Answer:

Abstraction focuses on what an object should do rather than exposing implementation details.

Python supports abstract base classes through `abc`.

```python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CardPayment(Payment):

    def pay(self, amount):
        print(f"Paid {amount} using card")
```

### Interview Answer:

> "Abstraction defines an interface or contract while hiding implementation details. Python supports this using abstract base classes and abstract methods."

### 🧠 Memory Trick:

## Q36. What is a decorator?

### Answer:

A decorator is a callable that wraps another callable to extend or modify its behavior.

```python
def logger(func):

    def wrapper():
        print("Function started")
        func()
        print("Function finished")

    return wrapper


@logger
def greet():
    print("Hello")


greet()
```

Common uses:

```text
Logging
Authentication
Authorization
Caching
Timing
Validation
```

For decorators that accept arbitrary arguments, `*args` and `**kwargs` are commonly used.

### Interview Answer:

> "A decorator wraps a function or callable to add behavior without modifying the original implementation. I commonly use decorators for cross-cutting concerns such as logging, authentication and caching."

### 🧠 Memory Trick:

**Decorator = Behavior Around Function**

## Q37. What is a generator?

### Answer:

This is useful for:

- Data streams
- Large datasets
- Memory-efficient pipelines

### Interview Answer:

> "A generator is a lazy iterator-producing function that uses `yield`. It is useful when I want to process large amounts of data without materializing the entire result in memory."

### 🧠 Memory Trick:

**Generator = yield + Lazy Evaluation**

## Q40. What is the difference between a shallow copy and deep copy?

### Answer:

A shallow copy creates a new outer object but nested objects may still be shared.

A deep copy recursively copies nested objects.

```python
import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)
```

Conceptually:

```text
Shallow:
New Outer Object
       ↓
Same Nested Objects

Deep:
New Outer Object
       ↓
New Nested Objects
```

### Interview Answer:

> "A shallow copy creates a new outer object while potentially sharing nested references. A deep copy recursively copies nested objects, making nested state independent."

### 🧠 Memory Trick:

**Shallow = Outer | Deep = Recursive**

## Q41. What is a closure?

### Answer:

A closure is an inner function that retains access to variables from its enclosing scope even after the enclosing function has returned.

```python
def multiplier(factor):

    def multiply(number):
        return number * factor

    return multiply


double = multiplier(2)

print(double(5))
```

The inner function retains `factor`.

Closures are useful for:

```text
Function factories
Callbacks
Decorators
State retention
Encapsulation
```

### Interview Answer:

> "A closure is a function together with its remembered enclosing environment. It allows an inner function to retain access to variables from an enclosing scope after the outer function has returned."

### 🧠 Memory Trick:

**Closure = Function + Remembered State**

## Q60. What is late binding in closures?

### Answer:

Python closures generally capture variables from the enclosing scope by reference, so the value can be looked up when the inner function is called rather than when it is created.

A classic example:

```python
functions = []

for i in range(3):
    functions.append(lambda: i)

print([f() for f in functions])
```

The result is:

```text
[2, 2, 2]
```

because the functions refer to the same `i`, whose final value is `2`.

A common fix is to bind the current value as a default argument:

```python
functions = []

for i in range(3):
    functions.append(lambda i=i: i)

print([f() for f in functions])
```

### Interview Answer:

> "Python closures use late binding for free variables, meaning the variable is resolved when the inner function executes. A default argument can be used when I want to capture the current value at function creation time."

### 🧠 Memory Trick:

**Late Binding = Lookup Later**

## Q98. What is the mutable default argument problem?

### Answer:

Default argument expressions are evaluated when the function is defined, not each time the function is called.

Problem:

```python
def add_item(item, items=[]):
    items.append(item)
    return items
```

Calling:

```python
print(add_item("A"))
print(add_item("B"))
```

can produce:

```text
["A"]
["A", "B"]
```

because the same default list is reused.

Correct approach:

```python
def add_item(item, items=None):

    if items is None:
        items = []

    items.append(item)

    return items
```

### Interview Answer:

> "Mutable default arguments are created once when the function is defined, so the same object can be reused across calls. I avoid this by using `None` as the default and creating the mutable object inside the function."

### 🧠 Memory Trick:

**Default Created Once**
