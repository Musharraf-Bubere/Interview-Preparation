## Q6. What is the difference between `==` and `is` in Python?

### Interview Answer

`==` checks whether two objects have equal values.

`is` checks whether two references point to the same object in memory.

### Example

    a = [1, 2]
    b = [1, 2]

    a == b
    → True

Because their values are equal.

But:

    a is b
    → False

because they are different list objects.

### Memory Trick

**`==` → Value**

**`is` → Identity**

### One-Line Answer

`==` compares values, while `is` checks object identity.

## Q9. What is the difference between `break`, `continue`, and `pass`?

### Interview Answer

`break` completely exits the loop.

`continue` skips the current iteration and moves to the next iteration.

`pass` does nothing and is used as a placeholder where Python requires a statement.

### Example

    for i in range(5):

        if i == 2:
            break

    for i in range(5):

        if i == 2:
            continue

        print(i)

### Memory Trick

**break → Stop**

**continue → Skip**

**pass → Do nothing**

### One-Line Answer

`break` exits a loop, `continue` skips the current iteration, and `pass` is a placeholder that performs no action.

## Q13. What is inheritance?

### Interview Answer

Inheritance allows one class to acquire properties and methods from another class.

The existing class is generally called the parent or base class, and the new class is called the child or derived class.

### Example

    class Animal:
        def speak(self):
            print("Animal sound")

    class Dog(Animal):
        pass

The `Dog` class inherits from `Animal`.

### Memory Trick

**Inheritance = Reuse from Parent**

### One-Line Answer

Inheritance allows a child class to reuse or extend the properties and methods of a parent class.

Polymorphism means that the same interface or method name can behave differently depending on the object or implementation.

### Example

    class Dog:
        def speak(self):
            return "Bark"

    class Cat:
        def speak(self):
            return "Meow"

Both classes have the same method name `speak()`, but the behavior is different.

### Memory Trick

**Poly = Many**

**Morphism = Forms**

**Polymorphism = Same interface, different behavior**

### One-Line Answer

Polymorphism allows the same interface or method to have different behavior depending on the object or implementation.

### Interview Answer

`except` handles the exception.

`else` runs when no exception occurs in the `try` block.

`finally` runs regardless of whether an exception occurs.

### Flow

    try
      ↓
    Exception?
     /      \
   Yes       No
    ↓         ↓
  except    else
     \       /
       finally

### Memory Trick

**try → Risky code**

**except → Error**

**else → Success**

**finally → Always**

### One-Line Answer

`try` contains risky code, `except` handles exceptions, `else` runs when no exception occurs, and `finally` executes regardless.

## Q23. What are iterators and generators?

### Interview Answer

An iterator is an object that provides values one at a time, typically through the iterator protocol using `__iter__()` and `__next__()`.

A generator is a convenient way to create an iterator using the `yield` keyword.

Generators produce values lazily, which means they do not need to store all generated values in memory at once.

### Example

    def numbers():
        yield 1
        yield 2
        yield 3

    for number in numbers():
        print(number)

### Memory Trick

**Iterator = Gives values one by one**

**Generator = Easy way to create an iterator**

### One-Line Answer

An iterator produces values one at a time, while a generator is a convenient way to create an iterator using `yield`.

A decorator is a function that modifies or extends the behavior of another function without changing its original source code.

Decorators are commonly used for:

    def logger(func):

A decorator can then be applied to another function.

### Memory Trick

**Decorator = Add behavior without changing original function**

### One-Line Answer

A decorator wraps a function to add or modify behavior without changing the function's original implementation.

## Q25. What is the difference between shallow copy and deep copy?

### Interview Answer

A shallow copy creates a new outer object but does not recursively copy nested objects.

A deep copy creates a new object and recursively copies the nested objects as well.

This difference becomes important when working with nested mutable data structures.

### Example

    import copy

    original = [[1, 2], [3, 4]]

    shallow = copy.copy(original)

    deep = copy.deepcopy(original)

If a nested list is modified through the shallow copy, the corresponding nested object can also be affected in the original.

With a deep copy, the nested objects are independent.

### Memory Trick

**Shallow → New outer object, shared nested objects**

**Deep → Independent nested objects**

### One-Line Answer

A shallow copy copies the outer object while sharing nested objects, whereas a deep copy recursively creates independent nested objects.
