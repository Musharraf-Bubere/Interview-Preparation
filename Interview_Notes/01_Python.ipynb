## Q32. How does a dictionary work internally?

### Interview Answer

When looking up a key, Python uses the key's hash and equality comparison to locate the corresponding entry.

### Memory Trick

**Dictionary → Hash → Lookup**

### One-Line Answer

A Python dictionary uses a hash-table-based implementation to provide efficient key-based lookup.

## Q33. Why must dictionary keys be hashable?

### Interview Answer

Dictionary keys need stable hash values so Python can efficiently locate entries in the underlying hash table.

Immutable built-in objects such as strings, numbers, and many tuples can be hashable.

Mutable containers such as lists are not hashable.

### Example

    data = {
        "name": "Ali"
    }

This works.

But:

    data = {
        [1, 2]: "value"
    }

raises an error because lists are unhashable.

### One-Line Answer

Dictionary keys must be hashable so the dictionary can use their hash values for efficient lookup.

## Q35. What happens when two dictionary keys have the same hash?

### Interview Answer

A hash collision occurs when different keys produce the same hash, and Python resolves it internally while checking key equality.

## Q45. What is the mutable default argument problem?

### Interview Answer

Using a mutable object such as a list as a default argument can cause unexpected behavior because the default object is created once when the function is defined and reused across calls.

### Problem

    def add_item(item, items=[]):
        items.append(item)
        return items

Repeated calls can share the same default list.

### Better Approach

    def add_item(item, items=None):
        if items is None:
            items = []

        items.append(item)
        return items

### Memory Trick

**Mutable Default → Shared State**

### One-Line Answer

Mutable default arguments can preserve state across function calls, so `None` is commonly used as the default instead.

## Q96. What is the Global Interpreter Lock (GIL)?

### Interview Answer

The Global Interpreter Lock, or GIL, is a mechanism in standard CPython that allows only one thread at a time to execute Python bytecode within a process.

This affects CPU-bound multithreaded Python programs.

However, threads can still be useful for many I/O-bound tasks, and native extensions may release the GIL during certain operations.

### Memory Trick

**GIL = One Thread Executing Python Bytecode at a Time in CPython**

### One-Line Answer

The GIL in CPython prevents multiple threads from executing Python bytecode simultaneously within the same process.


---

## Q97. What is the difference between multithreading and multiprocessing?

### Interview Answer

### Multithreading

Uses multiple threads within the same process.

It is commonly useful for I/O-bound workloads.

### Multiprocessing

Uses separate processes.

It can provide true parallel execution across CPU cores for CPU-bound work because each process has its own Python interpreter and memory space.

### Memory Trick

**Threads → I/O**

**Processes → CPU**

This is a useful rule of thumb, not an absolute rule.

### One-Line Answer

Multithreading is often useful for I/O-bound work, while multiprocessing can provide parallelism for CPU-bound work.
