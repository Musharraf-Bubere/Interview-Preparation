## Q9. What is the difference between DELETE, TRUNCATE, and DROP?

### 🧠 Memory Trick

## Q11. What is the difference between INNER JOIN and LEFT JOIN?

### 🎤 Interview Answer

INNER JOIN returns matching rows from both sides. LEFT JOIN returns every left-side row and matching right-side rows; unmatched right-side values become NULL.

### 📚 Detailed Explanation

```text
INNER → matches only
LEFT  → all left + matches
```

### 💻 Example / Diagram

LEFT JOIN is useful for finding users who have no orders, for example.

### 🌍 Real-World Use Case

INNER = matches; LEFT = keep left

### 🧠 Memory Trick

**INNER JOIN returns matches, while LEFT JOIN preserves all rows from the left table.**

### ⚡ One-Line Interview Answer

> INNER JOIN returns matching rows from both sides. LEFT JOIN returns every left-side row and matching right-side rows; unmatched right-side values become NULL.
