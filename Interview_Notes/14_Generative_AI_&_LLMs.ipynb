# 🧠 Generative AI & LLMs — Interview Questions

---

## Q1. What is Generative AI?

### Interview Answer

Generative AI is a type of Artificial Intelligence that can generate new content based on patterns learned from existing data.

It can generate:

- Text
- Images
- Audio
- Video
- Code

For example, an LLM can generate an answer to a user's question based on the prompt it receives.

### Memory Trick

**Generative AI = AI that generates content**

### One-Line Answer

Generative AI refers to AI systems that can generate new content such as text, images, audio, video, or code.

---

## Q2. What is an LLM?

### Interview Answer

LLM stands for **Large Language Model**.

It is a machine learning model trained on a large amount of text data to understand and generate human-like language.

LLMs can perform tasks such as:

- Question answering
- Summarization
- Translation
- Text generation
- Classification
- Code generation

Examples include models from OpenAI, Google, Anthropic, Meta, and other providers.

### Memory Trick

**LLM = Large Language Model**

### One-Line Answer

An LLM is a large neural network trained on extensive text data to understand and generate language.

---

## Q3. What is the difference between Generative AI and traditional Machine Learning?

### Interview Answer

Traditional machine learning is often used to make predictions or classifications from data.

Generative AI focuses on **generating new content**.

### Example

Traditional ML:

    Customer data → Predict customer churn

Generative AI:

    User prompt → Generate a response

### Memory Trick

**ML → Predict**

**GenAI → Generate**

### One-Line Answer

Traditional ML commonly predicts outcomes, while Generative AI generates new content based on learned patterns.

---

## Q4. What is a token in an LLM?

### Interview Answer

A token is a unit of text processed by a language model.

A token can represent:

- A complete word
- Part of a word
- Punctuation
- A special token

The exact tokenization depends on the model and tokenizer.

For example, a word may be represented by one token or several tokens.

### Memory Trick

**Token = Small unit of text processed by the model**

### One-Line Answer

A token is a unit of text that an LLM processes as part of its input or output.

---

## Q5. What is tokenization?

### Interview Answer

Tokenization is the process of converting text into tokens that a language model can process.

### Flow

    Text
      ↓
    Tokenizer
      ↓
    Tokens
      ↓
    Token IDs
      ↓
    Model

The tokenizer also maps tokens to numerical IDs that the model can process.

### Memory Trick

**Text → Tokens → IDs**

### One-Line Answer

Tokenization converts text into model-understandable tokens and their corresponding IDs.

---

## Q6. What are embeddings?

### Interview Answer

Embeddings are numerical vector representations of data such as text.

They capture semantic information so that items with similar meanings can have similar vector representations.

Embeddings are commonly used for:

- Semantic search
- Similarity search
- Recommendation systems
- RAG
- Clustering

### Example

    "How to reset my password?"
    
and

    "I forgot my password. How can I change it?"

can have similar embeddings because they have similar meaning.

### Memory Trick

**Embedding = Meaning represented as numbers**

### One-Line Answer

Embeddings represent data as numerical vectors that capture semantic relationships.

---

## Q7. What is the difference between an LLM and an embedding model?

### Interview Answer

An LLM is primarily designed to understand and generate language.

An embedding model converts content into numerical vectors representing semantic information.

### Example

LLM:

    Prompt → Generated Answer

Embedding model:

    Text → Vector

### Memory Trick

**LLM → Generate**

**Embedding Model → Represent**

### One-Line Answer

An LLM generates or processes language, while an embedding model converts content into numerical vectors for semantic comparison and retrieval.

---

## Q8. What is a Transformer?

### Interview Answer

A Transformer is a neural network architecture that is widely used in modern language models.

It uses mechanisms such as **self-attention** to determine relationships between different parts of a sequence.

Transformers allow models to process relationships between tokens efficiently and have become the foundation of many modern LLMs.

### Memory Trick

**Transformer = Attention-based architecture**

### One-Line Answer

A Transformer is a neural network architecture based heavily on attention mechanisms and is the foundation of many modern language models.

---

## Q9. What is self-attention?

### Interview Answer

Self-attention allows a model to determine how much importance different tokens in a sequence should have relative to one another.

It helps the model understand relationships between words or tokens based on their context.

### Example

In:

    "The animal didn't cross the road because it was tired."

Self-attention helps the model understand relationships between the tokens and determine what "it" refers to based on context.

### Memory Trick

**Self-Attention = Which tokens should I focus on?**

### One-Line Answer

Self-attention allows a model to weigh relationships between tokens in the same sequence based on context.

---

## Q10. What is a prompt?

### Interview Answer

A prompt is the input or instruction provided to a Generative AI model to guide its output.

A prompt can contain:

- Instructions
- Context
- Examples
- Constraints
- User input

### Example

    Summarize the following text in five bullet points.

This instruction guides the model toward a specific output format.

### Memory Trick

**Prompt = Instructions to the model**

### One-Line Answer

A prompt is the input or instruction given to a Generative AI model to guide its response.

---

## Q11. What is Prompt Engineering?

### Interview Answer

Prompt engineering is the process of designing and refining prompts to obtain more useful, accurate, and consistent outputs from an AI model.

It can involve:

- Clear instructions
- Context
- Examples
- Output constraints
- Role or task definition
- Structured formats

### Example

Instead of:

    Explain AI.

We can use:

    Explain Artificial Intelligence to a beginner in five simple bullet points with one real-world example.

The second prompt provides clearer requirements.

### Memory Trick

**Prompt Engineering = Better instructions → Better output**

### One-Line Answer

Prompt engineering is the process of designing effective prompts to guide an LLM toward useful and consistent outputs.

---

## Q12. What is zero-shot prompting?

### Interview Answer

Zero-shot prompting means asking the model to perform a task **without providing examples** of the desired task.

### Example

    Classify this review as positive or negative:

    "The product is excellent."

No example is provided.

### Memory Trick

**Zero-shot = Zero examples**

### One-Line Answer

Zero-shot prompting asks an LLM to perform a task without providing examples.

---

## Q13. What is few-shot prompting?

### Interview Answer

Few-shot prompting provides a small number of examples in the prompt to demonstrate the expected behavior or output format.

### Example

    Review: "Excellent product"
    Sentiment: Positive

    Review: "Very disappointing"
    Sentiment: Negative

    Review: "I really enjoyed it"
    Sentiment:

The examples guide the model toward the expected classification.

### Memory Trick

**Few-shot = Few examples**

### One-Line Answer

Few-shot prompting provides a small number of examples to guide the model's behavior.

---

## Q14. What is temperature in an LLM?

### Interview Answer

Temperature is a generation parameter that influences the randomness of model output.

Generally:

- Lower temperature → More predictable and focused output
- Higher temperature → More varied and creative output

The exact effect depends on the model and implementation.

### Example

For a factual extraction task, a lower temperature may be preferred.

For creative writing, a higher temperature may be useful.

### Memory Trick

**Low → Consistent**

**High → Creative**

### One-Line Answer

Temperature controls the degree of randomness or variation in generated output.

---

## Q15. What is hallucination in Generative AI?

### Interview Answer

A hallucination occurs when an AI model generates information that is incorrect, unsupported, or fabricated but presents it as if it were factual.

This can happen because the model generates responses based on learned patterns rather than inherently verifying every statement against a reliable source.

### Example

A model may invent:

- A nonexistent research paper
- A fake citation
- An incorrect fact
- A nonexistent API function

### Memory Trick

**Hallucination = Confident but unsupported answer**

### One-Line Answer

An LLM hallucination is an unsupported or incorrect generated response presented as if it were factual.

---

## Q16. How can we reduce hallucinations in LLM applications?

### Interview Answer

We can reduce hallucinations using techniques such as:

- Providing reliable context
- Retrieval-Augmented Generation
- Grounding responses in trusted data
- Clear prompts
- Structured outputs
- Validation
- Tool calling
- Source attribution where appropriate
- Limiting the model to available information

However, these techniques reduce risk rather than guaranteeing zero hallucinations.

### Memory Trick

**Ground → Constrain → Validate**

### One-Line Answer

Hallucinations can be reduced by grounding the model with reliable context, using tools or retrieval, constraining outputs, and validating results.

---

## Q17. What is context window?

### Interview Answer

The context window is the amount of input and output information that a model can process within a particular interaction, measured in tokens.

It can include:

- System instructions
- User messages
- Conversation history
- Retrieved documents
- Tool results
- Generated output

If the context becomes too large, information may need to be truncated, summarized, or managed differently depending on the application and model.

### Memory Trick

**Context Window = Model's available token space**

### One-Line Answer

A context window is the amount of tokenized information an LLM can process within a given interaction.

---

## Q18. What is fine-tuning?

### Interview Answer

Fine-tuning is the process of further training a pretrained model on a specific dataset to adapt its behavior or capabilities to a particular task or domain.

It can be useful when we need consistent task-specific behavior that is difficult to achieve through prompting alone.

Fine-tuning is different from RAG because RAG provides external information at inference time, while fine-tuning changes the model's learned parameters.

### Memory Trick

**Fine-tuning = Adapt the model**

### One-Line Answer

Fine-tuning further trains a pretrained model on task-specific data to adapt its behavior or capabilities.

---

## Q19. What is the difference between Prompt Engineering, RAG, and Fine-Tuning?

### Interview Answer

These solve different problems.

### Prompt Engineering

Changes the instructions given to the model.

    Prompt → Model → Response

### RAG

Provides relevant external information to the model at inference time.

    Query
      ↓
    Retrieve information
      ↓
    Context + Query
      ↓
    LLM
      ↓
    Response

### Fine-Tuning

Further trains the model on task-specific data.

    Base Model
        ↓
    Fine-tuning Data
        ↓
    Adapted Model

### Memory Trick

**Prompt → Change instructions**

**RAG → Add knowledge**

**Fine-tuning → Adapt behavior/model**

### One-Line Answer

Prompt engineering changes instructions, RAG provides external context, and fine-tuning adapts the model through additional training.

## Q20. What is structured output from an LLM?

### Interview Answer

Structured output means asking an LLM to return information in a predefined structure instead of unrestricted text.

For example:

    {
        "name": "Musharraf",
        "skills": ["Python", "SQL"],
        "experience": 2
    }

Structured outputs are useful when the response needs to be consumed by software.

They improve consistency and make downstream processing easier.

### Memory Trick

**Structured Output = LLM response with a defined schema**

### One-Line Answer

Structured output constrains an LLM response to a predefined format or schema so applications can process it reliably.

## Q21. What is tool calling or function calling?

### Interview Answer

Tool calling allows an LLM to request the execution of an external function or tool when it needs information or an action that it cannot perform directly.

The tool might:

- Search the web
- Query a database
- Call an API
- Perform calculations
- Retrieve information

### Flow

    User
      ↓
    LLM
      ↓
    Tool Call
      ↓
    External Tool/API
      ↓
    Tool Result
      ↓
    LLM
      ↓
    Final Response

### Memory Trick

**LLM → Tool → Result → LLM**

### One-Line Answer

Tool calling allows an LLM to invoke external tools or functions to retrieve information or perform actions.

## Q22. What is inference in Generative AI?

### Interview Answer

Inference is the process of using a trained model to generate predictions or outputs for new input data.

For an LLM:

    User Prompt
        ↓
    Trained Model
        ↓
    Generated Output

Training changes model parameters.

Inference uses the trained parameters to produce output.

### Memory Trick

**Training = Learn**

**Inference = Use**

### One-Line Answer

Inference is the process of using a trained model to generate an output for new input.

---

## Q23. What is model context versus model knowledge?

### Interview Answer

Model knowledge refers to information encoded in the model's learned parameters from its training.

Model context refers to information provided to the model during the current interaction, such as:

- User input
- Retrieved documents
- Conversation history
- Tool results

RAG is an example of providing external information through context rather than retraining the model.

### Memory Trick

**Knowledge = Learned**

**Context = Provided now**

### One-Line Answer

Model knowledge is information learned during training, while context is information provided to the model during inference.

## Q25. How would you troubleshoot an LLM application that is giving poor or incorrect responses?

### Interview Answer

I would troubleshoot it systematically:

1. Reproduce the issue.
2. Check the exact user input.
3. Inspect the prompt and instructions.
4. Check the model configuration.
5. Verify the retrieved context if RAG is being used.
6. Check tool calls and tool results.
7. Verify structured-output validation.
8. Check for missing or conflicting context.
9. Analyze whether the issue is hallucination, retrieval failure, prompt failure, or application logic.
10. Test with representative examples.
11. Make one change at a time.
12. Evaluate the result again.

I would avoid immediately changing the model without first identifying where the failure occurs.

### Example

If an LLM gives an incorrect answer in a RAG application, I would first check whether the correct document was retrieved. If retrieval is correct, I would inspect the prompt and context passed to the model. Then I would verify the model response and application-level validation.

### Memory Trick

**Input → Prompt → Retrieval → Tools → Model → Output → Validation**

### One-Line Answer

I would trace the complete LLM pipeline from input and prompt through retrieval, tools, model generation, and output validation to identify the root cause.
