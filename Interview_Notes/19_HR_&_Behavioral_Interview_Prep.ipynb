# 🧠 HR ANSWER FRAMEWORK

When you don't know how to answer an HR question, use:

**S → E → A → R**

### S — Situation

Briefly explain the situation.

### E — Explanation

Explain what happened.

### A — Action

Explain what you did.

### R — Result

Explain what you learned or achieved.

---

# 🎯 IMPORTANT HR RULES

## Rule 1 — Be Honest

Do not claim:

- Experience you don't have
- Technologies you have never used
- Responsibilities you never performed
- Results you cannot explain

Instead say:

**"I have worked with..."**

or

**"I have learned..."**

or

**"I have practical project experience with..."**

---

## Rule 2 — Don't Give Extremely Long Answers

For most HR questions:

**30–60 seconds** is enough.

For project or experience questions:

**1–2 minutes** is usually enough.

---

## Rule 3 — Don't Memorize Every Word

Understand the structure.

Your answer should sound natural rather than like you are reading a script.
