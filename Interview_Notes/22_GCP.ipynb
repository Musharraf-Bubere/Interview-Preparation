# ☁️ GCP / GOOGLE CLOUD PLATFORM INTERVIEW PREPARATION
# 25 Questions | Generic Technical Interview Preparation

---

## Q1. What is GCP?

### Interview Answer

GCP stands for Google Cloud Platform.

It is a cloud computing platform provided by Google that offers services for computing, storage, databases, networking, AI, monitoring, and application deployment.

Instead of managing all infrastructure on our own machine or physical servers, we can use cloud services based on the application's requirements.

### Example

    Application
        ↓
      GCP
        ↓
    Compute + Storage + Database + Networking

### Memory Trick

**GCP = Cloud Infrastructure + Services**

### One-Line Answer

GCP is Google's cloud platform that provides services for building, deploying, and managing applications.

---

## Q2. What is cloud computing?

### Interview Answer

Cloud computing means using computing resources such as servers, storage, databases, networking, and other services over the internet instead of managing physical infrastructure directly.

Cloud providers allow us to provision resources based on our requirements.

### Example

    Traditional:
    Company → Own Servers → Maintain Everything

    Cloud:
    Company → Cloud Provider → Use Required Services

### Memory Trick

**Cloud = Resources over Internet**

### One-Line Answer

Cloud computing provides computing resources and services over the internet instead of requiring users to manage physical infrastructure directly.

---

## Q3. What are the main benefits of using GCP?

### Interview Answer

Some important benefits are:

- Scalability
- Flexible resource management
- High availability options
- Managed services
- Global infrastructure
- Security features
- Monitoring and logging
- Pay-as-you-use pricing for many services

The exact benefits depend on which GCP services are used.

### Memory Trick

**Scale + Manage + Secure + Monitor**

### One-Line Answer

GCP provides scalable, managed, and secure cloud services for building and operating applications.

## Q4. What is a GCP project?

### Interview Answer

A GCP project is a logical container used to organize and manage cloud resources.

Resources such as compute instances, databases, APIs, and other services are associated with a project.

A project also provides an important boundary for configuration, permissions, billing, and resource management.

### Memory Trick

**Project = Resource Management Boundary**

### One-Line Answer

A GCP project is a logical container used to organize cloud resources, permissions, APIs, and billing.

## Q5. What is a GCP region?

### Interview Answer

A region is a specific geographical area where Google Cloud has cloud infrastructure.

When deploying an application, we can choose an appropriate region based on factors such as latency, availability requirements, and data-location requirements.

### Example

    User
      ↓
    Nearest Suitable Region
      ↓
    Application

### Memory Trick

**Region = Geographic Area**

### One-Line Answer

A GCP region is a geographical location containing Google Cloud infrastructure where resources can be deployed.

## Q6. What is a GCP zone?

### Interview Answer

A zone is an isolated deployment area within a GCP region.

A region can contain multiple zones.

Using multiple zones can help improve availability because an application does not have to depend on only one zone.

### Example

    Region
    ├── Zone A
    ├── Zone B
    └── Zone C

### Memory Trick

**Region → Zones**

### One-Line Answer

A zone is an isolated location within a GCP region where cloud resources can run.

## Q7. What is Compute Engine?

### Interview Answer

Google Compute Engine is a service that provides virtual machines on Google Cloud.

It gives us more control over the operating system, machine configuration, networking, and installed software.

It is useful when we need control similar to managing our own server.

### Example

    GCP
     ↓
    Compute Engine
     ↓
    Virtual Machine
     ↓
    Application

### Memory Trick

**Compute Engine = VM**

### One-Line Answer

Compute Engine is GCP's virtual machine service for running applications on configurable cloud servers.

---

## Q8. What is Cloud Run?

### Interview Answer

Cloud Run is a fully managed service for running containerized applications.

We provide a container image, and Google manages much of the underlying infrastructure.

It is useful for deploying containerized APIs and web applications without managing virtual machines directly.

### Example

    Dockerfile
        ↓
    Docker Image
        ↓
    Cloud Run
        ↓
    Running API

### Memory Trick

**Cloud Run = Managed Containers**

### One-Line Answer

Cloud Run is a managed GCP service for running containerized applications without directly managing servers.

---

## Q9. What is the difference between Compute Engine and Cloud Run?

### Interview Answer

Compute Engine provides virtual machines and gives us more control over the server environment.

Cloud Run is designed to run containers as a managed service, so Google handles more of the underlying infrastructure.

Therefore:

    Compute Engine
    → More infrastructure control

    Cloud Run
    → More managed container deployment

The choice depends on the application's requirements.

### Memory Trick

**VM Control → Compute Engine**

**Managed Container → Cloud Run**

### One-Line Answer

Compute Engine provides configurable virtual machines, while Cloud Run provides a managed environment for running containers.

---

## Q10. What is Cloud Storage?

### Interview Answer

Cloud Storage is GCP's object storage service.

It is used to store objects such as files, images, videos, backups, documents, and other unstructured data.

Data is stored in buckets.

### Example

    Application
        ↓
    Cloud Storage
        ↓
    Bucket
        ↓
    Files / Objects

### Memory Trick

**Cloud Storage = Objects in Buckets**

### One-Line Answer

Cloud Storage is GCP's object storage service used to store files and other unstructured data in buckets.

### Interview Answer

A bucket is a container used to organize and store objects in Google Cloud Storage.

For example, an application could use a bucket to store uploaded documents or generated files.

### Example

    Bucket
    ├── resume.pdf
    ├── report.pdf
    └── image.png

### Memory Trick

**Bucket = Container for Objects**

### One-Line Answer

A bucket is a storage container in Cloud Storage that holds objects such as files.

### Interview Answer

Cloud SQL is a managed relational database service provided by GCP.

It supports database engines such as PostgreSQL and MySQL.

Instead of manually managing the database server infrastructure, Google manages many operational tasks while we focus on the database and application.

### Example

    FastAPI
       ↓
    Cloud SQL
       ↓
    PostgreSQL

### Memory Trick

**Cloud SQL = Managed Relational Database**

### One-Line Answer

Cloud SQL is a managed GCP service for running relational databases such as PostgreSQL and MySQL.

## Q14. What is IAM in GCP?

### Interview Answer

IAM stands for Identity and Access Management.

It controls who can access which Google Cloud resources and what actions they are allowed to perform.

IAM uses identities, roles, and permissions to manage access.

### Example

    User / Service Account
            ↓
          Role
            ↓
       Permissions
            ↓
        GCP Resource

### Memory Trick

**IAM = Who + What Access**

### One-Line Answer

GCP IAM controls identities, roles, and permissions for accessing cloud resources.

---

## Q15. What is a service account?

### Interview Answer

A service account is an identity intended for applications or workloads rather than human users.

For example, a Cloud Run application may need permission to access another GCP service.

Instead of using a personal user account, the application can use a service account with the required permissions.

### Memory Trick

**User Account → Human**

**Service Account → Application**

### One-Line Answer

A service account is an identity used by applications and cloud workloads to access GCP resources securely.

---

## Q16. What is the principle of least privilege?

### Interview Answer

The principle of least privilege means giving a user or application only the permissions it actually needs.

For example, if an application only needs to read objects from a storage bucket, we should avoid giving it broad administrative permissions.

This reduces security risk.

### Memory Trick

**Only Required Access**

### One-Line Answer

Least privilege means granting only the minimum permissions required to perform a task.

---

## Q17. What is autoscaling?

### Interview Answer

Autoscaling means automatically adjusting the amount of computing resources based on demand.

When traffic increases, the platform can increase the number of running instances or resources.

When traffic decreases, it can reduce them.

This helps applications handle changing workloads efficiently.

### Example

    Low Traffic
        ↓
    Few Instances

    High Traffic
        ↓
    More Instances

### Memory Trick

**Demand ↑ → Resources ↑**

**Demand ↓ → Resources ↓**

### One-Line Answer

Autoscaling automatically adjusts application resources based on workload or traffic.

---

## Q18. What is the difference between scalability and availability?

### Interview Answer

Scalability is the ability of a system to handle increasing workload by adding resources.

Availability is the ability of a system to remain accessible and operational when users need it.

For example, adding more application instances can improve scalability, while distributing services across multiple failure domains can help improve availability.

### Memory Trick

**Scalability = Handle More**

**Availability = Stay Available**

### One-Line Answer

Scalability is about handling more workload, while availability is about keeping the system operational and accessible.

We can monitor things such as:

- CPU usage
- Memory usage
- Request count
- Latency
- Error rates
- Instance behavior

Monitoring helps identify performance problems and operational issues.

### Memory Trick

**Monitor = Health + Performance**

### One-Line Answer

GCP monitoring helps track application and infrastructure health, performance, and operational metrics.

## Q20. What are logs and why are they important in cloud applications?

### Interview Answer

Logs are records of events generated by an application or infrastructure.

They can contain information about:

- Requests
- Errors
- Exceptions
- Warnings
- Application behavior

Logs are important for troubleshooting because they help us understand what happened before and during a failure.

### Example

    User Request
        ↓
    Application
        ↓
    Error
        ↓
    Logs
        ↓
    Troubleshooting

### Memory Trick

**Logs = Evidence**

### One-Line Answer

Logs provide evidence about application behavior and are essential for troubleshooting production issues.

## Q21. Your application works locally but fails after deploying to GCP. What would you check?

### Interview Answer

I would compare the local environment with the cloud environment.

I would check:

- Environment variables
- Secrets
- Python dependencies
- Docker image
- Startup command
- Port configuration
- Database connectivity
- API credentials
- IAM permissions
- Network configuration
- Application logs

I would first identify whether the problem is related to configuration, infrastructure, application code, or an external dependency.

### Memory Trick

**Code → Config → Network → DB → IAM → Logs**

### One-Line Answer

I would compare the local and cloud environments and check configuration, dependencies, networking, database access, permissions, and logs.

---

## Q22. Your GCP application is returning a 500 error. How would you troubleshoot it?

### Interview Answer

First, I would reproduce the request and inspect the application logs.

Then I would identify the endpoint and trace the request through the application.

I would check:

- Request validation
- Application code
- Environment variables
- Database connection
- External API calls
- Authentication
- Recent deployment changes

If the error started after a deployment, I would compare the current version with the previous working version.

### Memory Trick

**500 → Logs → Trace → Compare**

### One-Line Answer

For a GCP 500 error, I would inspect logs, trace the failing request, check configuration and dependencies, and compare recent deployment changes.

---

## Q23. Your deployed API is very slow. What would you investigate?

### Interview Answer

I would first determine whether the latency comes from the application or an external dependency.

I would investigate:

- API response time
- CPU and memory usage
- Database query performance
- External API latency
- Network behavior
- Number of requests
- Application logs
- Recent code changes

I would use monitoring and logs to identify the bottleneck before making changes.

### Memory Trick

**API → CPU → DB → External API → Network**

### One-Line Answer

I would measure the latency and investigate the application, database, external APIs, networking, and resource usage to identify the bottleneck.

## Q24. How would you secure a Python application deployed on GCP?

### Interview Answer

I would use several layers of security.

First, I would avoid hard-coding secrets in source code or Docker images.

I would use secure secret management and appropriate IAM permissions.

I would apply least privilege to service accounts.

I would secure API endpoints using proper authentication and authorization.

I would also keep dependencies updated, restrict unnecessary network access, and monitor logs for suspicious or unexpected behavior.

### Memory Trick

**Secrets + IAM + Least Privilege + Auth + Monitoring**

### One-Line Answer

I would secure the application using secret management, IAM, least privilege, authentication, authorization, secure networking, and monitoring.
