# 🟢 BASIC LEVEL

## Q1. What is Microsoft Excel?

### 🎤 Interview Answer

Microsoft Excel is a spreadsheet application used to store, organize, calculate, analyze, visualize, and report data.

Excel provides features such as formulas, functions, tables, PivotTables, charts, data validation, conditional formatting, Power Query, and automation using VBA.

### 📚 Detailed Explanation

Excel organizes information into:

```text
Workbook
   ↓
Worksheet
   ↓
Rows + Columns
   ↓
Cells
   ↓
Data / Formulas
```

A workbook is the Excel file, while a worksheet is an individual sheet inside the workbook.

Excel can be used for simple calculations as well as business reporting and data analysis.

### 💻 Example

Suppose a company stores monthly sales:

| Month | Sales |
|---|---:|
| January | 100000 |
| February | 125000 |
| March | 140000 |

Excel can calculate total sales:

```excel
=SUM(B2:B4)
```

### 🌍 Real-World Use Case

A business analyst can use Excel to analyze sales, expenses, employee data, inventory, customer information, and operational KPIs.

### 🧠 Memory Trick

**Excel = Store → Calculate → Analyze → Visualize → Report**

### ⚡ One-Line Interview Answer

> Excel is a spreadsheet and data-analysis tool used for calculations, data manipulation, analysis, visualization, and reporting.

## Q5. What is the difference between relative, absolute, and mixed references?

### 🎤 Interview Answer

Relative references change when a formula is copied, absolute references remain fixed, and mixed references lock either the row or the column.

### 💻 Example

```excel
A1      Relative
$A$1    Absolute
$A1     Column fixed
A$1     Row fixed
```

Suppose:

```excel
=B2*$F$1
```

When copied downward, `B2` changes to `B3`, `B4`, etc., but `$F$1` remains fixed.

### 🌍 Real-World Use Case

This is useful when calculating sales tax, commission, exchange rates, or percentages using a fixed assumption.

### 🧠 Memory Trick

```text
A1    → moves
$A$1  → never moves
$A1   → column locked
A$1   → row locked
```

### ⚡ One-Line Interview Answer

> Relative references move, absolute references stay fixed, and mixed references lock either the row or column.

## Q16. What is VLOOKUP?

### 🎤 Interview Answer

VLOOKUP searches for a value in the first column of a table and returns a related value from another column.

### 💻 Example

```excel
=VLOOKUP(A2,$F$2:$H$100,3,FALSE)
```

If `A2` contains an employee ID, Excel searches the first column of the lookup table and returns the value from the third column.

### 🌍 Real-World Use Case

Finding:

```text
Employee ID → Employee Name
Product ID  → Product Price
Customer ID → Customer Region
```

### ⚠️ Limitation

VLOOKUP normally searches from left to right and depends on a column index.

### 🧠 Memory Trick

**VLOOKUP = Vertical search**

### ⚡ One-Line Interview Answer

> VLOOKUP searches vertically in the first column of a table and returns a corresponding value from another column.

---

## Q17. What is XLOOKUP and why is it preferred over VLOOKUP?

### 🎤 Interview Answer

XLOOKUP is a modern lookup function that provides more flexibility than VLOOKUP. It can search in either direction, has a separate lookup and return range, and provides a convenient not-found result.

### 💻 Example

```excel
=XLOOKUP(A2,F:F,H:H,"Not Found")
```

### Comparison

```text
VLOOKUP
Lookup column must be first
Uses column number
Usually left → right

XLOOKUP
Lookup and return ranges are separate
Can search left/right
Supports not-found result
```

### 🌍 Real-World Use Case

Looking up product prices from product IDs in large business datasets.

### 🧠 Memory Trick

**XLOOKUP = Flexible lookup**

### ⚡ One-Line Interview Answer

> XLOOKUP is a flexible lookup function that can search in either direction and directly specifies the return range.

## Q52. What are volatile functions?

### 🎤 Interview Answer

Volatile functions recalculate whenever Excel recalculates the workbook, which can increase calculation time in large workbooks.

Common examples include:

```excel
NOW()
TODAY()
RAND()
RANDBETWEEN()
OFFSET()
INDIRECT()
```

### 🌍 Real-World Use Case

Using thousands of volatile formulas in a large reporting workbook can significantly affect performance.

### ⚡ One-Line Interview Answer

## Q53. What is the difference between INDEX/MATCH and XLOOKUP?

### 🎤 Interview Answer

Both provide flexible lookup capabilities. XLOOKUP is generally easier to read and use, while INDEX/MATCH is a powerful combination with broad compatibility and strong control over lookup logic.

### Comparison

```text
XLOOKUP
Simple syntax
Separate lookup/return ranges
Built-in not-found option

INDEX/MATCH
Flexible
Powerful combination
Useful in older Excel environments
```

### ⚡ One-Line Interview Answer

> XLOOKUP provides a simpler modern lookup experience, while INDEX/MATCH remains a powerful and widely used lookup combination.

## Q56. What is query folding in Power Query?

### 🎤 Interview Answer

Query folding occurs when Power Query translates transformation steps into operations that the source system can execute, reducing the amount of data that must be transferred and processed locally.

### Diagram

```text
Power Query Transformation
          ↓
Can source execute it?
      ↙         ↘
    Yes          No
     ↓            ↓
Source does     Local
the work       processing
```

### 🌍 Real-World Use Case

When querying a database with millions of rows, filtering data at the source can dramatically reduce data movement.

### ⚡ One-Line Interview Answer

> Query folding pushes compatible transformations back to the source system to improve performance.

## Q64. How would you calculate month-over-month growth?

### 🎤 Interview Answer

Month-over-month growth compares the current month's value with the previous month's value.

### Formula

```excel
=(CurrentMonth-PreviousMonth)/PreviousMonth
```

Example:

```text
January = 100
February = 120

Growth = (120-100)/100
       = 20%
```

### ⚡ One-Line Interview Answer

> Month-over-month growth is calculated as the change from the previous month divided by the previous month's value.
