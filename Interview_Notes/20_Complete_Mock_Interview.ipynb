# 🧠 MOCK INTERVIEW RULES

## Rule 1 — One Question at a Time

I will ask **only one interview question at a time**.

You will answer it.

Then I will:

- Evaluate your answer
- Correct technical mistakes
- Improve your English if required
- Tell you what was good
- Tell you what needs improvement
- Give you a better natural answer
- Give you a memory trick
- Move to the next question

---

## Rule 2 — No Memorized Corporate English

Your answers should sound like **you are actually speaking**.

We will prefer:

**Simple + Clear + Natural**

instead of:

**Complex + Corporate + Memorized**

---

## Rule 3 — Technical Honesty

If you don't know an answer, say:

> "I am not completely sure about that, but my understanding is..."

or:

> "I haven't worked on that directly, but I have learned the concept."

Do not invent experience.

---

## Rule 4 — Project Honesty

For project questions, you should be able to explain:

**What → Why → How → Your Contribution → Challenge → Result**

---

## Rule 5 — Troubleshooting Framework

For technical scenarios, remember:

# ⭐ IMPORTANT

The **160 questions above are the preparation pool**.

The actual mock interview will NOT necessarily ask all 160 questions.

The goal is to simulate a realistic interview where questions can change based on your previous answer.

For example:

    Interviewer:
    "Tell me about your Hirely project."

            ↓

    You explain embeddings.

            ↓

    Interviewer:
    "What are embeddings?"

            ↓

    You explain embeddings.

            ↓

    Interviewer:
    "Why did you use embeddings instead of
     only keyword matching?"

This is how we will practice — **follow-up questions included**, rather than simply memorizing a fixed list.
