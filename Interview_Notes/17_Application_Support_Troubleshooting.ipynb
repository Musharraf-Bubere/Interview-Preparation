# 🛠️ Application Support & Troubleshooting — 25 Interview Questions

---

## Q1. What is Application Support?

### Interview Answer

Application Support is the process of maintaining, monitoring, troubleshooting, and resolving issues in software applications after they are deployed.

The main goal is to keep the application available, reliable, and working as expected.

Typical activities include:

- Monitoring applications
- Troubleshooting errors
- Investigating incidents
- Checking logs
- Identifying root causes
- Fixing issues
- Supporting users or teams
- Maintaining technical documentation
- Coordinating with developers and other teams

### Memory Trick

**Monitor → Troubleshoot → Fix → Maintain**

### One-Line Answer

Application Support focuses on keeping deployed applications reliable, available, and working correctly by monitoring and resolving issues.

---

## Q2. What is troubleshooting?

### Interview Answer

Troubleshooting is the systematic process of identifying the cause of a problem and resolving it.

I would not immediately change the code or configuration. I would first reproduce the issue, collect evidence, identify the root cause, apply a fix, and then verify the result.

### Basic Process

    Problem
       ↓
    Reproduce
       ↓
    Investigate
       ↓
    Find Root Cause
       ↓
    Fix
       ↓
    Test
       ↓
    Verify

### Memory Trick

**Reproduce → Investigate → Fix → Verify**

### One-Line Answer

Troubleshooting is a systematic process of finding the root cause of an issue and verifying that the fix resolves it.

---

## Q3. What is the difference between an incident and a problem?

### Interview Answer

An **incident** is an event where an application or service is not working as expected.

A **problem** is the underlying cause of one or more incidents.

### Example

    Incident:
    API is returning HTTP 500.

    Problem:
    Database connection pool is exhausted.

The incident describes what went wrong, while the problem focuses on why it happened.

### Memory Trick

**Incident = What went wrong**

**Problem = Why it went wrong**

### One-Line Answer

An incident is a disruption or failure, while a problem is the underlying cause of one or more incidents.

---

## Q4. What is root cause analysis?

### Interview Answer

Root Cause Analysis, or RCA, is the process of identifying the underlying reason why an issue occurred.

Instead of only fixing the visible symptom, I try to understand what caused the issue and how to prevent it from happening again.

### Example

    API returns 500
         ↓
    Check logs
         ↓
    Database timeout
         ↓
    Check database
         ↓
    Connection pool exhausted
         ↓
    Root Cause

### Memory Trick

**Don't fix only the symptom → Find the cause**

### One-Line Answer

Root Cause Analysis identifies the underlying cause of an issue so it can be properly fixed and prevented from recurring.

---

## Q5. What are logs and why are they important?

### Interview Answer

Logs are records generated by an application or system about events that occur during execution.

They can contain information such as:

- Errors
- Warnings
- Requests
- Responses
- Timestamps
- Exceptions
- Database operations
- Application events

Logs are important because they provide evidence about what happened and help with troubleshooting.

### Memory Trick

**Logs = Evidence of what happened**

### One-Line Answer

Logs record application events and are an important source of evidence during troubleshooting.

---

## Q6. What is the difference between INFO, WARNING, and ERROR logs?

### Interview Answer

They represent different levels of application events.

**INFO** is used for normal operational information.

**WARNING** indicates something unexpected or potentially problematic, but the application may continue working.

**ERROR** indicates that an operation failed or an important problem occurred.

### Example

    INFO:
    Application started successfully.

    WARNING:
    API response is slower than expected.

    ERROR:
    Database connection failed.

### Memory Trick

**INFO → Normal**

**WARNING → Concern**

**ERROR → Failure**

### One-Line Answer

INFO describes normal events, WARNING indicates potential issues, and ERROR indicates failures that require attention.

## Q7. How would you troubleshoot an application that suddenly stops working?

### Interview Answer

I would follow a systematic approach.

First, I would confirm the issue and identify its scope.

Then I would check:

1. Application status
2. Recent logs
3. Error messages
4. Recent deployments or configuration changes
5. API availability
6. Database connectivity
7. External dependencies
8. Server or resource health

After identifying the likely root cause, I would apply the appropriate fix and verify that the application is working again.

### Memory Trick

**Status → Logs → Changes → Dependencies → Fix → Verify**

### One-Line Answer

I would confirm the issue, check logs and recent changes, investigate dependencies and resources, identify the root cause, fix it, and verify the application.

---

## Q8. How would you troubleshoot a Python application throwing an exception?

### Interview Answer

I would first reproduce the exception and inspect the traceback.

Then I would identify:

- Exception type
- Error message
- File and line number
- Input that caused the problem
- Recent code changes
- Related logs

Then I would identify the root cause, fix it, and test both the failing case and relevant normal cases.

### Memory Trick

**Exception → Traceback → Cause → Fix → Test**

### One-Line Answer

I would inspect the traceback, reproduce the issue, identify the root cause, apply the fix, and retest the application.

---

## Q9. What is a traceback in Python?

### Interview Answer

A traceback is information provided by Python when an exception occurs.

It shows the sequence of function calls that led to the error and usually identifies the file and line where the exception occurred.

### Example

    Function A
       ↓
    Function B
       ↓
    Function C
       ↓
    Exception

The traceback helps identify where the failure originated.

### Memory Trick

**Traceback = Path to the error**

### One-Line Answer

A Python traceback shows the call path and location associated with an exception.

---

## Q10. How would you troubleshoot an API returning HTTP 500?

### Interview Answer

HTTP 500 indicates an internal server-side error.

I would:

1. Reproduce the request.
2. Check the request payload.
3. Check application logs.
4. Inspect the traceback.
5. Identify the failing endpoint.
6. Check database operations.
7. Check external API calls.
8. Check recent code or configuration changes.
9. Identify the root cause.
10. Apply the fix and retest.

I would not assume that every 500 error is caused by the database.

### Memory Trick

**500 → Server-side investigation**

### One-Line Answer

For a 500 error, I would inspect logs and traceback, trace the request through the application, identify the failing component, and verify the fix.

---

## Q11. What would you do if an API returns 404?

### Interview Answer

HTTP 404 means the requested resource or endpoint was not found.

I would check:

- URL
- HTTP method
- Endpoint path
- Path parameters
- Resource ID
- API version
- Routing configuration

I would also verify whether the resource actually exists.

### Memory Trick

**404 = Not Found**

### One-Line Answer

For a 404, I would verify the endpoint, URL, parameters, routing, and whether the requested resource exists.

---

## Q12. What would you do if an API returns 401?

### Interview Answer

HTTP 401 generally indicates that authentication is missing or invalid.

I would check:

- Whether the authentication token is present
- Whether the token is valid
- Whether it has expired
- Whether the authentication scheme is correct
- Whether the request contains the expected credentials

### Memory Trick

**401 = Authentication problem**

### One-Line Answer

A 401 response generally means the request has failed authentication.

---

## Q13. What would you do if an API returns 403?

### Interview Answer

HTTP 403 generally means the server understood the request but the authenticated user is not allowed to perform the requested action.

I would check:

- User identity
- Roles
- Permissions
- Access-control rules
- Resource ownership

### Memory Trick

**403 = Permission problem**

### One-Line Answer

A 403 response generally means the user is authenticated but does not have permission to perform the requested action.

---

## Q14. What is the difference between 400 and 500 errors?

### Interview Answer

A **400-level error** generally indicates a problem with the client request.

For example:

- Invalid input
- Missing required field
- Incorrect parameters

A **500-level error** generally indicates a problem on the server side.

### Example

    400:
    Invalid request data

    500:
    Unexpected server exception

### Memory Trick

**400 → Client/request side**

**500 → Server side**

### One-Line Answer

400-level errors generally indicate invalid client requests, while 500-level errors generally indicate server-side failures.

---

## Q15. How would you troubleshoot a slow API?

### Interview Answer

I would first measure where the time is being spent.

I would investigate:

- Application processing time
- Database query time
- External API latency
- Network latency
- Large payloads
- Expensive computations
- Resource usage
- Recent changes

I would use logs, timing information, profiling, and monitoring data where available.

After identifying the bottleneck, I would optimize that specific component and measure again.

### Memory Trick

**Measure → Find bottleneck → Optimize → Measure again**

### One-Line Answer

I would measure each stage of the request, identify the bottleneck, optimize it, and verify the performance improvement.

---

## Q16. How would you troubleshoot a database connection failure?

### Interview Answer

I would check the issue systematically.

First, I would verify:

- Database server availability
- Host and port
- Credentials
- Connection string
- Network connectivity
- Database permissions
- Connection pool
- Recent configuration changes

I would also check the application logs for the exact database error.

### Memory Trick

**Server → Network → Credentials → Configuration → Pool**

### One-Line Answer

I would verify database availability, connectivity, credentials, configuration, permissions, and connection-pool behavior.

## Q17. How would you troubleshoot a slow database query?

### Interview Answer

I would first identify the query causing the performance issue.

Then I would check:

- Query execution plan
- Indexes
- Table size
- Joins
- Filters
- Unnecessary columns
- Sorting
- Aggregations
- Database statistics

I would optimize the query or indexing strategy based on the actual bottleneck and then measure the result.

### Memory Trick

**Query → Plan → Index → Optimize → Measure**

### One-Line Answer

I would inspect the execution plan and query structure, identify the bottleneck, optimize the query or indexes, and measure the improvement.

---

## Q18. What would you do if an external API used by your application is unavailable?

### Interview Answer

I would first verify whether the external service is actually unavailable and check the response or timeout.

Then I would check:

- API status
- Timeout
- Authentication
- Rate limits
- Network connectivity
- Request format
- Logs

Depending on the application's requirements, I might use:

- Retry with limits
- Timeout handling
- Fallback behavior
- Cached data
- Graceful error messages

I would avoid unlimited retries because they can increase load and make the problem worse.

### Memory Trick

**Verify → Timeout → Retry carefully → Fallback → Inform**

### One-Line Answer

I would verify the external failure, handle timeouts and retries safely, use a fallback where appropriate, and provide a controlled error response.

## Q19. What is monitoring?

### Interview Answer

Monitoring is the continuous observation of an application's health and performance.

Common things to monitor include:

- Application availability
- Response time
- Error rate
- CPU and memory usage
- Database health
- API failures
- Request volume

Monitoring helps detect problems before or while they affect users.

### Memory Trick

**Monitoring = Know the application's health**

### One-Line Answer

Monitoring continuously tracks application health, performance, and failures so issues can be detected and addressed.

## Q20. What is alerting?

### Interview Answer

Alerting is the process of automatically notifying the responsible team when a monitored condition crosses a defined threshold or indicates a significant problem.

### Example

    Error rate > Threshold
          ↓
        Alert
          ↓
    Support/Engineering Team

A good alert should be actionable and avoid excessive false alarms.

### Memory Trick

**Monitoring observes → Alerting notifies**

### One-Line Answer

Alerting notifies the responsible team when monitored conditions indicate a significant issue.

## Q21. What is a health check endpoint?

### Interview Answer

A health check endpoint is an API endpoint used to determine whether an application or service is operating correctly.

For example:

    GET /health

It may return a simple successful response if the application is running.

More detailed health checks can also verify important dependencies such as databases, depending on the application's design.

### Memory Trick

**Health endpoint = Is the service alive/healthy?**

### One-Line Answer

A health check endpoint provides a way to verify the operational health of an application or service.

## Q22. How would you handle a production issue?

### Interview Answer

I would stay systematic and avoid making unverified changes.

My approach would be:

1. Understand the impact.
2. Confirm and reproduce the issue if possible.
3. Check logs and monitoring.
4. Identify the affected component.
5. Check recent changes.
6. Determine the safest immediate mitigation.
7. Fix the root cause where possible.
8. Test the solution.
9. Verify that the service has recovered.
10. Document the incident and root cause.

If the issue requires another team's expertise, I would provide them with clear evidence rather than simply saying that the application is not working.

### Memory Trick

**Impact → Evidence → Mitigation → Root Cause → Fix → Verify → Document**

### One-Line Answer

For a production issue, I would assess impact, collect evidence, mitigate safely, identify the root cause, fix and verify it, and document the incident.

## Q23. Why is documentation important in Application Support?

### Interview Answer

Documentation helps support teams understand how an application works and how common issues should be handled.

Useful documentation can include:

- Setup instructions
- API documentation
- Troubleshooting steps
- Known issues
- Deployment procedures
- Configuration details
- Incident resolutions

Good documentation reduces dependency on individual team members and makes troubleshooting faster.

### Memory Trick

**Document once → Help many times**

### One-Line Answer

Documentation makes application support more consistent, faster, and less dependent on individual knowledge.

---

## Q24. How would you handle an issue that you cannot solve immediately?

### Interview Answer

I would not guess or make risky changes.

I would:

1. Understand the issue.
2. Collect relevant evidence.
3. Check logs and documentation.
4. Try to reproduce it.
5. Narrow down the affected component.
6. Search existing known issues or solutions.
7. Ask the appropriate team member when necessary.
8. Share the evidence and troubleshooting already performed.
9. Keep track of the issue until it is resolved.
10. Document the final solution.

### Memory Trick

**Investigate → Document → Escalate with evidence → Follow through**

### One-Line Answer

If I cannot solve an issue immediately, I would investigate it systematically, document my findings, and escalate it with clear evidence to the appropriate person or team.

---

## Q25. What is your overall approach to troubleshooting any technical issue?

### Interview Answer

My general approach is:

    1. Understand the problem
    2. Reproduce the issue
    3. Collect evidence
    4. Check logs and errors
    5. Identify the affected component
    6. Compare expected vs actual behavior
    7. Find the root cause
    8. Apply the smallest appropriate fix
    9. Test the fix
    10. Verify the complete application
    11. Document the solution

I prefer a systematic approach because changing multiple things at once can make it difficult to identify the actual root cause.

### Memory Trick

**Understand → Reproduce → Investigate → Root Cause → Fix → Test → Verify → Document**

### One-Line Answer

I troubleshoot systematically by reproducing the issue, collecting evidence, finding the root cause, applying a controlled fix, testing it, and documenting the result.
