# 🧱 Object-Oriented Programming (OOP) — Interview Questions

---

## Q1. What is OOP?

### Interview Answer

OOP stands for Object-Oriented Programming.

It is a programming approach where we organize code around **objects and classes**.

A class defines the structure and behavior, while an object is an actual instance of that class.

OOP helps make code more reusable, maintainable, modular, and easier to manage.

### Example

    class Car:
        def start(self):
            print("Car started")

    car1 = Car()
    car1.start()

Here, `Car` is the class and `car1` is the object.

### Memory Trick

**OOP = Organize code using Objects**

### One-Line Answer

OOP is a programming paradigm that organizes code using classes and objects.

## Q2. What is a Class?

### Interview Answer

A class is a **blueprint or template** used to create objects.

It defines the attributes and methods that objects created from the class can have.

### Example

    class Student:

### Memory Trick

**Class = Blueprint**

### One-Line Answer

A class is a blueprint that defines the attributes and behavior of objects.

## Q5. What is Encapsulation?

### Interview Answer

Encapsulation means **bundling data and methods together inside a class** and controlling how that data can be accessed.

It helps protect the internal state of an object.

In Python, encapsulation is commonly implemented using naming conventions such as:

- Public
- Protected
- Private

### Example

    class BankAccount:
        def __init__(self):
            self.__balance = 1000

        def get_balance(self):
            return self.__balance

The `__balance` variable is intended to be private.

### Memory Trick

**Encapsulation = Protect and bundle data**

### One-Line Answer

Encapsulation bundles data and methods together and controls access to the object's internal data.

## Q6. What is Abstraction?

### Interview Answer

Abstraction means **hiding unnecessary implementation details and exposing only the required functionality**.

The user focuses on what an object does rather than how it internally works.

### Example

When we call:

    car.start()

we don't need to know every internal operation performed by the engine.

### Memory Trick

**Abstraction = Hide complexity**

### One-Line Answer

Abstraction hides implementation details and exposes only the necessary functionality.

## Q7. What is Inheritance?

### Interview Answer

Inheritance allows a child class to **reuse the properties and methods of a parent class**.

It promotes code reuse and creates a relationship between classes.

### Example

    class Animal:
        def speak(self):
            print("Animal speaks")

    class Dog(Animal):
        pass

    dog = Dog()
    dog.speak()

`Dog` inherits from `Animal`.

### Memory Trick

**Inheritance = Reuse from parent**

### One-Line Answer

Inheritance allows one class to acquire the properties and methods of another class.

## Q9. What is Polymorphism?

### Interview Answer

Polymorphism means **one interface or method name can have different behaviors depending on the object or context**.

For example, different classes can implement the same method differently.

### Example

    class Dog:
        def sound(self):
            print("Bark")

    class Cat:
        def sound(self):
            print("Meow")

    for animal in [Dog(), Cat()]:
        animal.sound()

The same `sound()` method behaves differently for different objects.

### Memory Trick

**Poly = Many**
**Morph = Forms**

### One-Line Answer

Polymorphism allows the same interface or method to behave differently for different objects.

## Q13. What is `self` in Python?

### Interview Answer

It allows us to access the object's attributes and methods inside the class.

## Q14. What is Method Overriding?

### Interview Answer

Method overriding occurs when a child class provides its **own implementation of a method that already exists in the parent class**.

### Example

    class Animal:
        def sound(self):
            print("Animal sound")

    class Dog(Animal):
        def sound(self):
            print("Bark")

Here, `Dog` overrides the `sound()` method of `Animal`.

### Memory Trick

**Override = Child changes parent behavior**

### One-Line Answer

Method overriding occurs when a child class provides a different implementation of a parent class method.

## Q15. Does Python support Method Overloading?

### Interview Answer

Python does not support traditional method overloading like some languages such as Java.

If we define multiple methods with the same name in a class, the latest definition replaces the previous one.

However, similar behavior can be achieved using:

- Default arguments
- `*args`
- `**kwargs`

### Example

    class Calculator:
        def add(self, a, b=0):
            return a + b

    calc = Calculator()

    print(calc.add(10))
    print(calc.add(10, 20))

### Memory Trick

**Python → No traditional overloading**

### One-Line Answer

Python does not support traditional method overloading, but similar behavior can be achieved using default arguments or variable-length arguments.

## Q17. What is Multiple Inheritance?

### Interview Answer

Multiple inheritance occurs when a class inherits from **more than one parent class**.

### Example

    class Father:
        def skills(self):
            print("Business")

    class Mother:
        def hobbies(self):

    class Child(Father, Mother):

    child.skills()
    child.hobbies()

`Child` inherits from both `Father` and `Mother`.

### Memory Trick

**Multiple = Multiple parents**

### One-Line Answer

Multiple inheritance occurs when one class inherits from more than one parent class.

## Q18. What is MRO in Python?

### Interview Answer

MRO stands for **Method Resolution Order**.

It defines the order in which Python searches for a method or attribute in a class hierarchy.

It becomes especially important when using multiple inheritance.

Python uses the **C3 linearization algorithm** to determine MRO.

### Example

    class A:
        pass

    class B(A):
        pass

    class C(B):
        pass

    print(C.mro())

This shows the order Python follows when searching for methods.

### Memory Trick

**MRO = Method search order**

### One-Line Answer

MRO defines the order in which Python searches classes for methods and attributes.

## Q19. What is `super()` in Python?

### Interview Answer

`super()` is used to access methods or functionality from a parent class without directly referring to the parent class by name.

It is commonly used with inheritance.

### Example

    class Animal:
        def __init__(self, name):
            self.name = name

    class Dog(Animal):
        def __init__(self, name, breed):
            super().__init__(name)
            self.breed = breed

Here, `super()` calls the parent class constructor.

### Memory Trick

**super() = Access parent functionality**

### One-Line Answer

`super()` is used to access methods or initialization logic from a parent class.

## Q20. What are Class Variables and Instance Variables?

### Interview Answer

An **instance variable** belongs to a specific object.

A **class variable** is shared by all objects of the class unless an instance overrides it.

### Example

    class Student:
        college = "ABC College"

        def __init__(self, name):
            self.name = name

Here:

`college` → Class variable

`name` → Instance variable

### Memory Trick

**Class variable → Shared**
**Instance variable → Object-specific**

### One-Line Answer

Class variables are shared at the class level, while instance variables belong to individual objects.

## Q21. What are Instance Methods, Class Methods, and Static Methods?

### Interview Answer

Python commonly provides three types of methods:

### 1. Instance Method

Works with object-specific data and normally uses `self`.

### 2. Class Method

Works with class-level data and uses `cls`.

It is defined using `@classmethod`.

### 3. Static Method

Does not depend on instance or class state.

It is defined using `@staticmethod`.

### Example

    class Student:

        college = "ABC"

        def show(self):
            print(self.college)

        @classmethod
        def change_college(cls, name):
            cls.college = name

        @staticmethod
        def greet():
            print("Hello")

### Memory Trick

**Instance → self**
**Class → cls**
**Static → neither**

### One-Line Answer

Instance methods use `self`, class methods use `cls`, and static methods do not require either.

## Q22. What is an Abstract Class?

### Interview Answer

An abstract class is a class that is designed to be a **base class and may contain abstract methods that child classes must implement**.

Python provides the `abc` module for creating abstract classes.

### Example

    from abc import ABC, abstractmethod

    class Animal(ABC):

        @abstractmethod
        def sound(self):
            pass

    class Dog(Animal):

        def sound(self):
            print("Bark")

The child class implements the abstract method.

### Memory Trick

**Abstract class = Template for child classes**

### One-Line Answer

An abstract class defines a common interface and can require child classes to implement specific methods.

## Q23. What is Composition in OOP?

### Interview Answer

Composition means building a class by **containing objects of other classes**.

It represents a **has-a relationship**.

### Example

    class Engine:
        def start(self):
            print("Engine started")

    class Car:
        def __init__(self):
            self.engine = Engine()

        def start(self):
            self.engine.start()

Here:

Car **has an** Engine.

### Memory Trick

**Composition = Has-a**

### One-Line Answer

Composition is an OOP relationship where one class contains an object of another class.

## Q24. What is the difference between Composition and Inheritance?

### Interview Answer

Inheritance represents an **is-a relationship**.

Composition represents a **has-a relationship**.

### Example

Inheritance:

    Dog is an Animal.

Composition:

    Car has an Engine.

Composition is often preferred when we want to build flexible systems by combining independent components.

### Memory Trick

**Inheritance → IS-A**
**Composition → HAS-A**

### One-Line Answer

Inheritance represents an is-a relationship, while composition represents a has-a relationship.
