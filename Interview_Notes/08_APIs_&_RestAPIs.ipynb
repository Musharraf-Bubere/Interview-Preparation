# 🌐 APIs & REST APIs — Interview Questions

---

## Q1. What is an API?

### Interview Answer

API stands for **Application Programming Interface**.

An API is a way for two different software systems or components to communicate with each other.

For example, a frontend application can send a request to a backend API to get user or product information.

### Example

A client sends:

    GET /users

The server processes the request and returns user data.

### Memory Trick

**API = Communication between applications**

### One-Line Answer

An API is an interface that allows different software applications or components to communicate with each other.

---

## Q2. What is a REST API?

### Interview Answer

REST stands for **Representational State Transfer**.

A REST API is an API that follows REST architectural principles and uses HTTP to communicate with resources.

For example:

    GET /users
    POST /users
    GET /users/10
    PUT /users/10
    DELETE /users/10

Each endpoint represents a resource or operation related to a resource.

### Memory Trick

**REST = Resources + HTTP**

### One-Line Answer

A REST API is an API that follows REST principles and commonly uses HTTP methods to interact with resources.

---

## Q3. What are the main HTTP methods used in REST APIs?

### Interview Answer

The commonly used HTTP methods are:

- `GET` → Retrieve data
- `POST` → Create data
- `PUT` → Replace or update a resource
- `PATCH` → Partially update a resource
- `DELETE` → Delete a resource

### Example

    GET /users
    POST /users
    PUT /users/10
    PATCH /users/10
    DELETE /users/10

### Memory Trick

**GET → Read**
**POST → Create**
**PUT/PATCH → Update**
**DELETE → Delete**

### One-Line Answer

GET retrieves, POST creates, PUT/PATCH updates, and DELETE removes resources.

---

## Q4. What is the difference between PUT and PATCH?

### Interview Answer

`PUT` is generally used to **replace the complete representation of a resource**.

`PATCH` is generally used to **partially update a resource**.

### Example

Suppose a user has:

    name
    email
    age

Using PUT, we generally send the complete updated resource.

Using PATCH, we can update only:

    {
        "age": 25
    }

### Memory Trick

**PUT = Full update**
**PATCH = Partial update**

### One-Line Answer

PUT generally replaces a resource, while PATCH is used for partial updates.

## Q5. What is an HTTP request?

### Interview Answer

An HTTP request is a message sent by a client to a server to perform an operation.

It can contain:

- HTTP method
- URL
- Headers
- Query parameters
- Path parameters
- Request body

### Example

    POST /users

    Content-Type: application/json

    {
        "name": "Musharraf",
        "age": 25
    }

### Memory Trick

**Request = Client → Server**

### One-Line Answer

An HTTP request is a message sent from a client to a server to request or perform an operation.

---

## Q6. What is an HTTP response?

### Interview Answer

An HTTP response is the message returned by the server after processing a client request.

It commonly contains:

- Status code
- Headers
- Response body

### Example

    HTTP 200 OK

    {
        "name": "Musharraf",
        "age": 25
    }

### Memory Trick

**Response = Server → Client**

### One-Line Answer

An HTTP response is the server's result returned to the client after processing a request.

---

## Q7. What are HTTP status codes?

### Interview Answer

HTTP status codes indicate the result of an HTTP request.

They are grouped into five categories:

- `1xx` → Informational
- `2xx` → Success
- `3xx` → Redirection
- `4xx` → Client-side error
- `5xx` → Server-side error

### Common Examples

    200 → OK
    201 → Created
    400 → Bad Request
    401 → Unauthorized
    403 → Forbidden
    404 → Not Found
    500 → Internal Server Error

### Memory Trick

**2 = Success**
**4 = Client problem**
**5 = Server problem**

### One-Line Answer

HTTP status codes indicate whether a request succeeded, failed, or requires further action.

---

## Q8. What is the difference between 401 and 403?

### Interview Answer

`401 Unauthorized` generally means the client has not provided valid authentication credentials.

`403 Forbidden` means the client is authenticated but does not have permission to access the requested resource.

### Example

401:

The user is not logged in.

403:

The user is logged in but tries to access an admin-only resource.

### Memory Trick

**401 → Authentication problem**
**403 → Permission problem**

### One-Line Answer

401 indicates an authentication issue, while 403 indicates that the authenticated user does not have sufficient permission.

---

## Q9. What is the difference between 400 and 500 errors?

### Interview Answer

`400`-series errors generally indicate a problem with the client's request.

`500`-series errors generally indicate a problem on the server side.

### Example

400:

    Missing required field
    Invalid input

500:

    Unexpected server exception
    Database failure

### Memory Trick

**4xx → Client**
**5xx → Server**

### One-Line Answer

4xx errors generally indicate client-side request problems, while 5xx errors indicate server-side failures.

They can contain information such as:

- Content type
- Authorization

- User agent

### Example

### Memory Trick

**Headers = Extra information**

## Q11. What is the purpose of `Content-Type`?

### Interview Answer

`Content-Type` tells the server or client what type of data is being sent.

For example:

    Content-Type: application/json

means the request body contains JSON data.

Other examples include:

    text/html
    application/xml
    multipart/form-data

### Memory Trick

**Content-Type = What is the data?**

### One-Line Answer

Content-Type specifies the media type of the data being sent in an HTTP request or response.

### Example

        "email": "example@email.com"

## Q13. What are path parameters?

### Interview Answer

Path parameters are values included directly in the URL path to identify a specific resource.

### Example

    GET /users/101

Here:

    101

is the user ID and acts as a path parameter.

Another example:

    GET /products/25

### Memory Trick

**Path parameter = Identify resource**

### One-Line Answer

A path parameter is a value in the URL path used to identify a specific resource.

---

## Q14. What are query parameters?

### Interview Answer

Query parameters are optional parameters included after `?` in a URL.

They are commonly used for:

- Filtering
- Searching
- Sorting
- Pagination

### Example

    GET /users?city=Mumbai&age=25

Here:

    city=Mumbai
    age=25

are query parameters.

### Memory Trick

**Query = Filter/Search/Sort**

### One-Line Answer

Query parameters are URL parameters commonly used for filtering, searching, sorting, and pagination.

---

## Q15. What is the difference between path parameters and query parameters?

### Interview Answer

Path parameters are generally used to **identify a specific resource**.

Query parameters are generally used to **filter, search, sort, or modify how resources are returned**.

### Example

Path parameter:

    GET /users/101

Query parameter:

    GET /users?city=Mumbai

### Memory Trick

**Path → Which resource?**
**Query → How to retrieve it?**

### One-Line Answer

Path parameters identify resources, while query parameters usually control filtering or retrieval behavior.

## Q19. What is JWT?

### Interview Answer

JWT stands for **JSON Web Token**.

It is a compact token format commonly used for authentication and securely transmitting claims between parties.

A JWT typically contains:

- Header
- Payload
- Signature

The server can validate the token to identify the user and its claims.

### Example

A client logs in and receives a JWT.

For later requests:

    Authorization: Bearer <JWT>

### Memory Trick

**JWT = Token-based authentication**

### One-Line Answer

JWT is a compact token format commonly used to carry authentication-related claims between a client and server.

## Q20. What does stateless mean in REST?

### Interview Answer

Stateless means that each request should contain the information necessary for the server to process it.

The server should not depend on stored session state from previous requests to understand the current request.

For example, if a request requires authentication, the client can send its authentication token with the request.

### Memory Trick

**Stateless = Every request carries what it needs**

### One-Line Answer

A stateless API does not rely on server-side session state to process each individual request.

---

## Q21. What is idempotency in REST APIs?

### Interview Answer

An operation is idempotent if making the same request multiple times has the **same intended final effect as making it once**.

For example, setting a user's status to "active" repeatedly should leave the resource in the same final state.

HTTP methods such as GET, PUT, and DELETE are generally designed to be idempotent, although actual application behavior depends on implementation.

POST is generally not considered idempotent.

### Memory Trick

**Idempotent = Repeat safely → Same final state**

### One-Line Answer

Idempotency means repeating the same request produces the same intended final state as making it once.

## Q22. What is API pagination?

### Interview Answer

Pagination means dividing a large set of results into smaller portions instead of returning everything in one response.

It helps reduce:

- Response size
- Memory usage
- Processing time
- Network load

### Example

    GET /users?page=2&limit=20

This can request the second page with 20 records.

### Memory Trick

**Pagination = Large data → Small pages**

### One-Line Answer

Pagination divides large API results into smaller pages to improve performance and usability.

## Q23. What is API versioning and why is it important?

### Interview Answer

API versioning allows us to introduce changes to an API while maintaining compatibility with existing clients.

For example:

    /api/v1/users
    /api/v2/users

If the response structure changes significantly, we can introduce a new version rather than unexpectedly breaking existing clients.

### Memory Trick

**Versioning = Change without breaking old clients**

### One-Line Answer

API versioning allows APIs to evolve while maintaining compatibility with existing clients.

## Q24. How would you troubleshoot an API that is not working?

### Interview Answer

I would follow a systematic approach:

1. Reproduce the issue.
2. Check the request URL and HTTP method.
3. Verify request parameters and body.
4. Check headers and authentication.
5. Check the HTTP status code.
6. Check application logs and traceback.
7. Verify database or external service dependencies.
8. Identify the root cause.
9. Apply the fix.
10. Retest with valid and invalid inputs.

### Example

If an API returns `500 Internal Server Error`, I would check the server logs and traceback first, identify the failing component, verify its dependencies, fix the root cause, and retest the endpoint.

### Memory Trick

**Request → Status → Logs → Dependency → Root Cause → Fix → Test**

### One-Line Answer

I troubleshoot APIs by checking the request, status code, logs, dependencies, root cause, and then testing the fix.

### Interview Answer

A good REST API should be:

- Clear and consistent
- Properly structured around resources
- Secure
- Well documented
- Validated
- Reliable
- Easy to maintain
- Properly versioned when needed
- Designed with meaningful status codes
- Able to handle errors gracefully

It should also provide predictable request and response formats.

### Memory Trick

**Good API = Clear + Secure + Reliable + Maintainable**

### One-Line Answer

A good REST API should be clear, consistent, secure, reliable, well-documented, properly validated, and easy to maintain.
