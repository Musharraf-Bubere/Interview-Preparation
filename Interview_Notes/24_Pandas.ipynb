## Q1. What is Pandas?

### Interview Answer

Pandas is an open-source Python library used for data manipulation, data analysis, and working with structured data.

It provides powerful data structures such as Series and DataFrame and offers functions for filtering, cleaning, transforming, aggregating, merging, and analyzing data.

It is commonly used in data science and machine learning workflows.

### Example

    Raw Data
       ↓
    Pandas
       ↓
    Clean → Transform → Analyze
       ↓
    ML / Visualization / Reporting

### Memory Trick

**Pandas = Load + Clean + Transform + Analyze**

### One-Line Answer

Pandas is a Python library used for manipulating, cleaning, transforming, and analyzing structured data.

## Q3. What is a Series in Pandas?

### Interview Answer

A Series is a one-dimensional labeled data structure that can store values of different types.

Each value has an associated index.

### Example

    import pandas as pd

    s = pd.Series([10, 20, 30])

    print(s)

    Output:

    0    10
    1    20
    2    30

### Memory Trick

**Series = One Column with Index**

### One-Line Answer

A Series is a one-dimensional labeled array in Pandas.

## Q4. What is a DataFrame?

### Interview Answer

A DataFrame is a two-dimensional labeled data structure consisting of rows and columns.

It is similar to a table in a database or a spreadsheet.

Each column can have its own data type.

### Example

    df = pd.DataFrame({
        "Name": ["A", "B"],
        "Age": [20, 25]
    })

### Memory Trick

**DataFrame = Table**

### One-Line Answer

A DataFrame is a two-dimensional tabular data structure containing labeled rows and columns.

## Q7. What is the difference between loc and iloc?

### Interview Answer

`loc` is primarily label-based indexing.

`iloc` is integer-position-based indexing.

### Example

    df.loc[2, "Age"]

    → Access using row/column labels


    df.iloc[2, 1]

    → Access using integer positions

### Memory Trick

**loc = Label**

**iloc = Integer Location**

### One-Line Answer

`loc` selects data using labels, while `iloc` selects data using integer positions.

## Q8. How do you select a column from a DataFrame?

### Interview Answer

A column can be selected using its column name.

### Example

    df["Age"]

For multiple columns:

    df[["Name", "Age"]]

The first returns a Series, while selecting multiple columns returns a DataFrame.

### Memory Trick

**One Column → Series**

**Multiple Columns → DataFrame**

### One-Line Answer

I can select a column using `df["column_name"]` and multiple columns using a list of column names.

## Q9. How do you filter rows in Pandas?

### Interview Answer

Boolean indexing can be used to filter rows based on conditions.

### Example

    df[df["Age"] > 25]

For multiple conditions, we can use `&` and `|` with parentheses.

### Example

    df[(df["Age"] > 25) & (df["City"] == "Mumbai")]

### Memory Trick

**Condition → Boolean Mask → Filter**

### One-Line Answer

I use boolean indexing to filter rows based on one or more conditions.

## Q10. What is the difference between drop and dropna?

### Interview Answer

`drop()` is used to remove specified rows or columns.

`dropna()` is specifically used to remove rows or columns containing missing values.

### Example

    df.drop(columns=["Age"])

    → Removes a specific column


    df.dropna()

    → Removes rows containing missing values

### Memory Trick

**drop = Specific Data**

**dropna = Missing Data**

### One-Line Answer

`drop()` removes specified rows or columns, while `dropna()` removes data containing missing values.

---

## Q11. How do you handle missing values in Pandas?

### Interview Answer

Pandas provides several methods for handling missing values.

Common approaches include:

    df.isna()
    df.notna()
    df.dropna()
    df.fillna()

For example, numerical missing values can sometimes be replaced with the mean or median depending on the problem.

### Example

    df["Age"] = df["Age"].fillna(df["Age"].median())

The appropriate method depends on the dataset and the reason for the missing values.

### Memory Trick

**Detect → Remove or Fill**

### One-Line Answer

I detect missing values using `isna()` and then either remove them with `dropna()` or replace them using `fillna()` based on the problem.

---

## Q12. What is the difference between fillna and interpolate?

### Interview Answer

`fillna()` replaces missing values using a specified value or strategy.

`interpolate()` estimates missing values based on surrounding data according to an interpolation method.

### Example

    df["Age"].fillna(25)

    → Fill with a fixed value


    df["Temperature"].interpolate()

    → Estimate missing values from surrounding values

### Memory Trick

**fillna = Replace**

**interpolate = Estimate**

### One-Line Answer

`fillna()` replaces missing values using a chosen value or strategy, while `interpolate()` estimates missing values from surrounding data.

---

## Q15. What is groupby in Pandas?

### Interview Answer

`groupby()` is used to divide data into groups based on one or more columns and then perform operations on each group.

It is commonly used for aggregation.

### Example

    df.groupby("Department")["Salary"].mean()

This calculates the average salary for each department.

### Memory Trick

**Group → Apply → Aggregate**

### One-Line Answer

`groupby()` groups data based on one or more columns so that we can perform aggregations or transformations on each group.

## Q16. What is the difference between merge, join, and concat?

### Interview Answer

These methods combine Pandas objects in different ways.

`merge()` combines DataFrames based on common columns or keys, similar to SQL joins.

`join()` is primarily designed to combine DataFrames using indexes or specified keys.

`concat()` combines objects along an axis, such as stacking rows or columns.

### Example

    merge
    → SQL-style key-based combination

    join
    → Index/key-based combination

    concat
    → Stack rows or columns

### Memory Trick

**merge = Keys**

**join = Index**

**concat = Stack**

### One-Line Answer

`merge()` performs key-based combinations, `join()` commonly works through indexes, and `concat()` combines objects along rows or columns.

---

## Q17. What is the difference between axis=0 and axis=1?

### Interview Answer

In Pandas, `axis=0` generally refers to the row axis, meaning operations are performed down the rows.

`axis=1` refers to the column axis, meaning operations are performed across columns.

### Example

    df.drop(columns=["Age"], axis=1)

    → Column operation


    df.drop(index=[0], axis=0)

    → Row operation

### Memory Trick

**axis=0 → Rows**

**axis=1 → Columns**

### One-Line Answer

In common DataFrame operations, `axis=0` refers to the row axis and `axis=1` refers to the column axis.

## Q18. What is apply() in Pandas?

### Interview Answer

`apply()` allows us to apply a function across a Series or along an axis of a DataFrame.

It can be useful when a transformation cannot be easily expressed using built-in vectorized operations.

### Example

    df["Age"] = df["Age"].apply(
        lambda x: x + 1
    )

Here, the function is applied to each value in the Age column.

### Memory Trick

**apply = Apply Function**

### One-Line Answer

`apply()` applies a function to values in a Series or along an axis of a DataFrame.

---

## Q19. What is the difference between apply, map, and applymap?

### Interview Answer

`map()` is primarily used with a Series to transform individual values.

`apply()` can operate on a Series or DataFrame and can apply functions along an axis.

For element-wise DataFrame transformations, modern Pandas uses `DataFrame.map()`; older code may use the name `applymap()`.

### Example

    Series:
    s.map(function)

    DataFrame:
    df.apply(function, axis=...)

    Element-wise DataFrame transformation:
    df.map(function)

### Memory Trick

**Series → map**

**Axis → apply**

**DataFrame Elements → map**

### One-Line Answer

`map()` performs element-wise Series transformations, `apply()` applies functions to Series or DataFrame axes, and DataFrame element-wise mapping is done with `DataFrame.map()` in current Pandas.

## Q20. What is vectorization in Pandas?

### Interview Answer

Vectorization means performing operations on an entire Series or DataFrame rather than explicitly looping through individual elements in Python.

Vectorized operations are generally preferred because they are concise and can be more efficient.

### Example

Instead of:

    df["total"] = [
        x * 2 for x in df["price"]
    ]

We can write:

    df["total"] = df["price"] * 2

### Memory Trick

**Vectorization = Operate on Whole Column**

### One-Line Answer

Vectorization performs operations on entire Pandas objects instead of manually looping through individual elements.

---

## Q21. How do you read and write CSV files using Pandas?

### Interview Answer

Pandas provides `read_csv()` to load CSV data into a DataFrame.

The `to_csv()` method can be used to write a DataFrame to a CSV file.

### Example

    df = pd.read_csv("data.csv")

    df.to_csv(
        "output.csv",
        index=False
    )

The same concept applies to many other file formats using Pandas' corresponding I/O functions.

### Memory Trick

**read_csv = Load**

**to_csv = Save**

### One-Line Answer

I use `pd.read_csv()` to read CSV data and `DataFrame.to_csv()` to write a DataFrame to a CSV file.

### Interview Answer

`sort_values()` is used to sort a DataFrame based on one or more columns.

### Example

    df.sort_values(
        by="Salary",
        ascending=False
    )

This sorts the data by salary in descending order.

We can also sort by multiple columns.

### Memory Trick

**sort_values = Sort by Column**

### One-Line Answer

I use `sort_values()` to sort DataFrame rows based on one or more columns.

## Q23. How do you change the data type of a Pandas column?

### Interview Answer

The `astype()` method can be used to convert a column to a desired data type.

### Example

    df["Age"] = df["Age"].astype(int)

For numeric conversion where invalid values may exist, `pd.to_numeric()` with appropriate error handling can be useful.

### Example

    df["Age"] = pd.to_numeric(
        df["Age"],
        errors="coerce"
    )

### Memory Trick

**astype = Convert Type**

### One-Line Answer

I use `astype()` for explicit type conversion and functions such as `pd.to_numeric()` when handling potentially invalid numeric values.

## Q24. How do you work with dates and times in Pandas?

### Interview Answer

Pandas provides datetime functionality through `pd.to_datetime()` and related datetime operations.

I can convert a column into datetime format and then extract components such as year, month, day, or day of the week.

### Example

    df["Date"] = pd.to_datetime(
        df["Date"]
    )

    df["Year"] = df["Date"].dt.year

    df["Month"] = df["Date"].dt.month

### Memory Trick

**to_datetime → .dt → Date Features**

### One-Line Answer

I use `pd.to_datetime()` to convert values into datetime objects and the `.dt` accessor to work with date components.
