# 📚 RAG — Retrieval-Augmented Generation
# 25 Interview Questions

---

## Q1. What is RAG?

### Interview Answer

RAG stands for **Retrieval-Augmented Generation**.

It is a technique where an application retrieves relevant information from an external knowledge source and provides that information to an LLM as context before generating the answer.

### Flow

    User Query
        ↓
    Retrieve Relevant Information
        ↓
    Add Retrieved Information to Prompt
        ↓
    LLM
        ↓
    Answer

RAG is useful when the application needs to answer questions using private, domain-specific, or frequently changing information.

### Memory Trick

**RAG = Retrieve → Augment → Generate**

### One-Line Answer

RAG retrieves relevant external information and provides it to an LLM as context to generate a more grounded answer.

---

## Q2. Why do we need RAG?

### Interview Answer

LLMs have limitations because their knowledge comes primarily from their training data and they may not have access to an application's latest or private information.

RAG allows us to provide relevant external information at inference time without retraining the model.

For example, a company can build a chatbot that answers questions from internal documents.

### Memory Trick

**LLM knowledge may be limited → RAG provides external knowledge**

### One-Line Answer

RAG is used to provide an LLM with relevant external information without retraining the model.

## Q4. What is document loading in RAG?

### Interview Answer

Document loading is the process of reading information from different sources and converting it into a format that the RAG pipeline can process.

Sources can include:

- PDF files
- Text files
- Websites
- Word documents
- Databases
- APIs

### Example

    PDF
      ↓
    Document Loader
      ↓
    Text/Documents
      ↓
    Chunking

### Memory Trick

**Loader = Bring data into the pipeline**

### One-Line Answer

Document loading is the process of importing information from sources such as PDFs, websites, or databases into a RAG pipeline.

## Q5. What is text splitting or chunking?

### Interview Answer

Chunking is the process of dividing large documents into smaller pieces called chunks.

Instead of embedding an entire large document, we split it into manageable sections.

### Example

    Large Document
          ↓
    ┌─────┼─────┐
    ↓     ↓     ↓
    C1    C2    C3

Chunking helps retrieval find the specific portion of a document that is relevant to the user's query.

### Memory Trick

**Chunking = Big document → Small useful pieces**

### One-Line Answer

Chunking divides large documents into smaller pieces so relevant information can be retrieved efficiently.

---

## Q6. Why is chunking important in RAG?

### Interview Answer

Chunking is important because retrieving an entire large document may provide too much irrelevant information to the LLM.

Good chunks should contain enough context to preserve meaning while remaining focused on a particular topic.

Poor chunking can cause:

- Missing context
- Irrelevant retrieval
- Poor answers
- Larger prompts
- Higher cost

### Memory Trick

**Good chunks → Better retrieval → Better answers**

### One-Line Answer

Chunking improves retrieval by creating focused pieces of information that can be provided to the LLM as relevant context.

---

## Q7. What is chunk size?

### Interview Answer

Chunk size determines approximately how much text is included in each chunk.

A smaller chunk can provide more precise retrieval, while a larger chunk can preserve more context.

There is no universally correct chunk size. It depends on:

- Document structure
- Query type
- Embedding model
- Context window
- Retrieval quality

### Memory Trick

**Small → Precise**

**Large → More context**

### One-Line Answer

Chunk size controls how much content each retrieved document chunk contains.

---

## Q8. What is chunk overlap?

### Interview Answer

Chunk overlap means sharing some text between consecutive chunks.

### Example

    Chunk 1:
    A B C D E

    Chunk 2:
        D E F G H

Here, D and E are shared.

Overlap helps prevent important information from being lost when a sentence or concept falls across chunk boundaries.

### Memory Trick

**Overlap = Shared context between chunks**

### One-Line Answer

Chunk overlap repeats some content between neighboring chunks to preserve context across chunk boundaries.

---

## Q9. What are embeddings in RAG?

### Interview Answer

Embeddings convert text into numerical vectors that represent semantic meaning.

During document ingestion, chunks are converted into embeddings and stored.

When a user asks a question, the query is also converted into an embedding.

The system then compares the query embedding with document embeddings to find semantically similar content.

### Flow

    Document Chunk → Embedding → Vector

    User Query → Embedding → Vector

    Compare Vectors
          ↓
    Relevant Chunks

### Memory Trick

**Text → Vector → Similarity**

### One-Line Answer

Embeddings represent document chunks and queries as vectors so semantically relevant information can be retrieved.

---

## Q10. What is a vector database?

### Interview Answer

A vector database is a database designed to store and search numerical vector representations efficiently.

In RAG, it typically stores:

- Embeddings
- Original or associated text
- Metadata

It allows the application to find vectors that are similar to a query vector.

Examples include:

- FAISS
- Chroma
- Pinecone
- Weaviate
- Milvus

### Memory Trick

**Vector DB = Store + Search embeddings**

### One-Line Answer

A vector database stores embeddings and enables efficient similarity-based retrieval.

---

## Q11. What is a retriever?

### Interview Answer

A retriever is the component responsible for finding relevant documents or chunks for a user's query.

The query is typically converted into an embedding, and the retriever searches the vector store for relevant content.

### Flow

    Query
      ↓
    Query Embedding
      ↓
    Vector Search
      ↓
    Top Relevant Chunks

### Memory Trick

**Retriever = Finds relevant information**

### One-Line Answer

A retriever searches the knowledge base and returns information relevant to the user's query.

---

## Q12. What is similarity search?

### Interview Answer

Similarity search finds documents or vectors that are most similar to a given query.

In RAG, the query embedding is compared with stored document embeddings.

Common similarity measures include:

- Cosine similarity
- Euclidean distance
- Dot product

The exact metric depends on the embedding model and vector database.

### Memory Trick

**Similarity Search = Find vectors closest in meaning**

### One-Line Answer

Similarity search retrieves vectors that are most similar to the query vector.

---

## Q13. What is cosine similarity?

### Interview Answer

Cosine similarity measures the similarity between two vectors based on the angle between them.

Its value is commonly interpreted as:

- Close to 1 → Highly similar direction
- Around 0 → Little directional similarity
- Close to -1 → Opposite direction

In embedding-based retrieval, cosine similarity is commonly used to compare semantic representations.

### Memory Trick

**Cosine similarity = Compare vector direction**

### One-Line Answer

Cosine similarity measures how similar two vectors are based on the angle between them.

---

## Q14. What is top-k retrieval?

### Interview Answer

Top-k retrieval means returning the **k most relevant results** for a query.

For example, if k = 5, the retriever returns the five highest-ranked relevant chunks.

### Example

    Query
      ↓
    Retrieval
      ↓
    5 most relevant chunks

The value of k affects both retrieval coverage and the amount of context sent to the LLM.

### Memory Trick

**k = Number of results returned**

### One-Line Answer

Top-k retrieval returns the k highest-ranked results for a query.

---

## Q15. What is metadata filtering in RAG?

### Interview Answer

Metadata filtering allows us to restrict retrieval using additional information stored with documents.

For example, documents may contain metadata such as:

- Department
- Date
- Document type
- User
- Category

We can retrieve only documents matching specific metadata conditions.

### Example

    Query:
    "What is our leave policy?"

    Filter:
    department = HR

The retriever can search only within HR documents.

### Memory Trick

**Semantic Search + Metadata Filter = More targeted retrieval**

### One-Line Answer

Metadata filtering restricts retrieval using document attributes such as category, date, department, or user.

---

## Q16. What is the difference between semantic search and keyword search?

### Interview Answer

Keyword search primarily matches exact or related words.

Semantic search focuses on meaning using embeddings.

### Example

Query:

    "How can I change my password?"

A keyword search may look for the word "password".

Semantic search can also find:

    "Steps to reset account credentials"

even though the wording is different.

### Memory Trick

**Keyword → Words**

**Semantic → Meaning**

### One-Line Answer

Keyword search matches terms, while semantic search retrieves information based on semantic meaning.

---

## Q17. What is hybrid search?

### Interview Answer

Hybrid search combines multiple retrieval approaches, commonly keyword-based search and semantic vector search.

This can improve retrieval because:

- Keyword search is useful for exact terms.
- Semantic search is useful for meaning and paraphrases.

### Flow

    Query
      ↓
    ┌──────────────┐
    ↓              ↓
    Keyword      Vector
    Search       Search
    ↓              ↓
    └──────┬───────┘
           ↓
       Combined Ranking
           ↓
       Relevant Results

### Memory Trick

**Hybrid = Keyword + Semantic**

### One-Line Answer

Hybrid search combines keyword and semantic retrieval to improve the chances of finding relevant information.

---

## Q18. What is reranking in RAG?

### Interview Answer

Reranking is a second-stage process that reorders retrieved documents based on their relevance to the query.

A common pipeline is:

    Query
      ↓
    Initial Retrieval
      ↓
    Top N Results
      ↓
    Reranker
      ↓
    Best Results
      ↓
    LLM

The initial retriever provides candidates, and the reranker performs a more focused relevance assessment.

### Memory Trick

**Retrieve first → Rank better**

### One-Line Answer

Reranking reorders initially retrieved documents to select the most relevant context for the LLM.

---

## Q19. What is the role of the LLM in RAG?

### Interview Answer

The LLM receives the user's question together with the retrieved context and generates the final response.

The LLM is not necessarily responsible for finding the documents itself.

### Flow

    User Query
        +
    Retrieved Context
        ↓
       LLM
        ↓
    Final Answer

This separation allows retrieval and generation to perform different responsibilities.

### Memory Trick

**Retriever finds → LLM generates**

### One-Line Answer

In RAG, the retriever finds relevant information and the LLM uses that information to generate the response.

---

## Q20. What is grounding in RAG?

### Interview Answer

Grounding means generating an answer based on trusted information supplied to the model rather than relying only on the model's learned knowledge.

In RAG, retrieved documents act as grounding context.

### Example

    Company Policy Documents
            ↓
        Retrieval
            ↓
      Relevant Policy
            ↓
           LLM
            ↓
      Grounded Answer

Grounding can reduce unsupported answers, although it does not guarantee that every generated answer will be correct.

### Memory Trick

**Grounding = Answer based on evidence**

### One-Line Answer

Grounding means using trusted external information as evidence for the model's response.

---

## Q21. What is the difference between RAG and fine-tuning?

### Interview Answer

RAG and fine-tuning solve different problems.

### RAG

RAG provides external information at inference time.

    Query
      ↓
    Retrieve Knowledge
      ↓
    LLM

It is useful when information changes frequently or comes from private documents.

### Fine-Tuning

Fine-tuning changes the model's parameters by training it further on task-specific data.

    Base Model
        ↓
    Fine-Tuning
        ↓
    Adapted Model

### Memory Trick

**RAG → Add knowledge**

**Fine-tuning → Adapt model**

### One-Line Answer

RAG provides external knowledge at inference time, while fine-tuning changes the model through additional training.

## Q22. What is the difference between RAG and simply giving a document to an LLM?

### Interview Answer

The main difference is how information is selected.

If we provide an entire document directly, the model receives all of that content as context.

In RAG, the system first searches the knowledge base and retrieves only the most relevant pieces.

### Direct Context

    Document
      ↓
    LLM

### RAG

    Large Knowledge Base
          ↓
       Retrieval
          ↓
    Relevant Chunks
          ↓
         LLM

RAG becomes especially useful when the knowledge base is large.

### Memory Trick

**Direct → Give everything**

**RAG → Retrieve what matters**

### One-Line Answer

RAG dynamically retrieves relevant information from a larger knowledge base instead of providing the entire knowledge base to the model.

## Q23. What can cause poor answers in a RAG application?

### Interview Answer

Poor RAG responses can come from different parts of the pipeline.

Possible causes include:

- Poor document quality
- Incorrect chunking
- Poor chunk size
- Incorrect embeddings
- Poor retrieval
- Wrong top-k value
- Missing metadata filters
- Poor reranking
- Insufficient context
- Irrelevant context
- Poor prompt
- LLM hallucination
- Output-processing errors

I would identify whether the problem is in **retrieval or generation** before changing the entire system.

### Memory Trick

**Data → Chunk → Embed → Retrieve → Prompt → Generate**

### One-Line Answer

Poor RAG responses can result from problems in the data, chunking, embeddings, retrieval, context, prompt, or generation stage.

## Q24. How would you troubleshoot a RAG application that gives incorrect answers?

### Interview Answer

I would troubleshoot the pipeline step by step.

First, I would reproduce the issue using the same query.

Then I would check:

1. The original documents
2. Document loading
3. Chunking
4. Embeddings
5. Retrieved chunks
6. Similarity scores or ranking
7. Metadata filters
8. Prompt and context sent to the LLM
9. LLM response
10. Output validation

The most important step is to determine whether the correct information was retrieved.

If the correct information was not retrieved, I would investigate the retrieval pipeline.

If the correct information was retrieved but the answer is still wrong, I would investigate the prompt, model behavior, context handling, or output processing.

### Memory Trick

**Retrieve first → Generate second**

### One-Line Answer

I would trace the RAG pipeline from documents and retrieval through prompt construction and generation to determine whether the failure is in retrieval or generation.

Possible improvements include:

- Improve document quality
- Optimize chunking
- Tune chunk size and overlap
- Select a suitable embedding model
- Tune top-k
- Add metadata filtering
- Use hybrid search
- Add reranking
- Improve the prompt
- Reduce irrelevant context
- Add citations or source references where appropriate
- Add evaluation datasets
- Monitor retrieval and generation quality

I would evaluate each change using representative queries and measurable metrics.

### Memory Trick

**Better Data → Better Retrieval → Better Context → Better Generation → Better Evaluation**

### One-Line Answer

I would improve RAG by optimizing document processing, retrieval, ranking, context quality, prompting, and evaluation while measuring the impact of each change.
