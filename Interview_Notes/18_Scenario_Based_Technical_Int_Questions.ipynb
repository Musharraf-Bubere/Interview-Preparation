# 🎯 Scenario-Based Technical Interview Questions — 25 Questions

---

## Q1. Your application is working on your machine but not working in another environment. How would you troubleshoot it?

### Interview Answer

I would first compare the two environments.

I would check:

- Python or runtime version
- Installed package versions
- Environment variables
- Configuration files
- Database connection
- API keys or credentials
- File paths
- Operating system differences
- Network access

Then I would reproduce the issue and compare the logs between the working and failing environments.

I would identify the difference causing the problem instead of randomly changing the application.

### Memory Trick

**Environment → Dependencies → Configuration → Database → Network → Logs**

### One-Line Answer

I would compare runtime, dependencies, configuration, environment variables, database, and network settings to identify the environment-specific difference.

---

## Q2. A user says the application is not working, but you cannot reproduce the issue. What would you do?

### Interview Answer

I would collect more information from the user.

I would ask for:

- Exact error message
- Time when the issue occurred
- Steps performed before the error
- Input used
- Screenshot if appropriate
- User or request details needed for investigation
- Whether the issue happens every time

Then I would check application logs and monitoring data around that time.

If necessary, I would try to reproduce the same scenario using the information collected.

### Memory Trick

**User Evidence → Logs → Reproduce**

### One-Line Answer

If I cannot reproduce the issue, I would collect exact reproduction details and correlate them with logs and monitoring data.

---

## Q3. An application suddenly became slow after a new deployment. How would you investigate?

### Interview Answer

I would first confirm that the performance degradation started after the deployment.

Then I would compare the current version with the previous version and check:

- Application response time
- Error rate
- Database queries
- External API calls
- CPU and memory usage
- New code paths
- Configuration changes
- Logs

If the new deployment is confirmed as the cause, I would identify the specific change responsible and determine whether rollback or a fix is the safest option.

### Memory Trick

**Before vs After Deployment**

### One-Line Answer

I would compare the old and new versions, identify what changed, measure the bottleneck, and then fix or safely roll back the problematic change.

---

## Q4. An API works for some users but returns errors for others. How would you troubleshoot it?

### Interview Answer

I would first compare a working request with a failing request.

I would check:

- Authentication
- User roles
- Permissions
- Request parameters
- Input data
- Resource ownership
- API response
- Logs

If the error is related to authorization, I would inspect the access-control logic.

If it is related to data, I would compare the input and database records.

### Memory Trick

**Working User vs Failing User**

### One-Line Answer

I would compare successful and failed requests and investigate authentication, authorization, input data, permissions, and resource-specific conditions.

---

## Q5. A database query works but takes too long. What would you do?

### Interview Answer

I would first measure the query execution time and inspect the execution plan.

Then I would check:

- Indexes
- Joins
- Filters
- Sorting
- Aggregations
- Number of rows
- Unnecessary columns
- Database statistics

I would identify the actual bottleneck before optimizing.

After making a change, I would run the query again and compare the performance.

### Memory Trick

**Measure → Explain → Optimize → Measure**

### One-Line Answer

I would inspect the execution plan and query structure, identify the bottleneck, optimize it, and measure the improvement.

---

## Q6. An external API is returning timeout errors. What would you check?

### Interview Answer

I would check whether the external API is available and whether the timeout is consistent.

I would investigate:

- Network connectivity
- API response time
- Timeout configuration
- Authentication
- Request payload
- Rate limits
- External service status
- Application logs

Depending on the requirement, I could use controlled retries, timeouts, fallback behavior, or cached data.

### Memory Trick

**Network → Timeout → Auth → Rate Limit → Logs**

### One-Line Answer

I would determine whether the timeout is caused by the network, external service, configuration, authentication, or rate limiting and handle it safely.

## Q7. Your application is returning incorrect data from the database. How would you investigate?

### Interview Answer

I would trace the data from the database to the final response.

I would check:

1. Database records
2. SQL query
3. Query parameters
4. Joins and filters
5. ORM logic if used
6. Data transformation
7. Business logic
8. API response

I would compare the expected result with the actual database result to identify where the data becomes incorrect.

### Memory Trick

**Database → Query → Processing → Response**

### One-Line Answer

I would trace the data from the database query through processing and business logic to determine where the incorrect value is introduced.

## Q8. An API suddenly starts returning HTTP 500 errors. What is your approach?

### Interview Answer

I would first reproduce the request and check the application logs and traceback.

Then I would trace:

    Request
      ↓
    Endpoint
      ↓
    Validation
      ↓
    Business Logic
      ↓
    Database / External API
      ↓
    Response

I would identify the exact failing component, fix the root cause, and test the endpoint again.

### Memory Trick

**Trace the request end-to-end**

### One-Line Answer

I would use the logs and traceback to trace the request through the application and identify the exact component causing the 500 error.

## Q9. A Python application is consuming too much memory. How would you troubleshoot it?

### Interview Answer

I would first confirm the memory increase using monitoring or system metrics.

Then I would investigate:

- Large objects
- Loading entire datasets into memory
- Unnecessary copies
- Caching
- Loops that continuously accumulate data
- Unclosed resources
- Memory leaks

I would use profiling tools if required.

Then I would optimize the specific source of memory usage and measure the result.

### Memory Trick

**Measure → Find large objects → Optimize → Measure**

### One-Line Answer

I would measure memory usage, identify what is consuming memory, optimize that component, and verify the improvement.

## Q10. A Python application is using too much CPU. What would you check?

### Interview Answer

I would first identify which process or operation is consuming CPU.

Then I would check:

- Infinite or inefficient loops
- Expensive calculations
- Repeated operations
- Large data processing
- Database queries
- External API retry loops
- Concurrency behavior

I would use profiling or monitoring to locate the CPU-intensive operation.

### Memory Trick

**Find the expensive operation**

### One-Line Answer

I would identify the CPU-intensive process using monitoring or profiling and optimize the specific bottleneck.

## Q11. An API is receiving invalid input. How should the application handle it?

### Interview Answer

The application should validate input before processing it.

Validation can include:

- Required fields
- Data types
- Allowed values
- String length
- Numeric ranges
- Business rules

The API should return a clear client-side error rather than allowing invalid data to reach deeper application logic.

### Memory Trick

**Validate early**

### One-Line Answer

I would validate incoming data at the API boundary and return a clear 4xx response for invalid input.

---

## Q12. An application is dependent on a third-party service that is temporarily unavailable. What should the application do?

### Interview Answer

The application should fail gracefully rather than crashing or waiting indefinitely.

Depending on the use case, I would consider:

- Timeout
- Limited retries
- Exponential backoff
- Fallback response
- Cached data
- Circuit breaker
- Clear error message

The exact strategy depends on how critical the external service is.

### Memory Trick

**Don't wait forever → Timeout + Controlled Recovery**

### One-Line Answer

I would use appropriate timeouts, controlled retries, and fallback behavior so the application can handle temporary dependency failures gracefully.

---

## Q13. What is a retry mechanism?

### Interview Answer

A retry mechanism attempts an operation again after a temporary failure.

It can be useful for transient failures such as:

- Temporary network issues
- Temporary service unavailability
- Rate limiting

Retries should have limits and appropriate delays.

For example, exponential backoff increases the waiting time between attempts.

### Memory Trick

**Retry ≠ Retry Forever**

### One-Line Answer

A retry mechanism repeats a failed operation for potentially temporary failures, using controlled limits and delays.

---

## Q14. What is a circuit breaker pattern?

### Interview Answer

A circuit breaker prevents an application from repeatedly calling a failing external service.

When failures exceed a configured threshold, the circuit can open and temporarily stop requests to that dependency.

After a recovery period, the system can test whether the dependency has recovered.

### Simple Flow

    Normal
      ↓
    Failures Increase
      ↓
    Circuit Opens
      ↓
    Stop Calls Temporarily
      ↓
    Recovery Check
      ↓
    Resume

### Memory Trick

**Failing dependency → Stop repeatedly calling it**

### One-Line Answer

A circuit breaker temporarily stops calls to an unhealthy dependency to prevent repeated failures and additional load.

### Interview Answer

I would not assume that a successful deployment means the application is healthy.

I would check:

- Application health
- Logs
- API responses
- Database connectivity
- Environment variables
- Configuration
- External dependencies
- Authentication
- Smoke tests
- Recent deployment changes

I would compare the behavior with the previous version.

### Memory Trick

**Deployment success ≠ Application success**

### One-Line Answer

I would verify application health after deployment by checking logs, dependencies, configuration, APIs, and smoke tests.

## Q16. What is a rollback?

### Interview Answer

Rollback means reverting an application or deployment to a previous known-good version.

It can be used when a new release introduces a serious issue and immediate recovery is required.

Rollback should be performed carefully, especially when database schema changes are involved.

### Memory Trick

**Rollback = Return to known-good version**

### One-Line Answer

Rollback restores a previous stable version when a new deployment causes a critical issue.

## Q17. How would you decide between fixing an issue immediately and escalating it?

### Interview Answer

I would consider:

- Severity
- Business impact
- My access and responsibility
- Risk of making the situation worse
- Whether I have enough evidence
- Required expertise
- Time sensitivity

If the issue is within my responsibility and I can safely fix it, I would proceed.

If it requires another team's expertise or has high risk, I would escalate with clear evidence and details of what I have already investigated.

### Memory Trick

**Impact + Risk + Ownership**

### One-Line Answer

I would decide based on business impact, risk, ownership, and whether I have the required knowledge and access to resolve the issue safely.

I would include:

- Problem description
- Impact
- Time of occurrence
- Steps to reproduce
- Error message
- Relevant logs
- Affected component
- Troubleshooting already performed
- Suspected root cause

I would escalate with the problem, impact, reproduction steps, evidence, investigation performed, suspected cause, and current status.

## Q19. How would you handle multiple technical issues at the same time?

### Interview Answer

I would prioritize them based on:

- Business impact
- Number of affected users
- Severity
- Urgency
- Dependencies

A production outage affecting many users would generally receive higher priority than a minor issue affecting one user.

I would keep track of each issue and communicate status clearly.

### Memory Trick

**Impact first**

### One-Line Answer

I would prioritize issues based on severity, business impact, affected users, and urgency.

## Q20. What would you do if a fix solves the issue but creates another problem?

### Interview Answer

I would first assess the new problem and determine whether the fix should be reverted.

I would avoid continuing to make random changes.

I would:

1. Assess impact.
2. Reproduce the new issue.
3. Identify what changed.
4. Check whether the original fix caused it.
5. Roll back if necessary.
6. Find a safer solution.
7. Test both the original and new scenarios.

### Memory Trick

**Fix one → Don't break another**

### One-Line Answer

I would assess the new impact, determine whether the fix caused it, roll back if necessary, and test a safer solution against both scenarios.

### Interview Answer

Intermittent bugs can be difficult because they may not reproduce consistently.

I would collect as much evidence as possible:

- Exact timestamps
- Request IDs
- User information where appropriate
- Input conditions
- Logs
- Trace information
- Application metrics
- Database state
- External service responses

I would look for patterns across successful and failed cases.

### Memory Trick

**Intermittent → Collect more evidence**

### One-Line Answer

For intermittent issues, I would correlate timestamps, logs, request information, inputs, and system metrics to identify a repeatable pattern.

## Q22. How would you troubleshoot a problem that started after a configuration change?

### Interview Answer

I would first compare the current configuration with the previous known-good configuration.

I would verify:

- Environment variables
- API endpoints
- Database settings
- Timeouts
- Authentication settings
- Feature flags
- Configuration files

If the timing strongly correlates with the change, I would test the previous configuration in a safe environment or roll back the configuration if appropriate.

### Memory Trick

**Configuration changed → Compare old vs new**

### One-Line Answer

I would compare the old and new configuration, identify the changed setting, and safely verify whether it caused the issue.

## Q23. How would you verify that a technical issue is completely fixed?

### Interview Answer

I would not only check that the original error disappeared.

I would:

1. Reproduce the original failure scenario.
2. Verify that it now works.
3. Test normal use cases.
4. Test relevant edge cases.
5. Check logs for new errors.
6. Check dependent functionality.
7. Monitor the application after the fix if appropriate.

### Memory Trick

**Original case + Normal case + Edge cases**

### One-Line Answer

I would verify the original failure is resolved and then test normal, edge, and related scenarios to ensure the fix did not introduce another issue.

### Interview Answer

After resolving the immediate issue, I would identify preventive actions.

Depending on the root cause, these could include:

- Automated tests
- Better input validation
- Monitoring
- Alerts
- Better error handling
- Documentation
- Code review
- Configuration validation
- Performance improvements
- Dependency checks

The preventive action should address the actual root cause.

### Memory Trick

**Fix now → Prevent later**

### One-Line Answer

I would address the root cause and add appropriate tests, monitoring, validation, documentation, or process improvements to prevent recurrence.

## Q25. Explain your complete troubleshooting approach with an example.

### Interview Answer

My general approach is to first understand and reproduce the issue.

For example, if an API suddenly starts returning a 500 error, I would:

1. Confirm the issue.
2. Reproduce the request.
3. Check the HTTP response.
4. Check application logs and traceback.
5. Identify the failing endpoint.
6. Trace the request through validation and business logic.
7. Check database or external API dependencies.
8. Review recent changes.
9. Identify the root cause.
10. Apply the smallest appropriate fix.
11. Test the original failing request.
12. Test normal and edge cases.
13. Verify application health.
14. Document the root cause and solution.

For me, the important point is to use evidence and follow a systematic process rather than making random changes.

### Memory Trick

**Reproduce → Evidence → Trace → Root Cause → Fix → Test → Verify → Document**

### One-Line Answer

I troubleshoot systematically by reproducing the issue, collecting evidence, tracing the failure to its root cause, applying a controlled fix, testing it, verifying the system, and documenting the solution.


# 🧠 MASTER SCENARIO FRAMEWORK

When the interviewer gives you an unexpected technical scenario, use:

**S → I → R → F → V → D**

### S — Situation

What exactly is happening?

### I — Investigation

What evidence will I collect?

### R — Root Cause

What is actually causing the problem?

### F — Fix

What is the safest appropriate solution?

### V — Verify

Does the fix actually solve the problem?

### D — Document

What should be recorded to prevent or resolve it faster next time?

# ⭐ GOLDEN INTERVIEW RULE

If you don't know the exact technical solution, **do not invent one**.

A strong answer is:

> "I would first reproduce the issue and collect evidence. Then I would check the logs, recent changes, and relevant dependencies to identify the root cause. If I couldn't resolve it safely with the information available to me, I would escalate it to the appropriate team with the evidence and troubleshooting steps I had already performed."
