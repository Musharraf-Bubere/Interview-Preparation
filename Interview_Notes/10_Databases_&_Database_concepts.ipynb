# 🗃️ Databases & Database Concepts — Interview Questions

---

## Q1. What is a Database?

### Interview Answer

A database is an organized collection of data that allows applications to store, retrieve, update, and manage information efficiently.

For example, an application can use a database to store:

- Users
- Products
- Orders
- Transactions
- Application records

### Memory Trick

**Database = Organized data storage**

### One-Line Answer

A database is an organized system used to store, manage, and retrieve data efficiently.

---

## Q2. What is a DBMS?

### Interview Answer

DBMS stands for **Database Management System**.

It is software that allows us to create, store, retrieve, update, and manage data in databases.

Examples include:

- MySQL
- PostgreSQL
- Oracle Database
- Microsoft SQL Server

### Memory Trick

**DBMS = Software that manages databases**

### One-Line Answer

A DBMS is software used to create, manage, and interact with databases.

---

## Q3. What is an RDBMS?

### Interview Answer

RDBMS stands for **Relational Database Management System**.

It stores data in the form of tables consisting of rows and columns and allows relationships to be established between tables.

Examples include:

- PostgreSQL
- MySQL
- Oracle
- SQL Server

### Memory Trick

**RDBMS = Tables + Relationships**

### One-Line Answer

An RDBMS stores data in related tables consisting of rows and columns.

---

## Q4. What is the difference between DBMS and RDBMS?

### Interview Answer

DBMS is a general term for software that manages databases.

RDBMS is a type of DBMS based on the relational model, where data is organized into tables and relationships can be established between them.

### Simple Difference

**DBMS → General database management**

**RDBMS → Relational/table-based database management**

### One-Line Answer

RDBMS is a relational type of DBMS that organizes data into related tables.

---

## Q5. What is a table?

### Interview Answer

A table is a structure in a relational database used to store data in:

- Rows
- Columns

Each row generally represents a record, while each column represents an attribute of that record.

### Example

    users

    id | name      | age
    ---|-----------|----
    1  | Musharraf | 25
    2  | Ahmed     | 27

### Memory Trick

**Table = Rows + Columns**

### One-Line Answer

A table stores related data in rows and columns.

---

## Q6. What is a Primary Key?

### Interview Answer

A primary key is a column or combination of columns that uniquely identifies each row in a table.

A primary key:

- Must uniquely identify records
- Cannot contain NULL values
- Is used to identify individual records

### Example

    CREATE TABLE users (
        id INT PRIMARY KEY,
        name VARCHAR(100)
    );

Here, `id` uniquely identifies each user.

### Memory Trick

**Primary Key = Unique identity**

### One-Line Answer

A primary key uniquely identifies each record in a table and cannot contain NULL values.

---

## Q7. What is a Foreign Key?

### Interview Answer

A foreign key is a column or set of columns that references a key in another table.

It is used to establish relationships between tables and maintain referential integrity.

### Example

    users
    id | name

    orders
    id | user_id

Here, `orders.user_id` can reference `users.id`.

### Memory Trick

**Foreign Key = Link between tables**

### One-Line Answer

A foreign key references a key in another table and helps establish relationships between tables.

---

## Q8. What is the difference between Primary Key and Foreign Key?

### Interview Answer

A primary key uniquely identifies a record within its own table.

A foreign key references a key in another table to establish a relationship.

### Example

    users.id → Primary Key

    orders.user_id → Foreign Key

### Memory Trick

**Primary = Identity**

**Foreign = Relationship**

### One-Line Answer

A primary key identifies a record, while a foreign key connects records between tables.

---

## Q9. What are database constraints?

### Interview Answer

Constraints are rules applied to table columns to maintain data integrity and consistency.

Common constraints include:

- PRIMARY KEY
- FOREIGN KEY
- UNIQUE
- NOT NULL
- CHECK
- DEFAULT

### Example

    CREATE TABLE users (
        id INT PRIMARY KEY,
        email VARCHAR(100) UNIQUE,
        age INT CHECK (age >= 18)
    );

### Memory Trick

**Constraints = Data rules**

### One-Line Answer

Database constraints are rules that help maintain data accuracy, consistency, and integrity.

---

## Q10. What is a UNIQUE constraint?

### Interview Answer

A UNIQUE constraint ensures that values in a column or combination of columns are not duplicated.

It is useful for fields such as email addresses or usernames where duplicates should generally not be allowed.

### Example

    email VARCHAR(100) UNIQUE

### Memory Trick

**UNIQUE = No duplicate values**

### One-Line Answer

A UNIQUE constraint prevents duplicate values in the constrained column or column combination.

---

## Q11. What is normalization?

### Interview Answer

Normalization is the process of organizing data into related tables to reduce:

- Data redundancy
- Duplicate information
- Update anomalies
- Inconsistencies

Common normal forms include:

- 1NF
- 2NF
- 3NF
- BCNF

### Memory Trick

**Normalization = Reduce unnecessary duplication**

### One-Line Answer

Normalization organizes database data to reduce redundancy and improve consistency.

---

## Q12. What is denormalization?

### Interview Answer

Denormalization is the intentional introduction of some redundancy into a database to improve read performance or simplify certain queries.

It may reduce the number of joins required, but it can increase storage and make updates more complex.

### Memory Trick

**Normalization → Less redundancy**

**Denormalization → Sometimes more redundancy for performance**

### One-Line Answer

Denormalization intentionally adds some redundancy to improve performance or simplify data retrieval.

---

## Q13. What is an index?

### Interview Answer

An index is a database structure that helps the database find rows more efficiently.

Indexes can improve query performance, especially for frequently searched or joined columns.

However, indexes also:

- Consume storage
- Add overhead to INSERT, UPDATE, and DELETE operations

### Example

    CREATE INDEX idx_users_email
    ON users(email);

### Memory Trick

**Index = Faster search**

### One-Line Answer

An index improves data retrieval performance but adds storage and write overhead.

---

## Q14. When should you create an index?

### Interview Answer

Indexes are useful on columns that are frequently used in:

- WHERE conditions
- JOIN conditions
- ORDER BY
- Certain GROUP BY operations

However, I would not create indexes blindly.

I would analyze query patterns and execution plans because unnecessary indexes can increase storage and slow down write operations.

### Memory Trick

**Frequent search → Consider index**

### One-Line Answer

I would create indexes based on query patterns and performance needs, especially for frequently searched or joined columns.

---

## Q15. What is a database transaction?

### Interview Answer

A transaction is a logical unit of work containing one or more database operations that should be treated as a single unit.

For example, transferring money between two accounts involves:

1. Deducting money from one account
2. Adding money to another account

Both operations should succeed together or the transaction should be rolled back.

### Memory Trick

**Transaction = One logical unit of work**

### One-Line Answer

A transaction groups database operations into a logical unit that should be completed consistently.

---

## Q16. What is ACID?

### Interview Answer

ACID represents four important properties of database transactions:

**A — Atomicity**

All operations in a transaction succeed or the transaction is rolled back.

**C — Consistency**

The database moves from one valid state to another valid state.

**I — Isolation**

Concurrent transactions should not improperly interfere with each other.

**D — Durability**

Once a transaction is committed, its changes should persist even after certain failures.

### Memory Trick

**A = All or Nothing**

**C = Correct State**

**I = Independent**

**D = Data Persists**

### One-Line Answer

ACID stands for Atomicity, Consistency, Isolation, and Durability, which describe important transaction guarantees.

---

## Q17. What is COMMIT?

### Interview Answer

`COMMIT` permanently saves the changes made during the current transaction.

### Example

    BEGIN;

    UPDATE accounts
    SET balance = balance - 100
    WHERE id = 1;

    COMMIT;

After the commit, the transaction's changes are saved.

### Memory Trick

**COMMIT = Save changes**

### One-Line Answer

COMMIT permanently saves the changes made by a transaction.

---

## Q18. What is ROLLBACK?

### Interview Answer

`ROLLBACK` reverses changes made during the current transaction that have not yet been committed.

### Example

    BEGIN;

    UPDATE accounts
    SET balance = balance - 100
    WHERE id = 1;

    ROLLBACK;

The uncommitted change is undone.

### Memory Trick

**ROLLBACK = Undo transaction changes**

### One-Line Answer

ROLLBACK undoes uncommitted changes made during a transaction.

---

## Q19. What is a database view?

### Interview Answer

A view is a virtual table based on the result of a SQL query.

It generally does not store the result as a separate physical copy of the underlying data.

Views can be useful for:

- Simplifying complex queries
- Providing controlled access to data
- Reusing common queries

### Example

    CREATE VIEW active_users AS
    SELECT id, name
    FROM users
    WHERE status = 'active';

### Memory Trick

**View = Saved query as a virtual table**

### One-Line Answer

A view is a virtual table based on a query over one or more underlying tables.

### Interview Answer

A stored procedure is a database-side program containing SQL statements and sometimes procedural logic that can be executed by calling it.

It can be useful for encapsulating reusable database operations.

The exact syntax and capabilities depend on the database system.

### Memory Trick

**Stored Procedure = Reusable database-side logic**

### One-Line Answer

A stored procedure is reusable logic stored and executed within the database system.

## Q21. What is a database connection pool?

### Interview Answer

A connection pool maintains a collection of reusable database connections.

Instead of creating a new connection for every request, an application can reuse an existing connection from the pool.

This can improve performance and reduce the overhead of repeatedly establishing connections.

### Flow

    Application
         ↓
    Connection Pool
      ↓    ↓    ↓
     DB   DB   DB

### Memory Trick

**Pool = Reuse connections**

### One-Line Answer

A connection pool reuses database connections to reduce connection overhead and improve application performance.

## Q22. What is ORM?

### Interview Answer

ORM stands for **Object-Relational Mapping**.

It allows developers to interact with database records using programming-language objects and models instead of writing SQL for every operation.

For example, in Python, SQLAlchemy is commonly used as an ORM toolkit.

### Concept

    Python Object
          ↓
         ORM
          ↓
       Database

### Memory Trick

**ORM = Objects ↔ Database tables**

### One-Line Answer

ORM maps application objects and database structures so developers can work with database data through programming-language constructs.

---

## Q23. What are SQLAlchemy and Alembic?

### Interview Answer

**SQLAlchemy** is a Python SQL toolkit and ORM that helps applications interact with relational databases.

It provides features for:

- Database connections
- SQL construction
- ORM mapping
- Transactions

**Alembic** is a database migration tool commonly used with SQLAlchemy.

It helps manage changes to the database schema over time.

### Example

If we add a new column to a user table, Alembic can be used to create and apply a migration for that schema change.

### Memory Trick

**SQLAlchemy = Database interaction**

**Alembic = Database schema migrations**

### One-Line Answer

SQLAlchemy helps Python applications interact with relational databases, while Alembic manages database schema migrations.

## Q24. How would you troubleshoot a slow database query?

### Interview Answer

I would follow a systematic approach:

1. Reproduce the slow query.
2. Check the query execution plan.
3. Identify expensive operations.
4. Check whether appropriate indexes exist.
5. Review joins and filtering conditions.
6. Check whether too much data is being retrieved.
7. Review database statistics and configuration where relevant.
8. Optimize the query or schema if required.
9. Test the query again.
10. Compare performance before and after the change.

### Memory Trick

**Measure → Execution Plan → Bottleneck → Optimize → Test**

### One-Line Answer

I would analyze the execution plan, identify the bottleneck, optimize the query or database design, and verify the performance improvement.

---

## Q25. How would you handle a database connection failure in an application?

### Interview Answer

I would first reproduce the issue and check the application logs.

Then I would verify:

1. Database server availability
2. Host and port
3. Database name
4. Username and credentials
5. Environment variables
6. Network connectivity
7. Connection pool configuration
8. Database limits
9. Recent configuration or deployment changes

After identifying the root cause, I would fix it and test the application again.

I would also ensure that the application handles the failure gracefully rather than exposing sensitive database details to users.

### Memory Trick

**Connection → Config → Network → Credentials → DB → Root Cause → Test**

### One-Line Answer

I would check logs, database availability, configuration, credentials, network connectivity, and connection limits to identify and fix the root cause.
