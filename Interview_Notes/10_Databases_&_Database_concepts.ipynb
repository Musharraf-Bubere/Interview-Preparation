# 🟢 BASIC LEVEL

## Q1. What is a database?

### Interview Answer

A database is an organized collection of data that can be stored, managed, retrieved, and updated efficiently.

For example, an e-commerce application may store:

```text
Customers
Products
Orders
Payments
Reviews
Inventory
```

### One-Line Answer

> A database is an organized collection of data designed for efficient storage, retrieval, modification, and management.

---

## Q2. What is a DBMS?

### Interview Answer

DBMS stands for Database Management System. It is software that allows users and applications to create, store, retrieve, update, secure, and manage data.

Examples:

```text
PostgreSQL
MySQL
Oracle Database
SQL Server
SQLite
MongoDB
```

A DBMS commonly provides:

```text
Data storage
Query processing
Transaction management
Concurrency control
Security
Backup and recovery
```

### One-Line Answer

> A DBMS is software that provides mechanisms for storing, managing, querying, securing, and recovering database data.

---

## Q3. What is an RDBMS?

### Interview Answer

RDBMS stands for Relational Database Management System. It organizes data into tables and represents relationships using keys and constraints.

Examples:

```text
PostgreSQL
MySQL
Oracle
SQL Server
SQLite
```

### One-Line Answer

> An RDBMS is a DBMS based on the relational model, where data is organized into related tables.

---

## Q4. What is the difference between DBMS and RDBMS?

### Interview Answer

DBMS is the broader category of database management systems, while RDBMS is a relational type of DBMS.

RDBMS emphasizes:

```text
Tables
Rows and columns
Relationships
Keys
Constraints
SQL
Transactions
```

### One-Line Answer

> RDBMS is a relational type of DBMS that organizes data into related tables and uses keys and constraints.

---

## Q5. What is a database schema?

### Interview Answer

A schema describes the logical organization of database objects such as:

```text
Tables
Columns
Data types
Constraints
Indexes
Views
Relationships
```

The exact meaning of schema can vary between database systems.

### One-Line Answer

> A schema describes the logical structure and organization of database objects.

## Q6. What is a table, row, and column?

### Interview Answer

A table stores related data. A row represents one record, and a column represents an attribute.

```text
employees

id | name | salary
---|------|-------
1  | Ali  | 70000
2  | Sara | 80000
```

### One-Line Answer

> A table stores records, rows represent individual records, and columns represent attributes.

---

## Q7. What is a data type?

### Interview Answer

A data type defines what kind of value a column can store.

Common examples:

```text
INTEGER
DECIMAL
VARCHAR
DATE
TIMESTAMP
BOOLEAN
```

Choosing appropriate data types improves correctness and storage efficiency.

### One-Line Answer

> A data type defines the kind of values a database column can store.

---

## Q8. What are database constraints?

### Interview Answer

Constraints are rules enforced by the database to maintain data integrity.

Common constraints:

```text
PRIMARY KEY
FOREIGN KEY
UNIQUE
NOT NULL
CHECK
DEFAULT
```

Example:

```sql
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    email VARCHAR(255) UNIQUE,
    salary DECIMAL(10,2) CHECK (salary >= 0)
);
```

### One-Line Answer

> Constraints enforce rules that keep database data valid and consistent.

---

## Q9. What is data integrity?

### Interview Answer

Data integrity means maintaining the accuracy, consistency, validity, and reliability of data.

Examples:

```text
Primary key → unique records
Foreign key → valid relationships
NOT NULL → required values
CHECK → valid values
Transactions → consistent updates
```

### One-Line Answer

> Data integrity ensures that stored data remains accurate, valid, consistent, and reliable.

---

## Q10. What is referential integrity?

### Interview Answer

Referential integrity ensures that relationships between related tables remain valid.

For example, an order should not reference a customer that does not exist when a foreign-key constraint is enforced.

```text
Customers
customer_id = 1
       ↑
       |
Orders.customer_id
```

### One-Line Answer

> Referential integrity ensures that foreign-key values reference valid related records.

# 🟡 INTERMEDIATE LEVEL

## Q11. What is a primary key?

### Interview Answer

A primary key uniquely identifies every row in a table.

Properties:

```text
Unique
Cannot be NULL
One primary-key constraint per table
```

Example:

```sql
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    name VARCHAR(100)
);
```

### One-Line Answer

> A primary key uniquely identifies each row in a table.

---

## Q12. What is a foreign key?

### Interview Answer

A foreign key references a key in another table and establishes a relationship.

```sql
CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id)
);
```

### One-Line Answer

> A foreign key references another table's key and maintains a valid relationship between tables.

---

## Q13. What is a candidate key?

### Interview Answer

A candidate key is a minimal set of columns capable of uniquely identifying every row.

If both `employee_id` and `email` are unique, both can be candidate keys. One is selected as the primary key.

### One-Line Answer

> A candidate key is a minimal attribute set that can uniquely identify a row.

---

## Q14. What is a composite key?

### Interview Answer

A composite key consists of multiple columns whose combination uniquely identifies a record.

Example:

```text
student_id
course_id
```

can form:

```sql
PRIMARY KEY (student_id, course_id)
```

This is common in many-to-many relationships.

### One-Line Answer

> A composite key uses multiple columns together to uniquely identify a record.

---

## Q15. What is a surrogate key?

### Interview Answer

A surrogate key is an artificial identifier created to identify a record.

Example:

```text
customer_id = 1001
```

It usually has no business meaning.

Benefits:

```text
Simple
Stable
Easy to reference
Often compact
```

### One-Line Answer

> A surrogate key is an artificial identifier used primarily to uniquely identify a record.

---

## Q16. What is a natural key?

### Interview Answer

A natural key is an existing business attribute that can uniquely identify an entity.

Examples:

```text
Email address
Product code
Government identifier
```

Natural keys may change or have business-specific constraints, so surrogate keys are often used as internal identifiers.

### One-Line Answer

> A natural key is a real-world business attribute that naturally identifies an entity.

---

Relationships describe how entities are associated.

Common types:

```text
One-to-One
One-to-Many
Many-to-Many
```

Example:

```text
Customer
   |
   | 1
   ↓
Orders
   | many
```

> Database relationships describe how records in different entities or tables are connected.

## Q18. Explain one-to-one, one-to-many, and many-to-many.

### Interview Answer

### One-to-One

```text
Person → Passport
```

### One-to-Many

### Many-to-Many

## Q19. What is normalization?

### Interview Answer

Normalization organizes relational data to reduce redundancy and avoid modification anomalies.

Common normal forms:

```text
1NF
2NF
3NF
BCNF
```

For example, customer information can be stored once in `customers` instead of being repeatedly copied into every order.

### One-Line Answer

> Normalization structures data to reduce redundancy and prevent insert, update, and delete anomalies.

---

## Q20. What are insertion, update, and deletion anomalies?

### Interview Answer

Poor database design can create:

**Insertion anomaly:** You cannot insert one fact without unrelated information.

**Update anomaly:** The same fact exists in multiple places and must be updated everywhere.

**Deletion anomaly:** Removing one record accidentally removes another important fact.

Normalization helps reduce these problems.

### One-Line Answer

> Modification anomalies are data-management problems caused by poorly structured and redundant database designs.

# 🔴 ADVANCED LEVEL

## Q21. Explain 1NF, 2NF, and 3NF.

### Interview Answer

### 1NF

Values should be atomic and repeating groups should be removed.

### 2NF

The table must be in 1NF, and non-key attributes must depend on the whole candidate key. This is especially important with composite keys.

### 3NF

The table must be in 2NF, and non-key attributes should not depend transitively on another non-key attribute.

### One-Line Answer

> 1NF addresses atomicity, 2NF addresses partial dependency, and 3NF addresses transitive dependency.

---

## Q22. What is BCNF?

### Interview Answer

BCNF stands for Boyce-Codd Normal Form.

It is stricter than 3NF. A relation is in BCNF when every determinant is a candidate key.

```text
Every determinant
       ↓
Candidate key
```

### One-Line Answer

> BCNF is a stronger normalization form where every determinant must be a candidate key.

---

## Q23. What is denormalization?

### Interview Answer

Denormalization intentionally introduces controlled redundancy to improve read performance or simplify queries.

Benefits:

```text
Fewer joins
Potentially faster reads
Simpler reporting
```

Costs:

```text
More storage
More complex writes
Possible consistency challenges
```

### One-Line Answer

> Denormalization deliberately introduces controlled redundancy when read performance or simplicity justifies the trade-off.

---

## Q24. What is an index?

### Interview Answer

An index is a data structure that can help a database locate rows more efficiently.

Example:

```sql
CREATE INDEX idx_customer_email
ON customers(email);
```

Indexes can improve reads but add:

```text
Storage cost
Write overhead
Maintenance cost
```

### One-Line Answer

> An index is a data structure that can speed up data retrieval while adding storage and write-maintenance overhead.

---

## Q25. What is a composite index?

### Interview Answer

A composite index contains multiple columns.

```sql
CREATE INDEX idx_orders_customer_date
ON orders(customer_id, order_date);
```

Column order matters because query predicates determine how effectively the index can be used.

### One-Line Answer

> A composite index contains multiple columns and should be designed around important query access patterns.

---

## Q26. What is a covering index?

### Interview Answer

A covering index contains all columns required by a query.

For:

```sql
SELECT order_date, amount
FROM orders
WHERE customer_id = 100;
```

an index containing:

```text
customer_id
order_date
amount
```

may allow the database to answer the query directly from the index.

### One-Line Answer

> A covering index contains the columns needed by a query and can potentially eliminate additional table lookups.

A database generally processes a query through stages such as:

The exact architecture differs between database systems.

### One-Line Answer

> Query processing transforms a database request into an execution strategy and executes it to produce results.

It may contain:

### Interview Answer

Query optimization attempts to reduce execution time and resource consumption while preserving the required result.

## Q30. What is a transaction?

### Interview Answer

A transaction is a logical unit of database work.

Example:

```text
Bank Transfer

Debit Account A
       ↓
Credit Account B
```

Both operations should behave as one logical unit.

```sql
BEGIN;

UPDATE accounts
SET balance = balance - 1000
WHERE account_id = 1;

UPDATE accounts
SET balance = balance + 1000
WHERE account_id = 2;

COMMIT;
```

If something fails:

```sql
ROLLBACK;
```

### One-Line Answer

> A transaction groups related database operations into a unit that can be committed or rolled back.

# 🟣 EXPERT LEVEL

## Q31. Explain ACID properties.

### Interview Answer

ACID means:

**Atomicity:** All operations succeed or the transaction is rolled back.

**Consistency:** The database moves between valid states according to its integrity rules.

**Isolation:** Concurrent transactions are controlled according to the selected isolation guarantees.

**Durability:** Committed changes are designed to survive failures according to the database's durability mechanisms.

### Memory Trick

```text
A → All or Nothing
C → Consistent State
I → Isolated Transactions
D → Durable Changes
```

### One-Line Answer

> ACID provides reliable transactional behavior through Atomicity, Consistency, Isolation, and Durability.

---

## Q32. What are isolation levels?

### Interview Answer

Isolation levels define how transactions interact with concurrent transactions.

Common levels:

```text
Read Uncommitted
Read Committed
Repeatable Read
Serializable
```

Some systems also support snapshot or MVCC-based approaches.

Possible anomalies include:

```text
Dirty Read
Non-repeatable Read
Phantom Read
```

### One-Line Answer

> Isolation levels define the visibility and concurrency guarantees between simultaneous transactions.

---

## Q33. What is a dirty read?

### Interview Answer

A dirty read occurs when one transaction reads uncommitted data written by another transaction.

```text
Transaction A
UPDATE balance = 500
Not committed

Transaction B
reads 500
```

If A rolls back, B read a value that was never committed.

### One-Line Answer

> A dirty read occurs when a transaction reads another transaction's uncommitted changes.

---

## Q34. What is a deadlock?

### Interview Answer

A deadlock occurs when transactions wait for resources held by each other.

```text
Transaction A
locks Row 1
   ↓
waits for Row 2

Transaction B
locks Row 2
   ↓
waits for Row 1
```

The database may detect the deadlock and abort one transaction.

Prevention strategies:

```text
Consistent lock ordering
Short transactions
Appropriate indexes
Avoid unnecessary locks
Retry logic
```

### One-Line Answer

> A deadlock is a circular wait between transactions that prevents them from progressing.

---

Concurrency control manages simultaneous database operations while maintaining correctness.

Common techniques:

### One-Line Answer

> Concurrency control coordinates simultaneous operations while maintaining database consistency and isolation.

## Q36. What is locking?

### Interview Answer

Locking controls concurrent access to database resources.

Conceptually:

```text
Shared / Read Lock
Exclusive / Write Lock
```

Exact lock behavior is DBMS-specific.

### One-Line Answer

> Locking prevents conflicting concurrent operations from accessing shared database resources unsafely.

---

## Q37. What is MVCC?

### Interview Answer

MVCC stands for Multi-Version Concurrency Control.

An MVCC system can maintain multiple versions or visibility information so transactions can read an appropriate consistent version without requiring every reader to wait for every writer.

```text
Version 1
Version 2
Version 3
    ↓
Visibility rules
```

Implementation differs between database systems.

### One-Line Answer

> MVCC manages concurrency using multiple data versions or visibility information to provide consistent reads.

---

## Q38. What is replication?

### Interview Answer

Replication maintains copies of database data on multiple database instances.

Example:

```text
Primary
   ↓
Replica
```

Benefits:

```text
High availability
Read scaling
Disaster recovery
Geographic distribution
```

Challenges:

```text
Replication lag
Failover complexity
Consistency considerations
Operational overhead
```

### One-Line Answer

> Replication maintains copies of data across multiple database nodes for availability, scaling, or recovery.

---

## Q39. What is sharding?

### Interview Answer

Sharding horizontally distributes data across multiple database nodes using a shard key.

```text
Shard 1 → Users A–M
Shard 2 → Users N–Z
```

Possible strategies:

```text
Hash
Range
Directory-based
Geographic
```

Challenges:

```text
Cross-shard queries
Rebalancing
Hotspots
Distributed transactions
Operational complexity
```

### One-Line Answer

> Sharding distributes different portions of data across multiple database nodes to enable horizontal scaling.

---

## Q40. What is partitioning?

### Interview Answer

Partitioning divides a logical table into smaller physical partitions.

Common strategies:

```text
Range
List
Hash
```

Example:

```text
orders
 ├── 2024
 ├── 2025
 └── 2026
```

Queries aligned with the partition key may benefit from partition pruning.

### One-Line Answer

> Partitioning divides a large logical table into smaller physical partitions for management and potentially improved performance.

---

## Q41. What is the CAP theorem?

### Interview Answer

CAP concerns distributed systems during network partitions.

The three properties are:

```text
C → Consistency
A → Availability
P → Partition Tolerance
```

When a network partition occurs, a distributed system cannot simultaneously guarantee both strong consistency and availability.

### One-Line Answer

> CAP theorem describes the consistency-versus-availability trade-off that arises when a distributed system experiences a network partition.

## Q47. What is connection pooling?

### Interview Answer

Creating a new database connection for every application request can be expensive.

Connection pooling maintains reusable database connections.

```text
Application
    ↓
Connection Pool
 ┌──┼──┐
 ↓  ↓  ↓
C1 C2 C3
```

A request borrows a connection, executes its work, and returns the connection to the pool.

### One-Line Answer

> Connection pooling reuses database connections to reduce connection overhead and improve resource utilization.

## Q48. What is a view and materialized view?

### Interview Answer

A **view** is a named query that can be queried like a virtual table.

```sql
CREATE VIEW active_users AS
SELECT *
FROM users
WHERE status = 'Active';
```

A **materialized view** stores the query result physically and must be refreshed.

```text
View
→ Query definition
→ Usually computes when queried

Materialized View
→ Stored result
→ Faster reads in suitable workloads
→ Requires refresh
```

Exact behavior depends on the DBMS.

### One-Line Answer

> A view stores a query definition, while a materialized view stores the query result for potentially faster reads.

## Q50. How would you troubleshoot a production database performance problem?

### Interview Answer

I would use an evidence-based process:

```text
Production Issue
      ↓
Identify affected queries
      ↓
Measure latency / throughput
      ↓
Check database health
      ↓
Inspect execution plans
      ↓
Check locks / blocking
      ↓
Check CPU / memory / I/O
      ↓
Check indexes
      ↓
Check connection usage
      ↓
Check recent changes
      ↓
Apply targeted optimization
      ↓
Benchmark
      ↓
Monitor
```

I would investigate:

```text
Slow queries
High CPU
High I/O
Lock contention
Deadlocks
Connection exhaustion
Replication lag
Missing indexes
Poor cardinality estimates
Large scans
Expensive joins
```

### Interview Example

> "If a production database became slow, I would first quantify the issue rather than immediately changing the schema. I would identify affected queries and inspect execution plans, database metrics, locks, connection usage, CPU, memory, and I/O. Then I would determine whether the bottleneck is query-related, indexing-related, concurrency-related, or infrastructure-related. I would make a targeted change, benchmark before and after, and monitor the production impact."

### One-Line Answer

> I troubleshoot database performance by measuring the bottleneck first, analyzing query plans and system behavior, applying targeted changes, and validating the improvement.

# ⭐ IMPORTANT DATABASE COMPARISONS

## Normalization vs Denormalization

```text
Normalization
→ Less redundancy
→ Better consistency
→ Potentially more joins

Denormalization
→ Controlled redundancy
→ Potentially faster reads
→ More write complexity
```

## Replication vs Sharding

```text
Replication
→ Copies data
→ Multiple nodes contain copies

Sharding
→ Splits data
→ Nodes contain different portions
```

## Partitioning vs Sharding

```text
Partitioning
→ Divides a logical table/database structure into partitions

Sharding
→ Distributes data across separate database nodes
```

## Backup vs Replication

```text
Backup
→ Recovery from data loss/corruption

Replication
→ Availability/scaling/redundancy
```

## SQL vs NoSQL

```text
SQL
→ Relational
→ Structured schema
→ Relationships
→ Complex joins
→ Strong transactional use cases

NoSQL
→ Non-relational models
→ Flexible structures in many systems
→ Distributed/specialized workloads
```

> **A strong database candidate does not only know SQL syntax. They understand how data should be modeled, how relationships and constraints maintain integrity, how transactions and concurrency work, how indexes affect performance, and how databases are secured, backed up, replicated, partitioned, and scaled in production.**

```text
DATABASE KNOWLEDGE
        ↓
DATA MODELING
        ↓
SQL
        ↓
INTEGRITY
        ↓
TRANSACTIONS
        ↓
CONCURRENCY
        ↓
PERFORMANCE
        ↓
SECURITY
        ↓
BACKUP & RECOVERY
        ↓
DISTRIBUTED SYSTEMS
        ↓
PRODUCTION DATABASE ENGINEERING
```
