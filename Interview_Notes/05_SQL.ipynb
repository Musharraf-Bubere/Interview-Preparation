## Q1. What is SQL?

### Interview Answer

SQL stands for Structured Query Language. It is used to interact with relational databases.

With SQL, we can:

```text
Create data structures
Insert data
Read data
Update data
Delete data
Control access
Manage transactions
```

Common relational database systems include:

```text
PostgreSQL
MySQL
SQL Server
Oracle
SQLite
```

### Example

```sql
SELECT *
FROM employees;
```

This retrieves rows from the `employees` table.

### One-Line Answer

> SQL is a language used to define, manipulate, query, and control data in relational database systems.

## Q2. What is a relational database?

### Interview Answer

A relational database stores data in structured tables consisting of rows and columns.

For example:

```text
employees

id | name    | department | salary
---|---------|------------|-------
1  | Ali     | IT         | 70000
2  | Sara    | HR         | 60000
3  | John    | IT         | 80000
```

Tables can be related using keys.

```text
employees
   │
   │ department_id
   ↓
departments
```

### One-Line Answer

> A relational database stores structured data in related tables using rows, columns, and keys.

## Q3. What is a table, row, and column?

### Interview Answer

A **table** stores related records.

A **row** represents one record.

A **column** represents an attribute of the records.

Example:

```text
students

id | name | age
---|------|----
1  | A    | 21
2  | B    | 22
```

Here:

```text
Table  → students
Row    → 1, A, 21
Column → age
```

### One-Line Answer

> A table stores records, rows represent individual records, and columns represent attributes.

## Q4. What is a primary key?

### Interview Answer

A primary key uniquely identifies each row in a table.

Properties:

```text
Unique
Not NULL
One primary-key constraint per table
```

Example:

```sql
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    name VARCHAR(100)
);
```

Here:

```text
employee_id
```

uniquely identifies each employee.

### One-Line Answer

> A primary key is a column or set of columns that uniquely identifies each row in a table.

## Q5. What is a foreign key?

### Interview Answer

A foreign key creates a relationship between tables by referencing a key in another table, usually a primary key.

Example:

```sql
CREATE TABLE departments (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(100)
);

CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    name VARCHAR(100),
    department_id INT,
    FOREIGN KEY (department_id)
        REFERENCES departments(department_id)
);
```

This helps enforce referential integrity.

### One-Line Answer

> A foreign key references a key in another table and helps maintain valid relationships between tables.

## Q7. What is NULL in SQL?

### Interview Answer

NULL represents missing, unknown, or unavailable information.

It is not the same as:

```text
0
''
'NULL'
FALSE
```

To check NULL:

```sql
SELECT *
FROM employees
WHERE manager_id IS NULL;
```

Not:

```sql
WHERE manager_id = NULL;
```

### Why?

NULL represents an unknown value, so normal equality comparison does not evaluate to TRUE.

### One-Line Answer

> NULL represents missing or unknown information and must be tested using IS NULL or IS NOT NULL.

## Q11. What is GROUP BY?

### Interview Answer

GROUP BY combines rows with the same grouping values so aggregate functions can be calculated per group.

Example:

```sql
SELECT department, AVG(salary)
FROM employees
GROUP BY department;
```

Result conceptually:

```text
department | avg_salary
-----------|-----------
HR         | 62000
IT         | 78000
Sales      | 70000
```

### One-Line Answer

> GROUP BY creates groups of rows based on one or more columns for aggregation.

## Q12. What are aggregate functions?

### Interview Answer

Aggregate functions calculate a result from multiple rows.

Common functions:

```text
COUNT()
SUM()
AVG()
MIN()
MAX()
```

Example:

```sql
SELECT
    COUNT(*) AS employee_count,
    AVG(salary) AS avg_salary,
    MAX(salary) AS max_salary
FROM employees;
```

### One-Line Answer

> Aggregate functions summarize multiple rows into a calculated result.

## Q13. What is HAVING and how is it different from WHERE?

### Interview Answer

WHERE filters rows before grouping.

HAVING filters groups after aggregation.

Example:

```sql
SELECT department, AVG(salary) AS avg_salary
FROM employees
WHERE status = 'Active'
GROUP BY department
HAVING AVG(salary) > 70000;
```

Conceptually:

```text
FROM
 ↓
WHERE
 ↓
GROUP BY
 ↓
HAVING
 ↓
SELECT
 ↓
ORDER BY
```

### One-Line Answer

> WHERE filters rows before grouping, while HAVING filters grouped results after aggregation.

For multiple columns:

## Q15. What is a JOIN?

### Interview Answer

A JOIN combines rows from multiple tables based on a related condition.

Example:

```sql
SELECT
    e.name,
    d.department_name
FROM employees e
JOIN departments d
    ON e.department_id = d.department_id;
```

Common joins:

```text
INNER JOIN
LEFT JOIN
RIGHT JOIN
FULL OUTER JOIN
CROSS JOIN
SELF JOIN
```

### One-Line Answer

> A JOIN combines related data from multiple tables using a join condition.

## Q16. Explain INNER JOIN vs LEFT JOIN.

### Interview Answer

### INNER JOIN

Returns only matching rows from both tables.

```text
A ∩ B
```

### LEFT JOIN

Returns every row from the left table plus matching rows from the right table.

If there is no match:

```text
Right-side columns → NULL
```

Example:

```sql
SELECT e.name, d.department_name
FROM employees e
LEFT JOIN departments d
    ON e.department_id = d.department_id;
```

### One-Line Answer

> INNER JOIN returns only matches, while LEFT JOIN preserves all rows from the left table and adds matching right-side data.

## Q17. What is a self join?

### Interview Answer

A self join joins a table to itself.

It is useful for hierarchical relationships such as employees and managers.

Example:

```sql
SELECT
    e.name AS employee,
    m.name AS manager
FROM employees e
LEFT JOIN employees m
    ON e.manager_id = m.employee_id;
```

The same table is referenced twice using different aliases.

### One-Line Answer

> A self join joins a table with itself to represent relationships within the same table.

## Q18. What is a subquery?

### Interview Answer

A subquery is a query nested inside another SQL statement.

Example:

```sql
SELECT name, salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
```

The inner query calculates average salary, and the outer query finds employees earning more than that average.

### One-Line Answer

> A subquery is a query nested inside another query and can provide values or sets used by the outer query.

---

## Q19. What is a correlated subquery?

### Interview Answer

A correlated subquery references a column from the outer query.

Example:

```sql
SELECT e.name, e.salary
FROM employees e
WHERE e.salary > (
    SELECT AVG(e2.salary)
    FROM employees e2
    WHERE e2.department_id = e.department_id
);
```

The inner query depends on the current outer employee's department.

### Performance Point

Correlated subqueries can sometimes be rewritten using joins or window functions for clearer or more efficient execution, depending on the database optimizer.

### One-Line Answer

> A correlated subquery depends on values from the outer query and is logically evaluated in relation to each outer row.

# 🔴 ADVANCED LEVEL

## Q21. What is a CTE?

### Interview Answer

CTE stands for Common Table Expression.

It creates a named temporary result set that can be referenced by the following SQL statement.

Example:

```sql
WITH avg_salary AS (
    SELECT AVG(salary) AS value
    FROM employees
)
SELECT name, salary
FROM employees
CROSS JOIN avg_salary
WHERE salary > avg_salary.value;
```

CTEs improve readability and can make complex queries easier to organize.

They can also be recursive.

### One-Line Answer

> A CTE is a named query result defined with WITH that makes complex SQL easier to structure and reuse within a statement.

---

## Q22. What is a recursive CTE?

### Interview Answer

A recursive CTE references itself and is useful for hierarchical or graph-like data.

Example use cases:

```text
Employee → Manager hierarchy
Category → Subcategory
Organization tree
Graph traversal
```

Conceptually:

```text
Anchor Query
     ↓
Recursive Query
     ↓
New Rows
     ↓
Recursive Query
     ↓
Stop Condition
```

Example structure:

```sql
WITH RECURSIVE employee_tree AS (
    SELECT employee_id, name, manager_id
    FROM employees
    WHERE manager_id IS NULL

    UNION ALL

    SELECT e.employee_id, e.name, e.manager_id
    FROM employees e
    JOIN employee_tree t
        ON e.manager_id = t.employee_id
)
SELECT *
FROM employee_tree;
```

Syntax varies slightly between database systems.

### One-Line Answer

> A recursive CTE repeatedly references its previous result and is useful for hierarchical data.

## Q23. What are window functions?

### Interview Answer

Window functions calculate values across related rows without collapsing them into one row per group.

Example:

```sql
SELECT
    name,
    department,
    salary,
    AVG(salary) OVER (
        PARTITION BY department
    ) AS department_avg
FROM employees;
```

Unlike GROUP BY:

```text
GROUP BY
→ Reduces rows

Window Function
→ Keeps individual rows
```

### One-Line Answer

> Window functions perform calculations across related rows while preserving the individual rows in the result.

## Q24. Explain ROW_NUMBER, RANK, and DENSE_RANK.

### Interview Answer

These functions assign rankings within a window.

```sql
SELECT
    name,
    salary,
    ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num,
    RANK() OVER (ORDER BY salary DESC) AS rank_num,
    DENSE_RANK() OVER (ORDER BY salary DESC) AS dense_rank_num
FROM employees;
```

Suppose salaries are:

```text
100
90
90
80
```

Then:

```text
ROW_NUMBER → 1, 2, 3, 4
RANK       → 1, 2, 2, 4
DENSE_RANK → 1, 2, 2, 3
```

### One-Line Answer

> ROW_NUMBER gives unique sequential numbers, RANK leaves gaps after ties, and DENSE_RANK does not leave gaps.

## Q25. How do you find the second-highest salary?

### Interview Answer

A robust approach is to use DENSE_RANK when duplicate salaries should share the same rank.

```sql
WITH ranked AS (
    SELECT
        salary,
        DENSE_RANK() OVER (
            ORDER BY salary DESC
        ) AS rnk
    FROM employees
)
SELECT DISTINCT salary
FROM ranked
WHERE rnk = 2;
```

This returns the second-highest distinct salary.

Another approach is:

```sql
SELECT MAX(salary)
FROM employees
WHERE salary < (
    SELECT MAX(salary)
    FROM employees
);
```

### Important Interview Point

Always clarify whether the interviewer wants:

```text
Second row
or
Second distinct salary
```

### One-Line Answer

> I prefer DENSE_RANK when the requirement is the second-highest distinct salary because it handles duplicate salary values correctly.

## Q26. How do you find duplicate records?

### Interview Answer

Use GROUP BY and HAVING.

Example:

```sql
SELECT email, COUNT(*) AS cnt
FROM customers
GROUP BY email
HAVING COUNT(*) > 1;
```

This identifies duplicate email values.

If you need complete duplicate rows, you can join this result back to the original table or use a window function.

### One-Line Answer

> I identify duplicates by grouping on the business key and filtering groups with COUNT(*) greater than one.

## Q27. How do you remove duplicate records while keeping one row?

### Interview Answer

A common approach is ROW_NUMBER.

Example:

```sql
WITH duplicates AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY email
            ORDER BY customer_id
        ) AS rn
    FROM customers
)
DELETE FROM customers
WHERE customer_id IN (
    SELECT customer_id
    FROM duplicates
    WHERE rn > 1
);
```

The exact DELETE syntax can vary by database system.

### Important Point

Before deleting, first run the ranking query as SELECT and verify which rows will be removed.

### One-Line Answer

> I use ROW_NUMBER partitioned by the duplicate key, keep rn = 1, and remove rows where rn > 1 after validating the result.

## Q33. Explain 1NF, 2NF, and 3NF.

### Interview Answer

### First Normal Form — 1NF

A table should have atomic values and no repeating groups.

Bad:

```text
customer | phones
---------|----------------
A        | 111, 222
```

Better:

```text
customer | phone
---------|------
A        | 111
A        | 222
```

### Second Normal Form — 2NF

The table should be in 1NF and non-key attributes should depend on the whole candidate key, especially relevant when the key is composite.

### Third Normal Form — 3NF

The table should be in 2NF and non-key attributes should not depend transitively on another non-key attribute.

### One-Line Answer

> 1NF addresses atomicity, 2NF removes partial dependency on composite keys, and 3NF removes inappropriate transitive dependencies.

## Q35. What is a transaction?

### Interview Answer

A transaction is a logical unit of work consisting of one or more database operations.

Example:

```text
Transfer ₹10,000
        ↓
Debit Account A
        ↓
Credit Account B
```

Both operations should succeed together.

```sql
BEGIN;

UPDATE accounts
SET balance = balance - 10000
WHERE account_id = 1;

UPDATE accounts
SET balance = balance + 10000
WHERE account_id = 2;

COMMIT;
```

If something fails:

```sql
ROLLBACK;
```

### One-Line Answer

> A transaction groups related database operations so they can be committed or rolled back as a unit.

## Q36. Explain ACID properties.

### Interview Answer

ACID describes important transaction properties.

### Atomicity

All operations succeed or none are applied.

### Consistency

A committed transaction preserves defined database constraints and integrity rules.

### Isolation

Concurrent transactions are controlled so their interactions follow the database's isolation guarantees.

### Durability

Once committed, changes are expected to persist even after failures, subject to the database's durability guarantees.

### Memory Trick

```text
A → All or Nothing
C → Consistent State
I → Isolated Transactions
D → Durable Changes
```

### One-Line Answer

> ACID means Atomicity, Consistency, Isolation, and Durability, which together provide reliable transactional behavior.

## Q37. What are transaction isolation levels?

### Interview Answer

Isolation levels control how transactions interact with concurrent transactions.

Common levels:

```text
Read Uncommitted
Read Committed
Repeatable Read
Serializable
```

Some databases also support:

```text
Snapshot / MVCC-based isolation
```

### Common anomalies

```text
Dirty Read
Non-repeatable Read
Phantom Read
```

### One-Line Answer

## Q38. What is a deadlock?

### Interview Answer

A deadlock occurs when transactions wait for resources held by each other.

Example:

```text
Transaction A
locks Row 1
 ↓
waits for Row 2

Transaction B
locks Row 2
 ↓
waits for Row 1
```

Neither can proceed.

The database may detect the deadlock and abort one transaction.

### Prevention Strategies

```text
Consistent lock ordering
Short transactions
Appropriate indexes
Avoid unnecessary locks
Retry failed transactions
```

### One-Line Answer

> A deadlock occurs when transactions form a circular wait for resources, preventing progress until the database resolves the conflict.

## Q43. What is SQL injection?

### Interview Answer

SQL injection occurs when untrusted user input is incorrectly incorporated into SQL statements, allowing an attacker to alter the intended query.

Unsafe conceptual pattern:

```text
"SELECT * FROM users WHERE name = '" + user_input + "'"
```

If input is not safely handled, the query structure may be manipulated.

### Prevention

Use:

```text
Parameterized Queries
Prepared Statements
ORM Parameter Binding
Input Validation
Least Privilege
```

Example concept:

```python
cursor.execute(
    "SELECT * FROM users WHERE name = %s",
    (username,)
)
```

The exact placeholder syntax depends on the database driver.

### One-Line Answer

> SQL injection is an attack caused by unsafe query construction and is primarily prevented using parameterized queries or prepared statements.

## Q45. What is a covering index?

### Interview Answer

A covering index contains all columns required by a query, allowing the database to answer the query directly from the index without fetching additional table data, when the optimizer chooses that plan.

Example query:

```sql
SELECT name, salary
FROM employees
WHERE department_id = 10;
```

A possible index:

```sql
CREATE INDEX idx_dept_name_salary
ON employees(department_id, name, salary);
```

The index contains:

```text
Filter column
+
Selected columns
```

### Benefit

Potentially fewer table lookups.

### Tradeoff

Larger index and additional write/storage cost.

### One-Line Answer

> A covering index contains all columns needed by a query, potentially allowing the database to satisfy the query directly from the index.

## Q46. How would you optimize a slow SQL query?

### Interview Answer

I would not blindly add indexes. I would first identify where the time is being spent.

My process:

```text
1. Reproduce the slow query
        ↓
2. Check execution plan
        ↓
3. Compare estimated vs actual rows
        ↓
4. Identify expensive operations
        ↓
5. Check indexes and selectivity
        ↓
6. Reduce unnecessary columns/rows
        ↓
7. Review JOIN conditions
        ↓
8. Review filtering and aggregation
        ↓
9. Check statistics
        ↓
10. Test an improved query
        ↓
11. Benchmark before vs after
```

Potential improvements:

```text
Better indexes
Sargable predicates
Early filtering where appropriate
Avoid unnecessary SELECT *
Better joins
Query rewriting
Partitioning for suitable large tables
Materialized views for repeated expensive analytics
```

### One-Line Answer

> I optimize SQL using execution plans and evidence, then improve indexing, query structure, filtering, joins, statistics, and data access based on the actual bottleneck.

## Q47. What is a SARGable query?

### Interview Answer

SARGable means a predicate is structured so the database can efficiently use an appropriate index where possible.

Example of potentially less index-friendly filtering:

```sql
WHERE YEAR(order_date) = 2026
```

A range predicate may be more index-friendly:

```sql
WHERE order_date >= '2026-01-01'
  AND order_date < '2027-01-01'
```

The exact optimizer behavior depends on the database, index, data distribution, and query plan.

### One-Line Answer

> A SARGable predicate is written in a way that allows the optimizer to efficiently use indexes when appropriate.

## Q48. What is partitioning?

### Interview Answer

Partitioning divides a large logical table into smaller physical partitions based on a partition key.

Common strategies:

```text
Range Partitioning
List Partitioning
Hash Partitioning
```

Example:

```text
orders
 ├── 2024
 ├── 2025
 └── 2026
```

A query for 2026 data may only need to access the relevant partition if partition pruning is possible.

### Benefits

- Manageability
- Potentially improved query performance
- Easier maintenance of large datasets

### Important Point

Partitioning is not a replacement for appropriate indexing.

### One-Line Answer

> Partitioning divides large tables into manageable pieces and can improve performance when queries align with the partition strategy.

## Q49. How do you handle NULL values in SQL?

### Interview Answer

I first determine what NULL means for the business requirement.

To find NULLs:

```sql
SELECT *
FROM employees
WHERE manager_id IS NULL;
```

To replace NULL:

```sql
SELECT COALESCE(phone, 'Not Available')
FROM employees;
```

To count non-NULL values:

```sql
SELECT COUNT(phone)
FROM employees;
```

To count all rows:

```sql
SELECT COUNT(*)
FROM employees;
```

### Important Interview Point

Many aggregate functions ignore NULL values, so the exact behavior should be understood rather than assumed.

### One-Line Answer

> I handle NULL explicitly using IS NULL, IS NOT NULL, COALESCE, and appropriate aggregate semantics depending on the business requirement.

# ⭐ IMPORTANT SQL COMPARISONS

## WHERE vs HAVING

```text
WHERE
→ Filters rows
→ Before GROUP BY aggregation

HAVING
→ Filters groups
→ After GROUP BY aggregation
```

---

## DELETE vs TRUNCATE vs DROP

```text
DELETE
→ Removes rows
→ Can filter with WHERE

TRUNCATE
→ Removes all table rows
→ Usually faster for clearing a table

DROP
→ Removes table definition and data
```

Exact transactional and identity behavior varies by DBMS.

---

## INNER JOIN vs LEFT JOIN

```text
INNER JOIN
→ Matching rows only

LEFT JOIN
→ All left rows
→ Matching right rows
→ NULL when no match
```

---

## GROUP BY vs Window Function

```text
GROUP BY
→ Collapses rows into groups

Window Function
→ Keeps individual rows
→ Calculates across related rows
```

---

## RANK vs DENSE_RANK vs ROW_NUMBER

```text
ROW_NUMBER
→ Unique sequence

RANK
→ Ties share rank
→ Gaps appear

DENSE_RANK
→ Ties share rank
→ No gaps
```

---

## UNION vs UNION ALL

```text
UNION
→ Combines results
→ Removes duplicates

UNION ALL
→ Combines results
→ Keeps duplicates
→ Usually avoids duplicate-elimination work
```

Example:

```sql
SELECT city FROM customers
UNION
SELECT city FROM suppliers;
```

---

## IN vs EXISTS

```text
IN
→ Tests membership in a set

EXISTS
→ Tests whether a matching row exists
```

Example:

```sql
SELECT *
FROM customers c
WHERE EXISTS (
    SELECT 1
    FROM orders o
    WHERE o.customer_id = c.customer_id
);
```

Neither is universally faster. The optimizer, data distribution, indexes, and query shape matter.

---

## PRIMARY KEY vs UNIQUE KEY

```text
PRIMARY KEY
→ Uniquely identifies rows
→ Cannot contain NULL in the key
→ One primary-key constraint per table

UNIQUE
→ Enforces uniqueness
→ NULL behavior can vary by DBMS
→ Multiple unique constraints may be allowed
```

```text
FROM
 ↓
JOIN / ON
 ↓
WHERE
 ↓
GROUP BY
 ↓
HAVING
 ↓
SELECT
 ↓
DISTINCT
 ↓
ORDER BY
 ↓
LIMIT / FETCH
```
