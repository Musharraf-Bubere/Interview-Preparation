# SQL Interview Preparation — 25 Questions

## Q1. What is SQL, and where is it used?

### Answer:

SQL stands for Structured Query Language.

It is a language used to communicate with and manage relational databases.

With SQL, we can:

- Retrieve data
- Insert data
- Update data
- Delete data
- Filter data
- Join tables
- Group and aggregate data
- Create and modify database objects

SQL is commonly used with relational database systems such as MySQL, PostgreSQL, Oracle, and SQL Server.

### Example:

    SELECT name, age
    FROM students
    WHERE age > 22;

This retrieves students whose age is greater than 22.

### Memory Trick:

SQL = **Retrieve → Insert → Update → Delete → Analyze**

### One-Line Answer:

"SQL is a language used to communicate with relational databases to store, retrieve, modify, and analyze structured data."

### DCL:

Examples:

### TCL:

Used to manage transactions.

DDL → **Structure**

### One-Line Answer:

"SQL commands are commonly categorized into DDL, DML, DQL, DCL, and TCL for defining structure, modifying data, querying data, controlling access, and managing transactions."

## Q3. What is the difference between WHERE and HAVING?

### Answer:

Both are used to filter data, but they work at different stages.

`WHERE` filters individual rows before grouping.

`HAVING` filters groups after `GROUP BY` and aggregation.

### Example:

    SELECT department, salary
    FROM employees
    WHERE salary > 50000;

Here, individual employees are filtered.

For grouped data:

    SELECT department, COUNT(*) AS employee_count
    FROM employees
    GROUP BY department
    HAVING COUNT(*) > 5;

Here, departments are grouped first and then filtered.

### Quick Difference:

WHERE → Filters rows

HAVING → Filters groups

### Memory Trick:

**WHERE = Rows**

**HAVING = Groups**

### One-Line Answer:

"`WHERE` filters individual rows before grouping, while `HAVING` filters groups after `GROUP BY`, commonly using aggregate conditions."

## Q4. What is the difference between DELETE, TRUNCATE, and DROP?

### Answer:

`DELETE` removes rows from a table and can use a `WHERE` condition.

Example:

    DELETE FROM employees
    WHERE employee_id = 101;

`TRUNCATE` removes all rows while keeping the table structure.

Example:

    TRUNCATE TABLE employees;

`DROP` removes the table itself, including its structure.

Example:

    DROP TABLE employees;

### Quick Comparison:

DELETE → Remove rows

TRUNCATE → Remove all rows

DROP → Remove table/object

### Memory Trick:

**DELETE = Rows**

**TRUNCATE = All Rows**

**DROP = Object**

### One-Line Answer:

"`DELETE` removes rows, `TRUNCATE` removes all rows while keeping the table structure, and `DROP` removes the table or database object itself."

## Q5. What is the difference between a Primary Key and a Foreign Key?

### Answer:

A Primary Key uniquely identifies each row in a table.

A Foreign Key is a column that creates a relationship between tables by referring to a key in another table.

### Example:

Suppose we have:

    customers
    ----------------
    customer_id
    name

And:

    orders
    ----------------
    order_id
    customer_id

Here:

`customers.customer_id` can be the Primary Key.

`orders.customer_id` can be a Foreign Key referencing the customer.

### Primary Key:

- Uniquely identifies a row
- Cannot contain NULL values
- A table normally has one primary key constraint, which can contain one or multiple columns

### Foreign Key:

- Creates a relationship between tables
- Can contain duplicate values
- Can reference a candidate/primary key depending on the database design
- NULL may be allowed unless restricted by another constraint

### Memory Trick:

Primary Key → **Identify**

Foreign Key → **Connect**

### One-Line Answer:

"A Primary Key uniquely identifies a row in a table, while a Foreign Key is used to establish a relationship between tables by referencing a key in another table."

## Q6. What is the difference between PRIMARY KEY and UNIQUE?

### Answer:

Both are used to enforce uniqueness, but they have important differences.

A Primary Key uniquely identifies each row and cannot contain NULL.

A UNIQUE constraint ensures that values in the constrained column or columns are unique, but its NULL behavior depends on the database system.

A table can have only one Primary Key constraint, while it can have multiple UNIQUE constraints.

### Example:

    CREATE TABLE users (
        user_id INT PRIMARY KEY,
        email VARCHAR(100) UNIQUE
    );

Here:

- `user_id` is the Primary Key.
- `email` must be unique.

### Memory Trick:

PRIMARY KEY → **Main Identity**

UNIQUE → **Unique Value**

### One-Line Answer:

"A Primary Key uniquely identifies rows and does not allow NULL, while a UNIQUE constraint enforces uniqueness and a table can have multiple UNIQUE constraints."

## Q7. What are constraints in SQL?

### Answer:

Constraints are rules applied to table columns to maintain data accuracy and integrity.

Common constraints include:

- `PRIMARY KEY`
- `FOREIGN KEY`
- `UNIQUE`
- `NOT NULL`
- `CHECK`
- `DEFAULT`

### Example:

    CREATE TABLE employees (
        id INT PRIMARY KEY,
        name VARCHAR(100) NOT NULL,
        email VARCHAR(100) UNIQUE,
        salary DECIMAL(10,2) CHECK (salary > 0),
        status VARCHAR(20) DEFAULT 'Active'
    );

Here:

- `PRIMARY KEY` → Uniquely identifies the employee
- `NOT NULL` → Name must have a value
- `UNIQUE` → Email must be unique
- `CHECK` → Salary must satisfy the condition
- `DEFAULT` → Provides a default status

### Why are constraints useful?

They prevent invalid data from being stored and help maintain database integrity.

### Memory Trick:

**PK → Identity**

**FK → Relationship**

**UNIQUE → No Duplicate**

**NOT NULL → Required**

**CHECK → Condition**

**DEFAULT → Default Value**

### One-Line Answer:

"Constraints are rules applied to database columns to maintain data integrity and prevent invalid or inconsistent data."

## Q8. What are joins in SQL?

### Answer:

A JOIN is used to combine data from two or more tables based on a related column or condition.

Common types of joins are:

- INNER JOIN
- LEFT JOIN
- RIGHT JOIN
- FULL OUTER JOIN
- CROSS JOIN

### Example:

Suppose we have:

    customers
    customer_id, name

    orders
    order_id, customer_id, amount

We can combine them using:

    SELECT c.name, o.amount
    FROM customers c
    INNER JOIN orders o
        ON c.customer_id = o.customer_id;

This returns customers who have matching orders.

### Memory Trick:

JOIN = **Combine Related Data**

### One-Line Answer:

"A JOIN combines data from multiple tables using a related column or condition."

## Q9. What is the difference between INNER JOIN and LEFT JOIN?

### Answer:

An `INNER JOIN` returns only rows where there is a match in both tables.

A `LEFT JOIN` returns all rows from the left table and matching rows from the right table. If there is no match, columns from the right table contain NULL.

### Example:

    SELECT c.name, o.amount
    FROM customers c
    INNER JOIN orders o
        ON c.customer_id = o.customer_id;

This returns only customers who have matching orders.

Using LEFT JOIN:

    SELECT c.name, o.amount
    FROM customers c
    LEFT JOIN orders o
        ON c.customer_id = o.customer_id;

This returns all customers, including customers who have no orders.

### Memory Trick:

INNER → **Only Matches**

LEFT → **Everything from Left**

### One-Line Answer:

"INNER JOIN returns only matching rows from both tables, while LEFT JOIN returns all rows from the left table and matching rows from the right table."

## Q10. What is GROUP BY in SQL?

### Answer:

`GROUP BY` is used to group rows that have the same values in one or more columns.

It is commonly used with aggregate functions such as:

- `COUNT()`
- `SUM()`
- `AVG()`
- `MIN()`
- `MAX()`

### Example:

    SELECT department, COUNT(*) AS employee_count
    FROM employees
    GROUP BY department;

This groups employees by department and counts the employees in each department.

### Another Example:

    SELECT department, AVG(salary) AS average_salary
    FROM employees
    GROUP BY department;

This calculates the average salary for each department.

### Memory Trick:

GROUP BY = **Group Similar Data**

### One-Line Answer:

"`GROUP BY` groups rows with the same values and is commonly used with aggregate functions to calculate results for each group."

## Q11. What are aggregate functions in SQL?

### Answer:

Aggregate functions perform calculations on multiple rows and return a single result for each group.

Common aggregate functions are:

- `COUNT()` → Counts rows or values
- `SUM()` → Calculates total
- `AVG()` → Calculates average
- `MIN()` → Finds minimum
- `MAX()` → Finds maximum

### Example:

    SELECT
        COUNT(*) AS total_employees,
        SUM(salary) AS total_salary,
        AVG(salary) AS average_salary,
        MIN(salary) AS minimum_salary,
        MAX(salary) AS maximum_salary
    FROM employees;

### Important Point:

Aggregate functions are commonly used with `GROUP BY` when we want separate results for different groups.

### Memory Trick:

**COUNT → How Many**

**SUM → Total**

**AVG → Average**

**MIN → Smallest**

**MAX → Largest**

### One-Line Answer:

"Aggregate functions perform calculations over multiple rows, such as counting records, calculating totals and averages, or finding minimum and maximum values."

## Q12. What is a subquery in SQL?

### Answer:

A subquery is a query written inside another SQL query.

The inner query is executed as part of the outer query.

### Example:

Suppose we want employees whose salary is greater than the average salary.

    SELECT name, salary
    FROM employees
    WHERE salary > (
        SELECT AVG(salary)
        FROM employees
    );

The inner query calculates the average salary, and the outer query finds employees whose salary is above that average.

### Where can subqueries be used?

Subqueries can appear in places such as:

- `WHERE`
- `FROM`
- `SELECT`
- `HAVING`

depending on the SQL requirement.

### Memory Trick:

Subquery = **Query Inside Query**

### One-Line Answer:

"A subquery is a query nested inside another query and is used when the result of one query is needed by another query."

## Q13. What is a correlated subquery?

### Answer:

A correlated subquery is a subquery that depends on a value from the outer query.

Unlike a simple independent subquery, it is logically evaluated with reference to the current row of the outer query.

### Example:

    SELECT e1.name, e1.salary, e1.department
    FROM employees e1
    WHERE e1.salary > (
        SELECT AVG(e2.salary)
        FROM employees e2
        WHERE e2.department = e1.department
    );

Here, the inner query refers to `e1.department` from the outer query.

So the average salary is calculated for the department of the current employee.

### Important Point:

Correlated subqueries can be useful, but for some problems a JOIN or window function may provide a more efficient or readable solution.

### Memory Trick:

Normal Subquery → **Independent**

Correlated Subquery → **Depends on Outer Query**

### One-Line Answer:

"A correlated subquery is a subquery that refers to values from the outer query, so its result depends on the current row being processed."

## Q14. What is the difference between UNION and UNION ALL?

### Answer:

Both are used to combine the results of multiple SELECT queries.

`UNION` removes duplicate rows from the combined result.

`UNION ALL` keeps duplicate rows.

### Example:

    SELECT city
    FROM customers

    UNION

    SELECT city
    FROM suppliers;

Duplicate city values are removed.

Using:

    SELECT city
    FROM customers

    UNION ALL

    SELECT city
    FROM suppliers;

Duplicate city values are retained.

### Important Point:

The SELECT statements used with UNION or UNION ALL must have compatible numbers and types of columns.

### Memory Trick:

UNION → **Combine + Remove Duplicates**

UNION ALL → **Combine + Keep Duplicates**

### One-Line Answer:

"`UNION` combines query results and removes duplicates, while `UNION ALL` combines the results and keeps duplicates."

## Q15. What is NULL in SQL, and how do you handle it?

### Answer:

`NULL` represents a missing, unknown, or not-applicable value.

It is not the same as:

- `0`
- An empty string
- `FALSE`

To check for NULL, we use `IS NULL` or `IS NOT NULL`.

### Example:

    SELECT *
    FROM employees
    WHERE manager_id IS NULL;

This returns employees whose manager ID is missing.

We should not normally write:

    WHERE manager_id = NULL

because NULL represents an unknown value and is not compared using normal equality.

### Handling NULL:

Functions such as `COALESCE()` can be used to provide an alternative value.

Example:

    SELECT name, COALESCE(phone, 'Not Available') AS phone
    FROM employees;

If `phone` is NULL, the query returns `"Not Available"`.

### Memory Trick:

NULL = **Missing / Unknown**

Check → `IS NULL`

Alternative → `COALESCE()`

### One-Line Answer:

"NULL represents a missing or unknown value in SQL. We check it using `IS NULL` or `IS NOT NULL`, and functions such as `COALESCE()` can be used to provide fallback values."

## Q16. What is the difference between IN and EXISTS?

### Answer:

Both can be used to filter rows based on whether related values exist, but they work differently.

`IN` checks whether a value matches any value in a list or subquery result.

`EXISTS` checks whether the subquery returns at least one row.

### Example using IN:

    SELECT name
    FROM employees
    WHERE department_id IN (
        SELECT department_id
        FROM departments
        WHERE location = 'Mumbai'
    );

This checks whether the employee's department ID is in the returned set.

### Example using EXISTS:

    SELECT e.name
    FROM employees e
    WHERE EXISTS (
        SELECT 1
        FROM departments d
        WHERE d.department_id = e.department_id
        AND d.location = 'Mumbai'
    );

This checks whether at least one matching department exists.

### Important Interview Point:

Neither is universally faster.

Performance depends on the database engine, query structure, indexes, data distribution, and optimizer.

### Memory Trick:

IN → **Match a Set of Values**

EXISTS → **Does a Match Exist?**

### One-Line Answer:

"`IN` checks whether a value belongs to a set of values, while `EXISTS` checks whether a related subquery returns at least one row."

## Q17. What is an index in SQL?

### Answer:

An index is a database structure that helps the database find rows more efficiently.

It is similar to an index in a book. Instead of scanning every page, we can use the index to locate relevant information faster.

### Example:

Suppose we frequently search employees by email.

    CREATE INDEX idx_employee_email
    ON employees(email);

Now the database can potentially use this index when executing queries such as:

    SELECT *
    FROM employees
    WHERE email = 'user@example.com';

### Advantages:

- Can improve read/query performance
- Useful for columns frequently used in filtering, joins, or sorting

### Disadvantages:

- Requires additional storage
- Can increase the cost of `INSERT`, `UPDATE`, and `DELETE` because indexes may also need to be maintained
- Too many indexes can hurt performance

### Important Interview Point:

An index does not automatically make every query faster. The database optimizer decides whether an index is beneficial.

### Memory Trick:

Index = **Faster Search**

### One-Line Answer:

"An index is a database structure that can improve query performance by allowing the database to find relevant rows more efficiently, but indexes also require storage and maintenance."

## Q18. What is normalization in SQL?

### Answer:

Normalization is the process of organizing data in a relational database to reduce unnecessary duplication and improve data integrity.

Instead of storing the same information repeatedly, we divide data into related tables.

### Example:

Instead of storing:

    order_id | customer_name | customer_phone | product

for every order, we can separate the information:

    Customers
    customer_id | name | phone

    Orders
    order_id | customer_id | product

Now customer information does not need to be repeated for every order.

### Common Normal Forms:

1. First Normal Form (1NF)
   - Data should contain atomic values.
   - Repeating groups should be removed.

2. Second Normal Form (2NF)
   - Must satisfy 1NF.
   - Non-key attributes should depend on the whole primary key.

3. Third Normal Form (3NF)
   - Must satisfy 2NF.
   - Non-key attributes should not depend on other non-key attributes.

### Benefits:

- Reduces data redundancy
- Improves data consistency
- Makes updates safer
- Improves database organization

### Memory Trick:

Normalization = **Reduce Duplication + Improve Integrity**

### One-Line Answer:

"Normalization is the process of organizing relational data into related tables to reduce redundancy and improve data integrity."

## Q19. What is a transaction in SQL?

### Answer:

A transaction is a sequence of one or more database operations that are treated as a logical unit of work.

A transaction should either complete successfully or be rolled back when appropriate.

### Example:

Suppose we transfer ₹1,000 from Account A to Account B.

Two operations are required:

1. Deduct ₹1,000 from Account A.
2. Add ₹1,000 to Account B.

Both operations should be treated as one transaction.

Conceptually:

    BEGIN;

    UPDATE accounts
    SET balance = balance - 1000
    WHERE account_id = 1;

    UPDATE accounts
    SET balance = balance + 1000
    WHERE account_id = 2;

    COMMIT;

If something goes wrong before the transaction is committed, the transaction can be rolled back where supported:

    ROLLBACK;

### Common Transaction Commands:

- `COMMIT`
- `ROLLBACK`
- `SAVEPOINT`

### Memory Trick:

Transaction = **One Logical Unit of Work**

### One-Line Answer:

"A transaction is a group of database operations treated as one logical unit of work, which can be committed or rolled back according to the transaction outcome."

## Q20. What are ACID properties?

### Answer:

ACID describes four important properties of reliable database transactions.

ACID stands for:

- Atomicity
- Consistency
- Isolation
- Durability

### 1. Atomicity

A transaction is treated as an all-or-nothing operation.

Either all required operations succeed, or the transaction is rolled back.

Example:

In a money transfer, both debit and credit should succeed as part of the transaction.

### 2. Consistency

A transaction should move the database from one valid state to another valid state while respecting defined rules and constraints.

### 3. Isolation

Concurrent transactions should be isolated according to the database's isolation guarantees, so intermediate states do not improperly interfere with other transactions.

### 4. Durability

Once a transaction is successfully committed, its changes should persist even if there is a system failure, subject to the database's durability guarantees.

### Memory Trick:

**A → All or Nothing**

**C → Correct State**

**I → Independent Transactions**

**D → Data Persists**

### One-Line Answer:

"ACID stands for Atomicity, Consistency, Isolation, and Durability, which are properties that help ensure reliable and consistent database transactions."

## Q21. What is a View in SQL?

### Answer:

A View is a virtual table based on the result of a SQL query.

It usually does not store the underlying result as a separate copy of the data; instead, the database uses the view definition when the view is queried, although implementation details can vary.

### Example:

Suppose we frequently need employee information for a particular department.

    CREATE VIEW sales_employees AS
    SELECT employee_id, name, salary
    FROM employees
    WHERE department = 'Sales';

Now we can query:

    SELECT *
    FROM sales_employees;

### Advantages:

- Simplifies complex queries
- Can provide a controlled interface to data
- Improves query reuse
- Can help restrict access to selected columns or rows

### Important Point:

A normal view is not the same as a materialized view.

A materialized view stores query results physically and generally needs to be refreshed.

### Memory Trick:

View = **Virtual Table**

### One-Line Answer:

"A view is a virtual table created from a SQL query. It can simplify complex queries, provide reusable logic, and help control access to underlying data."

A stored procedure is a named set of SQL statements stored in the database that can be executed when required.

The exact syntax and capabilities depend on the database system.

### Example:

Conceptually, a stored procedure could perform an operation such as retrieving employees from a particular department.

    CREATE PROCEDURE GetEmployeesByDepartment(IN dept_name VARCHAR(100))

        WHERE department = dept_name;

The procedure can then be called using the syntax supported by the database system.

### Advantages:

- Reusable database-side logic
- Can reduce repeated SQL code
- Can centralize certain database operations
- Can support controlled access patterns

### Important Point:

Stored procedures are database-specific in syntax and features.

### Stored Procedure vs Function:

A stored procedure generally performs a set of operations and may return result sets or output parameters depending on the database.

A function generally returns a value and can often be used as part of an expression, depending on the database system.

### Memory Trick:

Stored Procedure = **Stored SQL Logic**

### One-Line Answer:

"A stored procedure is a reusable set of SQL statements stored in the database and executed when needed."

## Q23. What are window functions in SQL?

### Answer:

A window function performs a calculation across a set of related rows without collapsing those rows into a single row like `GROUP BY` does.

Common window functions include:

- `ROW_NUMBER()`
- `RANK()`
- `DENSE_RANK()`
- `LAG()`
- `LEAD()`
- `SUM() OVER()`
- `AVG() OVER()`

### Example:

Suppose we want to rank employees by salary within each department.

    SELECT
        name,
        department,
        salary,
        RANK() OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS salary_rank
    FROM employees;

Here:

- `PARTITION BY department` creates a separate window for each department.
- `ORDER BY salary DESC` determines the ranking order.
- `RANK()` assigns the rank.

### GROUP BY vs Window Function:

`GROUP BY` → Combines rows into groups and returns fewer rows.

Window Function → Performs calculations across related rows while keeping individual rows.

### Example Use Cases:

- Ranking
- Running totals
- Comparing current and previous rows
- Finding top records within groups
- Calculating moving or partition-based averages

### Memory Trick:

Window Function = **Calculate Across Rows Without Collapsing Them**

### One-Line Answer:

"A window function performs calculations across related rows while keeping the individual rows in the result, making it useful for ranking, running totals, and row-to-row comparisons."

## Q24. What is a CTE in SQL?

### Answer:

CTE stands for Common Table Expression.

It allows us to define a temporary named result set that can be referenced within a SQL statement.

A CTE is created using the `WITH` keyword.

### Example:

    WITH high_salary_employees AS (
        SELECT employee_id, name, salary
        FROM employees
        WHERE salary > 50000
    )
    SELECT *
    FROM high_salary_employees;

Here, `high_salary_employees` is the CTE.

### Why use CTEs?

CTEs can:

- Make complex queries easier to read
- Break a large query into logical steps
- Improve query organization
- Support recursive queries in databases that provide recursive CTEs

### CTE vs Subquery:

A CTE and a subquery can sometimes solve the same problem.

A CTE can be easier to read and can be referenced multiple times within the same statement where supported by the SQL syntax.

### Important Point:

A CTE normally exists only for the duration of the statement in which it is defined.

### Memory Trick:

CTE = **WITH → Temporary Named Result**

### One-Line Answer:

"A CTE, or Common Table Expression, is a temporary named result set defined using `WITH` that helps organize and simplify complex SQL queries."

## Q25. How would you find the second-highest salary in SQL?

### Answer:

There are multiple ways to solve this problem.

One simple approach is using a subquery with `MAX()`.

Example:

    SELECT MAX(salary) AS second_highest_salary
    FROM employees
    WHERE salary < (
        SELECT MAX(salary)
        FROM employees
    );

The inner query finds the highest salary.

The outer query finds the maximum salary that is lower than the highest salary.

### Another Approach Using DENSE_RANK():

    SELECT salary
    FROM (
        SELECT
            salary,
            DENSE_RANK() OVER (ORDER BY salary DESC) AS salary_rank
        FROM employees
    ) ranked
    WHERE salary_rank = 2;

This approach returns the second distinct-highest salary.

### Important Interview Point:

We should clarify whether "second-highest salary" means:

- Second-highest distinct salary
- Second row after sorting

Usually, interview questions mean the second-highest distinct salary.

`DENSE_RANK()` is useful when duplicate salaries exist.

### Example:

Suppose salaries are:

100000
90000
90000
80000

The second-highest distinct salary is:

90000

### Memory Trick:

**Highest → MAX()**

**Second Highest → MAX below MAX**

or

**DENSE_RANK() = 2**

### One-Line Answer:

"To find the second-highest distinct salary, I can find the maximum salary below the overall maximum, or use `DENSE_RANK()` and select rank 2."
