## Q1. Can you explain your Hirely project?

### Answer:

Sure.

Hirely is an AI-powered recruitment platform that helps candidates and recruiters with different parts of the recruitment process.

### Problem Statement:

In traditional recruitment, recruiters may need to manually review many applications and compare candidate skills and resumes with job requirements.

Candidates also need to understand whether their resume matches a particular job and what areas they can improve.

So, the goal of Hirely is to bring these recruitment activities together and use AI where it can provide useful assistance.

### Solution:

I worked mainly on the backend and AI-related parts of the application.

The backend is built using Python and FastAPI. It provides REST APIs for features such as authentication, candidate profiles, jobs, applications and AI services.

For the AI part, I used Gemini for tasks such as resume analysis, cover letter generation and career assistance.

The project also uses embeddings for semantic comparison between a candidate's resume and a job.

For matching, I use deterministic logic for structured information such as skills, and AI/embeddings where semantic understanding is required.

### Architecture:

The main backend flow is:

```Client Request
      ↓

      ↓
Pydantic Validation
      ↓
Business / Domain Services
      ↓
 ┌───────────────┬────────────────┐
 ↓               ↓                ↓
PostgreSQL    AI Services      Business Logic
 ↓               ↓
SQLAlchemy     Gemini
ORM            Embeddings
               Resume Parsing
```

### Technologies I Worked With:

- Python
- FastAPI
- REST APIs
- Pydantic
- SQLAlchemy
- PostgreSQL
- Gemini
- Embeddings
- Resume parsing
- JWT authentication
- Role-based access control

### My Contribution:

My main focus was on backend and AI functionality.

I worked with FastAPI to build the API layer and used Pydantic for request and response validation.

I worked with PostgreSQL and SQLAlchemy for database operations.

On the AI side, I worked with Gemini, resume parsing, embeddings and AI-powered features such as ATS analysis and candidate-related assistance.

I also worked on deterministic skill matching and backend-controlled scoring logic instead of depending on the LLM for every decision.

### Overall:

The main idea of Hirely is to combine a reliable backend with AI capabilities.

The backend handles structured data and business rules, while the AI layer handles tasks that require language understanding or semantic analysis.

This makes the application more useful while keeping important business logic under backend control.

### Memory Trick:

Problem → Backend → AI → Matching → My Contribution → Result

### One-Line Version:

> "Hirely is an AI-powered recruitment platform where I mainly worked on the Python backend and AI layer, using FastAPI, PostgreSQL, Gemini and embeddings to build recruitment and AI-assisted features."

## Q2. What was your specific contribution to the Hirely project?

### Answer:

My main contribution to Hirely was on the backend and AI-related functionality.

I worked with Python and FastAPI to build and handle REST APIs. I also worked with Pydantic for request and response validation, and PostgreSQL with SQLAlchemy for database operations.

On the AI side, I worked with Gemini for features such as resume analysis, ATS analysis, cover letter generation and career assistance.

I also worked on the candidate-job matching logic. For structured information like skills, I used deterministic matching logic, while embeddings were used for semantic similarity between the candidate's resume and the job requirements.

One important thing I learned from this project was that we should not depend on an LLM for every decision. Business rules and important calculations should remain under backend control, while AI should be used where language understanding or semantic analysis is actually required.

### Memory Trick:

**Backend → Database → AI → Matching → Deterministic Control**

### Short Version:

"My main contribution was on the Python backend and AI layer, where I worked with FastAPI, PostgreSQL, Gemini, embeddings, resume analysis and candidate-job matching."

## Q3. Can you explain the architecture of the Hirely project?

### Answer:

Sure.

Hirely follows a layered backend architecture where the backend handles the main business logic and the AI layer is used for tasks that require language understanding or semantic analysis.

The flow is:

Client Request → FastAPI API → Pydantic Validation → Business Services → Database / AI Services

First, the client sends a request to the FastAPI backend.

FastAPI receives the request and Pydantic validates the input data.

Then the request goes to the appropriate business or domain service, where the main application logic is handled.

For normal operations, such as user management, jobs, applications and structured data, the backend communicates with PostgreSQL through SQLAlchemy.

For AI-related operations, the backend calls the AI services. These services can use Gemini, embeddings or resume-parsing components depending on the requirement.

The important design principle is that deterministic business logic stays in the backend, while AI is used only where it provides value, such as resume understanding, semantic matching or generating text.

So, the overall architecture keeps the backend reliable and separates AI functionality from the core application logic.

### Simple Architecture:

Client  
↓  
FastAPI  
↓  
Pydantic Validation  
↓  
Business / Domain Services  
↓  
├── PostgreSQL + SQLAlchemy  
│  
└── AI Services  
    ├── Gemini  
    ├── Embeddings  
    └── Resume Parsing

### Memory Trick:

**Request → Validate → Logic → Database / AI**

### One-Line Answer:

"Hirely uses a layered backend architecture where FastAPI handles APIs, Pydantic validates data, business services handle application logic, PostgreSQL stores structured data, and a separate AI layer handles Gemini, embeddings and resume-related intelligence."

## Q4. Why did you use FastAPI in the Hirely project?

### Answer:

I used FastAPI to build the backend REST APIs for Hirely.

One reason I chose FastAPI is that it works well with Python and provides a simple way to create API endpoints.

I used it to handle features such as authentication, candidate profiles, jobs, applications and AI-related services.

I also used Pydantic with FastAPI to validate the request and response data. This helped ensure that the API received data in the expected format.

Another benefit is that FastAPI supports automatic API documentation, which makes it easier to test and understand the available endpoints.

So, in Hirely, FastAPI acts as the main backend API layer between the client and the application services.

### Memory Trick:

**Python → APIs → Validation → Documentation**

### One-Line Answer:

"I used FastAPI to build the Python backend REST APIs, handle requests, validate data with Pydantic, and connect the application with database and AI services."

## Q5. Why did you use Pydantic in the Hirely project?

### Answer:

I used Pydantic with FastAPI for data validation and to define the structure of API requests and responses.

For example, when a client sends data to an API, Pydantic checks whether the required fields are present and whether the data has the expected types and format.

I used Pydantic schemas for different parts of the application, including candidate-related data, job-related data and AI responses.

This helped make the API more reliable and prevented invalid or unexpected data from moving further into the application.

It was also useful for structured AI output because instead of directly trusting the response from the LLM, I could validate the generated data against a defined schema.

### Memory Trick:

**Define → Validate → Structure → Reliable**

### One-Line Answer:

"I used Pydantic to define structured data models and validate API requests, responses and structured AI outputs before they were used by the application."

## Q6. Why did you use PostgreSQL and SQLAlchemy in Hirely?

### Answer:

I used PostgreSQL as the relational database for storing the structured data of the Hirely application.

For example, it stores information related to users, candidates, recruiters, jobs, applications, skills and resumes.

I used SQLAlchemy as the ORM to interact with PostgreSQL from Python.

Instead of writing raw SQL for every database operation, SQLAlchemy allowed me to work with Python models and queries while still interacting with the PostgreSQL database.

I also used Alembic for database migrations, so database schema changes could be managed in a controlled way.

This combination helped keep the database layer organized and maintainable.

### Memory Trick:

**PostgreSQL → Store Data**  
**SQLAlchemy → Python ↔ Database**  
**Alembic → Manage Schema Changes**

### One-Line Answer:

"I used PostgreSQL to store Hirely's structured application data, SQLAlchemy as the Python ORM to interact with the database, and Alembic to manage database schema migrations."

## Q7. What is the role of Gemini in the Hirely project?

### Answer:

I used Gemini as the main Large Language Model for the AI-related features of Hirely.

I used it for tasks that require understanding or generating natural language.

For example, Gemini is used for resume analysis, ATS analysis, cover letter generation and career assistance.

For the ATS analyzer, the backend first handles deterministic parts such as skill matching and scoring. Gemini is then used for the qualitative analysis, such as identifying strengths, improvements and providing a summary.

The important point is that I did not depend on Gemini for every decision. I kept important business rules and calculations in the backend and used Gemini where language understanding was required.

The output from Gemini is also handled in a structured way using Pydantic validation.

### Memory Trick:

**Gemini → Understand → Analyze → Generate → Structured Output**

### One-Line Answer:

"Gemini is used as the LLM for Hirely's AI features such as resume analysis, ATS analysis, cover letter generation and career assistance, while important business logic remains controlled by the backend."

## Q8. How does candidate-job matching work in Hirely?

### Answer:

In Hirely, candidate-job matching uses a combination of deterministic skill matching and semantic similarity.

First, the candidate's skills are compared with the skills required by the job using deterministic matching logic.

For example, if a job requires Python, SQL and FastAPI, the backend can directly check which of these skills the candidate has.

For information that requires more semantic understanding, I use embeddings to compare the candidate's resume with the job requirements.

The embedding similarity helps identify whether the candidate's resume is semantically relevant to the job, even when the exact words may not be the same.

The important part is that I do not let the LLM make the final matching decision. The backend controls the scoring and business logic, which makes the result more predictable and explainable.

### Memory Trick:

**Skills → Embeddings → Score → Backend Control**

### One-Line Answer:

"Hirely combines deterministic skill matching with embedding-based semantic similarity, while keeping the final scoring and business logic under backend control."

## Q9. Why did you use embeddings in Hirely?

### Answer:

I used embeddings to handle semantic similarity between a candidate's resume and a job description.

Traditional keyword matching only checks whether the exact same words are present. Embeddings are useful because they represent text as numerical vectors that capture semantic meaning.

For example, a resume may mention "building REST services with Python," while a job description may mention "Python backend API development." The exact words are different, but the meaning is related.

In Hirely, I used embeddings to calculate the semantic similarity between the candidate's resume and the job requirements.

This semantic score is then used as one part of the overall matching process, while the final scoring logic remains controlled by the backend.

### Memory Trick:

**Text → Vector → Similarity → Matching**

### One-Line Answer:

"I used embeddings to convert resume and job text into vectors and calculate semantic similarity, so Hirely can identify relevant matches even when the exact keywords are different."

## Q10. What is the difference between deterministic matching and semantic matching in Hirely?

### Answer:

Deterministic matching is based on fixed rules and exact data.

For example, if a job requires Python and the candidate has Python as a skill, the backend can directly identify it as a match.

Semantic matching is based on the meaning of the text rather than only exact words. In Hirely, I use embeddings to compare the semantic similarity between the candidate's resume and the job requirements.

So, deterministic matching is useful for structured information such as skills, while semantic matching is useful for understanding the meaning of resume and job text.

I used both approaches because they complement each other. The backend handles predictable business rules, while embeddings help with more flexible semantic understanding.

### Memory Trick:

**Deterministic = Exact Rules**  
**Semantic = Meaning**

### Simple Example:

Job requires:
**"Python backend development"**

Resume says:
**"Developed REST APIs using Python and FastAPI."**

- Deterministic matching → checks exact structured skills such as Python.
- Semantic matching → understands that the resume content is related to Python backend development.

### One-Line Answer:

"Deterministic matching checks structured information using fixed rules, while semantic matching uses embeddings to understand the meaning and similarity of resume and job content."

## Q11. How does the ATS Analyzer work in Hirely?

### Answer:

The ATS Analyzer evaluates how well a candidate's resume matches a particular job.

I designed it as a hybrid approach, combining deterministic backend logic, semantic similarity using embeddings, and qualitative analysis using Gemini.

First, the backend checks the candidate's skills against the job's required and preferred skills.

Then, embeddings are used to calculate the semantic relevance between the resume and the job.

Gemini is used for qualitative analysis, such as identifying the candidate's strengths, areas for improvement, suggestions and an overall summary.

The final ATS score is calculated by the backend rather than allowing Gemini to decide the score.

The scoring is based on:

- Required Skill Coverage — 45%
- Preferred Skill Coverage — 15%
- Semantic Relevance — 30%
- Resume Completeness — 10%

The result is then returned in a structured format, with Pydantic used to validate the AI-generated response.

### Memory Trick:

**Skills → Semantic → AI Analysis → Backend Score → Structured Result**

### One-Line Answer:

"Hirely's ATS Analyzer combines deterministic skill matching, embedding-based semantic relevance and Gemini qualitative analysis, while the final score is calculated and controlled by the backend."

## Q12. How did you make the ATS score reliable instead of depending completely on Gemini?

### Answer:

I kept the important scoring logic under backend control instead of asking Gemini to calculate the final score.

The ATS Analyzer uses different components for different purposes.

First, deterministic logic is used to calculate structured skill coverage.

Then, embeddings are used to calculate semantic relevance between the resume and job requirements.

Gemini is mainly used for qualitative analysis, such as strengths, improvements, suggestions and the summary.

After calculating these components, the backend applies the predefined scoring weights and calculates the final ATS score.

This approach makes the score more predictable and consistent because the LLM is not responsible for making the final numerical decision.

### Memory Trick:

**Deterministic → Embeddings → Gemini → Backend Score**

### One-Line Answer:

"I made the ATS score reliable by keeping the final scoring calculation in the backend, while using deterministic matching, embeddings and Gemini for their specific strengths."

## Q13. How did you use Gemini for resume analysis in Hirely?

### Answer:

In Hirely, Gemini is used to analyze resume content and extract useful information from it.

The resume first goes through the resume parsing process, where the document content is extracted and converted into usable text or structured information.

That information is then provided to the AI service with the relevant context required for the particular analysis.

Gemini can understand the resume content and help identify information such as skills, experience and other relevant details needed by the recruitment features.

I also use structured schemas with Pydantic so that the AI response follows an expected format before it is used by the application.

The important point is that the AI is used for understanding unstructured resume content, while the backend remains responsible for validation and business logic.

### Memory Trick:

**Resume → Parse → Context → Gemini → Structured Output → Backend**

### One-Line Answer:

"I used Gemini to understand and analyze parsed resume content, while Pydantic and the backend were used to structure, validate and control the resulting data."

## Q14. What is resume parsing, and how did you use it in Hirely?

### Answer:

Resume parsing is the process of extracting useful information from a resume document and converting it into structured or usable data.

In Hirely, resume parsing is used because resumes can come in different document formats and contain unstructured information.

I worked with the resume parsing component to extract the resume content, which could then be used by other parts of the application such as resume analysis and ATS analysis.

The parsed resume information can be combined with job requirements and passed to the appropriate AI or matching services.

So, resume parsing acts as an important first step between the uploaded resume document and the recruitment intelligence features.

### Memory Trick:

**Resume File → Extract → Structure → Analyze → Match**

### One-Line Answer:

"Resume parsing converts an uploaded resume into usable information, which Hirely can then use for resume analysis, ATS evaluation and candidate-job matching."

## Q15. How did you handle authentication and authorization in Hirely?

### Answer:

I used JWT-based authentication in Hirely to securely authenticate users.

When a user logs in successfully, the application generates a JWT token. The token is then used to access protected API endpoints.

For authorization, I used Role-Based Access Control, or RBAC.

Hirely has different types of users, such as candidates and recruiters. Their permissions are different, so the backend checks the user's role before allowing access to role-specific operations.

For example, candidate-related operations should be accessible to authorized candidates, while recruiter-specific operations should be protected for recruiters.

I also used password hashing with Argon2 instead of storing passwords directly.

The important point is that authentication and authorization are handled by the backend, not by the AI layer.

### Memory Trick:

**Login → JWT → Protected API → Role Check → Access**

### One-Line Answer:

"I used JWT for authentication, RBAC for authorization, and Argon2 for password hashing, with the backend controlling access to protected endpoints."

## Q16. What is RBAC, and why did you use it in Hirely?

### Answer:

RBAC stands for Role-Based Access Control.

It means that access to different features or API endpoints is controlled based on the user's role.

In Hirely, we have different types of users, mainly candidates and recruiters. They have different responsibilities, so they should not have access to all the same operations.

For example, a candidate can manage their profile, view jobs and apply for jobs, while a recruiter can create and manage job postings and review applications.

The backend checks the authenticated user's role before allowing access to protected operations.

I used RBAC to make sure users can access only the functionality that is appropriate for their role.

### Memory Trick:

**Role → Permission → Access**

### One-Line Answer:

"RBAC controls access based on user roles. In Hirely, it ensures candidates and recruiters can access only the operations allowed for their respective roles."

## Q17. What would you do if a Hirely API endpoint was not working?

### Answer:

First, I would try to reproduce the issue so I can understand whether it happens consistently or only for a specific request.

Then I would check the API request, including the endpoint, HTTP method, request body, headers and authentication token.

Next, I would check the application logs and the error message to identify where the request is failing.

I would verify the Pydantic validation, business logic and database operation depending on where the failure occurs.

If the endpoint involves an AI service, I would also check whether the external AI service is responding correctly and whether the input being sent to it is valid.

After identifying the root cause, I would fix the issue, test the endpoint again with valid and invalid inputs, and make sure the change has not affected other functionality.

### Memory Trick:

**Reproduce → Request → Logs → Logic → Database/AI → Fix → Test**

### One-Line Answer:

"If an API endpoint fails, I would reproduce the issue, check the request and logs, identify the root cause, fix it, and retest the complete API flow."

## Q18. What would you do if the Hirely application was giving a 500 Internal Server Error?

### Answer:

First, I would try to reproduce the issue and identify which API endpoint is returning the 500 error.

Then I would check the application logs and traceback to find the actual exception causing the server error.

Next, I would check the request data and follow the request flow through the application.

I would verify the Pydantic validation, business logic and database operations. If the endpoint uses an AI service, I would also check the AI request and response.

For example, the issue could be caused by invalid data, a database connection problem, an unhandled exception, or an external service failure.

Once I identify the root cause, I would fix the issue and test the endpoint again with different inputs.

I would also check related functionality to make sure the fix has not introduced another problem.

### Memory Trick:

**500 → Reproduce → Logs → Traceback → Find Root Cause → Fix → Retest**

### One-Line Answer:

"For a 500 error, I would reproduce the issue, check the logs and traceback, trace the request to find the root cause, fix it, and retest the complete flow."

## Q19. What was the most challenging part of the Hirely project?

### Answer:

One of the challenging parts was designing the AI-powered matching and ATS functionality in a way that was reliable and controllable.

A simple approach would be to send the resume and job description to an LLM and ask it to give a final score. But this can make the result less predictable.

So, I separated the responsibilities.

For structured information such as required skills, I used deterministic backend logic.

For semantic understanding between the resume and job description, I used embeddings.

For qualitative analysis, such as strengths, improvements and suggestions, I used Gemini.

Finally, the backend controlled the scoring and combined the different components.

This taught me that when building AI applications, we should not use AI for everything. We should use deterministic logic where the rules are clear and use AI where language understanding is actually required.

### Memory Trick:

**Challenge → Separate → Deterministic → Embeddings → Gemini → Backend Control**

### One-Line Answer:

"The main challenge was making AI-based matching reliable, so I separated deterministic business logic, semantic matching and LLM-based qualitative analysis instead of depending completely on the LLM."

## Q20. How would you improve Hirely in the future?

### Answer:

There are several areas where I could improve Hirely in the future.

First, I would improve the resume analysis and matching system by making the semantic matching more advanced and evaluating its performance using a proper evaluation dataset.

I could also introduce RAG for features where the AI needs access to a controlled knowledge base, instead of relying only on the LLM's general knowledge.

Another improvement would be adding better monitoring and observability for the AI features, so we can track failures, response quality and performance.

I would also consider background processing for expensive operations such as resume parsing, embedding generation and LLM calls, especially as the number of users increases.

Finally, I would continue improving testing, error handling and API performance to make the system more production-ready.

### Memory Trick:

**Matching → RAG → Observability → Background Jobs → Testing**

### One-Line Answer:

"I would improve Hirely by making matching more advanced, adding RAG where useful, improving AI observability, moving expensive operations to background processing, and strengthening testing and performance."

## Q21. What did you learn from building the Hirely project?

### Answer:

The main thing I learned from Hirely was how to combine traditional backend development with AI functionality.

I learned how to build REST APIs using FastAPI, validate data using Pydantic, work with PostgreSQL and SQLAlchemy, and implement authentication and role-based access control.

On the AI side, I learned how to use LLMs and embeddings for practical use cases such as resume analysis, ATS analysis and semantic matching.

I also learned that AI applications should not depend completely on an LLM. Deterministic business logic should remain under backend control, while AI should be used where it provides value.

Most importantly, the project improved my understanding of debugging, testing, API design and building a complete backend-oriented AI application.

### Memory Trick:

**Backend → Database → Security → AI → Debugging → Architecture**

### One-Line Answer:

"Hirely taught me how to combine reliable backend engineering with AI capabilities while keeping important business logic deterministic and controllable."

## Q22. Suppose a user says that the Hirely ATS Analyzer was working yesterday, but today it is failing. How would you troubleshoot it?

### Answer:

First, I would reproduce the issue and check whether the problem is happening for all users or only for a particular resume or job.

Then I would check the application logs and identify the exact error.

I would trace the complete ATS flow:

Resume → Resume Parsing → Data Processing → Skill Matching → Embeddings → Gemini Analysis → Final Score → API Response

I would check each stage to find where the failure is occurring.

If resume parsing is failing, I would check the uploaded file and parsing service.

If skill matching or embeddings are failing, I would check the input data and the related processing logic.

If Gemini is failing, I would check the API configuration, request, response and any service error.

If the AI analysis succeeds but the final response fails, I would check the Pydantic validation and API response handling.

After identifying the root cause, I would fix the issue, test the complete ATS workflow again, and verify that existing functionality is still working.

### Memory Trick:

**Reproduce → Logs → Trace Flow → Find Root Cause → Fix → Test**

### One-Line Answer:

"I would troubleshoot the ATS flow step by step, starting with reproduction and logs, then checking parsing, matching, embeddings, Gemini, validation and the API response until I identify and fix the root cause."
