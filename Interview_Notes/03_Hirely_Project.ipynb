# Hirely — Project Explanation for Interviews

## Project Introduction

Hirely is a startup-grade AI-powered recruitment platform designed to connect candidates and recruiters and make different parts of the recruitment process more intelligent and efficient.

The idea behind Hirely comes from a practical problem in recruitment. A recruiter may have to work with many candidates, resumes, job descriptions, applications, required skills, and candidate profiles. At the same time, candidates need to understand whether their profile is suitable for a particular job, how well their resume matches the job, what skills they are missing, and how they can improve their profile.

Instead of treating recruitment as only a traditional CRUD application, Hirely combines a structured recruitment backend with a dedicated AI/GenAI layer.

The overall concept can be represented as:

```text
                    HIRELY
                      |
          Candidate ↔ Recruiter
                      |
               AI Matching Engine
```

The important idea is that AI is not responsible for everything. Traditional backend logic handles structured application data, business rules, authentication, authorization, and database operations, while AI is used for tasks that benefit from language understanding, semantic comparison, analysis, ranking, and explanation.

---

## The Problem I Wanted to Solve

In a traditional recruitment workflow, a candidate may upload a resume and apply to different jobs, while recruiters create jobs and review multiple applications.

As the number of candidates increases, several activities become difficult to perform manually:

```text
Resume screening
Skill comparison
Candidate search
Candidate-job matching
Candidate ranking
Application management
Resume improvement
Understanding why a candidate matches a job
```

From the candidate's perspective, there is also a need for assistance with:

```text
Resume analysis
ATS checking
Resume improvement
Job matching
Career guidance
Cover-letter generation
Application tracking
```

So, I designed Hirely as a platform where these recruitment workflows can be managed through a backend application and AI can assist with the parts that require semantic and language-based intelligence.

---

# How I Designed the System

I chose a layered modular architecture with a dedicated AI/GenAI layer.

The high-level architecture is:

```text
                         React
                           |
                           | HTTP / JSON
                           ↓
                        FastAPI
                           |
                           ↓
                  Pydantic Validation
                           |
                           ↓
                   Business Services
                      /                               /                                ↓              ↓
             SQLAlchemy        AI Services
                    |              |
                    ↓              ├── LLM
             Relational DB         |
                                   ├── Embeddings
                                   |
                                   └── AI Processing
```

The frontend communicates with the FastAPI backend through HTTP/JSON.

FastAPI validates incoming requests using Pydantic and passes valid requests into the business/service layer.

From there, the application can communicate with the relational database through SQLAlchemy or invoke the appropriate AI service.

The AI service can work with an LLM, an embedding model, or other AI-processing components.

The AI result is then converted into a structured result and returned through the backend.

This separation is important because I don't want the AI layer to directly control the application's database or bypass the application's business rules and security model.

I designed the matching process as multiple components rather than asking an LLM one question and accepting its answer as the final result.

The AI engine contains concepts such as:

```text
Resume Parsing
Job Analysis
Candidate Representation
Job Representation
Embeddings
Semantic Similarity
Skill Matching
Matching Engine
Ranking
Match Explanation
AI Orchestration
```

The overall flow is:

```text
Candidate Resume
       |
       ↓
Resume Parsing
       |
       ↓
Candidate Representation
       |
       ↓
Embedding / Structured Information
       |
       |
       |                    Job Description
       |                           |
       |                           ↓
       |                     Job Analysis
       |                           |
       |                           ↓
       |                    Job Representation
       |                           |
       └──────────────┬────────────┘
                      ↓
              Matching Engine
                      |
             ┌────────┴────────┐
             ↓                 ↓
       Skill Matching    Semantic Similarity
             |                 |
             └────────┬────────┘
                      ↓
                   Ranking
                      ↓
              Match Explanation
```

# Why I Use Both Structured Matching and Semantic Matching

One of the important architectural ideas in Hirely is that I don't want to depend entirely on an LLM for candidate matching.

Some information is structured and can be evaluated using deterministic logic.

For example:

```text
Required Skill: Python
Candidate Skill: Python

→ Direct skill match
```

This type of information can be handled by normal backend logic.

But resumes and job descriptions also contain semantic information.

For example, two pieces of text may describe related experience without using exactly the same wording.

That is where embeddings and semantic similarity become useful.

Conceptually:

```text
Structured Information
        ↓
Deterministic Skill Matching

Unstructured Text
        ↓
Embeddings
        ↓
Semantic Similarity

Natural Language Understanding
        ↓
LLM / AI Processing
```

The matching engine can then combine these signals.

This is an important design decision because it gives the system more control than simply asking an LLM to produce a final recruitment score.

The system is designed to process the resume and convert it into useful information for downstream recruitment features.

The conceptual flow is:

```text
Resume Upload
      ↓
Resume Parsing
      ↓
Candidate Information
      ↓
Candidate Representation
      ↓
ATS / Resume Analysis
      ↓
Matching / Recommendations
```

Resume processing can support multiple features rather than being implemented only for one endpoint.

For example, parsed resume information can be useful for:

```text
ATS checking
Resume improvement
Job matching
Candidate representation
Career assistance
```

This is why I treat resume parsing as part of the AI processing pipeline rather than as an isolated feature.

# Job Analysis

The same principle applies to job descriptions.

A job is not simply stored as a block of text.

The system is designed to analyze the job and create a useful representation for matching.

The job-analysis side can identify concepts such as:

```text
Required Skills
Preferred Skills
Job Requirements
Job Description
```

The conceptual flow is:

```text
Job Description
       ↓
Job Analysis
       ↓
Required / Preferred Information
       ↓
Job Representation
       ↓
Matching Engine
```

This representation can then be compared against candidate information.

# Semantic Similarity Using Embeddings

Embeddings are useful because recruitment information is not always expressed using identical words.

The basic idea is to transform text into vector representations.

```text
Resume Text
    ↓
Embedding Model
    ↓
Candidate Vector

Job Text
    ↓
Embedding Model
    ↓
Job Vector

Candidate Vector
        +
Job Vector
        ↓
Semantic Similarity
```

The similarity becomes one of the signals that can help determine how closely the candidate and job are related.

The important point is that embeddings are used for semantic comparison, while structured matching can still be handled separately.

# Candidate Ranking

Hirely also includes candidate ranking as part of the recruitment workflow.

The idea is that matching tells us how compatible a candidate and job are, while ranking allows multiple candidates to be compared for a particular job.

Conceptually:

```text
Job
 ↓
Candidate A → Match Signals
Candidate B → Match Signals
Candidate C → Match Signals
Candidate D → Match Signals
 ↓
Ranking Component
 ↓
Ranked Candidates
```

The ranking component can use the outputs from the matching process.

This is more useful for recruiters than simply showing independent match results for every candidate.

---

# Match Explanation

Another important component is the explanation layer.

Instead of only producing a ranking or score, the system should be able to provide a human-readable explanation of why a candidate matches a job.

Conceptually:

```text
Candidate
    +
Job
    +
Matching Signals
    ↓
Explanation Component
    ↓
Human-readable Explanation
```

This makes the AI output more understandable to the recruiter.

For example, the explanation could communicate which areas contributed to a strong match and where the candidate may have gaps.

The goal is not just to generate an answer but to make the matching result easier for a human to understand.

# AI Orchestration

Hirely also separates AI capabilities through an AI orchestration concept.

Instead of putting every AI operation inside an API endpoint, the AI orchestration layer coordinates the different AI capabilities.

Conceptually:

```text
FastAPI
   ↓
Business / Service Layer
   ↓
AI Orchestration
   ↓
┌───────────────────────────┐
│ Resume Processing         │
│ Job Analysis              │
│ Embeddings                │
│ Matching                  │
│ Ranking                   │
│ Explanation               │
│ LLM Processing            │
└───────────────────────────┘
```

This keeps the application modular and makes it easier to change or extend individual AI capabilities later.

The backend follows a layered approach.

The request flow is:

```text
Client
  ↓
FastAPI
  ↓
Pydantic
  ↓
Business Services
  ↓
SQLAlchemy
  ↓
PostgreSQL
```

AI services are integrated alongside the business service layer.

This avoids mixing database logic, API handling, and AI processing into one large piece of code.

---

# Database Layer

PostgreSQL is used as the relational database.

SQLAlchemy provides the ORM layer, while Alembic is used for database migrations.

The relational database is responsible for persistent structured application data.

The general architecture is:

```text
Application
    ↓
Business Services
    ↓
SQLAlchemy
    ↓
PostgreSQL
```

The database should remain the source of truth for structured application information.

AI services can process information, but the application should not depend on an LLM to maintain the authoritative state of users, jobs, applications, or other transactional data.

---

# Authentication and Security

Security is treated as a boundary around the application and AI layer.

The security flow is:

```text
User
  ↓
Authentication
  ↓
Authorization
  ↓
AI / Application Endpoint
  ↓
Service
```

Hirely uses JWT authentication and role-based access control concepts.

Password security uses Argon2 hashing.

An important security principle in the architecture is that the AI layer must not bypass the application's existing security model.

The AI service should receive only the information required for its particular task.

Sensitive information should not unnecessarily be sent to external AI providers.

API keys and model credentials should remain on the backend and be supplied through environment variables or secure secret management.

---

# Why the AI Layer Is Kept Separate

I intentionally keep the AI layer separate from ordinary CRUD operations.

Traditional CRUD operations are generally:

```text
Create
Read
Update
Delete
```

AI operations can be more expensive or unpredictable in terms of processing time.

For example:

```text
Resume parsing
Embedding generation
LLM requests
Large-scale candidate matching
Batch candidate ranking
```

Because of this, I don't want expensive AI processing to be tightly coupled with every synchronous API operation.

The architecture can evolve toward:

```text
                    FastAPI
                       |
             ┌─────────┴─────────┐
             ↓                   ↓
      Synchronous APIs     Background AI
                                 |
                                 ↓
                            AI Services
                                 |
                                 ↓
                         LLM / Embeddings
```

As the system grows, technologies such as background workers, queues, caching, and vector databases can be introduced when the actual workload justifies them.

## Q11. What is the difference between deterministic matching and semantic matching?

### Interview Answer

Deterministic matching follows explicit rules.

For example:

    Job requires: Python
    Candidate has: Python

The skill is an exact match.

Semantic matching uses embeddings to compare meaning.

For example:

    Job: Machine Learning Engineer
    Resume: ML Engineer

Even though the wording is different, the meaning can be recognized as similar.

Using both approaches gives us more control and better coverage.

### Memory Trick

**Deterministic = Exact**

**Semantic = Meaning**

### One-Line Answer

Deterministic matching checks explicit rules or exact matches, while semantic matching compares the meaning of the information.
