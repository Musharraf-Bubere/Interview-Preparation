# Question 4: Can you explain your MindSense AI project?

## Answer:

“Sure.

MindSense AI is a machine learning-based mental health risk prediction web application.

### Problem Statement:

The idea behind the project was to use assessment data to identify whether a person may fall into a particular mental health risk category.

The challenge was to take the user's assessment information, process it correctly, use a machine learning model to predict the risk, and also provide an understandable explanation of the prediction.

### Solution:

To solve this, I built an end-to-end machine learning application.

First, I prepared the dataset by removing unnecessary features, handling structural missing values, and converting categorical data into numerical features using binary encoding, ordinal mapping and one-hot encoding.

Then I split the data into training and testing sets and trained a Logistic Regression model for risk classification.

After getting the prediction, I integrated SHAP for Explainable AI. This helps identify which features contributed to the model's prediction, instead of only showing the final result.

I then connected the machine learning workflow with a Flask web application. The application provides risk assessment, analytics, recommendations, assessment history and PDF reports.

SQLite is used to store assessment history.

### My Contribution:

I worked on the machine learning workflow, including data preprocessing, feature engineering, model training and evaluation, and SHAP-based explainability. I also contributed to the application features and data-processing tasks.

### Result:

The Logistic Regression model achieved around 96% accuracy, with 95% precision, 94% recall and 95% F1-score.

Overall, this project gave me practical experience in taking a machine learning model from data preprocessing to deployment in a web application and making the prediction explainable.

# Question 10: What do accuracy, precision, recall, and F1-score mean?

## Answer:

“Accuracy tells us how many predictions the model got correct out of all the predictions.

Precision tells us, out of all the cases the model predicted as positive, how many were actually positive.

Recall tells us, out of all the actual positive cases, how many the model correctly identified.

F1-score is the balance between precision and recall. It is useful when we want to consider both precision and recall together.

In my MindSense AI project, I used these metrics to evaluate the performance of my classification model.”

## 🧠 Memory Trick:

Remember:

**Accuracy → Overall Correct**

**Precision → Predicted Positive → Actually Positive**

**Recall → Actual Positive → Correctly Found**

**F1 → Precision + Recall Balance**

### Easy example:

Imagine the model predicts that 100 people are at high risk.

- **Precision:** How many of those 100 were actually high risk?
- **Recall:** Out of all people who were actually high risk, how many did the model find?
- **Accuracy:** How many total predictions were correct?
- **F1-score:** How well precision and recall are balanced.

### One-line memory:

“Accuracy measures overall correctness, precision measures the correctness of positive predictions, recall measures how many actual positives we identified, and F1-score balances precision and recall.”

# Question 12: What is feature engineering, and what did you do in MindSense AI?

In MindSense AI, I performed feature-related transformations during preprocessing.

# Question 13: Why did you use one-hot encoding for City and Profession?

## Answer:

“I used one-hot encoding because City and Profession are nominal categorical variables.

There is no natural order between categories like Mumbai, Pune or Delhi, and similarly there is no order between professions like Teacher, Accountant or Engineer.

If I directly assigned numbers to these categories, the model might interpret those numbers as having an order or numerical relationship.

So, I used one-hot encoding, which creates separate binary columns for each category.

For example, for City, it creates columns such as `City_Mumbai`, `City_Pune`, and `City_Thane`.

This allows the model to use the categorical information without creating an artificial ranking between the categories.”

## 🧠 Memory Trick:

**Nominal = No Order → One-Hot**

Example:

`Mumbai, Pune, Delhi`

↓

`City_Mumbai | City_Pune | City_Delhi`

**One-line memory:**

“I used one-hot encoding because City and Profession are nominal categories without any natural order.”

# Question 15: How did you handle missing values in MindSense AI?

## Answer:

“I first analyzed the dataset to understand where the missing values were present.

Some missing values were structural rather than random. For example, certain fields were not applicable depending on whether the person was a student or a working professional.

So, instead of blindly filling all missing values with mean or mode, I analyzed the relationship between the columns.

For the Profession column, for example, if the person was a student and Profession was missing, I assigned the value `Student`. If the person was a working professional and Profession was missing, I assigned `Unknown`.

After this processing, there were no missing values remaining in the Profession column.

The main idea was to handle missing values based on the meaning of the data rather than applying the same method to every column.”

## 🧠 Memory Trick:

**Analyze → Understand → Handle → Verify**

For Profession:

**Student + Missing → Student**

**Working Professional + Missing → Unknown**

**Final → No Missing Values**

# Question 16: Why didn't you simply fill all missing values with mean or mode?

## Answer:

“I did not use the same missing-value treatment for every column because missing values can have different meanings.

In my dataset, some missing values were structural. For example, Profession could be missing depending on whether the person was a student or a working professional.
