# MindSense AI — Project Explanation for Interviews

## Project Introduction

MindSense AI is an end-to-end machine learning-based mental health risk prediction web application.

The main idea behind the project was to take assessment information provided by a user, process that information through a machine learning pipeline, predict the user's risk category, and then make the prediction more understandable using Explainable AI.

I did not want the application to behave like a simple machine learning model where the user only receives a prediction.

Instead, I wanted to build a complete application around the model:

```text
User Assessment
      ↓
Data Processing
      ↓
Feature Engineering
      ↓
Machine Learning Model
      ↓
Risk Prediction
      ↓
SHAP Explanation
      ↓
Recommendations / Analytics
      ↓
Assessment History
      ↓
PDF Report
```

So the project combines three major areas:

```text
Machine Learning
      +
Explainable AI
      +
Web Application
```

---

# The Problem I Wanted to Solve

The project was designed around the idea of using assessment data to identify whether a person may fall into a particular mental health risk category.

A machine learning model can produce a prediction, but simply showing a result is not always sufficient.

For example:

```text
Prediction:
High Risk
```

The next question naturally becomes:

```text
Why did the model make this prediction?
Which factors influenced the prediction?
What information contributed positively or negatively?
```

Therefore, explainability became an important part of the project.

The goal was:

> Build an end-to-end machine learning application that can predict mental health risk from assessment data and provide understandable explanations for the model's prediction.

# Step 2 — Data Preprocessing

After understanding the dataset, I prepared it for machine learning.

The preprocessing workflow included:

```text
Raw Dataset
     ↓
Remove Unnecessary Features
     ↓
Handle Structural Missing Values
     ↓
Categorical Feature Processing
     ↓
Numerical Representation
     ↓
Model-ready Dataset
```

I removed features that were not useful for the prediction task and handled structural missing values appropriately.

# Step 3 — Feature Engineering

A major part of the machine learning workflow was converting categorical information into numerical representations that the model could understand.

I used different encoding approaches depending on the nature of the feature.

These included:

```text
Binary Encoding
Ordinal Mapping
One-Hot Encoding
```

For example, a binary feature can be represented as:

```text
Yes → 1
No  → 0
```

For an ordinal feature, where the categories have an inherent order, ordinal mapping can be used.

For nominal categories, one-hot encoding can create separate numerical columns.

Conceptually:

```text
Categorical Data
       ↓
Encoding Strategy
       ↓
Numerical Features
       ↓
Machine Learning Model
```

The important point is that I did not blindly apply the same encoding to every categorical feature.

The encoding approach depends on the meaning of the feature.

# Step 6 — Model Evaluation

After training the model, I evaluated it using classification metrics.

The reported performance was approximately:

```text
Accuracy  → 96%
Precision → 95%
Recall    → 94%
F1 Score  → 95%
```

I would not rely on accuracy alone.

For a classification problem, precision and recall are also important because they tell us different things about the model's behavior.

### Accuracy

Accuracy tells us the proportion of total predictions that were correct.

```text
Accuracy =
Correct Predictions / Total Predictions
```

### Precision

Precision tells us, among the instances predicted as a particular positive class, how many were actually positive.

```text
Precision =
True Positives / (True Positives + False Positives)
```

### Recall

Recall tells us how many of the actual positive cases the model successfully identified.

```text
Recall =
True Positives / (True Positives + False Negatives)
```

### F1 Score

F1 combines precision and recall into a single metric.

```text
F1 =
2 × (Precision × Recall)
------------------------
   Precision + Recall
```

So I looked at multiple metrics instead of depending only on accuracy.

# Step 7 — Explainable AI With SHAP

One of the most important parts of MindSense AI is Explainable AI.

A normal ML application might simply produce:

```text
Prediction → High Risk
```

But that does not tell the user or developer why the model reached that prediction.

So I integrated SHAP.

SHAP stands for:

> SHapley Additive exPlanations

The purpose of SHAP is to explain how individual features contributed to a model prediction.

The conceptual workflow is:

```text
Input Features
      ↓
Machine Learning Model
      ↓
Prediction
      ↓
SHAP
      ↓
Feature Contributions
      ↓
Explanation
```

For example, conceptually:

```text
Feature A → Strong contribution
Feature B → Moderate contribution
Feature C → Low contribution
Feature D → Contribution toward another class
```

This provides much more information than returning only the final prediction.

---

# Why SHAP Was Important in MindSense AI

The reason I included SHAP was because prediction alone is not always enough.

Imagine the application says:

```text
Predicted Risk: High
```

A user may reasonably ask:

```text
Why?
```

SHAP helps answer that question by showing which features contributed to the model's output.

This makes the model more transparent.

The important distinction is:

```text
Machine Learning
→ What did the model predict?

SHAP
→ Which features contributed to that prediction?
```

That was one of the main learning objectives of the project.

# The Most Important Technical Challenge

One of the interesting challenges was taking a machine learning model and turning it into an application that a non-technical user could interact with.

A notebook can easily do:

```text
X → Model → Prediction
```

But an actual application needs much more:

```text
User Input
    ↓
Validation
    ↓
Preprocessing
    ↓
Correct Feature Order
    ↓
Model Prediction
    ↓
Explanation
    ↓
User-friendly Result
    ↓
Storage
```

The preprocessing used during prediction must also remain consistent with the preprocessing used during training.

For example, if a categorical variable was encoded during training, the application must apply the same transformation when a new user submits an assessment.

Otherwise, the model may receive incorrectly represented features.

This taught me that deploying machine learning is not simply about saving a model file.

The entire preprocessing and inference pipeline has to be consistent.

# Why Logistic Regression?

If the interviewer asks why I selected Logistic Regression, I would explain:

I selected Logistic Regression because this is a classification problem and Logistic Regression provides a strong, simple, and relatively interpretable baseline.

It is computationally efficient and works well when the relationship between the features and the log-odds of the target is reasonably appropriate.

It also works well with SHAP-based explanations, which was important because explainability was one of the project requirements.

I would also mention that model selection should ultimately be based on validation and comparison with alternative models rather than assuming that one algorithm is always best.

---

# Why Flask?

I used Flask to expose the machine learning workflow through a web application.

The model by itself is not useful to an end user if it only exists inside a notebook.

Flask provides a lightweight Python web framework that allows the application to:

```text
Receive User Input
       ↓
Run ML Pipeline
       ↓
Generate Prediction
       ↓
Generate Explanation
       ↓
Return Result
```

It was suitable for turning the trained model into an interactive application.

---

# Why SQLite?

SQLite was appropriate for the assessment-history requirement because the project is a lightweight web application and the stored data is relatively straightforward.

It provides persistent storage without requiring a separate database server.

The application can therefore store:

```text
Assessment
Prediction
Assessment History
```

and retrieve the historical information when required.

For a larger production deployment with multiple application instances and higher concurrency, I would consider moving the persistence layer to PostgreSQL.

I selected the encoding method based on whether the categorical feature was binary, ordered, or nominal.

## Q17. What would happen if you used inconsistent preprocessing during prediction?

### Interview Answer

The model expects prediction data to have the same feature representation it saw during training.

If preprocessing is inconsistent, it can cause:

- Incorrect predictions
- Missing or unexpected features
- Shape mismatches
- Encoding problems

Therefore, the same preprocessing logic and feature structure should be applied during both training and inference.

### Memory Trick

**Training preprocessing = Prediction preprocessing**

### One-Line Answer

Inconsistent preprocessing can cause incorrect predictions or errors because the model expects the same feature representation used during training.
