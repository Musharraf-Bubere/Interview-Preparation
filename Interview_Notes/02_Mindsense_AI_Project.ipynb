# Question 4: Can you explain your MindSense AI project?

## Answer:

“Sure.

MindSense AI is a machine learning-based mental health risk prediction web application.

### Problem Statement:

The idea behind the project was to use assessment data to identify whether a person may fall into a particular mental health risk category.

The challenge was to take the user's assessment information, process it correctly, use a machine learning model to predict the risk, and also provide an understandable explanation of the prediction.

### Solution:

To solve this, I built an end-to-end machine learning application.

First, I prepared the dataset by removing unnecessary features, handling structural missing values, and converting categorical data into numerical features using binary encoding, ordinal mapping and one-hot encoding.

Then I split the data into training and testing sets and trained a Logistic Regression model for risk classification.

After getting the prediction, I integrated SHAP for Explainable AI. This helps identify which features contributed to the model's prediction, instead of only showing the final result.

I then connected the machine learning workflow with a Flask web application. The application provides risk assessment, analytics, recommendations, assessment history and PDF reports.

SQLite is used to store assessment history.

### My Contribution:

I worked on the machine learning workflow, including data preprocessing, feature engineering, model training and evaluation, and SHAP-based explainability. I also contributed to the application features and data-processing tasks.

### Result:

The Logistic Regression model achieved around 96% accuracy, with 95% precision, 94% recall and 95% F1-score.

Overall, this project gave me practical experience in taking a machine learning model from data preprocessing to deployment in a web application and making the prediction explainable.

## Q10. Why did you use precision, recall, and F1-score instead of only accuracy?

### Interview Answer

Accuracy alone may not provide enough information, especially when class distributions are uneven or when different types of errors have different importance.

Precision tells us how many predicted positive cases were actually positive.

Recall tells us how many actual positive cases the model successfully identified.

F1-score combines precision and recall into a single metric.

So using multiple metrics gives a more complete view of model performance.

### Memory Trick

**Precision → Prediction correctness**

**Recall → Detection ability**

**F1 → Balance**

### One-Line Answer

I used multiple metrics because accuracy alone does not fully describe classification performance.

## Q11. What is SHAP?

### Interview Answer

SHAP stands for **SHapley Additive exPlanations**.

It is an Explainable AI technique used to understand how individual features contribute to a model prediction.

It can help answer questions such as:

- Which features influenced the prediction?
- Did a feature increase or decrease the prediction?
- Which features were most important?

### Memory Trick

**SHAP = Why did the model make this prediction?**

### One-Line Answer

SHAP is an Explainable AI technique used to understand how features contribute to individual model predictions.

## Q13. How does SHAP work at a high level?

### Interview Answer

At a high level, SHAP assigns contribution values to features for a particular prediction.

A positive contribution can indicate that a feature pushed the prediction in one direction, while a negative contribution can indicate the opposite direction, depending on the model output being explained.

This allows us to understand the contribution of individual features to a prediction.

### Memory Trick

**Feature → Contribution → Prediction**

### One-Line Answer

SHAP explains a prediction by estimating how individual features contributed to the model output.

## Q17. What would happen if you used inconsistent preprocessing during prediction?

### Interview Answer

The model expects prediction data to have the same feature representation it saw during training.

If preprocessing is inconsistent, it can cause:

- Incorrect predictions
- Missing or unexpected features
- Shape mismatches
- Encoding problems

Therefore, the same preprocessing logic and feature structure should be applied during both training and inference.

### Memory Trick

**Training preprocessing = Prediction preprocessing**

### One-Line Answer

Inconsistent preprocessing can cause incorrect predictions or errors because the model expects the same feature representation used during training.
