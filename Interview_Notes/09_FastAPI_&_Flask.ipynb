# ⚡ FastAPI & Flask — Interview Questions

---

## Q1. What is FastAPI?

### Interview Answer

FastAPI is a modern Python web framework used to build APIs and backend applications.

It is built on **Starlette** for web functionality and **Pydantic** for data validation and serialization.

FastAPI supports asynchronous programming and provides automatic API documentation.

### Memory Trick

**FastAPI = Python + APIs + Validation + Async + Documentation**

### One-Line Answer

FastAPI is a modern Python framework for building fast, validated, and well-documented APIs.

---

## Q2. What is Flask?

### Interview Answer

Flask is a lightweight Python web framework used to build web applications and APIs.

It follows a minimal and flexible approach, allowing developers to add the components they need.

### Memory Trick

**Flask = Lightweight + Flexible**

### One-Line Answer

Flask is a lightweight and flexible Python web framework used for building web applications and APIs.

---

## Q3. What is the difference between FastAPI and Flask?

### Interview Answer

Both are Python web frameworks, but they have different design approaches.

**FastAPI:**
- Designed strongly around APIs
- Uses type hints
- Provides automatic validation through Pydantic
- Supports asynchronous endpoints
- Provides automatic API documentation

**Flask:**
- Lightweight and minimal
- Flexible
- Has a large ecosystem
- Requires additional libraries for many API-related features

### Memory Trick

**FastAPI → API-focused + validation + async**

**Flask → Minimal + flexible**

### One-Line Answer

FastAPI provides more built-in API-oriented features, while Flask is a lightweight and flexible framework that gives developers more freedom in choosing components.

## Q4. What is a route or endpoint?

### Interview Answer

A route or endpoint defines a URL and the operation that should be performed when a client sends a request to that URL.

### Example

    @app.get("/users")
    def get_users():
        return {"users": []}

Here:

    /users

is the endpoint.

### Memory Trick

**Endpoint = URL + Operation**

### One-Line Answer

An endpoint is a URL path associated with a specific operation in an API.

---

## Q5. What is a route decorator in FastAPI?

### Interview Answer

A route decorator connects an HTTP method and URL path to a Python function.

Examples include:

    @app.get("/users")
    @app.post("/users")
    @app.put("/users/{id}")
    @app.delete("/users/{id}")

When a request matches the method and path, FastAPI executes the associated function.

### Memory Trick

**Decorator = Connect route to function**

### One-Line Answer

A FastAPI route decorator maps an HTTP method and URL path to a Python function.

## Q6. What are Pydantic models and why are they used in FastAPI?

### Interview Answer

Pydantic models are used to define the expected structure and types of data.

FastAPI uses Pydantic for request validation and data serialization.

### Example

    from pydantic import BaseModel

    class User(BaseModel):
        name: str
        age: int

If an API expects a `User`, FastAPI can validate incoming data against this model.

### Memory Trick

**Pydantic = Structure + Validation**

### One-Line Answer

Pydantic models define and validate structured data used by FastAPI APIs.

## Q9. What is asynchronous programming in FastAPI?

### Interview Answer

Asynchronous programming allows an application to perform certain I/O-bound operations without blocking the execution flow while waiting for them to complete.

FastAPI supports asynchronous endpoints using `async` and `await`.

### Example

    @app.get("/data")
    async def get_data():
        result = await some_async_operation()
        return result

It can be useful when handling operations such as network requests or other asynchronous I/O.

### Memory Trick

**async = Don't block while waiting**

### One-Line Answer

Asynchronous programming allows FastAPI to efficiently handle suitable I/O-bound operations without blocking while waiting.

---

## Q10. What is the difference between `def` and `async def` in FastAPI?

### Interview Answer

`def` defines a normal synchronous function.

`async def` defines an asynchronous function that can use `await` for asynchronous operations.

The choice depends on the libraries and operations being used.

If the operation is synchronous, using `async def` does not automatically make that operation asynchronous.

### Memory Trick

**def → Synchronous**

**async def → Asynchronous**

### One-Line Answer

`def` creates a synchronous endpoint, while `async def` creates an asynchronous endpoint that can await asynchronous operations.

## Q11. What is dependency injection in FastAPI?

### Interview Answer

Dependency injection is a mechanism that allows an endpoint to receive reusable functionality or resources without creating them directly inside the endpoint.

FastAPI provides `Depends()` for dependency injection.

It can be used for:

- Authentication
- Database sessions
- Common parameters
- Reusable logic

### Example

    from fastapi import Depends

    def get_current_user():
        return {"name": "User"}

    @app.get("/profile")
    def profile(user=Depends(get_current_user)):
        return user

### Memory Trick

**Dependency Injection = Give required functionality automatically**

### One-Line Answer

FastAPI dependency injection allows reusable dependencies such as authentication or database sessions to be provided to endpoints automatically.

## Q12. What is middleware?

It can be used for cross-cutting concerns such as:

- Logging
- Authentication
- CORS
- Request processing
- Response processing

### Flow

    Client
       ↓
    Middleware
       ↓
    Endpoint
       ↓
    Response
       ↓
    Middleware
       ↓
    Client

### Memory Trick

**Middleware = Layer around requests**

### One-Line Answer

Middleware processes requests and responses before or after they reach the endpoint.

## Q13. What is CORS?

### Interview Answer

CORS stands for **Cross-Origin Resource Sharing**.

It is a browser security mechanism that controls whether a web application from one origin can access resources from another origin.

For example, a frontend running on one domain or port may need permission to call an API running on another origin.

### Memory Trick

**CORS = Cross-Origin Permission**

### One-Line Answer

CORS controls browser-based access to resources across different origins.

## Q14. How do you return a custom status code in FastAPI?

### Interview Answer

FastAPI allows us to specify an appropriate HTTP status code for an endpoint.

### Example

    from fastapi import FastAPI, status

    app = FastAPI()

    @app.post("/users", status_code=status.HTTP_201_CREATED)
    def create_user():
        return {"message": "User created"}

The endpoint returns `201 Created` for successful creation.

### Memory Trick

**status_code = Tell client the result**

### One-Line Answer

FastAPI allows developers to specify appropriate HTTP status codes for API responses.

## Q15. How do you handle exceptions in FastAPI?

### Interview Answer

FastAPI provides `HTTPException` for returning controlled HTTP error responses.

### Example

    from fastapi import HTTPException

    @app.get("/users/{user_id}")
    def get_user(user_id: int):

        if user_id <= 0:
            raise HTTPException(
                status_code=400,
                detail="Invalid user ID"
            )

This allows the API to return a meaningful error response.

### Memory Trick

**HTTPException = Controlled API error**

### One-Line Answer

FastAPI can use `HTTPException` to return controlled errors with appropriate HTTP status codes and messages.

## Q16. What is the difference between request validation and business validation?

### Interview Answer

**Request validation** checks whether the incoming data has the correct structure and basic types.

**Business validation** checks whether the data follows application-specific rules.

### Example

Request validation:

    age must be an integer

Business validation:

    age must be at least 18 to create an account

### Memory Trick

**Request validation = Is the data correctly formed?**

**Business validation = Is the data logically allowed?**

### One-Line Answer

Request validation checks data structure and types, while business validation checks application-specific rules.

### One-Line Answer

A maintainable FastAPI application usually separates routes, schemas, business logic, database models, and supporting utilities.

## Q18. Why should business logic not be placed entirely inside API routes?

### Interview Answer

Putting all business logic directly inside routes can make the application difficult to:

- Maintain
- Test
- Reuse
- Debug

A better approach is to keep routes focused on handling HTTP requests and move business logic into service or domain layers.

### Memory Trick

**Route = HTTP handling**

**Service = Business logic**

### One-Line Answer

Separating business logic from routes improves maintainability, testing, reuse, and debugging.

## Q20. What is the difference between request models and response models?

### Interview Answer

A request model defines the data that the API expects from the client.

A response model defines the data that the API returns to the client.

Separating them improves control over the API contract and helps prevent accidentally exposing internal fields.

### Example

Request:

    {
        "name": "Musharraf",
        "password": "secret"
    }

Response:

    {
        "name": "Musharraf"
    }

The password should not be returned.

### Memory Trick

**Request Model → Input**

**Response Model → Output**

### One-Line Answer

Request models validate incoming data, while response models define and control outgoing data.

## Q21. What is ASGI and why is it relevant to FastAPI?

### Interview Answer

ASGI stands for **Asynchronous Server Gateway Interface**.

It is a standard interface between asynchronous Python web applications and web servers.

FastAPI applications commonly run through an ASGI server such as Uvicorn.

### Flow

    Client
       ↓
    Web Server
       ↓
    ASGI Server
       ↓
    FastAPI Application
       ↓
    Endpoint

### Memory Trick

**ASGI = Interface for modern async Python web apps**

### One-Line Answer

ASGI is a standard interface for Python asynchronous web applications, and FastAPI commonly runs through an ASGI server such as Uvicorn.

## Q23. How would you troubleshoot a FastAPI endpoint returning a 500 error?

### Interview Answer

I would follow a systematic approach:

1. Reproduce the issue.
2. Check the request data.
3. Check the API response and status code.
4. Inspect application logs and traceback.
5. Identify the failing function.
6. Check database or external API dependencies.
7. Verify configuration and environment variables.
8. Identify the root cause.
9. Apply the fix.
10. Retest the endpoint.

### Example

If an endpoint worked previously but suddenly returns `500`, I would first check the traceback and logs to identify whether the failure is in request processing, business logic, database access, or an external dependency.

### Memory Trick

**Reproduce → Logs → Root Cause → Fix → Test**

### One-Line Answer

I would reproduce the issue, inspect logs and traceback, identify the failing component and root cause, fix it, and retest.

## Q25. When would you choose FastAPI over Flask, and when would you choose Flask?

### Interview Answer

I would choose FastAPI when I am primarily building APIs and want features such as:

- Type-based validation
- Pydantic models
- Automatic documentation
- Async support

I would consider Flask when I need a lightweight and highly flexible framework and want to choose additional components myself.

The decision should depend on the application's requirements rather than simply choosing one framework as universally better.

### Memory Trick

**FastAPI → API-focused**

**Flask → Minimal and flexible**

### One-Line Answer

I would choose FastAPI for API-focused applications needing built-in validation and documentation, while Flask is useful when I want a lightweight and flexible framework.
