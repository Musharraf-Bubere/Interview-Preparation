# Power BI — Comprehensive Interview Questions & Answers

> **Level:** Basic → Intermediate → Advanced → Expert  
> **Format:** Interview Answer → Detailed Explanation → Example / Diagram → Real-World Use Case → Memory Trick → One-Line Interview Answer

# 🟢 BASIC LEVEL

## Q1. What is Power BI?

### 🎤 Interview Answer

Power BI is Microsoft's business intelligence and data visualization platform used to connect to data sources, clean and transform data, build data models, create interactive reports and dashboards, and share insights with users.

A typical Power BI workflow is:

```text
Data Sources
    ↓
Power Query
    ↓
Data Model
    ↓
DAX
    ↓
Visualizations
    ↓
Report
    ↓
Power BI Service
```

### 🌍 Real-World Use Case

A company can connect sales data from Excel and databases, transform it, create KPIs such as revenue and profit, and provide management with an interactive sales dashboard.

### 🧠 Memory Trick

**Connect → Clean → Model → Calculate → Visualize → Share**

### ⚡ One-Line Interview Answer

> Power BI is a business intelligence platform used to connect, transform, model, analyze, visualize, and share data.

---

## Q2. What are the main components of Power BI?

### 🎤 Interview Answer

The main components I commonly work with are Power BI Desktop, Power BI Service, Power Query, the Power BI data model, DAX, gateways, and Power BI mobile applications.

### 📚 Detailed Explanation

```text
Power BI Desktop
    ↓
Build Model + Report
    ↓
Power BI Service
    ↓
Publish + Share + Refresh
```

Power Query handles data transformation, while DAX handles calculations in the model.

### ⚡ One-Line Interview Answer

> Power BI consists of tools for data preparation, modeling, analysis, visualization, publishing, sharing, and monitoring.

---

## Q3. What is Power BI Desktop?

### 🎤 Interview Answer

Power BI Desktop is the development application used to connect to data, transform data using Power Query, create relationships and data models, write DAX calculations, and build reports.

### Diagram

```text
Power BI Desktop
├── Get Data
├── Power Query
├── Model View
├── DAX
└── Report View
```

### ⚡ One-Line Interview Answer

> Power BI Desktop is primarily used to develop Power BI data models and reports.

---

## Q4. What is Power BI Service?

### 🎤 Interview Answer

Power BI Service is the cloud-based environment used to publish, share, collaborate on, refresh, and manage Power BI content.

### Diagram

```text
Power BI Desktop
       ↓ Publish
Power BI Service
       ↓
Workspace
       ↓
Reports / Dashboards / Semantic Models
```

### ⚡ One-Line Interview Answer

> Power BI Service is the cloud platform used to publish, share, manage, and consume Power BI content.

---

## Q5. What is the difference between a report and a dashboard?

### 🎤 Interview Answer

A Power BI report is generally a multi-page interactive analytical document, while a dashboard is a single-page canvas in Power BI Service that can contain pinned tiles and provide a high-level view of important metrics.

### Comparison

```text
Report
→ Multiple pages
→ Detailed analysis
→ Interactive visuals

Dashboard
→ Single page
→ High-level monitoring
→ Power BI Service
```

### ⚡ One-Line Interview Answer

> Reports are designed for detailed interactive analysis, while dashboards provide a consolidated high-level view.

---

## Q6. What is a semantic model in Power BI?

### 🎤 Interview Answer

A semantic model is the structured analytical model containing tables, relationships, columns, measures, and metadata that Power BI uses to answer analytical queries.

### Diagram

```text
Source Data
    ↓
Tables
    ↓
Relationships
    ↓
Measures
    ↓
Semantic Model
    ↓
Reports
```

### ⚡ One-Line Interview Answer

> A semantic model organizes business data and calculations so reports can analyze it consistently.

---

## Q7. What is Power Query?

### 🎤 Interview Answer

Power Query is Power BI's data preparation and transformation engine. It is used to connect to data sources, clean data, transform columns, merge or append datasets, and prepare data before loading it into the model.

### Diagram

```text
Excel / CSV / SQL / API
          ↓
     Power Query
          ↓
   Clean + Transform
          ↓
      Data Model
```

### 🌍 Real-World Use Case

If monthly sales files arrive with inconsistent column names and extra rows, Power Query can standardize and clean them automatically during refresh.

### ⚡ One-Line Interview Answer

> Power Query is used for connecting to, cleaning, transforming, and loading data.

---

## Q8. What is DAX?

### 🎤 Interview Answer

DAX, or Data Analysis Expressions, is the formula language used in Power BI for creating calculated columns, calculated tables, and especially measures.

### 💻 Example

```DAX
Total Sales = SUM(Sales[SalesAmount])
```

### ⚡ One-Line Interview Answer

> DAX is the formula language used to create calculations in the Power BI data model.

---

## Q9. What is a measure?

### 🎤 Interview Answer

A measure is a DAX calculation evaluated dynamically according to the current filter context.

### 💻 Example

```DAX
Total Sales = SUM(Sales[SalesAmount])
```

If a report is filtered to Mumbai, the measure calculates sales for Mumbai.

### 🧠 Memory Trick

**Measure = Dynamic KPI**

### ⚡ One-Line Interview Answer

> A measure is a dynamic DAX calculation evaluated according to the current filter context.

---

## Q10. What is a calculated column?

### 🎤 Interview Answer

A calculated column is a DAX expression evaluated row by row and stored as a column in the model.

### 💻 Example

```DAX
Profit =
Sales[SalesAmount] - Sales[Cost]
```

### Comparison

```text
Calculated Column → Row by row → Stored
Measure           → Context based → Calculated dynamically
```

### ⚡ One-Line Interview Answer

> A calculated column performs row-level calculations and stores the resulting values in the model.

---

## Q11. What is a relationship in Power BI?

### 🎤 Interview Answer

A relationship connects tables using related columns so that filters and calculations can propagate between them.

### Example

```text
Customers
CustomerID
    │
    │
    ↓
Sales
CustomerID
```

### 🌍 Real-World Use Case

A customer slicer can filter sales transactions through the relationship between Customers and Sales.

### ⚡ One-Line Interview Answer

> A relationship connects tables and allows filters to flow between related data.

---

## Q12. What is a primary key and foreign key?

### 🎤 Interview Answer

A primary key uniquely identifies records in a table. A foreign key stores the corresponding key used to connect that table to another table.

### Example

```text
DimCustomer
CustomerID ← Primary Key

FactSales
CustomerID ← Foreign Key
```

### ⚡ One-Line Interview Answer

> A primary key uniquely identifies a record, while a foreign key connects it to another table.

---

## Q13. What is a star schema?

### 🎤 Interview Answer

A star schema contains a central fact table connected to surrounding dimension tables.

### Diagram

```text
             DimDate
                |
DimProduct — FactSales — DimCustomer
                |
           DimLocation
```

The fact table contains measurable business events, while dimensions provide descriptive information.

### 🌍 Real-World Use Case

Sales analytics can use FactSales with dimensions such as Date, Product, Customer, and Region.

### 🧠 Memory Trick

**Fact in center, Dimensions around it**

### ⚡ One-Line Interview Answer

> A star schema uses a central fact table surrounded by related dimension tables.

---

## Q14. What is a fact table?

### 🎤 Interview Answer

A fact table stores business events and measurable values, usually at a defined grain.

### Example

```text
FactSales
---------
DateKey
ProductKey
CustomerKey
Quantity
SalesAmount
Cost
```

### ⚡ One-Line Interview Answer

> A fact table stores transactional or measurable business events.

---

## Q15. What is a dimension table?

### 🎤 Interview Answer

A dimension table contains descriptive attributes used to categorize, filter, and analyze facts.

### Example

```text
DimProduct
----------
ProductKey
ProductName
Category
Brand
```

### ⚡ One-Line Interview Answer

> A dimension table provides descriptive context for analyzing fact data.

# 🟡 INTERMEDIATE LEVEL

## Q16. What is filter context in DAX?

### 🎤 Interview Answer

Filter context is the set of filters applied to a DAX calculation by slicers, filters, rows, columns, relationships, or other calculations.

### Example

If a report has:

```text
Region = Mumbai
Year = 2026
```

then:

```DAX
Total Sales = SUM(Sales[SalesAmount])
```

is evaluated only within that context.

### Diagram

```text
Report Filters
     ↓
Filter Context
     ↓
Measure
     ↓
Result
```

### 🧠 Memory Trick

**Filter Context = What data is currently allowed**

### ⚡ One-Line Interview Answer

> Filter context determines which data is visible to a DAX calculation at evaluation time.

---

## Q17. What is row context?

### 🎤 Interview Answer

Row context means that DAX is evaluating an expression for the current row, most commonly in calculated columns and iterator functions.

### Example

```DAX
Profit =
Sales[SalesAmount] - Sales[Cost]
```

For each row, DAX uses that row's SalesAmount and Cost.

### ⚡ One-Line Interview Answer

> Row context represents the current row being evaluated by DAX.

---

## Q18. What is CALCULATE?

### 🎤 Interview Answer

CALCULATE is one of the most important DAX functions because it evaluates an expression after modifying the filter context.

### Example

```DAX
Mumbai Sales =
CALCULATE(
    [Total Sales],
    Customer[City] = "Mumbai"
)
```

### Diagram

```text
Existing Filter Context
          ↓
      CALCULATE
          ↓
Modified Filter Context
          ↓
       Result
```

### 🌍 Real-World Use Case

Calculating sales for a particular region, product category, or business condition.

### 🧠 Memory Trick

**CALCULATE = Change Context → Calculate**

### ⚡ One-Line Interview Answer

> CALCULATE evaluates an expression in a modified filter context.

---

## Q19. What is the difference between row context and filter context?

### 🎤 Interview Answer

Row context refers to the current row being evaluated, whereas filter context represents the set of rows currently allowed by filters.

### Comparison

```text
Row Context
→ Current row
→ Common in calculated columns
→ Iterators create row context

Filter Context
→ Set of allowed rows
→ Slicers / filters / relationships
→ Critical for measures
```

### ⚡ One-Line Interview Answer

> Row context is about the current row, while filter context determines the data included in a calculation.

---

## Q20. What are iterator functions in DAX?

### 🎤 Interview Answer

Iterator functions evaluate an expression row by row over a table and then aggregate or return the result.

Common examples include:

```text
SUMX
AVERAGEX
COUNTX
FILTER
MAXX
MINX
```

### Example

```DAX
Revenue =
SUMX(
    Sales,
    Sales[Quantity] * Sales[UnitPrice]
)
```

### ⚡ One-Line Interview Answer

> Iterator functions evaluate an expression for each row and then produce an aggregated or table result.

---

## Q21. What is SUMX and how is it different from SUM?

### 🎤 Interview Answer

SUM directly adds values from a column, while SUMX iterates through a table and evaluates an expression for each row before summing the results.

### Example

```DAX
Total Amount = SUM(Sales[Amount])
```

versus:

```DAX
Total Revenue =
SUMX(
    Sales,
    Sales[Quantity] * Sales[UnitPrice]
)
```

### 🧠 Memory Trick

**SUM = Existing column | SUMX = Calculate each row then sum**

### ⚡ One-Line Interview Answer

> SUM adds a column, while SUMX evaluates a row-level expression and then sums the results.

---

## Q22. What is the difference between ALL and ALLSELECTED?

### 🎤 Interview Answer

ALL can remove filters from a table or column, while ALLSELECTED generally preserves the broader external selection context while removing filters introduced within the current visual context.

### Use Case

ALL is often used for:

```text
Grand total
Percentage of total
```

ALLSELECTED is useful when a calculation should respect the user's broader selections.

### ⚡ One-Line Interview Answer

> ALL removes specified filters, while ALLSELECTED preserves relevant outer selections.

---

## Q23. What is REMOVEFILTERS?

### 🎤 Interview Answer

REMOVEFILTERS is a DAX function used to remove filters from specified tables or columns.

### Example

```DAX
Total All Products =
CALCULATE(
    [Total Sales],
    REMOVEFILTERS(Product)
)
```

### ⚡ One-Line Interview Answer

> REMOVEFILTERS explicitly clears filters from specified tables or columns.

---

## Q24. What is RELATED?

### 🎤 Interview Answer

RELATED retrieves a value from a related table when there is an appropriate relationship and row context.

### Example

If Sales is related to Product:

```DAX
Category =
RELATED(Product[Category])
```

### ⚡ One-Line Interview Answer

> RELATED retrieves a value from a related table using an existing relationship.

## Q25. What is RELATEDTABLE?

### 🎤 Interview Answer

RELATEDTABLE returns the related rows from another table and is commonly used when navigating from a dimension row to related fact rows.

### Concept

```text
Customer
   ↓
RELATEDTABLE(Sales)
   ↓
All Sales Rows for Customer
```

### ⚡ One-Line Interview Answer

> RELATEDTABLE returns rows related to the current row through an existing relationship.

## Q26. What is Power Query Merge vs Append?

### 🎤 Interview Answer

Merge combines tables horizontally based on matching columns, while Append combines tables vertically by stacking rows.

### Diagram

```text
MERGE
Table A + Table B
      ↓
More columns

APPEND
Table A
  +
Table B
  ↓
More rows
```

### Example

Merge:

```text
CustomerID → Customer Details
```

Append:

```text
January Sales
February Sales
March Sales
      ↓
All Sales
```

### 🧠 Memory Trick

**Merge = Join | Append = Stack**

### ⚡ One-Line Interview Answer

> Merge joins tables using matching fields, while Append stacks tables vertically.

---

## Q27. What is query folding?

### 🎤 Interview Answer

Query folding occurs when Power Query pushes supported transformations back to the source system so that the source performs the work.

### Diagram

```text
Power Query
    ↓
Transformation
    ↓
Can source execute it?
    ↓
Yes → Source performs operation
No  → Power Query performs it
```

### 🌍 Real-World Use Case

Filtering millions of SQL Server rows at the source can reduce the amount of data transferred into Power BI.

### ⚡ One-Line Interview Answer

> Query folding pushes compatible Power Query transformations to the source system for better performance.

---

## Q28. What is Import mode?

### 🎤 Interview Answer

Import mode loads data into Power BI's in-memory storage engine, allowing fast interactive analysis.

### Advantages

- Fast report interaction
- Rich modeling capabilities
- Strong DAX support

### Trade-off

Data must be refreshed to reflect changes in the source.

### ⚡ One-Line Interview Answer

> Import mode loads data into the Power BI model for fast in-memory analysis.

---

## Q29. What is DirectQuery?

### 🎤 Interview Answer

DirectQuery keeps data in the source system and sends queries to that source when users interact with the report.

### Comparison

```text
Import
Source → Power BI Model → Query

DirectQuery
Source ← Query from Power BI
```

### Advantages

- Can work with very large source datasets
- Data can remain in the source

### Trade-offs

- Source performance matters
- Query performance can be slower
- Some modeling features have limitations

### ⚡ One-Line Interview Answer

> DirectQuery queries the source system rather than importing the full dataset into the Power BI model.

---

> A composite model combines different storage modes to meet specific modeling and performance requirements.

## Q36. What is time intelligence in Power BI?

### 🎤 Interview Answer

Time intelligence refers to DAX calculations that analyze business metrics across time, such as year-to-date, previous year, month-to-date, and year-over-year growth.

### Examples

```DAX
Sales YTD =
TOTALYTD(
    [Total Sales],
    'Date'[Date]
)
```

```DAX
Sales LY =
CALCULATE(
    [Total Sales],
    SAMEPERIODLASTYEAR('Date'[Date])
)
```

### ⚡ One-Line Interview Answer

> Time intelligence enables calculations that compare and aggregate business metrics across time periods.

## Q37. Why do we need a Date table?

### 🎤 Interview Answer

A dedicated Date table provides a consistent calendar structure for time intelligence and date-based analysis.

### Example

```text
DimDate
-------
Date
Year
Quarter
Month
Month Number
Week
Day
```

It can be related to the fact table's date column.

### ⚡ One-Line Interview Answer

> A Date table provides a reliable calendar dimension for time-based analysis and DAX time intelligence.

## Q38. What is a measure for Year-over-Year growth?

### 🎤 Interview Answer

A common approach is to calculate current-year sales, previous-year sales, and then calculate the percentage change.

### Example

```DAX
Sales LY =
CALCULATE(
    [Total Sales],
    SAMEPERIODLASTYEAR('Date'[Date])
)
```

```DAX
YoY Growth % =
DIVIDE(
    [Total Sales] - [Sales LY],
    [Sales LY]
)
```

### ⚡ One-Line Interview Answer

> YoY growth compares the current period with the equivalent previous-year period and expresses the difference as a percentage.

## Q43. What is cardinality?

### 🎤 Interview Answer

Cardinality describes the number of unique values in a column and is also used when describing relationship types.

For example:

```text
CustomerID
1001
1002
1003
```

has relatively high uniqueness compared with a column such as:

```text
Gender
Male
Female
```

### 🌍 Real-World Use Case

High-cardinality columns can increase model size and affect compression and performance.

### ⚡ One-Line Interview Answer

> Cardinality describes the uniqueness of values in a column and is important for both relationships and model performance.

### 🎤 Interview Answer

Cross-filter direction determines how filters propagate through relationships.

Many-to-many relationships can introduce ambiguous filter paths and unexpected results if the model is not designed carefully.

> Many-to-many relationships require careful modeling because they can create ambiguity and unexpected filter behavior.

# 🔴 EXPERT LEVEL

## Q51. Explain the Power BI end-to-end architecture.

### 🎤 Interview Answer

A typical Power BI solution starts with source systems such as Excel, SQL Server, APIs, or cloud platforms.

Power Query connects to and transforms the data. The transformed data is loaded into a semantic model containing tables, relationships, and DAX measures. Reports are then created on top of the model and published to Power BI Service, where users can consume the reports and dashboards.

### Architecture

```text
Data Sources
   │
   ├── SQL Server
   ├── Excel
   ├── CSV
   ├── APIs
   └── Cloud Sources
          ↓
     Power Query
          ↓
    Data Preparation
          ↓
    Semantic Model
     ┌──────────────┐
     │ Facts        │
     │ Dimensions   │
     │ Relationships│
     │ Measures     │
     └──────────────┘
          ↓
       Reports
          ↓
   Power BI Service
          ↓
 Users / Dashboards
```

### ⚡ One-Line Interview Answer

> Power BI follows a flow from data sources through Power Query and the semantic model to reports and Power BI Service.

---

## Q52. How would you design an enterprise Power BI data model?

### 🎤 Interview Answer

I would begin with the business requirements and define the grain of each fact table. Then I would identify dimensions, keys, relationships, and required measures.

I would generally prefer a star schema, minimize unnecessary many-to-many relationships, use a dedicated Date dimension, and keep calculations in reusable measures where appropriate.

### Architecture

```text
                Date
                 │
Product ──── Sales Fact ─── Customer
                 │
              Region
                 │
             Employee
```

### ⚡ One-Line Interview Answer

> I would design the model around business grain, facts, dimensions, relationships, reusable measures, and a clean star schema.

## Q53. How would you troubleshoot a slow Power BI report?

### 🎤 Interview Answer

I would investigate the issue systematically rather than immediately rewriting all DAX.

### Process

```text
Slow Report
    ↓
Identify Slow Visual
    ↓
Check Query Performance
    ↓
Inspect DAX
    ↓
Inspect Model
    ↓
Inspect Power Query
    ↓
Inspect Source
    ↓
Optimize
    ↓
Retest
```

I would also check the number of visuals, high-cardinality columns, relationship design, unnecessary columns, storage mode, query folding, and source performance.

### ⚡ One-Line Interview Answer

> I troubleshoot Power BI performance layer by layer: visuals, DAX, model, Power Query, and source.

## Q56. How do you reduce Power BI model size?

### 🎤 Interview Answer

I would remove unnecessary columns and rows, use appropriate data types, reduce unnecessary high-cardinality fields, aggregate data where detailed granularity is not required, and ensure Power Query performs efficient transformations.

### Checklist

```text
Remove unused columns
Remove unused rows
Reduce cardinality
Use correct data types
Avoid unnecessary calculated columns
Optimize model relationships
Use aggregations where appropriate
```

### ⚡ One-Line Interview Answer

> I reduce model size by minimizing unnecessary data, cardinality, columns, and inefficient calculations.

## Q57. What is the difference between a calculated column and a measure from a performance perspective?

### 🎤 Interview Answer

Calculated columns are materialized in the model and can increase model size, while measures are calculated at query time according to context.

Therefore, when a value can be calculated dynamically as a measure, a measure is often preferable for analytical KPIs.

However, calculated columns are appropriate when a row-level attribute is needed for grouping, filtering, or relationships.

### ⚡ One-Line Interview Answer

> Measures are generally preferred for dynamic analytical calculations, while calculated columns are useful for persistent row-level attributes.

---

## Q58. How would you handle multiple date columns in a fact table?

### 🎤 Interview Answer

If a fact table contains multiple business dates such as Order Date, Ship Date, and Delivery Date, I would model the date relationships carefully.

One approach is to use a role-playing Date dimension concept, where the same date dimension serves different business roles, while ensuring the model's active and inactive relationships are handled correctly.

### Example

```text
Date
 │
 ├── Order Date
 ├── Ship Date
 └── Delivery Date
```

### ⚡ One-Line Interview Answer

> Multiple date roles require careful relationship design so time-based calculations use the intended business date.

---

## Q59. What is USERELATIONSHIP?

### 🎤 Interview Answer

USERELATIONSHIP activates an existing inactive relationship during a calculation.

### Example

Suppose Order Date is active and Ship Date is inactive:

```DAX
Sales by Ship Date =
CALCULATE(
    [Total Sales],
    USERELATIONSHIP(
        Sales[ShipDate],
        'Date'[Date]
    )
)
```

### ⚡ One-Line Interview Answer

> USERELATIONSHIP activates an inactive relationship for a specific DAX calculation.

---

## Q60. How would you implement a reusable KPI framework?

### 🎤 Interview Answer

I would create a consistent semantic layer with reusable base measures and derive business KPIs from those measures.

### Example

```DAX
Total Sales = SUM(Sales[Amount])

Total Cost = SUM(Sales[Cost])

Profit = [Total Sales] - [Total Cost]

Profit Margin =
DIVIDE([Profit], [Total Sales])
```

This is better than repeating complex logic inside every visual.

### ⚡ One-Line Interview Answer

> I build reusable base measures and derive higher-level KPIs from them to keep the model consistent and maintainable.

## Q70. How would you explain Power BI to an interviewer in one complete story?

### 🎤 Interview Answer

Power BI is an end-to-end business intelligence platform.

I see the process as starting with business requirements and source data. I first understand what the business needs to measure and then connect Power BI to sources such as SQL databases, Excel files, APIs, or cloud systems.

I use Power Query to clean and transform the data. After that, I build a semantic model with appropriate fact and dimension tables, relationships, and a star-schema structure where appropriate.

For business calculations, I use DAX measures. I pay particular attention to filter context, row context, CALCULATE, time intelligence, and reusable measures.

Then I build interactive reports using appropriate visuals, slicers, drill-downs, and drill-through pages.

Before publishing, I validate the numbers against trusted source calculations and test filters, edge cases, performance, and security.

Finally, I publish the report to Power BI Service, configure refresh and gateway requirements where needed, apply Row-Level Security if required, and make the report available to business users.
