# ⚠️ Exception Handling in Python — Interview Questions

---

## Q1. What is Exception Handling in Python?

### Interview Answer

Exception handling is a mechanism used to handle runtime errors without abruptly stopping the program.

Python provides keywords such as:

- `try`
- `except`
- `else`
- `finally`
- `raise`

We can use them to detect and handle exceptions properly.

### Example

    try:
        result = 10 / 0
    except ZeroDivisionError:
        print("Cannot divide by zero")

The program handles the error instead of terminating unexpectedly.

### Memory Trick

**Exception Handling = Detect → Handle → Continue**

### One-Line Answer

Exception handling allows us to handle runtime errors gracefully without unexpectedly terminating the program.

## Q2. What is an Exception?

### Interview Answer

An exception is an event that occurs during program execution and **interrupts the normal flow of the program**.

For example, dividing a number by zero raises a `ZeroDivisionError`.

### Example

    result = 10 / 0

This raises:

    ZeroDivisionError

### Memory Trick

**Exception = Runtime problem**

### One-Line Answer

An exception is a runtime event that interrupts the normal execution of a program.

## Q3. What is the difference between an Error and an Exception?

### Interview Answer

An error is a broader term for a problem in a program.

An exception is a specific runtime problem that can often be detected and handled using exception handling.

For example:

- Syntax errors prevent Python from correctly parsing the code.
- Runtime exceptions such as `ValueError` or `ZeroDivisionError` can be handled using `try-except`.

### Memory Trick

**Error = General problem**
**Exception = Handleable runtime problem**

### One-Line Answer

An exception is a runtime problem that can generally be handled using Python's exception-handling mechanisms.

## Q4. What is the purpose of the `try` block?

### Interview Answer

The `try` block contains code that may raise an exception.

Python first executes the code inside the `try` block. If an exception occurs, Python looks for a matching `except` block.

### Example

    try:
        number = int(input("Enter number: "))

### Memory Trick

**try = Try risky code**

### One-Line Answer

The `try` block contains code that may generate an exception.

---

## Q5. What is the purpose of the `except` block?

### Interview Answer

The `except` block is used to **handle an exception** raised inside the `try` block.

### Example

    try:
        result = 10 / 0
    except ZeroDivisionError:
        print("Division by zero is not allowed")

### Memory Trick

**except = Handle exception**

### One-Line Answer

The `except` block handles exceptions raised by the `try` block.

---

## Q6. What is the purpose of the `else` block in exception handling?

### Interview Answer

The `else` block executes only when **no exception occurs** in the `try` block.

It is useful for code that should run only after the risky operation succeeds.

### Example

    try:
        number = int("100")
    except ValueError:
        print("Invalid number")
    else:
        print("Conversion successful")

### Memory Trick

**else = No exception**

### One-Line Answer

The `else` block executes when the `try` block completes successfully without raising an exception.

---

## Q7. What is the purpose of the `finally` block?

### Interview Answer

The `finally` block is used for cleanup code that should execute regardless of whether an exception occurs.

It is commonly used for:

- Closing files
- Closing database connections
- Releasing resources

### Example

    try:
        file = open("data.txt", "r")
        data = file.read()
    except FileNotFoundError:
        print("File not found")
    finally:
        print("Cleanup completed")

### Memory Trick

**finally = Always execute**

### One-Line Answer

The `finally` block is used for cleanup code that should execute regardless of whether an exception occurs.

## Q8. What is the difference between `try`, `except`, `else`, and `finally`?

### Interview Answer

Each block has a different purpose:

- `try` → Code that may raise an exception
- `except` → Handles the exception
- `else` → Runs when no exception occurs
- `finally` → Runs regardless of the result

### Memory Trick

**Try → Except → Else → Finally**

**Risk → Handle → Success → Cleanup**

### One-Line Answer

`try` runs risky code, `except` handles errors, `else` runs on success, and `finally` performs cleanup.

### Memory Trick

## Q11. What happens if we use a bare `except`?

### Interview Answer

A bare `except` catches almost any exception.

However, it is generally not recommended because it can hide unexpected problems and make debugging difficult.

### Example

    try:
        risky_operation()
    except:
        print("Something went wrong")

A better approach is to catch specific exceptions whenever possible.

### Memory Trick

**Bare except = Too broad**

### One-Line Answer

A bare `except` catches broadly, but specific exception handling is usually preferred for better debugging and control.

## Q18. What is exception propagation?

### Interview Answer

Exception propagation means that when an exception is not handled inside the current function, it moves up through the **call stack** until a suitable `except` block handles it.

If nobody handles it, the program terminates with an error message.

### Example

    def function_a():
        function_b()

    def function_b():
        result = 10 / 0

    try:
        function_a()
    except ZeroDivisionError:
        print("Exception handled")

The exception propagates from `function_b()` to `function_a()` and finally reaches the `try-except`.

### Memory Trick

**Unhandled → Moves upward**

### One-Line Answer

Exception propagation is the movement of an unhandled exception up the call stack until it is handled.

## Q19. What happens if an exception is not handled?

### Interview Answer

If an exception is not handled, Python terminates the current program execution and displays a **traceback** containing information about where the exception occurred.

### Example

    result = 10 / 0

This produces a traceback showing the location and type of exception.

### Memory Trick

**Unhandled exception → Program stops**

### One-Line Answer

An unhandled exception causes Python to stop the affected program execution and display a traceback.

## Q20. What is a traceback?

### Interview Answer

A traceback is an error report that shows the sequence of function calls that led to an exception.

It helps developers identify:

- Where the error occurred
- Which functions were involved
- The exception type
- The error message

### Memory Trick

**Traceback = Error path**

### One-Line Answer

A traceback shows the execution path that led to an exception and helps identify its cause.

## Q21. Why should we avoid using exceptions for normal program flow?

### Interview Answer

Exceptions are designed for handling unexpected or exceptional situations.

Using them for normal control flow can make code:

- Less readable
- Harder to understand
- Harder to debug
- Potentially less efficient

It is generally better to use normal conditions when a situation is expected.

### Example

Instead of intentionally causing an exception to check something, use an appropriate condition first.

### Memory Trick

**Exception = Exceptional situation**

### One-Line Answer

Exceptions should generally handle exceptional situations rather than replace normal program logic.

## Q22. How should you handle exceptions in a production application?

### Interview Answer

In a production application, I would:

1. Catch expected exceptions specifically.
2. Log useful error information.
3. Avoid exposing sensitive internal details to users.
4. Return meaningful error messages or responses.
5. Clean up resources properly.
6. Monitor recurring failures.
7. Fix the root cause instead of simply hiding the exception.

For example, in a backend API, I would log the actual technical error while returning a safe and meaningful response to the client.

### Memory Trick

**Catch → Log → Respond → Cleanup → Fix**

### One-Line Answer

Production exception handling should catch expected errors, log them properly, provide safe responses, clean up resources, and address the root cause.

Logging can provide:

- Better monitoring and debugging

## Q24. How would you troubleshoot an exception in a real application?

### Interview Answer

I would follow a systematic approach:

1. Reproduce the issue.
2. Read the traceback or error message.
3. Check application logs.
4. Identify the failing component.
5. Inspect the input that caused the failure.
6. Trace the execution flow.
7. Identify the root cause.
8. Apply the fix.
9. Test the fix with valid and invalid inputs.
10. Verify that the fix did not introduce another issue.

### Example

If an API returns a `500 Internal Server Error`, I would first check the logs and traceback, identify the failing function or service, inspect the input and dependencies, fix the root cause, and then retest the endpoint.

### Memory Trick

**Reproduce → Logs → Root Cause → Fix → Test**

### One-Line Answer

I troubleshoot exceptions by reproducing the issue, analyzing logs and traceback, finding the root cause, fixing it, and testing the solution.

## Q25. What is the difference between handling an exception and fixing an exception?

### Interview Answer

Handling an exception means preventing the application from crashing and providing an appropriate response.

Fixing an exception means identifying and correcting the **underlying root cause**.

For example, if a database connection fails, catching the exception prevents the application from crashing, but fixing the database configuration or connection issue solves the actual problem.

### Memory Trick

**Handle = Manage the symptom**
**Fix = Solve the root cause**

### One-Line Answer

Exception handling manages an error safely, while fixing an exception addresses the underlying root cause.
