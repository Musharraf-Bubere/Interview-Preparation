## Q1. What is TensorFlow?

### Interview Answer

TensorFlow is an open-source machine learning framework developed by Google.

It is used to build, train, evaluate, and deploy machine learning and especially deep learning models.

TensorFlow provides APIs for tensors, neural networks, automatic differentiation, optimization, model training, and deployment.

### Example

    Data
      ↓
    TensorFlow
      ↓
    Neural Network
      ↓
    Training
      ↓
    Model
      ↓
    Prediction

### Memory Trick

**TensorFlow = Deep Learning Framework**

### One-Line Answer

TensorFlow is an open-source framework used to develop, train, evaluate, and deploy machine learning and deep learning models.

## Q2. What is a Tensor?

### Interview Answer

A tensor is a multidimensional array used by TensorFlow to represent and process data.

A tensor can have different numbers of dimensions.

For example:

    Scalar → 0D
    Vector → 1D
    Matrix → 2D
    Higher-dimensional array → 3D+

TensorFlow performs many computations using tensors.

### Example

    import tensorflow as tf

    x = tf.constant([
        [1, 2],
        [3, 4]
    ])

### Memory Trick

**Tensor = Multidimensional Data**

### One-Line Answer

A tensor is a multidimensional data structure that TensorFlow uses for numerical computation and model processing.

## Q3. What are the main properties of a TensorFlow tensor?

### Interview Answer

Important tensor properties include:

- Shape
- Data type
- Rank
- Number of elements

For example:

    x = tf.constant([
        [1, 2, 3],
        [4, 5, 6]
    ])

    x.shape
    → (2, 3)

    x.dtype
    → Integer type

The rank represents the number of dimensions.

### Memory Trick

**Shape + Rank + Dtype**

### One-Line Answer

A TensorFlow tensor has properties such as shape, rank, and data type that describe its structure and contents.

## Q4. What is the difference between a TensorFlow tensor and a NumPy array?

### Interview Answer

Both tensors and NumPy arrays represent multidimensional data.

NumPy is primarily designed for numerical computing, while TensorFlow tensors are designed to work within TensorFlow's machine learning and deep learning ecosystem.

TensorFlow tensors can participate in automatic differentiation and can be processed using TensorFlow operations on supported hardware.

### Memory Trick

**NumPy → Numerical Computing**

**TensorFlow Tensor → ML Computation**

### One-Line Answer

NumPy arrays are general numerical data structures, while TensorFlow tensors are designed to integrate with TensorFlow's machine learning computation and training system.

---

## Q5. What is a TensorFlow constant?

### Interview Answer

A constant is a tensor whose value cannot be changed after it has been created.

It can be created using `tf.constant()`.

### Example

    x = tf.constant([10, 20, 30])

A constant is useful when data should remain fixed during computation.

### Memory Trick

**Constant = Cannot Change**

### One-Line Answer

`tf.constant()` creates a tensor with an immutable value.

---

## Q6. What is a TensorFlow Variable?

### Interview Answer

A TensorFlow Variable is a mutable tensor whose value can be updated.

Variables are important in machine learning because model parameters such as weights and biases need to change during training.

### Example

    w = tf.Variable([1.0, 2.0])

    w.assign([3.0, 4.0])

### Memory Trick

**Variable = Trainable / Changeable Value**

### One-Line Answer

A TensorFlow Variable stores mutable values and is commonly used for model parameters such as weights and biases.

---

## Q7. What is the difference between tf.constant and tf.Variable?

### Interview Answer

`tf.constant()` creates an immutable tensor.

`tf.Variable()` creates a mutable tensor whose value can be updated.

During neural network training, weights and biases are generally represented as variables because optimization updates them.

### Example

    constant:
    x = tf.constant([1, 2])

    variable:
    w = tf.Variable([1.0, 2.0])

### Memory Trick

**Constant → Fixed**

**Variable → Update**

### One-Line Answer

A constant is immutable, while a Variable is mutable and can be updated during model training.

## Q8. What is Keras?

### Interview Answer

Keras is a high-level deep learning API that provides a convenient way to build and train neural networks.

In modern TensorFlow, Keras is integrated into TensorFlow and is commonly accessed through `tf.keras`.

It provides components such as layers, models, optimizers, losses, callbacks, and training utilities.

### Example

    import tensorflow as tf

    model = tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dense(1)
    ])

### Memory Trick

**TensorFlow = Framework**

**Keras = High-Level API**

### One-Line Answer

Keras is a high-level deep learning API integrated with TensorFlow that simplifies building and training neural networks.

## Q9. What is a neural network?

### Interview Answer

A neural network is a machine learning model composed of interconnected layers of neurons.

A typical neural network contains:

    Input Layer
         ↓
    Hidden Layer(s)
         ↓
    Output Layer

Each neuron applies a transformation to its inputs, typically involving weights, a bias, and an activation function.

During training, the model learns suitable parameter values from data.

### Memory Trick

**Input → Hidden → Output**

### One-Line Answer

A neural network is a layered machine learning model that learns patterns by adjusting parameters such as weights and biases.

---

## Q10. What is a Dense layer?

### Interview Answer

A Dense layer is a fully connected neural network layer.

Each neuron in the layer receives input from every neuron in the previous layer.

It performs a transformation based on weights and biases and may apply an activation function.

### Example

    tf.keras.layers.Dense(
        64,
        activation="relu"
    )

Here, `64` represents the number of neurons.

### Memory Trick

**Dense = Fully Connected**

### One-Line Answer

A Dense layer is a fully connected neural network layer where each neuron receives input from all neurons in the previous layer.

---

## Q11. What is an activation function?

### Interview Answer

An activation function introduces non-linearity into a neural network.

Without nonlinear activation functions, stacking multiple linear layers would still result in a linear transformation.

Common activation functions include:

- ReLU
- Sigmoid
- Tanh
- Softmax

### Example

    Dense(
        64,
        activation="relu"
    )

### Memory Trick

**Activation = Non-Linearity**

### One-Line Answer

An activation function introduces non-linearity so that neural networks can learn complex patterns.

---

## Q12. What is ReLU?

### Interview Answer

ReLU stands for Rectified Linear Unit.

It is commonly used as an activation function in hidden layers.

Its mathematical definition is:

    ReLU(x) = max(0, x)

Therefore:

    Negative value → 0
    Positive value → Same value

### Example

    [-2, -1, 0, 2, 5]

    ReLU

    [ 0,  0, 0, 2, 5]

### Memory Trick

**ReLU = Negative → 0**

### One-Line Answer

ReLU is an activation function that outputs zero for negative values and the input value for positive values.

---

## Q13. What is the difference between Sigmoid, Softmax, and ReLU?

### Interview Answer

ReLU is commonly used in hidden layers to introduce non-linearity.

Sigmoid maps values approximately between 0 and 1 and is commonly used for binary classification output.

Softmax converts a vector of logits into a probability distribution across multiple classes and is commonly used for multi-class classification.

### Typical Usage

    Hidden Layers
        ↓
      ReLU

    Binary Classification
        ↓
      Sigmoid

    Multi-Class Classification
        ↓
      Softmax

### Memory Trick

**ReLU → Hidden**

**Sigmoid → Binary**

**Softmax → Multi-Class**

### One-Line Answer

ReLU is commonly used in hidden layers, sigmoid is common for binary classification outputs, and softmax is common for multi-class classification outputs.

---

## Q14. What is forward propagation?

### Interview Answer

Forward propagation is the process of passing input data through the neural network to generate a prediction.

The data moves from:

    Input
      ↓
    Layer
      ↓
    Layer
      ↓
    Output

Each layer performs computations using its parameters and activation functions.

### Memory Trick

**Input → Prediction**

### One-Line Answer

Forward propagation is the process of passing input data through the neural network to produce an output or prediction.

---

## Q15. What is backpropagation?

### Interview Answer

Backpropagation is the process used to calculate how the model's loss changes with respect to its trainable parameters.

It applies the chain rule of calculus to compute gradients, which are then used by an optimizer to update the model parameters.

### Flow

    Input
      ↓
    Forward Pass
      ↓
    Prediction
      ↓
    Loss
      ↓
    Backpropagation
      ↓
    Gradients
      ↓
    Optimizer
      ↓
    Updated Weights

### Memory Trick

**Prediction → Loss → Gradients → Update**

### One-Line Answer

Backpropagation computes gradients of the loss with respect to model parameters so an optimizer can update them.

---

## Q16. What is a loss function?

### Interview Answer

A loss function measures how different the model's predictions are from the actual target values.

The training process attempts to minimize this loss.

Different problems use different loss functions.

Examples include:

    Binary Classification
    → Binary Cross-Entropy

    Multi-Class Classification
    → Categorical / Sparse Categorical Cross-Entropy

    Regression
    → Mean Squared Error

### Memory Trick

**Loss = How Wrong Is the Model?**

### One-Line Answer

A loss function measures prediction error and provides the objective that training attempts to minimize.

---

## Q17. What is an optimizer?

### Interview Answer

An optimizer updates the model's trainable parameters using gradients calculated from the loss.

Common optimizers include:

- SGD
- Adam
- RMSprop

Adam is commonly used because it adapts the learning rate for individual parameters and often works well as a general-purpose starting point.

### Example

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy"
    )

### Memory Trick

**Optimizer = Updates Weights**

### One-Line Answer

An optimizer uses gradients to update model parameters and minimize the loss function.

---

## Q18. What is the learning rate?

### Interview Answer

The learning rate controls the size of the parameter updates made by the optimizer during training.

If the learning rate is too high, training may become unstable or fail to converge.

If it is too low, training may become unnecessarily slow.

### Example

    Large Learning Rate
        ↓
    Large Updates

    Small Learning Rate
        ↓
    Small Updates

### Memory Trick

**Learning Rate = Step Size**

### One-Line Answer

The learning rate controls how much model parameters change during each optimization step.

---

## Q19. What are epochs, batch size, and steps?

### Interview Answer

An epoch represents one complete pass through the training dataset.

Batch size represents the number of training samples processed before the model performs an update.

A step generally refers to one batch processed during training.

### Example

If we have:

    10,000 samples
    Batch size = 100

Then approximately:

    10,000 / 100
    = 100 steps per epoch

### Memory Trick

**Epoch = Whole Dataset**

**Batch = Small Group**

**Step = One Batch Update**

### One-Line Answer

An epoch is one pass through the training data, a batch is a group of samples processed together, and a step is typically one batch iteration.

---

## Q20. What is overfitting in deep learning?

### Interview Answer

Overfitting occurs when a model learns the training data too closely, including patterns or noise that do not generalize well to unseen data.

Typical symptoms include:

    Training Performance → Very Good
    Validation Performance → Much Worse

Some techniques used to reduce overfitting include:

- Dropout
- Regularization
- Data augmentation
- Early stopping
- More or better training data
- Appropriate model complexity

### Memory Trick

**Overfitting = Memorize Training Data**

### One-Line Answer

Overfitting occurs when a model performs well on training data but fails to generalize to unseen data.

---

## Q21. What is dropout?

### Interview Answer

Dropout is a regularization technique used in neural networks.

During training, it randomly disables a fraction of activations in a layer for each training step.

This helps reduce reliance on specific neurons and can reduce overfitting.

### Example

    model = tf.keras.Sequential([
        tf.keras.layers.Dense(
            128,
            activation="relu"
        ),
        tf.keras.layers.Dropout(0.5),
        tf.keras.layers.Dense(1)
    ])

### Memory Trick

**Dropout = Randomly Disable Some Activations During Training**

### One-Line Answer

Dropout is a regularization technique that randomly drops a fraction of activations during training to help reduce overfitting.

---

## Q22. What are callbacks in TensorFlow/Keras?

### Interview Answer

Callbacks are objects that allow us to perform actions at specific points during training.

They can be used for:

- Early stopping
- Saving checkpoints
- Learning-rate scheduling
- Logging
- Monitoring training

### Example

    callback = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    )

    model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        callbacks=[callback]
    )

### Memory Trick

**Callback = Training Assistant**

### One-Line Answer

Callbacks execute custom actions during model training, such as early stopping, checkpointing, and learning-rate scheduling.

---

## Q23. What is the difference between Sequential and Functional API in Keras?

### Interview Answer

The Sequential API is useful when a model can be represented as a simple linear stack of layers.

The Functional API is more flexible and allows us to create models with:

- Multiple inputs
- Multiple outputs
- Branching architectures
- Shared layers
- More complex connections

### Example

    Sequential:

    Input
      ↓
    Dense
      ↓
    Dense
      ↓
    Output


    Functional:

          Dense
         ↗
    Input
         ↘
          Dense
            ↓
          Output

### Memory Trick

**Sequential = Simple Stack**

**Functional = Complex Graph**

### One-Line Answer

Sequential is suited to simple layer stacks, while the Functional API supports more complex neural network architectures.

## Q24. How do you build and train a model using TensorFlow/Keras?

### Interview Answer

A typical workflow is:

    1. Prepare Data
           ↓
    2. Build Model
           ↓
    3. Compile Model
           ↓
    4. Train Model
           ↓
    5. Evaluate Model
           ↓
    6. Predict
           ↓
    7. Save / Deploy

### Example

    import tensorflow as tf

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(10,)),
        tf.keras.layers.Dense(
            64,
            activation="relu"
        ),
        tf.keras.layers.Dense(
            1,
            activation="sigmoid"
        )
    ])

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    model.fit(
        X_train,
        y_train,
        epochs=10,
        validation_data=(X_val, y_val)
    )

    model.evaluate(X_test, y_test)

    predictions = model.predict(X_test)

### Memory Trick

**Build → Compile → Fit → Evaluate → Predict**

### One-Line Answer

I typically build the Keras model, compile it with an optimizer and loss, train it with `fit()`, evaluate it, and then use it for prediction.
